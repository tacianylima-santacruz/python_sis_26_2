```python
portao_aberto = True

caixa = 4
caixa_maca = 12
total_macas = caixa * caixa_maca

cavaleiro = 5
maca_por_cavaleiro = total_macas//cavaleiro

portao_aberto = maca_por_cavaleiro > 8

print(f"quantidade total de maças: { total_macas}")
print(f"quantidade de maças por cavaleiro: {maca_por_cavaleiro}")
print(f"Portão aberto {portao_aberto}")


cavaleiro_adicional = cavaleiro + 1
maca_com_adicional = total_macas//cavaleiro_adicional
portao_aberto_adicional= maca_com_adicional > 8

print(f"nova quantidade de maças por cavaleiro: {maca_com_adicional}")
print(f"Portão continua aberto? {portao_aberto_adicional}")
```
