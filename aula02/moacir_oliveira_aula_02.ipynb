```python
character_name = "Donald Trump Bolsonaro Lula da Silva"

character_level = 10
character_health = 75

is_alive = character_health > 0

print(f"Nome: {character_name}")
print(f"Level: {character_level}")
print(f"Vida: {character_health}")
print(f"Vivo: {is_alive}")

character_sheet = {
    "Nome": character_name,
    "Level": character_level,
    "Vida": character_health,
    "Esta_Vivo?": is_alive
}

print("\n--- Dicionário da Ficha de Personagem ---")
for key, value in character_sheet.items():
    print(f"{key.replace('_', ' ').title()}: {value}")
```
