# **Atividade Prática 01: Montando a Ficha de um Personagem**
Crie 4 variáveis, utilizando os 4 tipos de dados:

Texto (str): Ex: Nome do seu personagem.

Número Inteiro (int): Ex: Defina a idade ou nível dele.

Número Decimal (float): Ex: Defina a altura do personagem.

Verdadeiro ou Falso (boolean): Ex: Diga se ele possui uma armadura (True) ou não (False).

```python
nome = str("Shaolin Matador")
print(nome)
```

```python
idade = int(21)
print(idade)
```

```python
altura = float(1.75)
print(altura)
```

```python
espada = bool(True)
print(espada )
```

```python
class Personagem:
  """Esse personagem tem o poder de Cura de alto nível."""

  nome = "Matheus"    #Nome do Personagem
  idade = 21          #Idade do Personagem
  altura = 1.75       #Altura do Personagem
  espada = True       #Espada do Personagem
print(f'Descrição do Personagem: {Personagem.__doc__}')
print(f' Nome: {nome} \n Idade: {idade}\n altura: {altura} \n Possui espada?: {espada}')

```

# **Atividade Prática 02 - Atividade input() e print ()**
Anteriormente, definíamos as variáveis do personagem de forma estática (fixa no código).

Agora o sistema deve permitir o cadastro de dados de um personagem/usuário em variáveis (crie variáveis novas).

E mostre a ficha do personagem/usuário.

```python
nome = str(input('Digite o nome do seu personagem:'))
idade = int(input('Digite o a idade do seu personagem: '))
altura = float(input('Digite a altura do seu personagem: '))
espada = bool(input("Você deseja ultilzar uma espada ? (Sim - Digite algo | Não - Aperte Enter sem digitar nada. )" ))
print('')
print(f'Dados do seu personagem: \n Nome: {nome}\n Idade: {idade} \n Altura: {altura} \n Possui espada: {espada} (True - sim | False - Não)')
```

# **Atividade Prática 03 - Atividade operadores e conversão**
Agora que você já sabe capturar dados com input(), converter tipos, fazer cálculos e comparações com operadores relacionais, é hora inserir essa dinâmica no seu próprio personagem. Crie pelo menos 3 novas variáveis com operações aritiméticas e comparações. Mostre o resultado dos seus cálculos e testes usando print()

```python
nome = str(input('Digite o seu nome: '))
nivel = int(input('Digite o seu nivel: '))
idade = int(input('Digite a sua idade: '))

anos_futuros = idade + 5
dobro_do_nivel = nivel * 2
maior_idade = idade >= 18
print('')
print(f'Nome: {nome}')
print(f'Daqui a 5 anos você terá essa idade:  {anos_futuros}')
print(f'O dobro do seu nível é: {dobro_do_nivel}')
print(f'Maior de idade?: {maior_idade}')
```
