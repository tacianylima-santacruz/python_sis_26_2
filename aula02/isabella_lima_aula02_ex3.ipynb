```python
# AULA 02 | EXERCÍCIO 3

# Introdução
print("========================================")
print("      BATALHA SOBRENATURAL")
print("========================================")

print("\nDean e Sam Winchester encontraram várias criaturas sobrenaturais.")
print("Cada criatura possui uma pontuação de força.")
print("Vamos descobrir quem vence essa batalha!")

fantasmas = int(input("\nDigite a quantidade de fantasmas: "))
demonios = int(input("Digite a quantidade de demônios: "))
vampiros = int(input("Digite a quantidade de vampiros: "))

pontos_fantasma = 10
pontos_demonio = 30
pontos_vampiro = 20

forca_fantasmas = fantasmas * pontos_fantasma
forca_demonios = demonios * pontos_demonio
forca_vampiros = vampiros * pontos_vampiro

forca_criaturas = forca_fantasmas + forca_demonios + forca_vampiros

pontos_dean = int(input("\nDigite a pontuação de Dean: "))
pontos_sam = int(input("Digite a pontuação de Sam: "))

forca_winchester = pontos_dean + pontos_sam

winchester_venceu = forca_winchester > forca_criaturas
criaturas_venceram = forca_criaturas > forca_winchester
empate = forca_winchester == forca_criaturas

print("\n========================================")
print("       RESULTADO DA BATALHA")
print("========================================")

print("\nForça dos fantasmas: " + str(forca_fantasmas))
print("Força dos demônios: " + str(forca_demonios))
print("Força dos vampiros: " + str(forca_vampiros))

print("\nForça total das criaturas: " + str(forca_criaturas))
print("Força total dos Winchester: " + str(forca_winchester))

print("\nOs Winchester venceram: " + str(winchester_venceu))
print("As criaturas venceram: " + str(criaturas_venceram))
print("A batalha terminou empatada: " + str(empate))

print("\n========================================")
```
