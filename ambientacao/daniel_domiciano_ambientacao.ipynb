```python
caixas = 4
macas_cx = 12
total_macas = caixas * macas_cx

cavalheiros = 5
macas_por_cavalheiros = total_macas // cavalheiros

cavalheiro_desafio = 1
macas_por_cavalheiros_desafio = total_macas // (cavalheiros + cavalheiro_desafio)

portao_aberto = macas_por_cavalheiros > 8
portao_aberto_desafio = macas_por_cavalheiros_desafio > 8

print(f"Total de Maçãs: {total_macas}")
print(f"Maçãs por Cavaleiro: {macas_por_cavalheiros}")
print(f"Portão Aberto? {portao_aberto}")
print(f"Maçãs por Cavaleiro no Desafio: {macas_por_cavalheiros_desafio}")
print(f"Portão Desafio Aberto? {portao_aberto_desafio}")
```
