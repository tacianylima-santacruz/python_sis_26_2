**# Váriaveis e operadores matemáticos**

```python
# --- 1. DADOS (Variáveis) ---
qtd_spam = 3
preco_spam = 5.00

qtd_eggs = 2
preco_eggs = 2.50

# --- 2. PROCESSAMENTO (Operadores * e +) ---
total_spam = qtd_spam * preco_spam
total_eggs = qtd_eggs * preco_eggs

custo_total = total_spam + total_eggs

# --- 3. SAÍDA DE DADOS ---
print("=== CAFÉ DA MANHÃ PYTHON ===")
print(f"Spam ({qtd_spam}x): R$ {total_spam:}")
print(f"Eggs ({qtd_eggs}x): R$ {total_eggs:}")
print(f"Custo Total: R$ {custo_total:.2f}")
```

# Prática Operadores Lógicos

```python
# --- 1. DECLARAÇÃO DE VARIÁVEIS (Termos da comunidade) ---
spam = True
eggs = True
bacon = False

# --- 2. OPERADOR LÓGICO 'and' (E) ---
# O prato só é o clássico do Monty Python se tiver Spam E Eggs ao mesmo tempo.
# True and True resulta em True.
prato_tradicional = spam and eggs
print("O prato é o clássico da comunidade (Spam e Eggs)?")
print(prato_tradicional)
print("-" * 30)

# --- 3. OPERADOR LÓGICO 'or' (OU) ---
# O cliente aceita o café da manhã se tiver Eggs OU Bacon.
# True or False resulta em True (basta um ser verdadeiro).
cafe_da_manha_aceito = eggs or bacon
print("O cliente aceitou o café da manhã?")
print(cafe_da_manha_aceito)
print("-" * 30)

# --- 4. OPERADOR LÓGICO 'not' (NÃO / INVERSÃO) ---
# O Cavaleiro Negro está gravemente ferido, mas ele nega a realidade.
# O 'not' inverte o valor booleano: not True vira False.
esta_ferido = True
cavaleiro_negro_admite_dor = not esta_ferido
print("O Cavaleiro Negro admite que está machucado?")
print(cavaleiro_negro_admite_dor)
```

# Gestão em Tecnologia da Informação
# Exemplo:


```python
import pandas as pd
import plotly.express as px

# Lista de consumo em Terabytes para cada um dos 12 meses
consumo = [10, 11, 12, 14, 15, 15, 16, 18, 19, 21, 22, 25]

# Lista de custos de licenças em Reais para cada um dos 12 meses
licencas = [5000, 5000, 5200, 5200, 5200, 5500, 5500, 5500, 6000, 6000, 6000, 6500]

# Estrutura do dicionário usando as variáveis acima
dados_auditoria = {
    "Mes": ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun", "Jul", "Ago", "Set", "Out", "Nov", "Dez"],
    "Consumo_TB": consumo,
    "Custo_Licencas": licencas
}


import pandas as pd
import plotly.express as px

# Criando o DataFrame
df = pd.DataFrame(dados_auditoria)

# Definição de premissas financeiras
PRECO_POR_TB = 150.00

# Processamento de dados e cálculo de custos
df["Custo_Infraestrutura"] = df["Consumo_TB"] * PRECO_POR_TB
df["Custo_Total"] = df["Custo_Infraestrutura"] + df["Custo_Licencas"]
df = df.round(2)

# Exibição da tabela no terminal
print("\n--- RELATÓRIO CONSOLIDADO DE AUDITORIA DE TI ---")
print(df.to_string(index=False))

# Geração do Dashboard Interativo com Plotly
fig = px.line(
    df,
    x="Mes",
    y=["Custo_Infraestrutura", "Custo_Licencas", "Custo_Total"],
    title="Auditoria e Evolução Mensal de Custos de TI",
    labels={"value": "Valor (R$)", "Mes": "Mês", "variable": "Categoria"},
    markers=True,
)

fig.update_layout(
    hovermode="x unified",
    legend_title_text="Legenda de Custos",
    yaxis_tickprefix="R$ ",
)

fig.show()

```
