```python
pacotes = [12, 0, 25, 40, -1, 18, 50]

for pacote in pacotes:
    if pacote == 0:
        continue

    if pacote < 0:
        print("ALERTA: falha de segurança detectada!")
        break

    print(f"Pacote de {pacote} MB processado com sucesso.")
```

```python
def registrar_heroi(nome, classe, reino="Terra-média"):
    return f"Herói: {nome} | Classe: {classe} | Reino: {reino}"


print(registrar_heroi("Aragorn", "Guerreiro", "Gondor"))
print(registrar_heroi("Legolas", "Arqueiro"))


def preparar_equipamentos(nome_heroi, *itens):
    print(f"\nEquipamentos de {nome_heroi}:")

    for item in itens:
        print(f"- {item}")


preparar_equipamentos("Aragorn", "Espada", "Escudo", "Armadura")


def ficha_personagem(nome, **atributos):
    print(f"\nFicha de {nome}:")

    for atributo, valor in atributos.items():
        print(f"{atributo}: {valor}")


ficha_personagem(
    "Legolas",
    forca=80,
    agilidade=95,
    inteligencia=85,
    resistencia=75
)


perigos = ["Orcs", "Nevoeiro", "Aranhas", "Balrog", "Uruk-hai"]

print("\nIniciando jornada para Mordor...")

for perigo in perigos:
    if perigo == "Nevoeiro":
        print("Nevoeiro encontrado. Ignorando o combate e seguindo viagem.")
        continue

    if perigo == "Balrog":
        print("Balrog encontrado! Gandalf avisa: FUJAM!")
        break

    print(f"Perigo encontrado: {perigo}. Preparar para o combate!")
```

```python

servidores_online = 10


def alterar_servidores(quantidade):

    global servidores_online
    servidores_online += quantidade


alterar_servidores(-5)
alterar_servidores(5)

print(f"Servidores ativos: {servidores_online}")


def gerenciar_conexoes():
    limite_conexoes = 10

    def usar_conexao():
        nonlocal limite_conexoes
        limite_conexoes -= 1


    usar_conexao()
    usar_conexao()

    print(f"Limite de conexões restantes: {limite_conexoes}")


gerenciar_conexoes()

def alterar_cota(cota_atual, nova_cota):
    ajustar = lambda cota: nova_cota
    return ajustar(cota_atual)


cota = 100

cota = alterar_cota(cota, 200)

print(f"Cota de armazenamento atual: {cota} GB")
```

```python
def para_binario(numero):
    if numero < 2:
        return str(numero)

    return para_binario(numero // 2) + str(numero % 2)

def para_hexadecimal(numero):
    caracteres = "0123456789ABCDEF"

    if numero < 16:
        return caracteres[numero]

    return para_hexadecimal(numero // 16) + caracteres[numero % 16]

numero = int(input("Digite um número decimal: "))

binario = para_binario(numero)
hexadecimal = para_hexadecimal(numero)

print("Decimal:", numero)
print("Binário:", binario)
print("Hexadecimal:", hexadecimal)
```
