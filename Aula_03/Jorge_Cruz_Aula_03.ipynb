```python
#Atividade 01
"""
Análise do Gene HBB (Hemoglobina Beta)
Diagnóstico de Anemia Falciforme (HbS) via comparação de DNA e tradução de códons.
"""

# =========================================================
# PROCESSAMENTO 1: Padronização, formatação e comparação
# =========================================================

# Sequência de referência normal (Hemoglobina A)
referencia = "ATGGTGCACCTGACTCCTGAG"

# Solicita ao usuário a fita de DNA (21 nucleotídeos)
fita_usuario = input("Digite a sequência de DNA com 21 nucleotídeos: ")

# Padroniza para maiúsculas
fita_usuario = fita_usuario.upper()

# Exibe a fita formatada com f-string
print(f"\n {'=' * 50}")
print(f"Fita de DNA digitada (formatada): {fita_usuario}")
print(f"{'=' * 50}\n ")

# Compara a fita digitada com a referência (True = mutante, False = normal)
mutante = fita_usuario != referencia

# Lista de diagnósticos possíveis
diagnostico = [
    "Sequência NORMAL (Hemoglobina A)",
    "Sequência MUTANTE (Anemia Falciforme - HbS)"
]

# Exibe o resultado acessando o índice da lista com a variável booleana
print(f"Diagnóstico: {diagnostico[mutante]}\n")

# ============================================================
# PROCESSAMENTO 2: Tradução de trincas (códons) via .replace()
# ============================================================

# Fatia a fita em trincas (códons) de 3 em 3 nucleotídeos.
# Isso evita que .replace() confunda um códon com uma sequência
# "escondida" dentro de outro (ex: "CTG" aparece dentro de "CCTGAG").
trincas = [fita_usuario[i:i + 3] for i in range(0, len(fita_usuario), 3)]

# Traduz cada trinca aplicando .replace() encadeado
aminoacidos = []
for trinca in trincas:
  aminoacido = (
      trinca
      .replace("ATG", "met")
      .replace("GTG", "val")
      .replace("CAC", "his")
      .replace("CTG", "leu")
      .replace("ACT", "thr")
      .replace("CCT", "pro")
      .replace("GAG", "glu")
  )
  aminoacidos.append(aminoacido)
# Concatena os aminoácidos traduzidos com "-"
proteina = "".join(aminoacidos)

print(f"Cadeia de aminoacidos traduzida: {proteina}\n")

# =========================================================
# PROCESSAMENTO 3: Localização de padrões (.find())
# =========================================================

aminoacido_busca = input("Digite um aminoácido para buscar na proteína (ex: Val ou His): ")

posicao = proteina.find(aminoacido_busca)

# =========================================================
# RELATÓRIO FINAL
# =========================================================

print(f"\n{'=' * 50}")
print("RELATÓRIO FINAL - ANÁLISE DO GENE HBB")
print(f"{'=' * 50}")
print(f"DNA de referência   : {referencia}")
print(f"DNA digitado        : {fita_usuario}")
print(f"Tamanho da fita     : {len(fita_usuario)} nucleotídeos")
print(f"Proteína traduzida  : {proteina}")
print(f"Diagnóstico         : {diagnostico[mutante]}")
print(f"Posição de '{aminoacido_busca}' na proteína: {posicao}")
print(f"{'=' * 50}")



#Atividade 02
"""
A Sociedade do Anel - Higienização e Análise de Pergaminho
Objetivo: usar variáveis, métodos de string e índices para limpar,
analisar e extrair informações de um texto.
"""

# =========================================================
# ENTRADA DE DADOS
# =========================================================

# Dica: ao testar, digite espaços extras no início e no fim, ex:
#    "   A Sociedade do Anel partiu de Valfenda ao amanhecer   "
texto_bruto = input("Digite o texto do pergaminho (com espaços extras): ")

# =========================================================
# LIMPEZA E HIGIENIZAÇÃO
# =========================================================

# Remove espaços inúteis do início e do fim
texto_limpo = texto_bruto.strip()

print(f"\n{'=' * 60}")
print(f"Texto original : '{texto_bruto}'")
print(f"Texto limpo    : '{texto_limpo}'")
print(f"{'=' * 60}\n")

# =========================================================
# RELATÓRIO
# =========================================================

# Número total de caracteres do texto limpo
total_caracteres = len(texto_limpo)

# Divide o texto em uma lista de palavras (usa espaço como separador)
lista_palavras = texto_limpo.split()

# Quantidade total de palavras
total_palavras = len(lista_palavras)

# Primeira e última palavra (usando índices da lista)
primeira_palavra = lista_palavras[0]
ultima_palavra = lista_palavras[-1]

print("--- RELATÓRIO DO PERGAMINHO ---")
print(f"Total de caracteres : {total_caracteres}")
print(f"Total de palavras   : {total_palavras}")
print(f"Primeira palavra    : {primeira_palavra}")
print(f"Última palavra      : {ultima_palavra}")

# =========================================================
# DETECTOR DE AMEAÇAS
# =========================================================

palavra_procurada = "Anel"

# Verifica se a palavra existe no texto (case-insensitive)
existe = palavra_procurada.lower() in texto_limpo.lower()

# Conta quantas vezes a palavra aparece (case-insensitive)
quantidade_ocorrencias = texto_limpo.lower().count(palavra_procurada.lower())

print(f"\n--- DETECTOR DE AMEAÇAS ---")
print(f"A palavra '{palavra_procurada}' foi encontrada no texto? {existe}")
print(f"Quantidade de vezes que '{palavra_procurada}' aparece: {quantidade_ocorrencias}")
"""
A Jornada do Personagem - Operadores e Condicionais
Objetivo: usar operadores aritméticos e o comando if/else para dar
consequências às ações do personagem na história.
"""

print("=== A JORNADA DO GUERREIRO===")

# =========================================================
# CENA 1: O personagem é atacado por um monstro
# =========================================================

hp_atual = 40       # vida atual do personagem
dano_monstro = 45   # dano recebido no ataque

# Cálculo: vida restante após o ataque
hp_restante = hp_atual - dano_monstro

print(f"O monstro ataca! Vida antes: {hp_atual} | Dano recebido: {dano_monstro}")
print(f"Vida restante: {hp_restante}")

# Decisão 1: o personagem morreu?

if hp_restante <= 0:
    print("Guerreiro foi de arrasta. O monstro matou o guerreiro!\n")

else:
    print(f"Guerreiro sobreviveu com {hp_restante} de vida" )


# =========================================================
# CENA 2: O personagem contra-ataca o monstro
# =========================================================

forca_personagem = 25
forca_espada = 15
defesa_monstro = 30

# Cálculo: ataque total do personagem

ataque_total = forca_personagem + forca_espada

print(f"Força: {forca_personagem} + Bonus da espada: {forca_espada} = Ataque total: {ataque_total}")
print(f"Defesa do monstro: {defesa_monstro}")

# Decisão 2: o ataque foi crítico?
if ataque_total > defesa_monstro:
    print(f"Guerreiro matou o monstro:")

else:
    print(f"O monstro sobreviveu>")

# =============================================================
# CENA 3: O personagem compra uma poção e tenta pagar a taverna
# =============================================================

moedas_ouro = 50
custo_pocao = 35
custo_taverna = 20

# Cálculo: moedas restantes após comprar a poção
moedas_restantes = moedas_ouro - custo_pocao

print(f"Moedas de ouro: {moedas_ouro} | - custo da porção: {custo_pocao}")
print(f"Moedas restantes: {moedas_restantes }")

# Decisão 3: dá pra pagar a taverna?

if moedas_restantes < custo_taverna:
    print("Ferrou, vai lavar a louça")
else:
    print(f"Por pouco hein.")

print("====FIM DA JORNADA===")
```
