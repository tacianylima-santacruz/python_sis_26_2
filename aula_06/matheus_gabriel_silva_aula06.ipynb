```python
# Atividade Prática 01 - Aula 06
# Animais Favoritos (Listas)

animais = ["cachorro", "gato", "coelho", "hamster", "tucano", "peixe"]
favoritos = ["cachorro", "gato", "tucano"]

# 1. Mostrar os animais favoritos
print(f"Animais favoritos: {favoritos}")

# 2. Quantidade de animais e quantidade de favoritos
print(f"\nTotal de animais na lista: {len(animais)}")
print(f"Total de animais favoritos: {len(favoritos)}")

# 3. Primeiro e último animal favorito
print(f"\nPrimeiro favorito: {favoritos[0]}")
print(f"Último favorito: {favoritos[-1]}")

# 4. Remover um animal da lista
animais.remove("peixe")
print(f"\nLista após remover 'peixe': {animais}")

# 5. Adicionar um ou mais animais na lista
animais.append("papagaio")
animais.append("iguana")
print(f"\nLista após adicionar novos animais: {animais}")

# 6. Verificar quantos animais temos agora
print(f"\nTotal de animais agora: {len(animais)}")

# 7. Trocar a ordem dos favoritos e imprimir
favoritos.reverse()
print(f"\nFavoritos em nova ordem: {favoritos}")
```

```python
# Atividade Prática 02 - Aula 06
# O Cardápio Fixo do Axolote (Tuplas)

# 1. Tupla com três elementos de tipos variados
refeicao = ("08:00", "minhocas", 50)

# 2. Tentando alterar o alimento principal:
# refeicao[1] = "pizza de calabresa"
# ⬆️ Essa linha dá erro: TypeError: 'tuple' object does not support item assignment
# Tuplas são imutáveis, por isso a linha acima foi removida do código.

# Desempacotamento da tupla em três variáveis, em 1 linha:
horario, alimento, vitaminas = refeicao

# 3. Frase amigável usando as variáveis desempacotadas
print(f"Às {horario}, o axolote come {alimento} com {vitaminas} vitaminas.")
```

```python
# Atividade Prática 03 - Aula 06
# Agenda Telefônica (Dicionários)

# 1. Dicionário de contatos
contatos = {
    "Ana": "9999-1111",
    "Bruno": "9999-2222",
    "Carla": "9999-3333"
}
print(f"1. Contatos iniciais: {contatos}")

# 2. Adicionar um novo contato sem alterar a declaração original acima
contatos["Diego"] = "9999-4444"
print(f"2. Contatos após adicionar 'Diego': {contatos}")

# 3. Verificar se "Chai" existe; se não, adicionar com número inventado
if "Chai" not in contatos:
    contatos["Chai"] = "9999-5555"
print(f"3. Contatos após verificar/adicionar 'Chai': {contatos}")

# 4. Excluir um contato usando um método que não gera erro se ele não existir
contatos.pop("Bruno", None)
print(f"4. Contatos após remover 'Bruno': {contatos}")

# 5. Imprimir todos os nomes presentes no dicionário
print(f"5. Nomes na agenda: {list(contatos.keys())}")
```

```python
# Atividade Prática 04 - Aula 06
# A Sacola do Clube do Livro (Conjuntos)

# 1. Lista inicial com títulos repetidos
livros_da_sacola = ["Duna", "O Hobbit", "Duna", "Harry Potter", "O Hobbit", "1984"]
print(f"Lista original (com repetidos): {livros_da_sacola}")

# 2. Convertendo a lista em um conjunto para eliminar as duplicatas
livros_sem_repetir = set(livros_da_sacola)

# 3. Imprimindo o conjunto (sem repetições e sem ordem garantida)
print(f"Conjunto final (sem duplicatas): {livros_sem_repetir}")
```

```python
# Atividade Prática 05 - Aula 06
# Banco de Dados de Biblioteca (Combinação de Tipos)

# Dicionário principal: chave = nome do livro (str) | valor = dicionário com os detalhes
biblioteca = {
    "Duna": {
        "autor": "Frank Herbert",
        "ano": 1965,
        "generos": ["Ficção Científica", "Aventura"],
        "disponivel": True
    },
    "O Hobbit": {
        "autor": "J.R.R. Tolkien",
        "ano": 1937,
        "generos": ["Fantasia"],
        "disponivel": False
    }
}

# Imprimindo um resultado para cada tipo de dado usado na estrutura
livro_escolhido = "Duna"
detalhes = biblioteca[livro_escolhido]

print(f"Livro (str - chave do dicionário principal): {livro_escolhido}")
print(f"Autor (str): {detalhes['autor']}")
print(f"Ano de publicação (int): {detalhes['ano']}")
print(f"Gêneros (list): {detalhes['generos']}")
print(f"Disponível para empréstimo (bool): {detalhes['disponivel']}")
print(f"\nRegistro completo do livro (dict aninhado): {detalhes}")
```
