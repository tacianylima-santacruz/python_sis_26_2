```python
# Atividade Prática 05 - Aula 06
# Banco de Dados de Biblioteca (Combinação de Tipos)

biblioteca = {
    "Jane Eyre": {
        "autores": [
            {"nome": "Charlotte Brontë", "ano_nascimento": 1816, "nacionalidade": "Britânica"}
        ],
        "num_paginas": 659,
        "disponivel": True,
        "emprestado_para": None,
        "ano_publicacao": 1847,
        "generos": ["Ficção Vitoriana", "Romance de Formação (Bildungsroman)", "Histórico"],
        "avaliacao": 4.6
    },
    "Python Illustrated": {
        "autores": [
            {"nome": "Zia van Putten", "ano_nascimento": 2023, "nacionalidade": "Holandesa"},
            {"nome": "Maaike van Putten", "ano_nascimento": 1991, "nacionalidade": "Holandesa"},
            {"nome": "Imke van Putten", "ano_nascimento": 1997, "nacionalidade": "Holandesa"}
        ],
        "num_paginas": 432,
        "disponivel": False,
        "emprestado_para": "Jacob",
        "ano_publicacao": 2026,
        "generos": ["Informativo", "Educação"],
        "avaliacao": 5
    },
    "Se os gatos desaparecessem do mundo": {
        "autores": [
            {"nome": "Genki Kawamura", "ano_nascimento": None, "nacionalidade": None}
        ],
        "num_paginas": 144,
        "disponivel": False,
        "emprestado_para": "Chai",
        "ano_publicacao": 2012,
        "generos": ["Ficção Literária", "Realismo Mágico", "Ficção Filosófica"],
        "avaliacao": 4.2
    }
}

# 1. Imprimir todos os títulos de livros da biblioteca
print("Títulos disponíveis na biblioteca:")
print(list(biblioteca.keys()))
print()

# 2. Verificar se "Romance" é um gênero de "Python Illustrated"
tem_romance = "Romance" in biblioteca["Python Illustrated"]["generos"]
print(f'"Romance" é um gênero de "Python Illustrated"? {tem_romance}')
print()

# 3. Imprimir a nacionalidade do autor de Jane Eyre
nacionalidade_jane_eyre = biblioteca["Jane Eyre"]["autores"][0]["nacionalidade"]
print(f"Nacionalidade do autor de Jane Eyre: {nacionalidade_jane_eyre}")
print()

# 4. Chai devolveu o livro - atualizar "Se os gatos desaparecessem do mundo"
biblioteca["Se os gatos desaparecessem do mundo"]["disponivel"] = True
biblioteca["Se os gatos desaparecessem do mundo"]["emprestado_para"] = None
print("Livro 'Se os gatos desaparecessem do mundo' atualizado:")
print(biblioteca["Se os gatos desaparecessem do mundo"])
print()

# 5. Verificar se "Python Illustrated" está disponível e, se não, quem pegou emprestado
livro = biblioteca["Python Illustrated"]
if livro["disponivel"]:
    print('"Python Illustrated" está disponível.')
else:
    print(f'"Python Illustrated" não está disponível. Emprestado para: {livro["emprestado_para"]}')
```
