Atividade 01 - Aula 06

```python


animais = ["cachorro", "gato", "coelho", "axolote", "tartaruga", "coruja"]
animais_favoritos = ["axolote", "cachorro", "coruja"]

print("--- Animais favoritos ---")
print(animais_favoritos)

print("\nQuantidade de animais:", len(animais))
print("Quantidade de animais favoritos:", len(animais_favoritos))

print("\nPrimeiro animal favorito:", animais_favoritos[0])
print("Último animal favorito:", animais_favoritos[-1])

animais.remove("gato")
print("\nLista após remover o gato:", animais)

animais.append("golfinho")
animais.extend(["lobo", "pinguim"])
print("Lista após adicionar animais:", animais)

print("\nQuantidade atual de animais:", len(animais))

animais_favoritos.reverse()
print("Favoritos em nova ordem:", animais_favoritos)
```

Atividade 03 - Aula 06

```python


contatos = {
    "Frodo": "99999-1111",
    "Sam": "98888-2222",
    "Gandalf": "97777-3333"
}

print("--- Contatos iniciais ---")
print(contatos)


contatos["Aragorn"] = "96666-4444"
print("\nApós adicionar Aragorn:")
print(contatos)


if "Chai" not in contatos:
    contatos["Chai"] = "95555-5555"
    print("\nChai não existia e foi adicionada.")
else:
    print("\nChai já existe nos contatos.")

print(contatos)

contato_removido = contatos.pop("Sam", None)
print("\nContato removido:", contato_removido)
print("Contatos após a remoção:", contatos)

print("\nNomes presentes na agenda:")
for nome in contatos.keys():
    print(nome)
```
