Animais Favoritos

Objetivo: Crie uma lista com vários animais para realizar filtros:
Filtros:
1. Mostre em tela seus animais favoritos;
2. Imprima a quantidade de animais da lista e a quantidade de animais favoritos;
3. Mostre qual é o primeiro animail favorito e o último;
4. Remova um dos animais da lista e imprima o resultado;
5. Adicione um ou mais animais na lista;
6. Verifique quantos animais temos agora;
8. Troque a ordem dos animais favoritos de acordo com sua preferência e imprima.

```python
animais = ["Cachorro", "Gato", "Papagaio", "Tartaruga", "Coelho", "Hamster"]
favoritos = ["Cachorro", "Gato", "Coelho"]

print(favoritos)

print(len(animais))
print(len(favoritos))

print(favoritos[0])
print(favoritos[-1])

animais.remove("Hamster")
print(animais)

animais.append("Calopsita")
print(animais)

print(len(animais))

favoritos.reverse()
print(favoritos)
```

O Cardápio Fixo do Axolote (Tuplas)
* Objetivo: Como sabemos, as tuplas são coleções imutáveis, sendo o formato ideal para armazenar dados que não devem ser modificados acidentalmente. O axolote de estimação tem uma rotina de alimentação muito restrita!
Atividades:
1. Crie uma tupla contendo três elementos de tipos variados: o horário da refeição (texto), o alimento principal (texto) e a quantidade de vitaminas (número).
2. Tente alterar o alimento principal da tupla para "pizza de calabresa".
Delete a linha que causou o erro. Agora, utilize a técnica de desempacotamento de tuplas para extrair esses três valores e salvá-los em três variáveis independentes usando apenas 1 linha de código.
3. Imprima uma frase amigável usando as variáveis desempacotadas. Por exemplo: "Às 08:00, o axolote come minhocas com 50 vitaminas."



```python
refeicao_axolote = ("08:00", "minhocas", 50)

horario, alimento, vitaminas = refeicao_axolote

print(f"Às {horario}, o axolote come {alimento} com {vitaminas} vitaminas.")
```

A Sacola do Clube do Livro (Conjuntos)

Objetivo: Um conjunto (set) é uma estrutura que, por natureza, não permite elementos duplicados, sendo útil para remover duplicatas de coleções de dados automaticamente. Fui à feira de troca de livros com meus amigos e jogamos todos os livros que arrecadamos em uma grande sacola. Acabamos pegando várias cópias do mesmo livro!

Atividades:

1. Crie uma lista inicial com cerca de 6 ou 7 títulos de livros, mas coloque de propósito uns 3 títulos repetidos (por exemplo: "Duna", "O Hobbit", "Duna", "Harry Potter", "O Hobbit").
2. Converta essa sua lista em um Conjunto para forçar o sistema a eliminar todas as cópias duplicadas em um passe de mágica.
3. Imprima o conjunto na tela. Note que o resultado final perdeu as repetições e que os itens provavelmente mudaram de lugar, já que os conjuntos não mantêm a ordem de inserção dos elementos!



```python
livros = ["Duna", "O Hobbit", "Duna", "Harry Potter", "O Hobbit", "Fundação", "1984"]
print(livros)

conjunto_livros = set(livros)

print(conjunto_livros)
```
