### Atividade 01 Aula 06
**Estruturas de dados recomendadas:**
1. **Lista de Compras:** `list` (Lista - preserva a ordem e permite duplicados/alterações).
2. **Registro de Alunos e Notas:** `dict` (Dicionário - ideal para mapear chave-valor, ex: nome do aluno como chave e nota como valor).
3. **Datas de eventos históricos:** `dict` ou `list` de `tuples` (para associar o evento ao ano de forma imutável).
4. **Identificar alunos em comum:** `set` (Conjunto - permite usar operações de interseção `&` de forma extremamente rápida).

### Atividade Prática 01 Aula 06 - Animais Favoritos

```python
# 1. Criação da lista e favoritos
animais = ["Gato", "Cachorro", "Lontra", "Axolote", "Leão", "Panda"]
favoritos = ["Axolote", "Lontra"]
print("Animais favoritos:", favoritos)

# 2. Quantidade de animais
print(f"Total de animais: {len(animais)} | Favoritos: {len(favoritos)}")

# 3. Primeiro e último favorito
print(f"Primeiro favorito: {favoritos[0]} | Último favorito: {favoritos[-1]}")

# 4. Remover um animal
animais.remove("Leão")
print("Lista após remover Leão:", animais)

# 5. Adicionar animais
animais.extend(["Tigre", "Coelho"])

# 6. Nova quantidade
print("Quantidade atual de animais:", len(animais))

# 7. Trocar ordem dos favoritos
favoritos[0], favoritos[1] = favoritos[1], favoritos[0]
print("Favoritos em nova ordem:", favoritos)
```

### Atividade Prática 02 Aula 06 - O Cardápio Fixo do Axolote (Tuplas)

```python
# 1. Criar tupla
refeicao_axolote = ("12:00", "Minhocas", 50)

# 2. Desempacotamento de tuplas em 1 linha
horario, alimento, vitaminas = refeicao_axolote

# 3. Impressão amigável
print(f"Às {horario}, o axolote come {alimento} com {vitaminas} vitaminas.")
```

### Atividade Prática 03 Aula 06 - Agenda Telefônica (Dicionários)

```python
# 1. Criar dicionário de contatos
contatos = {"Ana": "99999-1111", "Bruno": "99999-2222"}
print("Contatos iniciais:", contatos)

# 2. Adicionar novo contato sem alterar diretamente a declaração original
contatos["Carlos"] = "99999-3333"
print("Após adicionar Carlos:", contatos)

# 3. Verificar se "Chai" existe, se não, adicionar
if "Chai" not in contatos:
    contatos["Chai"] = "98888-7777"
print("Após verificar/adicionar Chai:", contatos)

# 4. Excluir contato de forma segura (sem gerar erro se não existir)
contatos.pop("Bruno", None)
print("Após tentar excluir Bruno:", contatos)

# 5. Imprimir todos os nomes (chaves)
print("Nomes na agenda:", list(contatos.keys()))
```

### Atividade Prática 04 Aula 06 - A Sacola do Clube do Livro (Conjuntos)

```python
# 1. Lista inicial com duplicados
livros_sacola = ["Duna", "O Hobbit", "Duna", "Harry Potter", "O Hobbit", "1984", "Dom Casmurro"]
print("Lista original:", livros_sacola)

# 2. Conversão para Conjunto (Set)
livros_unicos = set(livros_sacola)

# 3. Exibir o resultado
print("Conjunto sem duplicatas:", livros_unicos)
```

### Atividade Prática 05 Aula 06 - Banco de Dados de Biblioteca

```python
# Estrutura combinada de dados da biblioteca
biblioteca = {
    "Jane Eyre": {
        "autores": [{"nome": "Charlotte Brontë", "nascimento": 1816, "nacionalidade": "Britânica"}],
        "paginas": 659,
        "publicacao": 1847,
        "generos": ["Ficção Vitoriana", "Romance de Formação", "Histórico"],
        "avaliacao": 4.6,
        "disponivel": True,
        "emprestado_para": None
    },
    "Python Illustrated": {
        "autores": [
            {"nome": "Zia van Putten", "nascimento": 2023, "nacionalidade": "Holandesa"},
            {"nome": "Maaike van Putten", "nascimento": 1991, "nacionalidade": "Holandesa"},
            {"nome": "Imke van Putten", "nascimento": 1997, "nacionalidade": "Holandesa"}
        ],
        "paginas": 432,
        "publicacao": 2026,
        "generos": ["Informativo", "Educação"],
        "avaliacao": 5.0,
        "disponivel": False,
        "emprestado_para": "Jacob"
    },
    "Se os gatos desaparecessem do mundo": {
        "autores": [{"nome": "Genki Kawamura", "nascimento": None, "nacionalidade": "Japonesa"}],
        "paginas": 144,
        "publicacao": 2012,
        "generos": ["Ficção Literária", "Realismo Mágico", "Ficção Filosófica"],
        "avaliacao": 4.2,
        "disponivel": False,
        "emprestado_para": "Chai"
    }
}

# Tarefa 1: Imprimir todos os títulos
print("Títulos na biblioteca:", list(biblioteca.keys()))

# Tarefa 2: Verificar se 'Romance' é gênero de 'Python Illustrated'
tem_romance = "Romance" in biblioteca["Python Illustrated"]["generos"]
print("¿'Romance' é gênero de 'Python Illustrated'?: ", tem_romance)

# Tarefa 3: Nacionalidade do autor de Jane Eyre
nac_autor = biblioteca["Jane Eyre"]["autores"][0]["nacionalidade"]
print("Nacionalidade do autor de Jane Eyre:", nac_autor)

# Tarefa 4: Chai devolveu o livro
biblioteca["Se os gatos desaparecessem do mundo"]["disponivel"] = True
biblioteca["Se os gatos desaparecessem do mundo"]["emprestado_para"] = None
print("Status pós devolução de 'Se os gatos desaparecessem do mundo':", biblioteca["Se os gatos desaparecessem do mundo"])

# Tarefa 5: Verificar disponibilidade de Python Illustrated e quem pegou emprestado
livro_py = biblioteca["Python Illustrated"]
if not livro_py["disponivel"]:
    print(f"'Python Illustrated' não está disponível. Foi emprestado para: {livro_py['emprestado_para']}")
```
