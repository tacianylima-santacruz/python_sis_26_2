````python
# Criar uma lista com vários animais
animais = ["cão", "gato", "papagaio", "tartaruga", "coelho", "hamster"]

# 1. Mostre em ecrã os seus animais favoritos
animais_favoritos = ["cão", "gato", "coelho"]
print("Animais favoritos:", animais_favoritos)

# 2. Imprima a quantidade de animais da lista e a quantidade de animais favoritos
print(f"Total de animais na lista: {len(animais)}")
print(f"Total de animais favoritos: {len(animais_favoritos)}")

# 3. Mostre qual é o primeiro animal favorito e o último
print(f"Primeiro animal favorito: {animais_favoritos[0]}")
print(f"Último animal favorito: {animais_favoritos[-1]}")

# 4. Remova um dos animais da lista e imprima o resultado
animais.remove("hamster")
print("Lista após remoção:", animais)

# 5. Adicione um ou mais animais na lista
animais.append("cavalo")
animais.extend(["peixe", "tigre"])
print("Lista após adições:", animais)

# 6. Verifique quantos animais temos agora
print(f"Agora temos {len(animais)} animais na lista.")

# 7. Troque a ordem dos animais favoritos de acordo com a sua preferência e imprima
animais_favoritos.reverse()
print("Animais favoritos com ordem invertida:", animais_favoritos)
````

````python
# 1. Criar um dicionário de contactos
agenda = {
    "Ana": "912345678",
    "Bruno": "923456789",
    "Carla": "934567890"
}

# 2. Adicionar um novo contacto sem alterar diretamente a declaração original
agenda.update({"Daniel": "945678901"})

# 3. Verificar se o contacto "Chail" existe. Se não, adicione-o com um número inventado
if "Chail" not in agenda:
    agenda["Chail"] = "960000000"

# 4. Excluir um dos contactos usando um método que não cause erro se não existir (pop)
agenda.pop("Bruno", None)

# 5. Imprimir todos os nomes presentes no dicionário
print("Nomes na agenda:", list(agenda.keys()))
````
