```python
# Etapa 1: Estrutura em Memória

configuracoes_cluster = {
    "nome_cluster": "cluster-producao",
    "versao_so": "Ubuntu Server 24.04 LTS",
    "servidores": [
        {
            "Servidor": "srv-app-01",
            "IP": "192.168.1.10",
            "Portas": [22, 80, 443]
        },
        {
            "Servidor": "srv-db-01",
            "IP": "192.168.1.20",
            "Portas": [22, 3306, 5432]
        },
        {
            "Servidor": "srv-web-01",
            "IP": "192.168.1.30",
            "Portas": [22, 80, 443]
        }
    ]
}

print("ETAPA 1 - ESTRUTURA EM MEMÓRIA")
print(configuracoes_cluster)
```

```python
# Etapa 2: Exportação JSON

import json

with open("cluster_config.json", "w", encoding="utf-8") as arquivo_json:
    json.dump(configuracoes_cluster, arquivo_json, indent=4, ensure_ascii=False)

print("Arquivo cluster_config.json criado com sucesso!")

# Conferindo o conteúdo do arquivo
with open("cluster_config.json", "r", encoding="utf-8") as arquivo_json:
    print(arquivo_json.read())
```

```python
# Etapa 3: Exportação CSV

import csv

with open("servidores_cluster.csv", "w", newline="", encoding="utf-8") as arquivo_csv:
    escritor = csv.writer(arquivo_csv, delimiter=";")

    escritor.writerow(["Servidor", "IP", "Portas"])

    for servidor in configuracoes_cluster["servidores"]:
        portas = ", ".join(map(str, servidor["Portas"]))
        escritor.writerow([
            servidor["Servidor"],
            servidor["IP"],
            portas
        ])

print("Arquivo servidores_cluster.csv criado com sucesso!")

# Conferindo o conteúdo do CSV
with open("servidores_cluster.csv", "r", encoding="utf-8") as arquivo_csv:
    leitor = csv.reader(arquivo_csv, delimiter=";")
    for linha in leitor:
        print(linha)

```
