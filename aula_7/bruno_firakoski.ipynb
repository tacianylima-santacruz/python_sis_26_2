Parte 01 - Caça aos Bugs - Soneca de gato com bugs
Nós escrevemos um programa para calcular o tempo total que o gato dorme em um dia. Mas algo parece estranho... O total que a nossa função retorna não corresponde ao tempo real que o gato passou tirando sonecas:

Mesmo que você identifique o erro neste código logo de cara, pratique usar comandos print, um ponto de interrupção (breakpoint) ou o registro de rastreamento (traceback log) para descobrir o que está errado e com que tipo de bug estamos lidando.
Além disso, corrija o problema para que a função retorne o tempo total de soneca correto.

```python
def nap_time(hours_list):
    total = 0
    counter = 0
    while counter < len(hours_list):
        print("counter:", counter, "| soneca:", hours_list[counter], "| total:", total)
        total += hours_list[counter]
        counter += 1
    return total
cat_naps = [2, 3, 1.5, 4]
print("Total nap time:", nap_time(cat_naps))

## Tipo de bug: erro de lógica. O programa não quebra, mas devolve o resultado errado (8.5 em vez de 10.5).
## Com o print dentro do while, dá para ver que o counter começa em 1 e a primeira soneca (índice 0, valor 2) nunca é somada.
## Correção: começar o counter em 0, porque o primeiro item de uma lista tem índice 0.
```

Atividade Prática 02 Aula 07 -
Parte 02 - Quanto o seu gato mia?
Imagine que você escreveu este programa para calcular com que frequência o seu gato mia, em média, por hora acordado. Você manteve um registro de quanto ele miou e de quanto ele dormiu a cada dia. Você arredondou os tempos de soneca; afinal, você é um desenvolvedor de software, não um veterinário. Tudo no seu código parece estar perfeitamente bem. Mas adivinha? O programa trava mesmo assim. Hora de depurar (debug)! Use o traceback, declarações print e/ou pontos de interrupção (breakpoints) para descobrir o que está causando o travamento e com que tipo de bug você está lidando. Pense em uma maneira inteligente de modificar o código para que ele não trave mais:

```python
def average_meows(logs):
    total_meows_per_hour = 0
    valid_days = 0

    for log in logs:
        nap_hours = 0

        for nap in log["naps"]:
            nap_hours += nap

        awake_hours = 24 - nap_hours
        print(log["day"], "| horas dormindo:", nap_hours, "| horas acordado:", awake_hours)

        if awake_hours == 0:
            continue

        total_meows_per_hour += log["meows"] / awake_hours
        valid_days += 1

    return total_meows_per_hour / valid_days

logs = [
    {"day": "Monday", "naps": [1, 2, 5], "meows": 90},
    {"day": "Tuesday", "naps": [4, 3, 2], "meows": 75},
    {"day": "Wednesday", "naps": [8, 6, 9, 1], "meows": 9},
    {"day": "Thursday", "naps": [1, 1, 5, 7], "meows": 38},
    {"day": "Friday", "naps": [3, 3, 2, 8], "meows": 94},
    {"day": "Saturday", "naps": [4, 8, 3], "meows": 64},
    {"day": "Sunday", "naps": [3, 3, 2, 8], "meows": 37}
]

print("Average meows per awake hour:", average_meows(logs))

## Tipo de bug: erro de execução (ZeroDivisionError), na linha "total_meows_per_hour += log["meows"] / awake_hours".
## Com o print, dá para ver que na quarta-feira o gato dorme 24 horas (8 + 6 + 9 + 1), então awake_hours vale 0 e a divisão quebra.
## Correção: pular os dias em que o gato não ficou acordado (if awake_hours == 0: continue).
## Também contei só os dias válidos (valid_days) na média final, em vez de usar len(logs).
```

Atividade Prática 02 Aula 07 -
Parte 03 - Bugs na Trilha
Você está escrevendo um programa que calcula a distância total de uma trilha. Os dados são flexíveis, e cada segmento da trilha é passado como uma tupla que consiste na distância e na unidade. A função deve imprimir o total de todos esses segmentos. O programa roda sem travar (não quebra), mas algo sobre esse total parece estar errado... Use técnicas de depuração (debugging) para descobrir o que há de errado com este código e conserte o bug:

```python
def total_distance(*segments):
    total = 0
    for length, unit in segments:
        unit = unit.lower().strip()
        print("segmento:", length, repr(unit), "| total até agora:", total)
        if unit == "m":
            total += length
        elif unit == "km":
            total += length * 1000

    print("Total distance:", total, "meters")

total_distance(
    (2.5, "km"),
    (400, "M"),
    (1.0, "km"),
    (800, "m"),
    (1.3, "KM"),
    (900, "m "),
    (1.1, "km")
)

## Tipo de bug: erro de lógica. O programa não quebra, mas devolve o total errado (0.0059 em vez de 8000 metros).
## Com o print, dá para ver três problemas:
## 1) unit.lower() == "M" nunca é verdadeiro, porque o texto já está em minúsculas e é comparado com "M" maiúsculo.
## 2) Para converter km em metros é preciso multiplicar por 1000, e o código dividia.
## 3) A unidade "m " tem um espaço no fim e não batia com "m".
## Correção: usar unit.lower().strip(), comparar com "m" minúsculo e trocar length / 1000 por length * 1000.
```

Atividade Prática 03 Aula 07 -
Caça aos Bugs e Tratamento de Erros - Ingressos do museu
Você está ajudando um museu local a configurar uma máquina de ingressos. Você escreveu um programa em Python para ela: o visitante digita a sua idade, e a máquina imprime o tipo de ingresso que ele recebe. A lógica é simples:

Crianças menores de 12 anos têm acesso gratuito;
Adolescentes menores de 18 anos recebem um desconto para jovens;
Adultos menores de 65 anos pagam o valor integral;
Pessoas com 65 anos ou mais recebem um desconto para idosos;
Porém você deve tratar o código para que independente doque o visitante digite o programa não dê erro.

Caso seja uma idade válida o programa deve dar as boas vindas. Caso não deve pedir redigitação.

E independente do usuário digita certo ou errado o sistema deve limpar a tela para aguardar pelo próximo visitante.

```python
import os
import time

def ticket_type():
    while True:
        try:
            age = int(input("Enter your age: "))
            if age < 0:
                raise ValueError
        except ValueError:
            print("Invalid age. Please type a valid number.")
        else:
            print("Welcome to the museum!")
            if age < 12:
                print("You get in for free. Lucky you!")
            elif age < 18:
                print("You get a youth discount!")
            elif age < 65:
                print("That's a full-price ticket for you.")
            else:
                print("You get a senior discount!")
        finally:
            time.sleep(2)
            os.system("cls" if os.name == "nt" else "clear")

ticket_type()

## Bug 1: input() devolve texto (str), e o código comparava texto com número (age < 12), o que dá TypeError.
## Correção: converter com int(input(...)).
## Bug 2: se o visitante digitar letras, vazio ou decimal, o int() dá ValueError.
## Correção: try/except ValueError, que mostra a mensagem de erro e volta a pedir a idade.
## Idade negativa também é inválida: raise ValueError dentro do try.
## else: só roda se a idade for válida, então dá as boas-vindas e mostra o ingresso.
## finally: roda sempre, digitando certo ou errado, e limpa a tela para o próximo visitante.
## time.sleep(2) dá 2 segundos para ler a mensagem antes de limpar.
## O while True mantém a máquina esperando o próximo visitante, e para com Ctrl+C.
```
