# 🧬 Atividade Prática 01 — Aula 03: Analisando o Código Genético

## 🧬 Gene HBB — Hemoglobina

### O que é o Gene HBB?

Trecho de **DNA responsável por sintetizar a Beta-Hemoglobina**, proteína essencial das hemácias para o transporte de oxigênio no sangue.

### Por que este trecho?

É uma **sequência real**. Mudanças de apenas uma letra neste gene podem alterar o aminoácido e causar doenças como a **Anemia Falciforme**.

---

## 🎯 Objetivo da Aula

Criar um script em **Python** para:

* Traduzir o DNA em aminoácidos;
* Buscar aminoácidos específicos dentro da proteína;
* Comparar uma sequência de DNA com uma sequência de referência;
* Identificar se a sequência é **normal ou mutante**.

### 📥 Entrada

Sequência de DNA do gene HBB:

```text
ATGGTGCACCTGACTCCTGAG
```

---

# ⚙️ Processamento 1 — Padronização e Comparação

Solicite que o usuário digite uma **fita de DNA com 21 nucleotídeos** utilizando a função `input()`.

### Conceitos

* `.upper()`
* `f-strings`
* Operador de diferença `!=`
* Valores booleanos (`True` / `False`)
* Listas e índices

### Instruções

1. Padronize a fita digitada pelo usuário para **letras maiúsculas**.
2. Exiba a fita formatada utilizando uma **f-string**.
3. Defina a sequência de referência normal da Hemoglobina A:

```python
"ATGGTGCACCTGACTCCTGAG"
```

4. Compare a fita digitada pelo usuário com a fita de referência utilizando o operador `!=`.

O resultado deverá ser um valor **Booleano**:

5. Crie uma lista com as duas opções de diagnóstico:

---

# 🧪 Processamento 2 — Comparação e Tradução de Códons

### Conceitos

* `.replace()`
* `.replace()` encadeado
* Concatenação de strings

### Instruções

Crie a variável `proteina` aplicando métodos `.replace()` sucessivos na fita de DNA.

Traduza a fita de DNA substituindo as **trincas (códons)** pelos seus respectivos aminoácidos utilizando o método `.replace()` encadeado.

Ao final, **imprima a cadeia de aminoácidos** resultante na tela.

### 🧬 Tabela de Códons do Gene HBB

| Códon | Aminoácido |
| ----- | ---------- |
| `ATG` | Met        |
| `GTG` | Val        |
| `CAC` | His        |
| `CTG` | Leu        |
| `ACT` | Thr        |
| `CCT` | Pro        |
| `GAG` | Glu        |

---

# 🔎 Processamento 3 — Localização de Padrões

### Conceito

Utilização do método `.find()` para localizar uma sequência dentro de uma string.

### Instruções

Solicite ao usuário um **aminoácido para busca**.

Exemplos:

```text
Val
His
```

Utilize `.find()` para localizar o **índice do aminoácido** dentro da proteína traduzida.

---

# 📊 Relatório Final

Ao final do programa, imprima um relatório contendo:

* **DNA de referência**
* **DNA digitado pelo usuário**
* **Tamanho da fita** utilizando `len()`
* **Proteína traduzida**
* **Resultado do diagnóstico**
* **Posição do aminoácido pesquisado**


````python
Hemoglobina_A = "ATGGTGCACCTGACTCCTGAG"

print(f"A fita de referência é: {Hemoglobina_A}")

# Entrada do usuário
DNA = input("Digite uma fita de DNA com 21 nucleotídeos: ")[:21]
DNA_padrao = DNA.upper()
print(f"O DNA digitado foi: {DNA_padrao}")

# Processamento 1: Diagnóstico
DNA_mutante = DNA_padrao != Hemoglobina_A
diagnosticos = [
    "✅ Sequência NORMAL (Hemoglobina A)",
    "⚠️ Sequência MUTANTE (Anemia Falciforme - HbS)"
]

# Processamento 2: Tradução
proteina = DNA_padrao.replace("ATG", "Met-").replace("GTG", "Val-").replace("CAC", "His-").replace("CTG", "Leu-").replace("ACT", "Thr-").replace("CCT", "Pro-").replace("GAG", "Glu-")

# Processamento 3: Busca e Relatório
busca = input("Digite um aminoácido para buscar (ex: Met, Val, His): ")
posicao = proteina.find(busca)

print("\n" + "="*50)
print("      RELATÓRIO FINAL")
print("="*50)
print(f"DNA de Referência:      {Hemoglobina_A}")
print(f"DNA Digitado:          {DNA_padrao}")
print(f"Tamanho da fita:       {len(DNA_padrao)} nucleotídeos")
print(f"Proteína Traduzida:    {proteina}")
print(f"Diagnóstico:           {diagnosticos[DNA_mutante]}")
print(f"Busca do aminoácido:   O aminoácido '{busca}' está na posição {posicao}")
print("="*50)
````

# 🧙‍♂️ Atividade Prática 02 — Aula 03: O Analisador de Pergaminhos da Terra-Média

## 📜 A Missão

A **Sociedade do Anel** encontrou um pergaminho antigo, mas o texto está bagunçado!

Seu trabalho é criar um programa em **Python** que receba esse texto, faça a **higienização dos dados** e gere um relatório.

---

# ⚙️ Passo a Passo — Requisitos do Sistema

## 📥 1. Entrada de Dados

O programa deve:

* Solicitar que o usuário digite o **texto do pergaminho** utilizando `input()`.
* A frase digitada deve conter **espaços extras no começo e no final**.

---

## 🧹 2. Limpeza e Higienização

Remova os **espaços inúteis no início e no final** do texto digitado e armazene o resultado em uma nova variável.

> 💡 Dica: utilize o método `.strip()`.

---

# 📊 3. O Relatório

O programa deve calcular e exibir na tela:

* 🔢 O **número total de caracteres** do texto limpo.
* 📝 A **quantidade total de palavras** presentes no texto.
* 🥇 A **primeira palavra** do texto.
* 🏁 A **última palavra** do texto.

---

# ⚠️ 4. Detector de Ameaças

O sistema deve verificar se uma determinada palavra existe dentro do texto.

### 🔎 Verificação

Utilize, por exemplo, a palavra:

```text
Anel
```

Ou escolha **outra palavra** de sua preferência.

O programa deve:

* Verificar se a palavra existe no texto;
* Imprimir diretamente o resultado da verificação;
* Contar **quantas vezes a palavra aparece** no texto.

> 💡 Dica: utilize os métodos e operadores de strings estudados em aula.


````python
# Analisador de Pergaminhos

# Entrada do texto
pergaminho = input("Digite o texto do pergaminho: ")

# Limpeza do texto
texto_limpo = pergaminho.strip()

# Quantidade de caracteres
quantidade_caracteres = len(texto_limpo)

# Separação das palavras
palavras = texto_limpo.split()

# Quantidade de palavras
quantidade_palavras = len(palavras)

# Primeira e última palavra
primeira_palavra = palavras[0]
ultima_palavra = palavras[-1]

# Detector de ameaças
palavra_ameaca = "Anel"

existe_ameaca = palavra_ameaca in texto_limpo
quantidade_ameaca = texto_limpo.count(palavra_ameaca)

# Exemplo para preencher o input:
# O Anel está perdido na Terra-Média. O Anel precisa ser encontrado.

# Relatório
print(f"""
===== RELATÓRIO DO PERGAMINHO =====

Texto: {texto_limpo}

Quantidade de caracteres: {quantidade_caracteres}
Quantidade de palavras: {quantidade_palavras}
Primeira palavra: {primeira_palavra}
Última palavra: {ultima_palavra}

===== DETECTOR DE AMEAÇAS =====

A palavra "{palavra_ameaca}" existe? {existe_ameaca}
Quantidade de vezes encontrada: {quantidade_ameaca}
""")
````

## 🎯 A Missão

Crie um pequeno script onde seu personagem realiza ações que envolvem **cálculos e tomadas de decisão**.

## 💡 Ideias para os Testes Condicionais

Você pode utilizar algumas das situações abaixo como inspiração:

### ❤️ Vida do personagem

> A vida (`HP`) do personagem chegou a zero ou menos?

**Resultado:** `Game Over!`

### ⚔️ Ataque

> A força do personagem somada ao bônus da espada é maior que a defesa do monstro?

**Resultado:** `Acerto crítico!`

---

### 🪙 Dinheiro

> O dinheiro que sobrou após comprar a poção é suficiente para pagar a taverna?

**Resultado:**

* Pode dormir na taverna;
* Ou vai dormir na rua!

## ✅ Requisitos

Seu programa deve conter:

* [ ] Pelo menos **3 novas variáveis**;
* [ ] Operações **aritméticas**;
* [ ] Pelo menos **3 testes condicionais**;
* [ ] Uso de `if / else`;
* [ ] Uso de operadores relacionais;
* [ ] Resultados apresentados com `print()`.


````python
# Jornada do Personagem

# Dados do personagem
vida = 100
forca = 20
dinheiro = 50

# Cálculos

# Dano causado pelo personagem
dano = forca + 10

# Vida depois de receber um ataque
vida_restante = vida - 30

# Dinheiro depois de comprar uma poção
dinheiro_restante = dinheiro - 20


# Decisão 1 - Ataque
if dano > 25:
    print("O ataque foi forte! O personagem derrotou o monstro.")
else:
    print("O ataque foi fraco! O monstro continua de pé.")


# Decisão 2 - Vida
if vida_restante > 0:
    print("O personagem sobreviveu ao ataque.")
else:
    print("Game Over! O personagem morreu.")


# 🪙 Decisão 3 - Dinheiro
if dinheiro_restante >= 30:
    print("O personagem tem dinheiro suficiente para pagar a taverna.")
else:
    print("O personagem ficou sem dinheiro e vai dormir na rua.")
````
