Animais Favoritos

Objetivo: Crie uma lista com vários animais para realizar filtros: Filtros:

Mostre em tela seus animais favoritos;
Imprima a quantidade de animais da lista e a quantidade de animais favoritos;
Mostre qual é o primeiro animail favorito e o último;
Remova um dos animais da lista e imprima o resultado;
Adicione um ou mais animais na lista;
Verifique quantos animais temos agora;
Troque a ordem dos animais favoritos de acordo com sua preferência e imprima.

```python
animais = ["cachorro", "gato", "cavalo", "vaca", "papagaio", "leão"]
favoritos = ["cachorro", "gato", "leão"]

# 1. Mostrar favoritos
print(favoritos)

# 2. Quantidade de animais e de favoritos
print(len(animais), len(favoritos))

# 3. Primeiro e último favorito
print(favoritos[0], favoritos[-1])

# 4. Remover um animal
animais.remove("vaca")
print(animais)

# 5. Adicionar animais
animais.append("panda")
animais.append("lobo")

# 6. Quantidade atual
print(len(animais))

# 7. Trocar a ordem dos favoritos
favoritos = ["leão", "cachorro", "gato"]
print(favoritos)
```

Atividade Prática 02 Aula 06
O Cardápio Fixo do Axolote (Tuplas)

Objetivo: Como sabemos, as tuplas são coleções imutáveis, sendo o formato ideal para armazenar dados que não devem ser modificados acidentalmente. O axolote de estimação tem uma rotina de alimentação muito restrita! Atividades:
Crie uma tupla contendo três elementos de tipos variados: o horário da refeição (texto), o alimento principal (texto) e a quantidade de vitaminas (número).
Tente alterar o alimento principal da tupla para "pizza de calabresa". Delete a linha que causou o erro. Agora, utilize a técnica de desempacotamento de tuplas para extrair esses três valores e salvá-los em três variáveis independentes usando apenas 1 linha de código.
Imprima uma frase amigável usando as variáveis desempacotadas. Por exemplo: "Às 08:00, o axolote come minhocas com 50 vitaminas."

```python
# 1. Criar a tupla (horário, alimento, vitaminas)
refeicao = ("08:00", "minhocas", 50)

# 2. Desempacotar em 1 linha
horario, alimento, vitaminas = refeicao

# 3. Frase amigável
print(f"Às {horario}, o axolote come {alimento} com {vitaminas} vitaminas.")
```

Agenda Telefônica (Dicionários)

Objetivo:

Crie um dicionário de contatos com alguns nomes (como chaves) e números de telefone (como valores);
Adicione um novo contato ao dicionário sem alterá-lo diretamente (na declaração original);
Verifique se o contato "Chai" existe nos contatos. Se não, adicione-a com um número de telefone inventado.;
Exclua um dos seus contatos do dicionário; use um método que não cause um erro se o contato não existir;
Imprima todos os nomes presentes no dicionário.
OBS.:Para cada ação imprima o resultado.# Atividade Prática 03 Aula 06

```python
# 1. Criar o dicionário de contatos
contatos = {"Bruno": "11 91111-1111", "Ana": "11 92222-2222", "Carlos": "11 93333-3333"}
print("1.", contatos)

# 2. Adicionar um novo contato (fora da declaração original)
contatos["Marina"] = "11 94444-4444"
print("2.", contatos)

# 3. Verificar se "Chai" existe; se não, adicionar
if "Chai" in contatos:
    print("3. Chai já existe:", contatos["Chai"])
else:
    contatos["Chai"] = "11 95555-5555"
    print("3. Chai adicionada:", contatos)

# 4. Excluir um contato sem causar erro se não existir
contatos.pop("Carlos", None)
print("4.", contatos)

# 5. Imprimir todos os nomes
print("5.", list(contatos.keys()))
```

Atividade Prática 04 Aula 06
A Sacola do Clube do Livro (Conjuntos)

Objetivo: Um conjunto (set) é uma estrutura que, por natureza, não permite elementos duplicados, sendo útil para remover duplicatas de coleções de dados automaticamente. Fui à feira de troca de livros com meus amigos e jogamos todos os livros que arrecadamos em uma grande sacola. Acabamos pegando várias cópias do mesmo livro!

Atividades:

Crie uma lista inicial com cerca de 6 ou 7 títulos de livros, mas coloque de propósito uns 3 títulos repetidos (por exemplo: "Duna", "O Hobbit", "Duna", "Harry Potter", "O Hobbit").
Converta essa sua lista em um Conjunto para forçar o sistema a eliminar todas as cópias duplicadas em um passe de mágica.
Imprima o conjunto na tela. Note que o resultado final perdeu as repetições e que os itens provavelmente mudaram de lugar, já que os conjuntos não mantêm a ordem de inserção dos elementos!

```python
# 1. Lista com títulos repetidos
livros = ["Duna", "O Hobbit", "Duna", "Harry Potter", "O Hobbit", "1984", "Duna"]
print("Lista original:", livros)

# 2. Converter em conjunto (remove duplicatas)
livros_unicos = set(livros)

# 3. Imprimir o conjunto
print("Conjunto:", livros_unicos)
```
