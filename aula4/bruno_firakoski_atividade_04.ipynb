Atividade Prática 03 Aula 04 - If, elif, else
Escreva um programa em Python que:

Pergunte ao usuário qual foi a sua nota (lembre-se de converter para float, pois as notas podem ter decimais).

Utilize as estruturas condicionais if, elif e else para avaliar a nota e exibir na tela o título conquistado pelo aluno, para cada uma das condições abaixo deve-se exibir uma mensagem diferente:

Nota 9.0 ou mais;

Nota de 7.0 até 8.9;

Nota de 5.0 até 6.9;

Nota abaixo de 5.0;

```python
nota_user = float(input("Digite sua nota (formato: n.n): "))
print(f"Nota do usuário: {nota_user}")

if nota_user >= 9.0:
  print(f"Nota: {nota_user} - Aprovado pra caramba! Farmou muita aura.")

elif nota_user >= 7.0 and nota_user <= 8.9:
  print(f"Nota: {nota_user} - Aprovado, mas pode farmar mais aura.")

elif nota_user >= 5.0 and nota_user <= 6.9:
  print(f"Nota: {nota_user} - Faltou aura meu mano.")

else:
  print(f"Nota: {nota_user} - Perdeu aura, foi moggado.")

```

Atividade Prática 04 Aula 04 - A Jornada do Personagem (Operadores e Condicionais)
Agora que você já sabe capturar dados com input(), converter tipos e fazer cálculos, chegou a hora de dar inteligência e consequências para a história do seu personagem usando o comando if!

Anteriormente, usar operadores relacionais (>, <, ==) apenas mostrava True ou False na tela. Agora, você vai usar essas comparações para fazer o computador tomar decisões.

A Missão:

Crie um pequeno script onde o seu personagem toma ações que envolvem cálculos e decisões.

Cálculos e Variáveis: Crie pelo menos 3 variáveis que envolvam operações aritméticas (ex: calcular o dano de um ataque, subtrair moedas de ouro após uma compra, somar pontos de experiência).

Tomada de Decisão (if / else): Baseado nos resultados desses cálculos, crie 3 testes condicionais para verificar o estado do seu personagem.

Mostre o que aconteceu na história usando o print().

Exemplos de ideias que você pode testar com o if:

A vida (HP) do personagem chegou a zero ou menos? (Game Over)

A força somada com o bônus da espada é maior que a defesa do monstro? (Acerto crítico)

O dinheiro que sobrou após comprar a poção é suficiente para pagar a taverna? (Pode dormir ou vai pra rua)

```python
# Atividade Prática 04 - A Jornada do Personagem

# Dados do personagem
vida = 100
forca = 20
dinheiro = 50

# Cálculos

# Dano do ataque
dano = forca + 10

# Vida restante depois de receber um ataque
vida_restante = vida - 30

# Dinheiro restante depois de comprar uma poção
dinheiro_restante = dinheiro - 20


# Decisão 1 - Ataque
if dano > 25:
    print("Acerto crítico! O personagem derrotou o monstro.")
else:
    print("O ataque foi fraco. O monstro continua vivo.")


# Decisão 2 - Vida
if vida_restante > 0:
    print("O personagem sobreviveu ao ataque.")
else:
    print("Game Over! O personagem morreu.")


# Decisão 3 - Dinheiro
if dinheiro_restante >= 30:
    print("O personagem tem dinheiro suficiente para pagar a taverna.")
else:
    print("O personagem ficou sem dinheiro e vai dormir na rua.")
```

Atividade Prática 05 Aula 04 - match...case
Sistema de Chamados (Help Desk) Você foi contratado para desenvolver a triagem de um sistema de chamados de TI. O programa deve pedir ao usuário para digitar o número correspondente ao problema que ele está enfrentando. Utilizando a estrutura match...case, direcione o chamado para a equipe correta e exiba uma mensagem confirmando o direcionamento, o menu de escolhas deve ter pelo menos 4 opções.

Exemplos de Opções: 1 - Problema com Computador/Hardware ➞ Direcionar para: Manutenção

2 - Instalação de Software➞ Direcionar para: Equipe de Suporte

3 - Sem acesso à Internet/Rede ➞ Direcionar para: Infraestrutura

4 - Reset de Senha ➞ Direcionar para: Atendimento Automático (Envio de link de redefinição)

Qualquer outra opção ➞ Exibir: Opção inválida. Chamado encaminhado para a Triagem Geral.

```python
# Atividade Prática 05 - Sistema de Chamados

print("""
===== SISTEMA DE CHAMADOS =====

[1] Problema com Computador/Hardware
[2] Instalação de Software
[3] Sem acesso à Internet/Rede
[4] Reset de Senha
""")

opcao = int(input("Digite o número do problema: "))

match opcao:

    case 1:
        print("Chamado direcionado para a equipe de Manutenção.")

    case 2:
        print("Chamado direcionado para a equipe de Suporte.")

    case 3:
        print("Chamado direcionado para a equipe de Infraestrutura.")

    case 4:
        print("Chamado direcionado para o Atendimento Automático.")
        print("Um link de redefinição de senha será enviado.")

    case _:
        print("Opção inválida.")
        print("Chamado encaminhado para a Triagem Geral.")
```

Atividade Prática 06 Aula 04 - While e For
Sistema de Caixa e Parcelamento Objetivo: Desenvolver um programa que utilize os dois tipos de laço de repetição (while e for).

Cenário: Você está programando o sistema de caixa de uma loja. O programa deve funcionar em duas etapas:

Passando as compras: O sistema deve pedir ao caixa para digitar o preço de cada produto. Como não sabemos quantos produtos o cliente vai levar, o programa deve continuar pedindo os preços e somando o total até que o caixa digite o valor 0 (zero), que significa que a compra acabou.

Opções de Parcelamento: Após calcular o total, o sistema deve mostrar ao cliente as opções de parcelamento, mostrando o valor de cada parcela.

Ao final do código, adicione um comentário (#) respondendo: Qual laço de repetição foi a melhor escolha para a etapa 1 e 2 e porque?

```python
# Atividade Prática 06 - Sistema de Caixa e Parcelamento

total = 0

# Etapa 1 - Passando as compras
preco = float(input("Digite o preço do produto (0 para finalizar): "))

while preco != 0:
    total = total + preco
    preco = float(input("Digite o preço do produto (0 para finalizar): "))


print(f"\nTotal da compra: R$ {total:.2f}")


# Etapa 2 - Opções de parcelamento
parcelas = int(input("Digite a quantidade de parcelas: "))

valor_parcela = total / parcelas

for i in range(1, parcelas + 1):
    print(f"Parcela {i}: R$ {valor_parcela:.2f}")


# O while foi melhor para a etapa 1 porque não sabemos quantos produtos serão comprados.
# O for foi melhor para a etapa 2 porque sabemos a quantidade de parcelas.
```
