```python
# Pontuação dos jogadores
pontuacao_jogador_1 = 850
pontuacao_jogador_2 = 700

# Operadores relacionais
print("A pontuação do Jogador 1 é maior que a do Jogador 2?", pontuacao_jogador_1 > pontuacao_jogador_2)

print("A pontuação do Jogador 1 é menor que a do Jogador 2?", pontuacao_jogador_1 < pontuacao_jogador_2)

print("As pontuações dos jogadores são iguais?", pontuacao_jogador_1 == pontuacao_jogador_2)

print("As pontuações dos jogadores são diferentes?", pontuacao_jogador_1 != pontuacao_jogador_2)

print("A pontuação do Jogador 1 é maior ou igual à do Jogador 2?", pontuacao_jogador_1 >= pontuacao_jogador_2)

print("A pontuação do Jogador 1 é menor ou igual à do Jogador 2?", pontuacao_jogador_1 <= pontuacao_jogador_2)
```

```python
# Dados cadastrados no sistema
usuario_cadastrado = "Vitor"
senha_cadastrada = "12345"

# Dados digitados pelo usuário
usuario_digitado = "Vitor"
senha_digitada = "12345"

# Situação da conta
conta_bloqueada = False

# Verificação das credenciais
credenciais_corretas = (
    usuario_digitado == usuario_cadastrado
    and senha_digitada == senha_cadastrada
)

# Verificação da conta
conta_esta_bloqueada = conta_bloqueada

# Verificação do acesso final
acesso_liberado = credenciais_corretas and not conta_bloqueada

# Resultados
print("As credenciais estão corretas?", credenciais_corretas)
print("A conta está bloqueada?", conta_esta_bloqueada)
print("O acesso final está liberado?", acesso_liberado)
```

```python


# Dados iniciais
vida = 100
forca = 25
ouro = 200

# Cálculos
dano_do_ataque = forca * 2
vida_do_monstro = 80 - dano_do_ataque
ouro_restante = ouro - 50
experiencia = forca + 75

# Decisão 1 - Vida do monstro
if vida_do_monstro <= 0:
    print("O monstro foi derrotado!")
else:
    print("O monstro ainda está vivo!")

# Decisão 2 - Ouro restante
if ouro_restante >= 100:
    print("Você tem ouro suficiente para descansar na taverna.")
else:
    print("Você não tem ouro suficiente para a taverna.")

# Decisão 3 - Experiência
if experiencia >= 100:
    print("Você ganhou experiência suficiente para subir de nível!")
else:
    print("Você ainda precisa de mais experiência.")

# Resultado dos cálculos
print("\n===== RESULTADO DA JORNADA =====")
print("Dano causado:", dano_do_ataque)
print("Vida restante do monstro:", vida_do_monstro)
print("Ouro restante:", ouro_restante)
print("Experiência:", experiencia)
```

```python


print("===== SISTEMA DE CHAMADOS DE TI =====")
print("1 - Problema com Computador/Hardware")
print("2 - Instalação de Software")
print("3 - Sem acesso à Internet/Rede")
print("4 - Reset de Senha")

opcao = int(input("Digite o número do problema: "))

match opcao:
    case 1:
        print("Chamado direcionado para a equipe de Manutenção.")

    case 2:
        print("Chamado direcionado para a equipe de Suporte.")

    case 3:
        print("Chamado direcionado para a equipe de Infraestrutura.")

    case 4:
        print("Chamado direcionado para o Atendimento Automático.")
        print("Um link de redefinição de senha será enviado.")

    case _:
        print("Opção inválida. Chamado encaminhado para a Triagem Geral.")
```

```python

# Etapa 1 - Passando as compras
total = 0

preco = float(input("Digite o preço do produto (0 para finalizar): "))

while preco != 0:
    total = total + preco
    preco = float(input("Digite o preço do produto (0 para finalizar): "))

print("\nTotal da compra: R$", total)

# Etapa 2 - Opções de parcelamento
print("\n===== OPÇÕES DE PARCELAMENTO =====")

for parcelas in range(1, 7):
    valor_parcela = total / parcelas
    print(f"{parcelas}x de R$ {valor_parcela:.2f}")

# Resposta:
# O while foi a melhor escolha para a etapa 1 porque não sabemos
# quantos produtos serão cadastrados e o laço termina quando o usuário digita 0.
# O for foi a melhor escolha para a etapa 2 porque sabemos que queremos
# mostrar exatamente 6 opções de parcelamento.
```
