````python
# 1. Erro de Nome (NameError)
nome_heroi = "Batman"
# Correção: Alterado de 'nome_heroi_cidade' para 'nome_heroi'
print(f"O herói que vai salvar a cidade é o {nome_heroi}")


# 2. Erro de Tipo (TypeError)
idade_pet = 5
# Correção: Uso de f-string para converter o inteiro automaticamente para string
mens_agem = f"A idade do meu cachorro é {idade_pet} anos."
print(mens_agem)


# 3. Erro de Valor (ValueError)
print("Calculadora de Dobro!")
numero_digitado = "10"  # Correção: Alterado de "dez" para um valor numérico válido em string
dobro = int(numero_digitado) * 2
print(f"O dobro é: {dobro}")


# 4. Erro de Chave (KeyError)
cat_attributes = {
    "name": "Ita",
    "age": 1,
    "color": "grey",
    "breed": "Desconhecida"  # Correção: Adicionada a chave 'breed' ao dicionário
}
raca_gato = cat_attributes["breed"]
print(f"A raça do gato é: {raca_gato}")


# 5. Erro de Índice (IndexError)
some_numbers = [10, 20, 30]
# Correção: Índice alterado para 2 (dentro do tamanho da lista)
quarto_numero = some_numbers[2]
print(f"O número escolhido foi: {quarto_numero}")


# 6. Erro de Divisão por Zero (ZeroDivisionError)
tesouro_moedas = 100
aventureiros_sobreviventes = 4  # Correção: Alterado de 0 para 4
print("Calculando a divisão do tesouro...")
moedas_por_pessoa = tesouro_moedas / aventureiros_sobreviventes
print(f"Cada aventureiro vai receber: {moedas_por_pessoa}")


# 7. Erro de Atribuição em Tupla (TypeError)
print("*** Sistema de Zoológico ***")
dieta_axolote = ["08:00", "minhocas", 50]  # Correção: Convertido para lista para permitir alteração
print("O axolote enjoou de minhocas. Vamos alterar a dieta para pizza!")
dieta_axolote[1] = "pizza de calabresa"
dieta_axolote = tuple(dieta_axolote)  # Retornado para tupla, se necessário
print(f"Nova dieta: {dieta_axolote}")


# 8. Erro de Excesso de Recursão (RecursionError)
def contagem_regressiva(tempo):
    if tempo <= 0:  # Correção: Adicionado caso base para parar a recursão
        print("Fogo!")
        return
    print(f"Faltam {tempo} segundos...")
    contagem_regressiva(tempo - 1)

print("Iniciando contagem para a explosão do Anel...")
contagem_regressiva(5)
```[cite: 4]

---

### Atividade Prática 02 Aula 07[cite: 5]

#### Parte 01 - Soneca de gato com bugs
```python
def nap_time(hours_list):
    total = 0
    # Correção: Substituído o while com índices incorretos por um ciclo for direto
    for hours in hours_list:
        total += hours
    return total

cat_naps = [1, 2, 4]
print("Total nap time:", nap_time(cat_naps))
````

````python
def average_meows(logs):
    total_meows_per_hour = 0
    for log in logs:
        soma_naps = sum(log["naps"])
        # Correção: Evita ZeroDivisionError caso a soma dos naps seja 0
        if soma_naps > 0:
            average_meows_per_hour = log["meows"] / soma_naps
            total_meows_per_hour += average_meows_per_hour
    return total_meows_per_hour / len(logs) if len(logs) > 0 else 0

logs = [
    {"day": "Monday", "naps": [1, 2, 4], "meows": 50},
    {"day": "Tuesday", "naps": [4, 3, 2], "meows": 75}
]

print("Average meows per audio hour:", average_meows(logs))
````

````python
def total_distance(segments):
    total = 0
    for length, unit in segments:
        if unit.lower() == "m":
            total += length
        elif unit.lower() == "km":
            total += length * 1000  # Correção: Multiplica por 1000 para converter km em metros
    print(f"Total distance: {total} meters")

total_distance([
    (2.5, "km"),
    (400, "m")
])
```[cite: 5]

---

### Atividade Prática 03 Aula 07: Ingressos do Museu[cite: 6]

```python
def ticket_type():
    try:
        age_input = input("Enter your age: ")
        age = int(age_input)

        if age < 12:
            print("You get in for free. Lucky you!")
        elif age < 18:
            print("You get a youth discount!")
        elif age < 65:
            print("That's a full price ticket for you.")
        else:
            print("You get a senior discount!")

    except ValueError:
        print("Erro: Digite um número válido para a idade.")
        ticket_type()  # Pede redigitados caso o valor seja inválido

ticket_type()
```[cite: 6]
````
