```python
# Atividade Prática 02 - Aula 07 - Parte 01
# Caça aos Bugs - Soneca de gato

def nap_time(hours_list):
    total = 0
    counter = 0  # começa em 0, para não pular a primeira soneca da lista
    while counter < len(hours_list):
        total += hours_list[counter]
        counter += 1
    return total


cat_naps = [2, 3, 1.5, 4]
print("Total nap time:", nap_time(cat_naps))

# Bug encontrado: era um erro lógico (o programa rodava sem quebrar, mas dava
# a resposta errada). O laço "while" começava em counter = 1, o que fazia o
# código pular a primeira soneca da lista (2 horas) e somar só
# 3 + 1.5 + 4 = 8.5, em vez de 2 + 3 + 1.5 + 4 = 10.5.
# Corrigido iniciando o counter em 0, para percorrer a lista desde o início.
```

```python
# Atividade Prática 02 - Aula 07 - Parte 02
# Quanto o seu gato mia?

def average_meows(logs):
    total_meows_per_hour = 0

    for log in logs:
        nap_hours = 0

        for nap in log["naps"]:
            nap_hours += nap

        awake_hours = 24 - nap_hours

        if awake_hours <= 0:
            # dia em que a soma das sonecas deu 24h (ou mais) - não há horas
            # acordado nesse dia, então pulamos ele para não dividir por zero
            print(f"Aviso: dia '{log['day']}' ignorado no cálculo (0 horas acordado registradas).")
            continue

        total_meows_per_hour += log["meows"] / awake_hours

    return total_meows_per_hour / len(logs)


logs = [
    {"day": "Monday", "naps": [1, 2, 5], "meows": 90},
    {"day": "Tuesday", "naps": [4, 3, 2], "meows": 75},
    {"day": "Wednesday", "naps": [8, 6, 9, 1], "meows": 9},
    {"day": "Thursday", "naps": [1, 1, 5, 7], "meows": 38},
    {"day": "Friday", "naps": [3, 3, 2, 8], "meows": 94},
    {"day": "Saturday", "naps": [4, 8, 3], "meows": 64},
    {"day": "Sunday", "naps": [3, 3, 2, 8], "meows": 37}
]

print("Average meows per awake hour:", average_meows(logs))

# Bug encontrado: ZeroDivisionError. Na quarta-feira ("Wednesday"), a soma das
# sonecas é 8 + 6 + 9 + 1 = 24, então awake_hours = 24 - 24 = 0, e o código
# tentava dividir log["meows"] por 0. Corrigido checando se awake_hours é
# menor ou igual a 0 e, nesse caso, pulando o dia (continue) em vez de travar.
```

```python
# Atividade Prática 02 - Aula 07 - Parte 03
# Bugs na Trilha

def total_distance(*segments):
    total = 0
    for length, unit in segments:
        unidade = unit.strip().lower()  # remove espaços extras e padroniza em minúsculas

        if unidade == "m":
            total += length
        elif unidade == "km":
            total += length * 1000  # converte km para metros (multiplica, não divide!)

    print("Total distance:", total, "meters")


total_distance(
    (2.5, "km"),
    (400, "M"),
    (1.0, "km"),
    (800, "m"),
    (1.3, "KM"),
    (900, "m "),
    (1.1, "km")
)

# Bugs encontrados (o código rodava sem quebrar, mas o resultado estava errado):
# 1) A comparação era "unit.lower() == 'M'". Como .lower() sempre devolve
#    letras minúsculas, essa comparação com 'M' maiúsculo nunca era
#    verdadeira, então os segmentos em metros nunca eram somados.
# 2) Para quilômetros, o código fazia "total += length / 1000", só que o
#    resultado final é anunciado em metros - então km deveria ser
#    multiplicado por 1000 para virar metros, e não dividido.
# 3) O segmento (900, "m ") tem um espaço extra depois do "m", por isso
#    usamos .strip() além de .lower() para essa comparação não falhar.
# Corrigido comparando com "m" minúsculo (após strip + lower) e convertendo
# km para metros multiplicando por 1000.
```

```python
# Atividade Prática 03 - Aula 07
# Caça aos Bugs e Tratamento de Erros - Ingressos do museu

import os


def limpar_tela():
    os.system("cls" if os.name == "nt" else "clear")


def ticket_type():
    while True:
        idade_texto = input("Enter your age: ")

        try:
            age = int(idade_texto)
        except ValueError:
            print("Idade inválida! Digite apenas números.")
            input("Pressione ENTER para tentar novamente...")
            limpar_tela()
            continue  # pede para redigitar, sem quebrar o programa

        if age < 0:
            print("Idade inválida! A idade não pode ser negativa.")
            input("Pressione ENTER para tentar novamente...")
            limpar_tela()
            continue

        if age < 12:
            print("You get in for free. Lucky you!")
        elif age < 18:
            print("You get a youth discount!")
        elif age < 65:
            print("That's a full-price ticket for you.")
        else:
            print("You get a senior discount!")

        print("Bem-vindo(a) ao museu! 🏛️")
        input("Pressione ENTER para liberar a máquina para o próximo visitante...")
        limpar_tela()
        break  # idade válida processada, encerra a função


ticket_type()

# Bugs encontrados:
# 1) TypeError: "age" vinha do input() como string, e o código comparava
#    "age < 12" (texto com número), o que o Python não permite. Corrigido
#    convertendo para número com int(age).
# 2) Sem tratamento de erro: se o visitante digitasse algo que não é um
#    número (ex: "abc"), o int(age) quebrava com ValueError. Corrigido com
#    try/except, que agora pede para o visitante redigitar a idade.
# 3) Faltava a limpeza de tela: adicionado import os + limpar_tela(), chamada
#    tanto quando a idade é válida quanto quando é inválida, deixando a tela
#    pronta para o próximo visitante.
```
