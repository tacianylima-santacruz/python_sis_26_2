```python
import os
from IPython.display import clear_output

# =====================================================================
# Atividade Prática 01 Aula 07: Caça aos Bugs
# =====================================================================

# --- BUG 1: NameError ---
# Código original causava erro pois a variável 'nome_heroi_cidade' não havia sido definida.
nome_heroi = "Batman"
# Correção: Alterado 'nome_heroi_cidade' para 'nome_heroi'
print("O herói que vai salvar a cidade é o " + nome_heroi)


# --- BUG 2: TypeError ---
idade_pet = 5
# Correção: Convertido 'idade_pet' de int para str para realizar a concatenação
mensagem = "A idade do meu cachorro é " + str(idade_pet) + " anos."
print(mensagem)


# --- BUG 3: ValueError ---
print("Calculadora de Dobro!")
numero_digitado = "dez"
# Correção: Usando try/except para capturar o erro caso o valor digitado não seja conversível para inteiro
try:
    dobro = int(numero_digitado) * 2
    print("O dobro é:", dobro)
except ValueError:
    print("Erro: O valor digitado não é um número válido. Usando o padrão 10.")
    numero_digitado = "10"
    dobro = int(numero_digitado) * 2
    print("O dobro de 10 é:", dobro)


# --- BUG 4: KeyError ---
cat_attributes = {
    "name": "Zia",
    "age": 1,
    "color": "grey"
}
# Correção: Usamos o método .get() para evitar o KeyError caso a chave 'breed' não exista
raca_gato = cat_attributes.get("breed", "Vira-lata (não informada)")
print("A raça da gata é:", raca_gato)


# --- BUG 5: IndexError ---
some_numbers = [10, 20, 30]
# Correção: O índice máximo para uma lista de 3 elementos é 2. Alterado para acessar o último elemento válido ou tratar o erro.
try:
    quarto_numero = some_numbers[3]
except IndexError:
    print("Erro: Tentativa de acessar índice fora do limite. Exibindo o último número disponível:")
    quarto_numero = some_numbers[-1]
print("O número escolhido foi:", quarto_numero)


# --- BUG 6: ZeroDivisionError ---
tesouro_moedas = 100
aventureiros_sobreviventes = 0
print("Calculando a divisão do tesouro...")
# Correção: Verificamos se há sobreviventes antes de realizar a divisão para evitar divisão por zero
if aventureiros_sobreviventes > 0:
    moedas_por_pessoa = tesouro_moedas / aventureiros_sobreviventes
    print("Cada aventureiro vai receber:", moedas_por_pessoa)
else:
    print("Infelizmente não restaram aventureiros para dividir o tesouro.")


# --- BUG 7: TypeError (Imutabilidade de Tuplas) ---
print("=== Sistema do Zoológico ===")
dieta_axolote = ("08:00", "minhocas", 50)
print("O axolote enjoou de minhocas. Vamos alterar a dieta para pizza!")
# Correção: Convertido para lista para fazer a alteração, depois convertido de volta para tupla
dieta_lista = list(dieta_axolote)
dieta_lista[1] = "pizza de calabresa"
dieta_axolote = tuple(dieta_lista)
print("Nova dieta:", dieta_axolote)


# --- BUG 8: RecursionError ---
def contagem_regressiva(tempo):
    # Correção: Adicionado caso base para parar a recursão quando o tempo chegar a zero
    if tempo <= 0:
        print("BUM! O Anel foi destruído!")
        return
    print(f"Faltam {tempo} segundos...")
    contagem_regressiva(tempo - 1)

print("Iniciando contagem para a explosão do Anel...")
contagem_regressiva(3)


# =====================================================================
# Atividade Prática 02 Aula 07 - Parte 01 - Soneca de gato
# =====================================================================

def nap_time(hours_list):
    total = 0
    # Correção: O counter deve iniciar em 0 para somar o primeiro elemento da lista
    counter = 0
    while counter < len(hours_list):
        total += hours_list[counter]
        counter += 1
    return total

cat_naps = [2, 3, 1.5, 4]
print("\nTotal nap time (Corrigido):", nap_time(cat_naps))


# =====================================================================
# Atividade Prática 02 Aula 07 - Parte 02 - Quanto o seu gato mia?
# =====================================================================

def average_meows(logs):
    total_meows_per_hour = 0

    for log in logs:
        nap_hours = 0
        for nap in log["naps"]:
            nap_hours += nap

        awake_hours = 24 - nap_hours

        # Correção: Se o gato dormiu 24 horas, as horas acordado são 0.
        # Evitamos divisão por zero assumindo que os miados por hora acordado serão 0 nesse dia.
        if awake_hours > 0:
            total_meows_per_hour += log["meows"] / awake_hours
        else:
            print(f"Aviso: No dia {log['day']}, o gato dormiu 24 horas!")

    return total_meows_per_hour / len(logs)

logs = [
    {"day": "Monday", "naps": [1, 2, 5], "meows": 90},
    {"day": "Tuesday", "naps": [4, 3, 2], "meows": 75},
    {"day": "Wednesday", "naps": [8, 6, 9, 1], "meows": 9},
    {"day": "Thursday", "naps": [1, 1, 5, 7], "meows": 38},
    {"day": "Friday", "naps": [3, 3, 2, 8], "meows": 94},
    {"day": "Saturday", "naps": [4, 8, 3], "meows": 64},
    {"day": "Sunday", "naps": [3, 3, 2, 8], "meows": 37}
]

print("Average meows per awake hour (Corrigido):", average_meows(logs))


# =====================================================================
# Atividade Prática 02 Aula 07 - Parte 03 - Bugs na Trilha
# =====================================================================

def total_distance(*segments):
    total = 0
    for length, unit in segments:
        # Correção: Convertido para minúsculas e removido espaços com .strip()
        cleaned_unit = unit.lower().strip()
        # Correção das condições físicas: 'm' são metros (soma direto), 'km' multiplica por 1000
        if cleaned_unit == "m":
            total += length
        elif cleaned_unit == "km":
            total += length * 1000

    print("Total distance:", total, "meters")

total_distance(
    (2.5, "km"),
    (400, "M"),
    (1.0, "km"),
    (800, "m"),
    (1.3, "KM"),
    (900, "m "),
    (1.1, "km")
)


# =====================================================================
# Atividade Prática 03 Aula 07 - Ingressos do museu com Tratamento de Erros
# =====================================================================

def ticket_type():
    while True:
        try:
            age_input = input("Enter your age (ou digite 'sair' para finalizar): ")
            if age_input.lower() == 'sair':
                break

            # Conversão para inteiro para garantir a validação numérica
            age = int(age_input)
            if age < 0:
                print("Por favor, digite uma idade válida (maior ou igual a zero).")
                continue

            print("\n--- Seja Bem-vindo ao Museu! ---")
            if age < 12:
                print("You get in for free. Lucky you!")
            elif age < 18:
                print("You get a youth discount!")
            elif age < 65:
                print("That's a full-price ticket for you.")
            else:
                print("You get a senior discount!")

        except ValueError:
            print("Entrada inválida! Por favor, insira apenas números inteiros para a idade.")
        finally:
            # Espera um breve input antes de limpar a tela para o próximo visitante ver o resultado
            input("\nPressione Enter para preparar a tela para o próximo visitante...")
            clear_output(wait=True)
            print("=== Sistema de Ingressos Limpo para o Próximo Visitante ===")

# Para testar o sistema interativo do museu, descomente a linha abaixo:
# ticket_type()
```
