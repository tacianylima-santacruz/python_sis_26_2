## Parte 02 - Quanto o seu gato mia?
Imagine que você escreveu este programa para calcular com que frequência o seu gato mia, em média, por hora acordado. Você manteve um registro de quanto ele miou e de quanto ele dormiu a cada dia. Você arredondou os tempos de soneca; afinal, você é um desenvolvedor de software, não um veterinário. Tudo no seu código parece estar perfeitamente bem. Mas adivinha? O programa trava mesmo assim.
Hora de depurar (debug)!
Use o traceback, declarações print e/ou pontos de interrupção (breakpoints) para descobrir o que está causando o travamento e com que tipo de bug você está lidando. Pense em uma maneira inteligente de modificar o código para que ele não trave mais:

```python
def average_meows(logs):
    total_meows_per_hour = 0

    for log in logs:
        nap_hours = 0

        for nap in log["naps"]:
            nap_hours += nap

        awake_hours = 24 - nap_hours
        total_meows_per_hour += log["meows"] / awake_hours

    return total_meows_per_hour / len(logs)

logs = [
    {"day": "Monday", "naps": [1, 2, 5], "meows": 90},
    {"day": "Tuesday", "naps": [4, 3, 2], "meows": 75},
    {"day": "Wednesday", "naps": [8, 6, 9, 1], "meows": 9},
    {"day": "Thursday", "naps": [1, 1, 5, 7], "meows": 38},
    {"day": "Friday", "naps": [3, 3, 2, 8], "meows": 94},
    {"day": "Saturday", "naps": [4, 8, 3], "meows": 64},
    {"day": "Sunday", "naps": [3, 3, 2, 8], "meows": 37}
]

print("Average meows per awake hour:", average_meows(logs))
```

## Parte 03 - Bugs na Trilha
Você está escrevendo um programa que calcula a distância total de uma trilha. Os dados são flexíveis, e cada segmento da trilha é passado como uma tupla que consiste na distância e na unidade. A função deve imprimir o total de todos esses segmentos. O programa roda sem travar (não quebra), mas algo sobre esse total parece estar errado...
Use técnicas de depuração (debugging) para descobrir o que há de errado com este código e conserte o bug:

```python
def total_distance(*segments):
    total = 0
    for length, unit in segments:
        if unit.lower() == "M":
            total += length
        elif unit.lower() == "km":
            total += length / 1000

    print("Total distance:", total, "meters")

total_distance(
    (2.5, "km"),
    (400, "M"),
    (1.0, "km"),
    (800, "m"),
    (1.3, "KM"),
    (900, "m "),
    (1.1, "km")
)
```

```python
def total_distance(*segments):
    total = 0
    for length, unit in segments:
        if unit.lower() == "M":
            total += length
        elif unit.lower() == "km":
            total += length * 1000
        elif unit.lower() == "KM":
            total += length * 1000
        else :
            unit.lower() == "m"
            total += length

    print("Total distance:", total, "meters")

total_distance(
    (2.5, "km"),
    (400, "M"),
    (1.0, "km"),
    (800, "m"),
    (1.3, "KM"),
    (900, "m "),
    (1.1, "km")
)

#Foram adicionadas novas operações para converter os possíveis números
#inseridos por usuários em uma única métrica, nesse caso metros
```

## Caça aos Bugs e Tratamento de Erros - Ingressos do museu

Você está ajudando um museu local a configurar uma máquina de ingressos. Você escreveu um programa em Python para ela: o visitante digita a sua idade, e a máquina imprime o tipo de ingresso que ele recebe. A lógica é simples:

* Crianças menores de 12 anos têm acesso gratuito;
* Adolescentes menores de 18 anos recebem um desconto para jovens;
* Adultos menores de 65 anos pagam o valor integral;
* Pessoas com 65 anos ou mais recebem um desconto para idosos;

Porém você deve tratar o código para que independente doque o visitante digite o programa não dê erro.

Caso seja uma idade válida o programa deve dar as boas vindas. Caso não deve pedir redigitação.

E independente do usuário digita certo ou errado o sistema deve limpar a tela para aguardar pelo próximo visitante.


```python
def ticket_type():
    age = input("Enter your age: ")
    if age < 12:
        print("You get in for free. Lucky you!")
    elif age < 18:
        print("You get a youth discount!")
    elif age < 65:
        print("That's a full-price ticket for you.")
    else:
        print("You get a senior discount!")

ticket_type()
```
