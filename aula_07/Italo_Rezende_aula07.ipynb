Atividade 02 - Parte 2 - Aula 07

```python
def average_meows(logs):
    total_meows_per_hour = 0
    valid_days = 0  # Contador para saber quantos dias válidos entram na média

    for log in logs:
        nap_hours = 0

        for nap in log["naps"]:
            nap_hours += nap

        awake_hours = 24 - nap_hours

        # Só faz o cálculo se o gato ficou acordado alguma hora no dia
        if awake_hours > 0:
            total_meows_per_hour += log["meows"] / awake_hours
            valid_days += 1

    # Evita divisão por zero caso todos os dias tenham 24h de sono
    if valid_days == 0:
        return 0

    return total_meows_per_hour / valid_days

logs = [
    {"day": "Monday", "naps": [1, 2, 5], "meows": 90},
    {"day": "Tuesday", "naps": [4, 3, 2], "meows": 75},
    {"day": "Wednesday", "naps": [8, 6, 9, 1], "meows": 9},
    {"day": "Thursday", "naps": [1, 1, 5, 7], "meows": 38},
    {"day": "Friday", "naps": [3, 3, 2, 8], "meows": 94},
    {"day": "Saturday", "naps": [4, 8, 3], "meows": 64},
    {"day": "Sunday", "naps": [3, 3, 2, 8], "meows": 37}
]

print("Average meows per awake hour:", average_meows(logs))
```

```python


def total_distance(*segments):
    total = 0

    for length, unit in segments:
        # strip() remove espaços extras, como em "m "
        unidade_limpa = unit.strip().lower()

        if unidade_limpa == "m":
            total += length
        elif unidade_limpa == "km":
            # 1 km possui 1000 metros
            total += length * 1000
        else:
            print("Unidade inválida:", unit)

    print("Distância total:", total, "metros")


total_distance(
    (2.5, "km"),
    (400, "M"),
    (1.0, "km"),
    (800, "m"),
    (1.3, "KM"),
    (900, "m "),
    (1.1, "km")
)
```

```python


import os


def limpar_tela():
    """Limpa a tela no Windows ou em outros sistemas operacionais."""
    os.system("cls" if os.name == "nt" else "clear")


def ticket_type():
    while True:
        try:
            idade_digitada = input("Digite a idade do visitante (ou 'sair'): ").strip()

            if idade_digitada.lower() == "sair":
                limpar_tela()
                print("Sistema de ingressos encerrado.")
                break

            idade = int(idade_digitada)

            if idade < 0 or idade > 130:
                print("Idade inválida. Digite uma idade entre 0 e 130.")
            elif idade < 12:
                print("Bem-vindo! Você tem acesso gratuito.")
            elif idade < 18:
                print("Bem-vindo! Você recebe desconto para jovens.")
            elif idade < 65:
                print("Bem-vindo! Seu ingresso é de valor integral.")
            else:
                print("Bem-vindo! Você recebe desconto para idosos.")

        except ValueError:
            print("Entrada inválida. Digite apenas uma idade inteira.")

        input("\nPressione Enter para atender o próximo visitante...")
        limpar_tela()


ticket_type()
```
