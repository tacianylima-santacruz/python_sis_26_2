Atividade Pratica 3

```python
# Variável global
servidores_online = 0

# Função para adicionar servidores online
def ativarServidor(quantidade):
  """
  Nesta função adicionamos os servidores ativos
  onde a quantidade é a quantidade que definimos
  e os servidores online é nossa variavel global
  fazemos as somas delas para definir e mudar seu valor
  """
  global servidores_online
  servidores_online += quantidade
  print(f"{quantidade} servidor(es) ativado(s).")

# Função para remover servidores online
def desativarServidor(quantidade):
  """
  Ja nesta função usamos para removemos os servidores inativos
  usandos as variaveis da mesma forma que na função anterios
  mas realizando a subtração da varivavel global
  """
  global servidores_online
  servidores_online -= quantidade
  print(f"{quantidade} servidor(es) desativado(s).")

# Simulação do sistema
ativarServidor(5)
ativarServidor(2)
desativarServidor(1)

# Saldo final
print(f"\nTotal de servidores ativos: {servidores_online}")


def gerenciarConexao():
  conexao = 5
  def alterarConexao():
    nonlocal conexao
    conexao -= 1
    print("\n Uma conexão foi usada.")

  alterarConexao()
  alterarConexao()

  print(f"\n Conexões restantes: {conexao}")

gerenciarConexao()



```

```python

# Cota inicial de armazenamento (em GB)
cota = 100

# Função lambda para alterar a cota
alterarCota = lambda cota_atual, aumento: cota_atual + aumento

# Testando a função
nova_cota = alterarCota(cota, 50)

print("Cota inicial:", cota, "GB")
print("Nova cota:", nova_cota, "GB")
```
