```python
pacotes = [12, 0, 25, 40, -1, 18, 50]

for pacote in pacotes:
    if pacote == 0:
        continue  # ignora pacotes vazios
    if pacote < 0:
        print(f"ALERTA: pacote com tamanho inválido ({pacote} MB). Falha de segurança detectada! Processamento interrompido.")
        break  # interrompe o laço
    print(f"Pacote de {pacote} MB processado com sucesso.")
```

```python
# Etapa 1: Registro do Herói
def registrar_heroi(nome, classe, reino="Terra-média"):
    return f"{nome} é um(a) {classe} do reino de {reino}."

print(registrar_heroi("Aragorn", "Guardião", "Gondor"))
print(registrar_heroi("Frodo", "Portador do Anel"))


# Etapa 2: A Chamada das Armas
def chamada_das_armas(heroi, *itens):
    print(f"Equipamentos de {heroi}:")
    for item in itens:
        print(f"- {item}")

chamada_das_armas("Legolas", "Arco de Lothlórien", "Flechas élficas", "Adaga")


# Etapa 3: Ficha Detalhada do Personagem
def ficha_personagem(nome, **atributos):
    print(f"Ficha de {nome}:")
    for atributo, valor in atributos.items():
        print(f"{atributo}: {valor}")

ficha_personagem("Gimli", forca=80, agilidade=95, resistencia=90)


# Desafio Bônus: A Jornada
perigos = ["Orcs", "Nevoeiro", "Lobos", "Nevoeiro", "Balrog", "Torre de Isengard"]

for perigo in perigos:
    if perigo == "Nevoeiro":
        print("Nevoeiro à frente... ignorando o combate e seguindo viagem.")
        continue
    if perigo == "Balrog":
        print("Gandalf grita: 'Fujam, seus tolos!' Jornada interrompida!")
        break
    print(f"Enfrentando o perigo: {perigo}!")
```

```python
# Etapa 1: Registro de Servidor (global e Docstring)
servidores_online = 0  # variável global

def atualizar_servidores(quantidade):
    """
    Atualiza o total de servidores online.

    Parâmetros:
        quantidade (int): número de servidores a ativar (positivo)
                          ou desativar (negativo).

    Impacto na variável global:
        Modifica diretamente 'servidores_online', somando a quantidade
        recebida ao total atual.
    """
    global servidores_online
    servidores_online += quantidade

# 5 operações: +1 ativa um servidor, -1 desativa um servidor
operacoes = [1, 1, 1, -1, 1]

for op in operacoes:
    atualizar_servidores(op)
    acao = "ativado" if op > 0 else "desativado"
    print(f"Servidor {acao}. Total online: {servidores_online}")

print(f"Saldo de servidores ativos: {servidores_online}")


# Etapa 2: Configuração de conexão ao banco de dados (função aninhada)
def gerenciar_conexoes():
    limite_conexoes = 10  # variável local da função externa

    def abrir_conexao():
        nonlocal limite_conexoes
        limite_conexoes -= 1
        print(f"Conexão aberta. Limite restante: {limite_conexoes}")

    abrir_conexao()
    abrir_conexao()

    print(f"Limite de conexões restantes: {limite_conexoes}")

gerenciar_conexoes()


# Etapa 3: Ajuste temporário de cota (lambda)
alterar_cota = lambda cota_atual, ajuste: cota_atual + ajuste

cota = 500  # GB
cota = alterar_cota(cota, 200)
print(f"Cota aumentada: {cota} GB")

cota = alterar_cota(cota, -150)
print(f"Cota reduzida: {cota} GB")
```

```python
DIGITOS = "0123456789ABCDEF"  # a posição da letra no texto é o valor do dígito

def para_binario(n):
    """Converte um decimal em binário (texto), de forma recursiva."""
    if n < 2:                      # caso base: 0 ou 1
        return DIGITOS[n]
    return para_binario(n // 2) + DIGITOS[n % 2]

def para_hexadecimal(n):
    """Converte um decimal em hexadecimal (texto), de forma recursiva."""
    if n < 16:                     # caso base: um único dígito hexadecimal
        return DIGITOS[n]
    return para_hexadecimal(n // 16) + DIGITOS[n % 16]

numero = int(input("Digite um número decimal: "))

if numero < 0:
    print("Digite um número maior ou igual a zero.")
else:
    print(f"Decimal:      {numero}")
    print(f"Binário:      {para_binario(numero)}")
    print(f"Hexadecimal:  {para_hexadecimal(numero)}")
```
