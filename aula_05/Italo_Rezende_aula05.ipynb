```python
def registrar_heroi(nome,classe, reino="Reino desconhecido"):
  return f"{nome}, da classe {classe}, vem do reino: {reino}."

print(registrar_heroi("Jão", "Saqueador", "Cwb"))
print(registrar_heroi("Cleiton", "Guerreiro"))


def chamada_das_armas(nome_heroi, *itens):
  print(f"Equipamentos de {nome_heroi}:")
  for item in itens:
    print(f"- {item}")

chamada_das_armas("The cuiudo", "Faca", "9mm","Katana")

def ficha_personagem(nome, **atributos):
  print(f"Ficha de {nome}:")
  for atributo, valor in atributos.items():
    print(f"{atributo}: {valor}")

ficha_personagem("Zézim", forca=90, velocidade=55, stamina=10)


perigos = ["Orcs", "Nevoeiro", "Aranhas", "Balrog", "Nazgul"]

print("\nA jornada começa!")

for perigo in perigos:
    if perigo == "Nevoeiro":
        print("Nevoeiro encontrado. A Comitiva segue viagem sem combater.")
        continue

    if perigo == "Balrog":
        print("Balrog encontrado! Gandalf grita: Fujam, seus tolos!")
        break

    print(f"A Comitiva enfrentou: {perigo}")

print("Fim da jornada.")
```

Atividade 03 - Aula 05

```python

servidores_online = 10


def alterar_servidores(quantidade):
    """
    Altera a quantidade de servidores online.

    Parâmetro:
    quantidade (int): número de servidores a ativar (positivo)
    ou desativar (negativo).

    Impacto:
    Modifica a variável global servidores_online.
    """
    global servidores_online
    servidores_online += quantidade


print("--- Monitoramento de servidores ---")
print("Servidores ativos no início:", servidores_online)

alterar_servidores(-5)
print("Após desativar 5 servidores:", servidores_online)

alterar_servidores(5)
print("Após ativar 5 servidores:", servidores_online)



def gerenciar_conexoes():
    limite_conexoes = 10

    def abrir_conexao():
        nonlocal limite_conexoes

        if limite_conexoes > 0:
            limite_conexoes -= 1
            print("Conexão aberta com sucesso.")
        else:
            print("Limite de conexões atingido.")

    abrir_conexao()
    abrir_conexao()

    print("Conexões restantes:", limite_conexoes)


print("\n--- Banco de Dados ---")
gerenciar_conexoes()


ajustar_cota = lambda cota_atual, valor: cota_atual + valor

cota_armazenamento = 100
cota_armazenamento = ajustar_cota(cota_armazenamento, 50)

print("\n--- Armazenamento em Nuvem ---")
print("Nova cota de armazenamento:", cota_armazenamento, "GB")
```

Atividade 04 - Aula 05

```python

digitos_hexadecimais = "0123456789ABCDEF"


def decimal_para_binario(numero):
    """Converte um número decimal inteiro não negativo para binário."""
    if numero < 2:
        return str(numero)

    return decimal_para_binario(numero // 2) + str(numero % 2)


def decimal_para_hexadecimal(numero):
    """Converte um número decimal inteiro não negativo para hexadecimal."""
    if numero < 16:
        return digitos_hexadecimais[numero]

    return decimal_para_hexadecimal(numero // 16) + digitos_hexadecimais[numero % 16]


numero_decimal = int(input("Digite um número decimal inteiro: "))

if numero_decimal < 0:
    print("Digite apenas um número inteiro positivo ou zero.")
else:
    binario = decimal_para_binario(numero_decimal)
    hexadecimal = decimal_para_hexadecimal(numero_decimal)

    print("\n--- Resultado da conversão ---")
    print("Número decimal:", numero_decimal)
    print("Número binário:", binario)
    print("Número hexadecimal:", hexadecimal)
```
