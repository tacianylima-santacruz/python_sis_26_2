# Atividade Prática 01 Aula 05
Sistema de Monitoramento de Tráfego de Rede
Objetivo: Você está desenvolvendo um script para analisar pacotes de dados que transitam por um servidor da empresa. O sistema recebe uma lista com o tamanho dos pacotes (em Megabytes) e precisa processá-los individualmente seguindo algumas regras de segurança.
* Dados de Entrada:
Utilize a seguinte lista de pacotes recebidos:
pacotes = [12, 0, 25, 40, -1, 18, 50]
* Escreva um programa usando um laço para percorrer a lista de pacotes e aplique as seguintes regras:
* Ignore pacotes com tamanho igual a 0, pois eles não contêm informação relevante.
* Se o sistema encontrar um pacote com valor menor que zero (ex: -1), exiba uma mensagem de alerta e interrompa o processamento (pois indica uma falha de segurança).
* Para os demais pacotes, exiba a mensagem: "Pacote de [X] MB processado com sucesso."

```python
pacotes = [12, 0, 25, 40, -1, 18, 50]

for pacote in pacotes:
    if pacote == 0:
        continue
    if pacote < 0:
        print("Alerta: Falha de segurança detectada! Processamento interrompido.")
        break
    print(f"Pacote de {pacote} MB processado com sucesso.")
```

Crie um laço que percorra uma lista de perigos no caminho para Mordor:

```python
def registrar_heroi(nome, classe, reino="Desconhecido"):
    return f"Herói: {nome} | Classe: {classe} | Reino: {reino}"

print(registrar_heroi("Aragorn", "Guerreiro", "Gondor"))
print(registrar_heroi("Legolas", "Arqueiro"))

print("-" * 30)

def chamar_armas(nome_heroi, *itens):
    print(f"Herói: {nome_heroi}")
    for item in itens:
        print(f"- {item}")

chamar_armas("Gimli", "Machado de Batalha", "Elmo de Ferro", "Cota de Malha")

print("-" * 30)

def ficha_detalhada(nome, **atributos):
    print(f"Personagem: {nome}")
    for atributo, valor in atributos.items():
        print(f"{atributo.capitalize()}: {valor}")

ficha_detalhada("Frodo", forca=40, agilidade=85, determinacao=100)

print("-" * 30)

perigos = ["Lobo", "Nevoeiro", "Orque", "Balrog", "Aranha"]
for perigo in perigos:
    if perigo == "Nevoeiro":
        continue
    if perigo == "Balrog":
        print("Gandalf avisa: Fujam, seus tolos! Jornada interrompida!")
        break
    print(f"Enfrentando o perigo: {perigo}!")
```

```python

servidores_ativos = 10

def alterar_servidores(quantidade):
    """
    Altera o número total de servidores online.

    Parâmetros:
    quantidade (int): Número de servidores a serem adicionados (positivo) ou removidos (negativo).

    Modifica a variável global 'servidores_ativos'.
    """
    global servidores_ativos
    servidores_ativos += quantidade

alterar_servidores(-5)
print(f"Saldo de servidores ativos: {servidores_ativos}")

print("-" * 30)

def gerenciar_banco():
    conexoes_restantes = 5

    def usar_conexao():
        nonlocal conexoes_restantes
        conexoes_restantes -= 1

    usar_conexao()
    usar_conexao()
    print(f"Limite de conexões restantes: {conexoes_restantes}")

gerenciar_banco()

print("-" * 30)

alterar_cota = lambda cota_atual, multiplicador: cota_atual * multiplicador
nova_cota = alterar_cota(100, 1.5)
print(f"Nova cota de armazenamento: {nova_cota} GB")
```

# Atividade Prática 04 Aula 05
Tradutor de Máquinas (Binário e Hexadecimal)
Na área de TI, os computadores processam dados em Binário (0 e 1), mas para facilitar a leitura humana de grandes blocos de dados (como endereços de memória, códigos de cores ou endereços físicos MAC das placas de rede), usamos o sistema Hexadecimal (Base 16, que vai de 0 a 9 e depois de A a F).
Objetivo: Criar um programa que receba um único número decimal digitado pelo usuário e mostre suas versões em Binário e Hexadecimal, usando apenas funções recursivas e concatenação (junção) de textos. Nenhuma lista é permitida!
A Função Binária: Crie uma função recursiva
Dica: Como o hexadecimal usa letras, crie uma variável de texto,  você pode usar a posição da letra.

```python
def decimal_para_binario(numero):
    if numero == 0:
        return "0"
    if numero == 1:
        return "1"
    return decimal_para_binario(numero // 2) + str(numero % 2)

def decimal_para_hexadecimal(numero):
    simbolos = "0123456789ABCDEF"
    if numero < 16:
        return simbolos[numero]
    return decimal_para_hexadecimal(numero // 16) + simbolos[numero % 16]

numero_digitado = int(input("Digite um número decimal: "))

if numero_digitado == 0:
    binario_resultado = "0"
    hexadecimal_resultado = "0"
else:
    binario_resultado = decimal_para_binario(numero_digitado)
    hexadecimal_resultado = decimal_para_hexadecimal(numero_digitado)

print(f"Binário: {binario_resultado}")
print(f"Hexadecimal: {hexadecimal_resultado}")
```
