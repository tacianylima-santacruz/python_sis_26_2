```python
# Atividade Prática 01 - Aula 05
# Sistema de Monitoramento de Tráfego de Rede

pacotes = [12, 0, 25, 40, -1, 18, 50]

for pacote in pacotes:
    if pacote == 0:
        continue  # ignora pacotes vazios

    if pacote < 0:
        print(f"🚨 ALERTA! Pacote inválido detectado ({pacote} MB). Falha de segurança! Processamento interrompido.")
        break  # para o processamento imediatamente

    print(f"Pacote de {pacote} MB processado com sucesso.")
```

```python
# Atividade Prática 02 - Aula 05
# O Sistema da Comitiva do Anel

# --- Etapa 1: Registro do Herói (parâmetro com valor padrão + return) ---
def registrar_heroi(nome, classe, reino="uma terra desconhecida"):
    return f"{nome} é um(a) {classe} de {reino}."


print(registrar_heroi("Aragorn", "Guerreiro", "Gondor"))  # com reino
print(registrar_heroi("Legolas", "Arqueiro"))  # sem reino (usa o valor padrão)

# --- Etapa 2: A Chamada das Armas (*args) ---
def chamada_das_armas(nome_heroi, *itens):
    print(f"\nItens entregues a {nome_heroi}:")
    for item in itens:
        print(f"- {item}")


chamada_das_armas("Frodo", "Espada Curta", "Cota de Malha", "Capa Élfica")

# --- Etapa 3: Ficha Detalhada do Personagem (**kwargs) ---
def ficha_detalhada(nome, **atributos):
    print(f"\nFicha de combate de {nome}:")
    for atributo, valor in atributos.items():
        print(f"{atributo}: {valor}")


ficha_detalhada("Gimli", forca=80, agilidade=45, resistencia=90)

# --- Desafio Bônus: A Jornada (continue e break) ---
perigos = ["Orcs", "Nevoeiro", "Aranha Gigante", "Balrog", "Orcs"]

print("\n===== A JORNADA PARA MORDOR =====")
for perigo in perigos:
    if perigo == "Nevoeiro":
        print("🌫️ Nevoeiro à frente... a Comitiva ignora e segue viagem.")
        continue

    if perigo == "Balrog":
        print("🔥 UM BALROG! Gandalf grita: 'Fujam, insensatos!' A jornada é interrompida.")
        break

    print(f"⚔️ A Comitiva enfrenta: {perigo}")
```

```python
# Atividade Prática 03 - Aula 05
# Monitoramento de TI

# --- Etapa 1: Registro de Servidor (global + Docstring) ---
servidores_online = 20  # variável global


def alterar_servidores(quantidade):
    """
    Altera a quantidade de servidores online.

    Parâmetro:
        quantidade (int): valor a ser somado ao total de servidores.
                           Use um número negativo para desativar servidores
                           e um número positivo para ativar novos servidores.

    Impacto: modifica diretamente a variável global 'servidores_online'.
    """
    global servidores_online
    servidores_online += quantidade


print(f"Servidores online antes: {servidores_online}")
alterar_servidores(-5)  # desativa 5 servidores
print(f"Servidores online depois: {servidores_online}")

# --- Etapa 2: Configuração de Conexão ao Banco de Dados (função aninhada + nonlocal) ---
def gerenciar_conexoes():
    limite_conexoes = 100  # variável local da função externa

    def abrir_conexao():
        nonlocal limite_conexoes
        limite_conexoes -= 1
        print(f"Conexão aberta. Limite restante: {limite_conexoes}")

    abrir_conexao()
    abrir_conexao()


print("\n===== CONEXÕES AO BANCO DE DADOS =====")
gerenciar_conexoes()

# --- Etapa 3: Ajuste Temporário de Cota (função lambda) ---
ajustar_cota = lambda cota_atual, valor: cota_atual + valor

cota_inicial = 500  # GB
nova_cota = ajustar_cota(cota_inicial, 250)

print(f"\nCota inicial: {cota_inicial} GB")
print(f"Nova cota após ajuste: {nova_cota} GB")
```
