### Resolução: Atividade Prática 01
Sistema de Monitoramento de Tráfego de Rede

```python
pacotes = [12, 0, 25, 40, -1, 18, 50]

for pacote in pacotes:
    if pacote == 0:
        continue
    elif pacote < 0:
        print(f"ALERTA: Pacote com tamanho inválido ({pacote} MB) detectado! Interrompendo processamento por segurança.")
        break
    else:
        print(f"Pacote de {pacote} MB processado com sucesso.")
```

### Resolução: Atividade Prática 02
O Sistema da Comitiva do Anel

```python
# Etapa 1: Registro do Herói
def registrar_heroi(nome, classe, reino="Desconhecido"):
    return f"Herói: {nome} | Classe: {classe} | Reino: {reino}"

print(registrar_heroi("Aragorn", "Guerreiro", "Gondor"))
print(registrar_heroi("Legolas", "Arqueiro"))

# Etapa 2: A Chamada das Armas
def chamar_armas(nome_heroi, *itens):
    print(f"\nArmas de {nome_heroi}:")
    for item in itens:
        print(f" - {item}")

chamar_armas("Gimli", "Machado Duplo", "Elmo de Ferro", "Cota de Malha")

# Etapa 3: Ficha Detalhada
def ficha_personagem(nome, **atributos):
    print(f"\nFicha de {nome}:")
    for atr, val in atributos.items():
        print(f"{atr.capitalize()}: {val}")

ficha_personagem("Frodo", forca=40, agilidade=85, furtividade=99)

# Desafio Bônus: A Jornada
perigos = ["Lobo", "Nevoeiro", "Orque", "Balrog", "Troll"]
print("\nIniciando a Jornada...")
for perigo in perigos:
    if perigo == "Nevoeiro":
        print("Nevoeiro detectado: Ignorando combate e seguindo viagem.")
        continue
    elif perigo == "Balrog":
        print("Balrog detectado! Gandalf grita: 'Fujam, seus tolos!' Interrompendo jornada.")
        break
    else:
        print(f"Combatendo e vencendo o perigo: {perigo}")
```

### Resolução: Atividade Prática 03
Monitoramento de TI

```python
# Etapa 1: Registro de Servidor
servidores_ativos = 10

def alterar_servidores(quantidade):
    """
    Altera a quantidade de servidores online utilizando escopo global.
    Parâmetros: quantidade (int) -> valor positivo para ativar, negativo para desativar.
    """
    global servidores_ativos
    servidores_ativos += quantidade

# Desativando/ativando servidores (saldo líquido de -5 servidores)
alterar_servidores(-5)
print(f"Servidores ativos restantes: {servidores_ativos}")

# Etapa 2: Configuração de Conexão Aninhada
def gerenciar_conexoes():
    limite_conexoes = 5

    def usar_conexao():
        nonlocal limite_conexoes
        if limite_conexoes > 0:
            limite_conexoes -= 1

    usar_conexao()
    usar_conexao()
    print(f"Limite de conexões restantes: {limite_conexoes}")

gerenciar_conexoes()

# Etapa 3: Ajuste de Cota (Lambda)
ajustar_cota = lambda cota_atual, multiplicador: cota_atual * multiplicador
nova_cota = ajustar_cota(100, 1.5)
print(f"Nova cota de armazenamento: {nova_cota} GB")
```

### Resolução: Atividade Prática 04
Tradutor de Máquinas (Binário e Hexadecimal Recursivos)

```python
def dec_para_binario(n):
    if n == 0:
        return "0"
    if n == 1:
        return "1"
    return dec_para_binario(n // 2) + str(n % 2)

def dec_para_hexadecimal(n):
    caracteres_hex = "0123456789ABCDEF"
    if n < 16:
        return caracteres_hex[n]
    return dec_para_hexadecimal(n // 16) + caracteres_hex[n % 16]

# Teste com número digitado ou fixo
numero = 254
print(f"Número Decimal: {numero}")
print(f"Em Binário: {dec_para_binario(numero)}")
print(f"Em Hexadecimal: {dec_para_hexadecimal(numero)}")
```
