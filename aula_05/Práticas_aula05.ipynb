# Atividade Prática 01 Aula 05
Sistema de Monitoramento de Tráfego de Rede
Objetivo: Você está desenvolvendo um script para analisar pacotes de dados que transitam por um servidor da empresa. O sistema recebe uma lista com o tamanho dos pacotes (em Megabytes) e precisa processá-los individualmente seguindo algumas regras de segurança.
* Dados de Entrada:
Utilize a seguinte lista de pacotes recebidos:
pacotes = [12, 0, 25, 40, -1, 18, 50]
* Escreva um programa usando um laço para percorrer a lista de pacotes e aplique as seguintes regras:
* Ignore pacotes com tamanho igual a 0, pois eles não contêm informação relevante.
* Se o sistema encontrar um pacote com valor menor que zero (ex: -1), exiba uma mensagem de alerta e interrompa o processamento (pois indica uma falha de segurança).
* Para os demais pacotes, exiba a mensagem: "Pacote de [X] MB processado com sucesso."

```python
#prática01

pacotes = [12, 0, 25, 40, -1, 18, 50]

print("Iniciando o processamento de pacotes")

for tamanho_pacote in pacotes:
    if tamanho_pacote == 0:
        print(f"Pacote de {tamanho_pacote} MB ignorado (não há informação relevante).")
        continue
    elif tamanho_pacote < 0:
        print(f"FALHA DE SEGURANÇA: Pacote inválido detectado. Processamento interrompido.")
        break
    else:
        print(f"Pacote de {tamanho_pacote} MB processado com sucesso.")

print("Processamento de pacotes concluído.")
```

* Etapa 3: Ajuste Temporário de Cota (Função lambda)
Crie uma função com lambda para alterar a cota de armazenamento de arquivos em nuvem no servidor.
Teste: chame a função alterando a cota.


```python
#prática03
#variável global controlando o total de servidores online
total_servidores_online = 0

def gerenciar_servidores(acao, quantidade):

    global total_servidores_online
    if acao == 'ativar':
        total_servidores_online += quantidade
        print(f"{quantidade} servidores ativados.")
    elif acao == 'desativar':
        if total_servidores_online - quantidade >= 0:
            total_servidores_online -= quantidade
            print(f"{quantidade} servidores desativados.")
        else:
            print("Não é possível deativar mais servidores do que o total online.")
    else:
        print("Ação inválida. Use 'ativar' ou 'desativar'.")
    print(f"Saldo atual de servidores online: {total_servidores_online}")

#teste desatiando servidores
print("Gerenciamento de Servidores")
gerenciar_servidores('ativar', 3)
gerenciar_servidores('desativar', 1)
gerenciar_servidores('ativar', 5)
gerenciar_servidores('desativar', 4)
gerenciar_servidores('desativar', 5)
```

```python
#prática03
def gerenciar_conexoes_db(limite_inicial):
    limite_conexoes = limite_inicial # Variável local da função externa

    def consumir_conexao():
        nonlocal limite_conexoes # Declara que limite_conexoes não é local, mas também não global
        if limite_conexoes > 0:
            limite_conexoes -= 1
            print(f"Conexão utilizada. Limite de conexões restantes: {limite_conexoes}")
        else:
            print("Limite de conexões atingido! Não é possível estabelecer mais conexões.")

    return consumir_conexao


print("Gerenciamento de Conexões DB")

conexao_db_gerenciador = gerenciar_conexoes_db(3)

conexao_db_gerenciador() #Chama a função interna pela primeira vez
conexao_db_gerenciador() #Chama a função interna pela segunda vez
conexao_db_gerenciador() #Chama a função interna pela terceira vez
conexao_db_gerenciador() #Tenta chamar a função interna novamente
```

```python
#prática03
def criar_ajustador_cota(cota_base_mb):
    ajustador_cota = lambda multiplicador: cota_base_mb * multiplicador
    return ajustador_cota

print("Ajuste Temporário de Cota")

ajustador_cota2 = criar_ajustador_cota(100)

cota_atualizada_1 = ajustador_cota2(1.5)
print(f"Cota de armazenamento atualizada para 1.5x: {cota_atualizada_1} MB")

cota_atualizada_2 = ajustador_cota2(2.0)
print(f"Cota de armazenamento atualizada para 2.0x: {cota_atualizada_2} MB")

cota_atualizada_3 = ajustador_cota2(0.8)
print(f"Cota de armazenamento atualizada para 0.8x: {cota_atualizada_3} MB")
```

# Atividade Prática 04 Aula 05
Tradutor de Máquinas (Binário e Hexadecimal)
Na área de TI, os computadores processam dados em Binário (0 e 1), mas para facilitar a leitura humana de grandes blocos de dados (como endereços de memória, códigos de cores ou endereços físicos MAC das placas de rede), usamos o sistema Hexadecimal (Base 16, que vai de 0 a 9 e depois de A a F).
Objetivo: Criar um programa que receba um único número decimal digitado pelo usuário e mostre suas versões em Binário e Hexadecimal, usando apenas funções recursivas e concatenação (junção) de textos. Nenhuma lista é permitida!
A Função Binária: Crie uma função recursiva
Dica: Como o hexadecimal usa letras, crie uma variável de texto,  você pode usar a posição da letra.

```python
from ast import arguments
def decimal_para_binario(n):
    # Caso base
    if n == 0:
        return "0"
    elif n == 1:
        return "1"
    # Chamada recursiva concatenando o resto da divisão por 2 ao final
    return decimal_para_binario(n // 2) + str(n % 2)

def decimal_para_hexadecimal(n):
    caracteres_hex = "0123456789ABCDEF"
    # Caso base
    if n < 16:
        return caracteres_hex[n]

    return decimal_para_hexadecimal(n // 16) + caracteres_hex[n % 16]

numero_decimal = int(input("Digite um número decimal: "))

print(f"Número decimal: {numero_decimal}")

# Tratamento para o caso do número zero
if numero_decimal == 0:
    binario = "0"
    hexadecimal = "0"
elif numero_decimal > 0:
    binario = decimal_para_binario(numero_decimal)
    hexadecimal = decimal_para_hexadecimal(numero_decimal)

print(f"Binário: {binario}")
print(f"Hexadecimal: {hexadecimal}")
```
