Atividade Prática 01 Aula 05
Sistema de Monitoramento de Tráfego de Rede Objetivo: Você está desenvolvendo um script para analisar pacotes de dados que transitam por um servidor da empresa. O sistema recebe uma lista com o tamanho dos pacotes (em Megabytes) e precisa processá-los individualmente seguindo algumas regras de segurança.

Dados de Entrada: Utilize a seguinte lista de pacotes recebidos: pacotes = [12, 0, 25, 40, -1, 18, 50]
Escreva um programa usando um laço para percorrer a lista de pacotes e aplique as seguintes regras:
Ignore pacotes com tamanho igual a 0, pois eles não contêm informação relevante.
Se o sistema encontrar um pacote com valor menor que zero (ex: -1), exiba uma mensagem de alerta e interrompa o processamento (pois indica uma falha de segurança).
Para os demais pacotes, exiba a mensagem: "Pacote de [X] MB processado com sucesso."

```python
pacotes = [12, 0, 25, 40, -1, 18, 50]

for i in pacotes:
  if i == 0:
    continue
  elif i < 0:
    print(f"Mensagem de segurança? {i}")
    break
  else:
    print(f"Pacote de [{i}] MB processado com sucesso.")
```

O Sistema da Comitiva do Anel Objetivo: Criar um script para gerenciar os membros da Comitiva e preparar seus equipamentos utilizando funções.

Etapa 1: Registro do Herói (Parâmetros com valor padrão e return): O Crie uma função que receba o nome e a classe do personagem e, opcionalmente, o seu reino. A função deve retornar um texto formatado Teste: invoque a função com e sem o parâmetro reino.

Etapa 2: A Chamada das Armas: Crie uma função que receba o nome do herói e uma quantidade indeterminada de itens. A função deve imprimir o nome do herói e listar cada item recebido com um marcador (-). Teste: chame a função passando 3 itens.

Etapa 3: Ficha Detalhada do Personagem: Crie uma função para registrar estatísticas de combate dinâmicas. A função deve exibir o nome do personagem e percorrer todos os atributos passados (ex: forca=80, agilidade=95) imprimindo o nome do atributo e o seu valor. Teste: chame a função usando argumentos nomeados.

Desafio Bônus (Controle de Fluxo): A Jornada: Crie um laço que percorra uma lista de perigos no caminho para Mordor: Se o perigo for "Nevoeiro": ignorar o combate e seguir viagem. Se o perigo for "Balrog": interromper a jornada imediatamente (Gandalf avisa para fugir!).

```python
# Atividade Prática 02 - Sistema da Comitiva do Anel

# Etapa 1 - Registro do Herói

def registrar_heroi(nome, classe, reino="Reino Desconhecido"):
    return f"Nome: {nome} | Classe: {classe} | Reino: {reino}"


print(registrar_heroi("Aragorn", "Guerreiro", "Gondor"))
print(registrar_heroi("Legolas", "Arqueiro"))


# Etapa 2 - A Chamada das Armas

def armas(heroi, *itens):
    print(f"\nEquipamentos de {heroi}:")

    for item in itens:
        print(f"- {item}")


armas("Aragorn", "Espada", "Arco", "Escudo")


# Etapa 3 - Ficha Detalhada

def ficha(nome, **atributos):
    print(f"\nFicha de {nome}:")

    for atributo, valor in atributos.items():
        print(f"{atributo}: {valor}")


ficha("Legolas", forca=80, agilidade=95, resistencia=70)


# Desafio Bônus - A Jornada

perigos = ["Orcs", "Nevoeiro", "Aranhas", "sem ideia de perigo", "Monstro da CLT"]

for perigo in perigos:

    if perigo == "Nevoeiro":
        print("Nevoeiro encontrado. Seguindo viagem.")

    elif perigo == "sem ideia de periogo":
        print("sem ideia de periogo encontrado! Gandalf avisa para fugir!")
        break

    else:
        print(f"{perigo} encontrado. Preparar para o combate.")
```

Monitoramento de TI Objetivo: Criar um script para registrar alertas, aplicar configurações dinâmicas e processar chamados da infraestrutura.

Etapa 1: Registro de Servidor (global e Docstring) Crie uma variável global e uma função que use a palavra-chave global para controlar o total de servidores online. Adicione uma Docstring explicando a função, parâmetros e o impacto na variável global. Desative/ative 5 servidores e imprima o saldo de servidores ativos.
Etapa 2: Configuração de Conexao ao Banco de Dados Aninhada Crie uma função para gerenciar conexoes e dentro dela, criar uma variável local. Crie uma função interna que use nonlocal para alterar a quantidade de conexoes. Execução: Chame a função interna 2 vezes e exiba o limite de conexões restantes.
Etapa 3: Ajuste Temporário de Cota (Função lambda) Crie uma função com lambda para alterar a cota de armazenamento de arquivos em nuvem no servidor. Teste: chame a função alterando a cota.

```python
# Atividade Prática 03 - Monitoramento de TI


# Etapa 1 - Registro de Servidor

servidores_online = 0


def alterar_servidores(quantidade):
    """
    Altera a quantidade de servidores online.
    A variável global servidores_online é modificada.
    """
    global servidores_online
    servidores_online = servidores_online + quantidade


alterar_servidores(5)
alterar_servidores(-5)

print(f"Servidores ativos: {servidores_online}")


# Etapa 2 - Conexões com Banco de Dados

def gerenciar_conexoes():
    conexoes = 5

    def usar_conexao():
        nonlocal conexoes
        conexoes = conexoes - 1

    usar_conexao()
    usar_conexao()

    print(f"Conexões restantes: {conexoes}")


gerenciar_conexoes()


# Etapa 3 - Ajuste Temporário de Cota

def alterar_cota(cota):
    nova_cota = lambda valor: cota + valor
    return nova_cota(50)


cota_atual = 100
cota_atual = alterar_cota(cota_atual)

print(f"Cota atual: {cota_atual} GB")
```

Tradutor de Máquinas (Binário e Hexadecimal) Na área de TI, os computadores processam dados em Binário (0 e 1), mas para facilitar a leitura humana de grandes blocos de dados (como endereços de memória, códigos de cores ou endereços físicos MAC das placas de rede), usamos o sistema Hexadecimal (Base 16, que vai de 0 a 9 e depois de A a F). Objetivo: Criar um programa que receba um único número decimal digitado pelo usuário e mostre suas versões em Binário e Hexadecimal, usando apenas funções recursivas e concatenação (junção) de textos. Nenhuma lista é permitida! A Função Binária: Crie uma função recursiva Dica: Como o hexadecimal usa letras, crie uma variável de texto, você pode usar a posição da letra.

```python
# Atividade Prática 04 - Tradutor de Máquinas

numero = int(input("Digite um número decimal: "))


# Função para converter para binário
def binario(numero):
    if numero == 0:
        return "0"

    if numero == 1:
        return "1"

    return binario(numero // 2) + str(numero % 2)


# Função para converter para hexadecimal
def hexadecimal(numero):
    caracteres = "0123456789ABCDEF"

    if numero < 16:
        return caracteres[numero]

    return hexadecimal(numero // 16) + caracteres[numero % 16]


print(f"Binário: {binario(numero)}")
print(f"Hexadecimal: {hexadecimal(numero)}")
```
