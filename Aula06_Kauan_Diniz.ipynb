```python
#ATIVIDADE 01 AULA 06

animais = ["Leão", "Águia", "Jacaré", "gato", "Lhama"]
favoritos = ["Leão", "Jacaré", "Lhama"]

print("Animais favoritos")
for animal in favoritos:
    print(animal)

print(f"\nTotal de animais: {len(animais)}")
print(f"Total de favoritos: {len(favoritos)}")

print(f"Primeiro animal favorito: {favoritos[0]}")
print(f"Último animal favorito: {favoritos[-1]}")

animais.remove("Águia")
print(f"\nLista após remover 'Águia': {animais}")

animais.append("Macaco")
print(f"\nLista após adicionar 'Macaco': {animais}")


favoritos.reverse()
print(f"\nFavoritos na nova ordem: {favoritos}")
```

```python
#ATIVIDADE 03 AULA 06

contatos = {"Kauan": "9999-1111", "KKauann": "9888-2222"}
print(f"Contatos originais: {contatos}")

contatos_atualizados = {**contatos, "KKKauannnn": "9777-3333"}
print(f"Após adicionar KKKauannnn: {contatos_atualizados}")

if "Chai" not in contatos_atualizados:
    contatos_atualizados["Chai"] = "9666-4444"
print(f"Após verificar/adicionar KKKKauannnn: {contatos_atualizados}")

contatos_atualizados.pop("Kauan", None)
print(f"Após remover Kauan: {contatos_atualizados}")

print(f"Nomes no dicionário: {list(contatos_atualizados.keys())}")
```
