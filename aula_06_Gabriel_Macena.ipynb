```python
# Atividade 01

animais = ["cachorro", "gato", "leão", "tigre", "elefante", "macaco", "coelho", "girafa"]
favoritos = ["cachorro", "gato", "leão"]

print("Animais favoritos:", favoritos)
print("Total de animais:", len(animais))
print("Total de favoritos:", len(favoritos))

print("Primeiro favorito:", favoritos[0])
print("Último favorito:", favoritos[-1])

animais.pop(5)
print("Após remover um animal:", animais)

animais.append(["urso", "panda"])
print("Após adicionar animais:", animais)

print("Total de animais agora:", len(animais))

favoritos.reverse()
print("Favoritos em nova ordem:", favoritos)
```

```python
# Atividade 03

contatos = {
    "João": "99999-1111",
    "Maria": "98888-2222",
    "Pedro": "97777-3333"
}

print("Contatos iniciais:")
print(contatos)

contatos["Ana"] = "96666-4444"

print("\nApós adicionar Ana:")
print(contatos)

if "Chai" not in contatos:
    contatos["Chai"] = "95555-5555"
    print("\nChai não estava nos contatos. Chai foi adicionada:")
else:
    print("\nChai já existe nos contatos:")

print(contatos)



contatos.pop("Pedro", None)

print("\nApós excluir Pedro:")
print(contatos)
print("\nNomes presentes na agenda:")

for nome in contatos:
    print(nome)
```

```python
# Atividade 04

livros = [
    "Duna",
    "O Hobbit",
    "Duna",
    "Harry Potter",
    "O Hobbit",
    "1984",
    "Duna"
]

print("Lista original:")
print(livros)

livros_unicos = set(livros)

print("\nConjunto de livros sem repetições:")
print(livros_unicos)

print("\nQuantidade de livros únicos:", len(livros_unicos))
```

```python
# Atividade Prática 05

biblioteca = {
    "Duna": {
        "autor": "Frank Herbert",
        "ano_publicacao": 1965,
        "generos": ["Ficção Científica", "Aventura"],
        "disponivel": True
    },

    "O Hobbit": {
        "autor": "J.R.R. Tolkien",
        "ano_publicacao": 1937,
        "generos": ["Fantasia", "Aventura"],
        "disponivel": False
    }
}

print("Catálogo da biblioteca:")
print(biblioteca)
print("\nAutor de Duna:")
print(biblioteca["Duna"]["autor"])
print("\nAno de publicação de Duna:")
print(biblioteca["Duna"]["ano_publicacao"])
print("\nGêneros de Duna:")
print(biblioteca["Duna"]["generos"])
print("\nDuna está disponível?")
print(biblioteca["Duna"]["disponivel"])
```
