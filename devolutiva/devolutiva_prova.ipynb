

(      ) O loop while repete um bloco de código enquanto uma certa condição for avaliada como Verdadeira (True).

O loop while (que significa "enquanto" em inglês) é uma estrutura de repetição baseada exatamente nessa regra: ele verifica a condição e, enquanto ela resultar em um valor verdadeiro, o bloco de código interno continuará sendo executado repetidas vezes.

Assim que a condição for avaliada como falsa, o ciclo é quebrado e o programa continua sua execução normal fora do loop.

```python
contador = 1

# O bloco abaixo será repetido enquanto a condição (contador <= 3) for Verdadeira
while contador <= 3:
    print(f"O contador vale: {contador}")

    # É fundamental modificar a variável testada para que uma hora ela se torne Falsa,
    # caso contrário, você criará um "loop infinito".
    contador += 1

print("O loop terminou porque o contador chegou a 4 (a condição se tornou Falsa).")
```

```python
contador = 1

print("Iniciando o loop...\n")

while contador <= 3:
    # Imprime a avaliação da condição ANTES de executar a lógica
    print(f"Comparação atual ({contador} <= 3) -> {contador <= 3}")

    # Lógica do programa
    print(f"Executando ação para o contador {contador}")

    # Atualiza o contador para a próxima iteração
    contador += 1
    print("-" * 30) # Apenas uma linha divisória para organizar a leitura

# Imprimindo a comparação final que fez o loop ser quebrado
print(f"Comparação final ({contador} <= 3) -> {contador <= 3}")
print("O loop foi encerrado!")
```

# Questão 06
Como você especifica múltiplos valores para o mesmo case em uma instrução match?

Com o operador lógico and.



```python
valor = 2

match valor:
    # Isso vai gerar um SyntaxError: invalid syntax
    case 1 and 2:
        print("Valor é 1 ou 2")
```


Com o operador lógico or.



```python
valor = 2

match valor:
    # Isso vai gerar um SyntaxError: invalid syntax
    case 1 or 2:
        print("Valor é 1 ou 2")
```

Assim como o and, o uso da palavra or não é permitido na sintaxe de verificação de padrões do case. Ele também resultará em um erro de sintaxe imediato.



Com o símbolo de barra vertical ou pipe (|).


```python

valor = 2

match valor:
    case 1 | 2 | 3:
        # Como o valor é 2, este bloco será executado com sucesso!
        print("O valor é 1, 2 ou 3.")
    case _:
        print("Outro valor.")

# Saída: O valor é 1, 2 ou 3.
```



Colocando vários cases uns sobre os outros antes do bloco indentado.

O Python não permite isso. Cada case exige obrigatoriamente um bloco de código indentado abaixo dele. Se você tentar empilhar, receberá um erro de indentação.

```python
valor = 2

match valor:
    # Isso vai gerar um IndentationError: expected an indented block
    case 1:
    case 2:
        print("Valor é 1 ou 2")
```

Nota extra: Se você colocasse um pass no case 1, o código rodaria, mas não faria o que você quer. Se o valor fosse 1, ele bateria no case 1, faria o pass e ignoraria o case 2, pois no Python o bloco match é interrompido assim que encontra o primeiro caso verdadeiro

Por que está errada: Argumentos nomeados em uma chamada não mudam a quantidade de variáveis que uma função aceita. Se ela pede 3 parâmetros, você deve passar 3.
O conceito real: O que torna uma função capaz de receber uma quantidade infinita e variada de dados são os parâmetros *args e **kwargs.

```python
# O *args permite que a função aceite qualquer quantidade de números
def somar_tudo(*numeros):
    print(sum(numeros))

somar_tudo(10, 20)          # 2 entradas
somar_tudo(1, 2, 3, 4, 5)   # 5 entradas
```

Exportar variáveis de um arquivo para outro.  

Acessar e modificar variáveis globais dentro de funções.   




```python
pontuacao = 10

def zerar_pontuacao_errado():
    # O Python acha que você quer criar uma NOVA variável LOCAL chamada "pontuacao".
    # A variável global original lá de fora ficará intacta.
    pontuacao = 0
    print(f"Dentro da função: {pontuacao}")

zerar_pontuacao_errado()
print(f"Fora da função: {pontuacao}")

# Saída:
# Dentro da função: 0
# Fora da função: 10 (A global não foi alterada!)
```

```python
pontuacao = 10

def zerar_pontuacao_certo():
    # Avisamos ao Python: "Use a variável pontuacao que já existe lá fora!"
    global pontuacao
    pontuacao = 0  # Agora isso altera o valor real da variável global

zerar_pontuacao_certo()
print(f"Fora da função: {pontuacao}")

# Saída:
# Fora da função: 0 (A global foi modificada com sucesso!)
```
