```python
pacotes = [12, 0, 25, 40, -1, 18, 50]

for pacote in pacotes:
    if pacote == 0:
        continue
    elif pacote < 0:
        print("Alerta: Pacote com valor menor que zero encontrado! Interrompendo o processamento por falha de segurança.")
        break
    else:
        print(f"Pacote de [{pacote}] MB processado com sucesso.")
```

```python
# Etapa 1: Registro do Heroi
def registrar_heroi(nome, classe, reino="Desconhecido"):
    return f"Herói: {nome}, Classe: {classe}, Reino: {reino}"

print(registrar_heroi("Aragorn", "Guerreiro", "Gondor"))
print(registrar_heroi("Frodo", "Ladrão"))

# Etapa 2: A Chamada das Armas
def chamada_armas(nome_heroi, *itens):
    print(f"Herói: {nome_heroi}")
    print("Itens recebidos:")
    for item in itens:
        print(f"- {item}")

chamada_armas("Gandalf", "Cajado", "Chapéu", "Pipa")

# Etapa 3: Ficha Detalhada do Personagem
def ficha_personagem(nome, **estatisticas):
    print(f"Personagem: {nome}")
    for atributo, valor in estatisticas.items():
        print(f"- {atributo}: {valor}")

ficha_personagem("Legolas", forca=80, agilidade=95)

# Desafio Bônus: Controle de Fluxo
perigos = ["Floresta", "Nevoeiro", "Montanha", "Balrog", "Fim"]

for perigo in perigos:
    if perigo == "Nevoeiro":
        print("Nevoeiro encontrado: ignorar o combate e seguir viagem.")
        continue
    elif perigo == "Balrog":
        print("Balrog encontrado: interromper a jornada imediatamente (Gandalf avisa para fugir!).")
        break
    else:
        print(f"Passando por: {perigo}")
```

```python
# Etapa 1: Registro de Servidor
servidores_online = 10

def gerenciar_servidores(quantidade):
    """
    Função para gerenciar servidores online.
    Parâmetros: quantidade (int) - número de servidores a somar ou subtrair.
    Impacto: Modifica a variável global servidores_online.
    """
    global servidores_online
    servidores_online += quantidade
    return servidores_online

gerenciar_servidores(-5) # Desativa 5
gerenciar_servidores(5)  # Ativa 5
print(f"Saldo de servidores ativos: {servidores_online}")

# Etapa 2: Configuração de Conexao ao Banco de Dados Aninhada
def gerenciar_conexoes():
    limite_conexoes = 50

    def alterar_conexao():
        nonlocal limite_conexoes
        limite_conexoes += 10

    alterar_conexao()
    alterar_conexao()
    print(f"Limite de conexões restantes: {limite_conexoes}")

gerenciar_conexoes()

# Etapa 3: Ajuste Temporário de Cota
ajustar_cota = lambda cota_atual, ajuste: cota_atual + ajuste
nova_cota = ajustar_cota(500, 150)
print(f"Nova cota de armazenamento: {nova_cota} GB")
```

```python
def converter_binario(n):
    if n == 0:
        return "0"
    elif n == 1:
        return "1"
    else:
        return converter_binario(n // 2) + str(n % 2)

def converter_hexadecimal(n):
    letras = "0123456789ABCDEF"
    if n < 16:
        return letras[n]
    else:
        return converter_hexadecimal(n // 16) + letras[n % 16]

# Teste do programa
numero_decimal = int(input("Digite um número decimal: "))
print(f"Versão em Binário: {converter_binario(numero_decimal)}")
print(f"Versão em Hexadecimal: {converter_hexadecimal(numero_decimal)}")
```
