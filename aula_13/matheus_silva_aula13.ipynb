### Atividade Prática 01 - Etapa 1: Importar Pandas e Criar/Carregar Dados

```python
import pandas as pd
import numpy as np

# Criando um dataset simples de exemplo (Vendas de Loja de Eletrônicos)
dados = {
    'Produto': ['Smartphone', 'Notebook', 'Tablet', 'Fone de Ouvido', 'Smartwatch', 'Notebook', None],
    'Preco': [2500, 4500, 1200, 150, None, 4500, 800],
    'Quantidade': [10, 5, 8, 50, 15, 5, 3],
    'Categoria': ['Eletrônicos', 'Informática', 'Eletrônicos', 'Acessórios', 'Acessórios', 'Informática', 'Eletrônicos']
}

df = pd.DataFrame(dados)
display(df)
```

### Atividade Prática 01 - Etapa 2: Inspecionar e Tratar Dados (Duplicatas e Nulos)

```python
# 1. Inspeção básica
print("--- Primeiras Linhas ---")
display(df.head())
print("\n--- Informações do DataFrame ---")
df.info()

# 2. Tratamento de duplicatas
df = df.drop_duplicates()

# 3. Tratamento de valores nulos (NaN)
# Substituindo Preço nulo pela média e preenchendo Produto nulo com 'Desconhecido'
df['Preco'] = df['Preco'].fillna(df['Preco'].mean())
df['Produto'] = df['Produto'].fillna('Desconhecido')

print("\n--- Dados Após o Tratamento ---")
display(df)
```

### Atividade Prática 01 - Etapa 3: Filtrar Categoria e Salvar CSV

```python
# Filtrando apenas produtos com preço acima de R$ 500
df_filtrado = df[df['Preco'] > 500]

# Salvando em um novo arquivo CSV limpo
df_filtrado.to_csv('produtos_limpos.csv', index=False)
print("Arquivo 'produtos_limpos.csv' salvo com sucesso!")
display(df_filtrado)
```

### Atividade Prática 02 - Etapas 1, 2 e 3: Carregar Dados, Criar e Customizar o Gráfico

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Etapa 1: Importar a base de dados limpa
df_limpo = pd.read_csv('produtos_limpos.csv')

# Pergunta: Qual o faturamento estimado (Preço x Quantidade) por produto de alto valor?
df_limpo['Faturamento'] = df_limpo['Preco'] * df_limpo['Quantidade']

# Etapas 2 e 3: Criar e profissionalizar o gráfico
plt.figure(figsize=(10, 6))
sns.barplot(x='Produto', y='Faturamento', data=df_limpo, palette='viridis', hue='Produto', legend=False)

# Customização visual
plt.title('Faturamento Estimado por Produto (Acima de R$ 500)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Produtos', fontsize=12)
plt.ylabel('Faturamento Estimado (R$)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
