Exemplo de tratamento: preenchendo valores nulos em uma coluna hipotética 'box_office' com 0

```python
if 'budget' in dataframe_imdb.columns:
    dataframe_imdb['budget'].fillna(0, inplace=True)

# Removendo linhas que tenham o 'genre' (gênero) nulo, pois é essencial para nossa análise
if 'genres' in dataframe_imdb.columns:
    dataframe_imdb.dropna(subset=['genres'], inplace=True)
```

Ordenação

Estatística

```python
# Média (mean) - Qual a duração média dos filmes do Top 250?
media_duracao = dataframe_imdb['runtimeMinutes'].mean()
print(f"Média de duração: {media_duracao:.2f} minutos")

# Mediana (median) - Qual é o valor exato do meio se ordenarmos as durações?
mediana_duracao = dataframe_imdb['runtimeMinutes'].median()
print(f"Mediana de duração: {mediana_duracao} minutos")

# Moda (mode) - Qual foi o ano que mais emplacou filmes no Top 250?
moda_ano = dataframe_imdb['startYear'].mode()[0] # [0] pega o primeiro resultado
print(f"Ano com mais filmes (Moda): {moda_ano}")
```

```python
# Desvio Padrão (std) - O quão dispersas estão as notas dos filmes?
desvio_notas = dataframe_imdb['averageRating'].std()
print(f"Desvio padrão das notas: {desvio_notas:.2f}")
```

```python
# Mínimo (min) - Qual é o ano do filme mais antigo da lista?
ano_mais_antigo = dataframe_imdb['startYear'].min()
print(f"Filme mais antigo lançado em: {ano_mais_antigo}")

# Máximo (max) - Qual foi a maior bilheteria mundial registrada na base?
maior_bilheteria = dataframe_imdb['grossWorldwide'].max()
print(f"Maior bilheteria: $ {maior_bilheteria:,.2f}")

# Soma (sum) - Se somarmos a bilheteria de todos os filmes da base, quanto dá?
soma_bilheterias = dataframe_imdb['grossWorldwide'].sum()
print(f"Soma de todas as bilheterias: $ {soma_bilheterias:,.2f}")
```

```python
# Contagem (count) - Quantos filmes têm a informação de Orçamento (budget) preenchida?
# (A função count ignora valores nulos/NaN)
qtd_orcamentos = dataframe_imdb['budget'].count()
print(f"Filmes com orçamento preenchido: {qtd_orcamentos} de {len(dataframe_imdb)}")

# Quantil (quantile) - Qual nota delimita os 25% melhores filmes da lista (Top 25%)?
# 0.75 representa o 3º quartil (75% dos dados estão abaixo dele)
nota_top_25 = dataframe_imdb['averageRating'].quantile(0.75)
print(f"Para estar entre os 25% melhores do ranking, o filme precisa ter nota acima de: {nota_top_25}")
```

Agrupamento de Dados

```python
# 1. Agrupamos pela coluna 'contentRating' (Classificação Indicativa)
# 2. Selecionamos a coluna 'averageRating' (Nota Média)
# 3. Calculamos a média (mean)
media_por_classificacao = dataframe_imdb.groupby('contentRating')['averageRating'].mean()

# Ordenando os resultados do maior para o menor para facilitar a leitura
media_por_classificacao = media_por_classificacao.sort_values(ascending=False)

print("Média de notas por Classificação Indicativa:")
print(media_por_classificacao)
```

```python
# 1. Agrupamos pela coluna 'startYear' (Ano)
# 2. Selecionamos a coluna 'primaryTitle' (Título) para contar
# 3. Usamos a função de contagem (count)
filmes_por_ano = dataframe_imdb.groupby('startYear')['primaryTitle'].count()

# Ordenando para ver quais anos tiveram MAIS filmes no Top 250
filmes_por_ano_ordenado = filmes_por_ano.sort_values(ascending=False)

print("Top 5 Anos com mais filmes na lista:")
print(filmes_por_ano_ordenado.head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Carrega a base e ordena pelas 5 maiores bilheterias
dataframe_imdb = pd.read_csv("/imdb_top_250_movies.csv")
top_10_bilheteria = dataframe_imdb.sort_values(by='grossWorldwide', ascending=False).head(10)

# Exibe apenas a Tabela
print("--- TABELA DE DADOS ---")
tabela_exibicao = top_10_bilheteria[['primaryTitle', 'grossWorldwide','startYear']]
print(tabela_exibicao)
```

```python
# Definindo o tamanho da figura (Slide 44)
plt.figure(figsize=(8, 6))

# Criando o gráfico de barras (Slide 40)
# Vamos colocar os Títulos no eixo X e as Bilheterias no eixo Y
plt.bar(top_10_bilheteria['primaryTitle'], top_10_bilheteria['grossWorldwide'], color=['#e63946', '#f4a261', '#2a9d8f', '#264653', '#e9c46a'])

# Adicionando Títulos e Eixos (Slide 42)
plt.title('Top 10 Maiores Bilheterias (IMDB Top 250)', fontsize=14)
plt.xlabel('Nome do Filme', fontsize=10)
plt.ylabel('Bilheteria Mundial (em Bilhões de $)', fontsize=10)

# Rotacionando os nomes dos filmes para não ficarem encavalados
plt.xticks(rotation=45, ha='right')

# Adicionando a grade para facilitar a leitura (Slide 44)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Exibindo o gráfico
plt.tight_layout() # Ajusta as margens automaticamente
plt.show()
```

```python
# Conta a quantidade por classificação e plota a pizza mostrando a porcentagem com 1 casa decimal
dataframe_imdb['contentRating'].value_counts().head(5).plot(kind='pie', autopct='%1.1f%%')

plt.title('Proporção de Classificações Indicativas (Top 5)')
plt.ylabel('') # Remove o rótulo lateral para ficar mais estético
plt.show()
```

Dispersão

```python
# Cruza o eixo X (Orçamento) com o eixo Y (Bilheteria)
dataframe_imdb.plot(kind='scatter', x='budget', y='grossWorldwide', alpha=0.5, color='green')

plt.title('Orçamento vs Bilheteria Mundial')
plt.show()
```

Boxplot

```python
# Cria o boxplot da coluna 'runtimeMinutes' agrupado pela classificação 'contentRating'
# Para ficar legível, vamos filtrar apenas 3 categorias principais antes
filmes_filtrados = dataframe_imdb[dataframe_imdb['contentRating'].isin(['R', 'PG', 'PG-13'])]

filmes_filtrados.boxplot(column='runtimeMinutes', by='contentRating', grid=False)

plt.title('Distribuição de Duração por Classificação')
plt.suptitle('') # Remove um título automático extra que o Pandas gera
plt.xlabel('Classificação Indicativa')
plt.ylabel('Duração (Minutos)')
plt.show()
```
