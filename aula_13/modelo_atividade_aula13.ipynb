# Atividade Prática 01 Aula 13

Importação de  Dados e Tratamento

Nesta atividade, você atuará como um Analista de Dados preparando a geração de insights. Sua missão é estruturar uma base de dados bruta:

Requisitos:

*Crie scripts em Python que executem as etapas a seguir:

Etapa 1:

importe a biblioteca pandas e outras que necessitar e carregue  uma base de dados reais de sua escolha.

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/tacianylima-santacruz/python_sis_26_2/main/aula_13/imdb_top_250_movies.csv"

df = pd.read_csv(url)

print("Base importada com sucesso!")
print("Total de registros:", len(df))
print("Total de colunas:", len(df.columns))
```

Etapa 2:

Inspecione a base (use pelo menos head() e info()) e execute um tratamento de dados inconsistentes e preencha ou remova valores ausentes (NaN).

```python
# Visualizar os primeiros registros
display(df.head())

# Verificar informações da base
df.info()

# Verificar valores ausentes
print("Valores ausentes antes do tratamento:")
print(df.isnull().sum())

# Remover registros duplicados
df = df.drop_duplicates()

# Remover colunas totalmente vazias
df = df.dropna(axis=1, how="all")

# Limpar textos
for coluna in df.select_dtypes(include="object").columns:
    df[coluna] = df[coluna].str.strip()
    df[coluna] = df[coluna].replace(
        ["", "NA", "N/A", "null", "None"],
        pd.NA
    )

# Preencher valores numéricos ausentes com a mediana
for coluna in df.select_dtypes(include="number").columns:
    if df[coluna].isnull().any():
        df[coluna] = df[coluna].fillna(df[coluna].median())

# Preencher valores de texto ausentes com a moda
for coluna in df.select_dtypes(include="object").columns:
    if df[coluna].isnull().any():
        moda = df[coluna].mode()
        if not moda.empty:
            df[coluna] = df[coluna].fillna(moda.iloc[0])

print("\nTRATAMENTO CONCLUÍDO!")
print("Valores ausentes restantes:", df.isnull().sum().sum())
print("Total de registros:", len(df))
print("Total de colunas:", len(df.columns))
```

Etapa 3:

Escolha uma categoria de interesse, crie um filtro (ex: apenas itens com valor acima de X, ou apenas do ano Y) e salve este novo DataFrame limpo em um novo arquivo .csv.


```python
# Filtrar filmes com avaliação igual ou superior a 8
df_filtrado = df[df["averageRating"] >= 8.0].copy()

# Salvar arquivo CSV
df_filtrado.to_csv(
    "imdb_filmes_filtrados.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Arquivo salvo com sucesso!")
print("Total de filmes selecionados:", len(df_filtrado))
print("Nome do arquivo: imdb_filmes_filtrados.csv")
```

# Atividade Prática 02 Aula 13

Gráficos de base de dados
Agora que seus dados estão estruturados, é hora de contar histórias com eles. Baseando-se no que vimos sobre boas práticas de visualização.

Requisitos:  
*Crie scripts em Python que executem as etapas a seguir:

Etapa 1:

Importe sua base de dados limpa utilizando o pandas.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("imdb_filmes_filtrados.csv")

print("Base de dados carregada com sucesso!")
print(df.head())
print("Total de filmes:", len(df))
```

Etapa 2:

Pense em uma pergunta que seus dados podem responder, escolha os tipos de gráficos adequados e crie o gráfico.

```python
# Selecionar os 10 filmes com maiores avaliações
top_10 = df.nlargest(10, "averageRating").sort_values(
    "averageRating",
    ascending=True
)

# Criar gráfico de barras
fig, ax = plt.subplots(figsize=(12, 6))

barras = ax.barh(
    top_10["primaryTitle"],
    top_10["averageRating"]
)

print("Gráfico dos 10 filmes mais bem avaliados criado!")
```

Etapa 3:

Profissionalize sua visualização: adicione um título, rótulos para os eixos X e Y, e ajuste as cores para facilitar o entendimento.

```python
# Personalizar as cores das barras
for barra in barras:
    barra.set_color("#2878B5")

# Adicionar título e rótulos
ax.set_title(
    "Os 10 Filmes Mais Bem Avaliados do IMDb",
    fontsize=16,
    fontweight="bold"
)

ax.set_xlabel("Avaliação média")
ax.set_ylabel("Nome do filme")

# Ajustar visualização
ax.set_xlim(0, 10)
ax.grid(axis="x", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```
