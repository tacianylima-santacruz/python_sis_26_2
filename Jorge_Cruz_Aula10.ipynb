```python
#Atividade Sala
#Etapa 1:

import csv
import json

# =====================================================================
# ETAPA 1: Estrutura em Memória (Dicionário único do Cluster)
# =====================================================================

cluster_config = {
    "nome_cluster": "Cluster-Prod-01",
    "versao_so": "Ubunto Server 26.11 BRA",
    "servidores": [
        {
            "servidor": "srv-web-01  (Rayane)",
            "ip": "190.210.10.00",
            "portas": [10, 20, 30]
        },
        {
            "servidor": "srv-web-02  (Maya)",
            "ip": "191.211.11.01",
            "portas": [40, 50, 60]
        },
        {
            "servidor": "srv-web-03  (Apollo)",
            "ip": "192.212.12.02",
            "portas": [70, 80, 90]
        }
    ]
}

# =====================================================================
# ETAPA 2: Exportação JSON (cluster_config.json)
# =====================================================================

with open("cluster_config.json", "w", encoding="UTF=8") as arq_json:
  json.dump(cluster_config, arq_json, indent=4, ensure_ascii=False)
print("Arquivo 'clustera_config.json' exportado com sucesso.")

# =====================================================================
# ETAPA 3: Exportação CSV (servidores_cluster.csv)
# =====================================================================

with open("servidores_sluster.csv", "w", newline="", encoding="UTF=8") as arq_csv:
  writer = csv.writer(arq_csv, delimiter=";")

  # Grava o cabeçalho exigido no enunciado
  writer.writerow(["Servidor", "IP", "Portas"])

  # Percorre a lista de servidores de dentro do dicionário

  for srv in cluster_config["servidores"]:
    # Converte a lista de números [10, 20, 30] para texto "10, 20, 30"
    portas_formatadas = " ;  ".join(str(p) for p in srv["portas"])

    writer.writerow([srv["servidor"], srv["ip"], portas_formatadas])

print("Arquivo 'servidores_cluster.csv' exportado com sucesso.")

```

```python

#Etapa 2

import openpyxl
from openpyxl.styles import Font, PatternFill

# =====================================================================
# ETAPA 2: Crie uma planilha (Workbook, Aba e Cabeçalho)
# =====================================================================
wb = openpyxl.Workbook()
ws = wb.active
ws.title = "Servidores"

cabecalho = ["Hostname", "IP", "Memória (GB)", "Uso de Disco (%)", "Status"]
ws.append(cabecalho)

# Estilizando o cabeçalho em negrito
for celula in ws[1]:
    celula.font = Font(bold=True)

# =====================================================================
# ETAPA 3: Inserção de dados (Pelo menos 4 servidores simulados)
# =====================================================================
servidores = [
    ["srv-web-01", "192.168.10.11", 16, 45, "Ativo"],
    ["srv-db-01", "192.168.10.12", 64, 88, "Ativo"],
    ["srv-cache-01", "192.168.10.13", 32, 62, "Ativo"],
    ["srv-backup-01", "192.168.10.14", 32, 94, "Alerta"],
    ["srv-dev-01", "192.168.10.15", 8, 25, "Inativo"],
]

for srv in servidores:
    ws.append(srv)

# =====================================================================
# ETAPA 4: Avisos (Destaque visual + Alerta no terminal + Salvar)
# =====================================================================
preenchimento_alerta = PatternFill(
    start_color="FFCCCC", end_color="FFCCCC", fill_type="solid"
)
fonte_alerta = Font(color="9C0006", bold=True)

print("--- Auditoria de Armazenamento ---")

# Percorre da linha 2 até a última linha preenchida (pulando o cabeçalho)
for linha in range(2, ws.max_row + 1):
    hostname = ws.cell(row=linha, column=1).value
    celula_disco = ws.cell(row=linha, column=4)
    uso_disco = celula_disco.value

    if uso_disco >= 80:
        # Pinta a célula de vermelho claro e deixa o texto vermelho escuro
        celula_disco.fill = preenchimento_alerta
        celula_disco.font = fonte_alerta

        # Exibe o alerta no terminal
        print(
            f"ALERTA: O servidor '{hostname}' está com {uso_disco}% de disco "
            "e precisa de expansão de armazenamento!"
        )

# Salva o arquivo Excel na pasta do projeto
nome_arquivo = "inventario_servidores.xlsx"
wb.save(nome_arquivo)
print(f"\nPlanilha '{nome_arquivo}' gerada e salva com sucesso!")
```

```python
#Etapa 3

import openpyxl
from openpyxl.styles import Font

# ---------------------------------------------------------------------
# OPCIONAL: Cria um arquivo log.txt de exemplo caso ele não exista na pasta
# (Se você já subiu o log.txt da aula anterior no Colab, pode apagar ou comentar este bloco)
# ---------------------------------------------------------------------
conteudo_exemplo = """2026-11-24 08:00:10 - INFO - Sistema inicializado com sucesso
2026-11-24 08:15:42 - INFO - Usuario admin realizou login
2026-11-24 09:05:19 - AVISO - Uso de memoria acima de 75%
2026-11-24 10:30:00 - ERRO - Falha de conexao com o banco de dados
2026-11-24 10:32:15 - INFO - Reconexao estabelecida
"""
with open("log.txt", "w", encoding="utf-8") as f:
    f.write(conteudo_exemplo)

# =====================================================================
# ETAPA 3: Criação da Planilha e Cabeçalho
# =====================================================================
wb = openpyxl.Workbook()
ws = wb.active
ws.title = "Auditoria de Logs"

cabecalho = ["Data/Hora", "Tipo de Evento", "Mensagem"]
ws.append(cabecalho)

# Deixa o cabeçalho em negrito
for celula in ws[1]:
    celula.font = Font(bold=True)

# =====================================================================
# ETAPAS 1, 2 e 4: Leitura segura, Extração de dados e Salvamento
# =====================================================================
try:
    with open("log.txt", "r", encoding="utf-8") as arquivo_log:
        for linha in arquivo_log:
            linha_limpa = linha.strip()

            # Ignora linhas em branco no arquivo
            if not linha_limpa:
                continue

            # Divide a linha em 3 partes usando " - " como separador
            partes = linha_limpa.split(" - ", maxsplit=2)

            if len(partes) == 3:
                data_hora = partes[0].strip()
                tipo_evento = partes[1].strip()  # INFO, AVISO ou ERRO
                mensagem = partes[2].strip()
            else:
                # Fallback caso alguma linha do log não siga o padrão com " - "
                data_hora = "N/A"
                tipo_evento = "DESCONHECIDO"
                for nivel in ["INFO", "AVISO", "ERRO"]:
                    if nivel in linha_limpa.upper():
                        tipo_evento = nivel
                        break
                mensagem = linha_limpa

            # Etapa 4: Preenche a linha na planilha
            ws.append([data_hora, tipo_evento, mensagem])

    # Salva o arquivo Excel no diretório de execução
    wb.save("auditoria_logs.xlsx")
    print("Conversão concluída! Planilha 'auditoria_logs.xlsx' salva com sucesso.")

except FileNotFoundError:
    print("Erro: O arquivo 'log.txt' não foi encontrado no diretório atual. "
          "Verifique se ele foi enviado para a pasta do Google Colab.")
```
