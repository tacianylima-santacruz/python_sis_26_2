```python
nome = input("Digite o nome do personagem: ")
nivel = int(input("Digite o nível do personagem: "))
altura = float(input("Digite a altura do personagem: "))
arma = input("O personagem possui uma arma? (sim/não): ")

print("\n--- FICHA DO PERSONAGEM ---")
print("Nome:", nome)
print("Nível:", nivel)
print("Altura:", altura)
print("Possui arma:", arma)

# Calcula o dobro do nível do personagem
nivel_dobrado = nivel * 2

# Calcula uma pontuação usando o nível e a altura
pontuacao = nivel * altura

# Verifica se o personagem possui nível suficiente
nivel_alto = nivel >= 1000

print("\n--- RESULTADOS ---")
print("Dobro do nível:", nivel_dobrado)
print("Pontuação:", pontuacao)
print("Possui nível alto:", nivel_alto)
```
