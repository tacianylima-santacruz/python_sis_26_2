```python
#atividade 01

# ===== SISTEMA DE MONITORAMENTO DE TRÁFEGO DE REDE =====

pacotes = [12, 0, 25, 40, -1, 18, 50]

print("=== INICIANDO PROCESSAMENTO DE PACOTES ===")
print()

for pacote in pacotes:
    # Regra 1: Ignorar pacotes com tamanho igual a 0

    if pacote == 0:
      print(">>Pacote vazio (0 MB), ignorado")
      continue
    # Regra 2: Pacote negativo indica falha de segurança - interrompe o processamento
    if pacote < 0:
      print(f">> ALERTA DE SEGURANÇA! Pacote inválido detectado ({pacote} MB). Processamento interrompido.")
      break

    # Regra 3: Processamento normal dos demais pacotes
    print(f"Pacote de {pacote} MB processado com sucesso.")

print()
print("=== FIM DA EXECUÇÃO ===")


#Atividade 02

# ===== O SISTEMA DA COMITIVA DO ANEL =====

# ===== ETAPA 1: Registro do Herói (parâmetro com valor padrão + return) =====
def registrar_heroi(nome, classe, reino="Reino Desconhecido"):
    return f"{nome}, o {classe}, natural de {reino}."

print("=== ETAPA 1: REGISTRO DO HERÓI ===")

# Teste COM o parâmetro reino
ficha1 = registrar_heroi("Aragorn", "Guardião", "Gondor")
print(ficha1)

# Teste SEM o parâmetro reino (usa o valor padrão)
ficha2 = registrar_heroi("Legolas", "Arqueiro Élfico")
print(ficha2)

print()

# ===== ETAPA 2: A Chamada das Armas (*args) =====
def chamada_das_armas(nome_heroi, *itens):
    print(f"Equipamentos de {nome_heroi}:")
    for item in itens:
        print(f"- {item}")

print("=== ETAPA 2: A CHAMADA DAS ARMAS ===")
chamada_das_armas("Frodo", "Anel de Poder", "Espada Ferroada", "Capa Élfica")

print()

# ===== ETAPA 3: Ficha Detalhada do Personagem (**kwargs) =====
def ficha_detalhada(nome, **atributos):
    print(f"Ficha de combate de {nome}:")
    for atributo, valor in atributos.items():
        print(f"{atributo}: {valor}")

print("=== ETAPA 3: FICHA DETALHADA ===")
ficha_detalhada("Gimli", forca=90, agilidade=60, resistencia=95)

print()

# ===== DESAFIO BÔNUS: A Jornada (continue e break) =====
print("=== DESAFIO BÔNUS: A JORNADA PARA MORDOR ===")

perigos = ["Orcs", "Nevoeiro", "Aranha Gigante", "Balrog", "Nazgûl"]

for perigo in perigos:
    if perigo == "Nevoeiro":
        print(f">> {perigo} encontrado. Ignorando combate e seguindo viagem.")
        continue

    if perigo == "Balrog":
        print(f">> {perigo} apareceu! Gandalf grita: 'FUJAM, TOLOS!' Jornada interrompida.")
        break

    print(f">> Enfrentando {perigo} na jornada.")

print("=== FIM DA EXECUÇÃO ===")


#Atividade 03

# ===== SISTEMA DE MONITORAMENTO DE TI =====

# ===== ETAPA 1: Registro de Servidor (global + Docstring) =====

servidores_online = 20  # variável global - total de servidores ativos no momento

def alterar_servidores(quantidade):
    """
    Altera o número de servidores online no sistema.

    Parâmetros:
        quantidade (int): Quantidade de servidores a ativar ou desativar.
                           Use valor positivo para ATIVAR servidores
                           e valor negativo para DESATIVAR servidores.

    Impacto na variável global:
        Esta função modifica diretamente a variável global 'servidores_online',
        somando (ou subtraindo) a quantidade informada ao total atual.
        Como 'servidores_online' é declarada com a palavra-chave 'global' dentro
        da função, a alteração feita aqui persiste fora da função também.
    """
    global servidores_online
    servidores_online = servidores_online + quantidade

print("=== ETAPA 1: REGISTRO DE SERVIDOR ===")
print(f"Servidores online antes do ajuste: {servidores_online}")

alterar_servidores(-5)  # desativando 5 servidores

print(f"Servidores online depois do ajuste: {servidores_online}")

print()

# ===== ETAPA 2: Configuração de Conexão ao Banco de Dados (nonlocal) =====

def gerenciar_conexoes():
    limite_conexoes = 10  # variável local da função externa

    def abrir_conexao():
        nonlocal limite_conexoes
        limite_conexoes = limite_conexoes - 1
        print(f"Conexão aberta. Limite restante: {limite_conexoes}")

    # Chamando a função interna 2 vezes
    abrir_conexao()
    abrir_conexao()

    print(f"Limite final de conexões disponíveis: {limite_conexoes}")

print("=== ETAPA 2: CONFIGURAÇÃO DE CONEXÃO AO BANCO DE DADOS ===")
gerenciar_conexoes()

print()

# ===== ETAPA 3: Ajuste Temporário de Cota (função lambda) =====

ajustar_cota = lambda cota_atual, ajuste: cota_atual + ajuste

print("=== ETAPA 3: AJUSTE TEMPORÁRIO DE COTA ===")

cota_armazenamento = 500  # cota atual em GB
nova_cota = ajustar_cota(cota_armazenamento, 250)  # aumentando a cota em 250 GB

print(f"Cota de armazenamento antes do ajuste: {cota_armazenamento} GB")
print(f"Cota de armazenamento depois do ajuste: {nova_cota} GB")


#Atividade 04
# ===== TRADUTOR DE MÁQUINAS (BINÁRIO E HEXADECIMAL) =====

# String com os "dígitos" hexadecimais - a posição de cada caractere
# corresponde ao seu valor (posição 10 = 'A', posição 11 = 'B', etc.)
digitos_hex = "0123456789ABCDEF"


def decimal_para_binario(numero):
    """
    Converte um número decimal para sua representação em binário,
    usando recursão e concatenação de textos (sem listas).
    """
    # Caso base: se o número for 0 ou 1, ele já é o próprio dígito binário
    if numero < 2:
        return str(numero)
    else:
        # Caso recursivo: converte a parte inteira da divisão por 2,
        # e concatena com o resto da divisão (0 ou 1)
        return decimal_para_binario(numero // 2) + str(numero % 2)


def decimal_para_hexadecimal(numero):
    """
    Converte um número decimal para sua representação em hexadecimal,
    usando recursão e concatenação de textos (sem listas).
    """
    # Caso base: se o número for menor que 16, ele corresponde a um único dígito hex
    if numero < 16:
        return digitos_hex[numero]
    else:
        # Caso recursivo: converte a parte inteira da divisão por 16,
        # e concatena com o dígito correspondente ao resto da divisão
        return decimal_para_hexadecimal(numero // 16) + digitos_hex[numero % 16]


# ===== PROGRAMA PRINCIPAL =====

numero_decimal = int(input("Digite um número decimal: "))

resultado_binario = decimal_para_binario(numero_decimal)
resultado_hex = decimal_para_hexadecimal(numero_decimal)

print()
print(f"Número decimal: {numero_decimal}")
print(f"Em binário: {resultado_binario}")
print(f"Em hexadecimal: {resultado_hex}")
```
