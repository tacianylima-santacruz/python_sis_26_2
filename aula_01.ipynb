```python
quantidade_caixas = 4
macas_por_caixa = 12
total_macas = quantidade_caixas * macas_por_caixa

total_cavaleiros = 5
macas_por_cavaleiro = total_macas // total_cavaleiros

portao_aberto = macas_por_cavaleiro > 8

print("Quantidade de maçãs por cavaleiro:")
print(macas_por_cavaleiro)

print("O portão do castelo se abriu?")
print(portao_aberto)
```
