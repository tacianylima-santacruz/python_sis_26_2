```python
# AULA 05 | EXERCÍCIO 4

# Tradutor de Máquinas (Binário e Hexadecimal)
# Na área de TI, os computadores processam dados em Binário (0 e 1),
# mas para facilitar a leitura humana de grandes blocos de dados
# (como endereços de memória, códigos de cores ou endereços físicos MAC das placas de rede),
# usamos o sistema Hexadecimal (Base 16, que vai de 0 a 9 e depois de A a F).

# Objetivo: Criar um programa que receba um único número decimal digitado pelo usuário
# e mostre suas versões em Binário e Hexadecimal, usando apenas funções recursivas
# e concatenação (junção) de textos. Nenhuma lista é permitida!

# A Função Binária: Crie uma função recursiva.
def binario(numero):
    if numero < 2:
        return str(numero)
    return binario(numero // 2) + str(numero % 2)

# Dica: Como o hexadecimal usa letras, crie uma variável de texto,
# você pode usar a posição da letra.
def hexadecimal(numero):
    digitos = "0123456789ABCDEF"
    if numero < 16:
        return digitos[numero]
    return hexadecimal(numero // 16) + digitos[numero % 16]

# Recebe um único número decimal digitado pelo usuário.
numero = int(input("Digite um número decimal: "))

# Mostra suas versões em Binário e Hexadecimal.
print("Binário:", binario(numero))
print("Hexadecimal:", hexadecimal(numero))
```
