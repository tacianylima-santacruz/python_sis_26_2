```python

# Esta é uma VARIÁVEL GLOBAL.
servidores_online = 0


def alterar_servidores(quantidade):
    """
    Altera a quantidade de servidores online.

    Parâmetros:
        quantidade (int): quantidade de servidores que será
        adicionada ou removida.

        Valores positivos ATIVAM servidores.
        Valores negativos DESATIVAM servidores.

    Impacto na variável global:
        A função altera diretamente a variável global
        'servidores_online' usando a palavra-chave global.
    """
    global servidores_online

    servidores_online += quantidade


# Ativando 5 servidores
alterar_servidores(5)

# Desativando 2 servidores
alterar_servidores(-2)

# Mostra quantos servidores continuam online
print("ETAPA 1")
print(f"Servidores online: {servidores_online}")


def gerenciar_conexoes():

    # Esta variável é LOCAL da função gerenciar_conexoes.
    # Ela começa com um limite de 5 conexões.
    conexoes = 5

    # Esta é uma FUNÇÃO INTERNA.
    # Ela existe dentro de gerenciar_conexoes().
    def usar_conexao():

        # 'nonlocal' permite alterar a variável 'conexoes'
        # que pertence à função externa.
        nonlocal conexoes

        # Cada vez que uma conexão é usada,
        # diminuímos 1 do limite disponível.
        conexoes -= 1

    # Chamando a função interna pela primeira vez
    usar_conexao()

    # Chamando a função interna pela segunda vez
    usar_conexao()

    # Retorna quantas conexões ainda estão disponíveis
    return conexoes


print("\nETAPA 2")

# Executamos a função e guardamos o resultado.
conexoes_restantes = gerenciar_conexoes()

print(f"Limite de conexões restantes: {conexoes_restantes}")

# Lambda é uma função pequena e anônima.
# Aqui ela recebe uma cota e acrescenta 100 GB.
alterar_cota = lambda cota: cota + 100


print("\nETAPA 3")

# Cota atual do servidor
cota_atual = 500

print(f"Cota atual: {cota_atual} GB")

# Chamamos a lambda para aumentar a cota.
nova_cota = alterar_cota(cota_atual)

print(f"Nova cota: {nova_cota} GB")
```
