```python
# O parâmetro "reino" possui um valor padrão.
# Isso significa que podemos chamar a função passando ou não o reino.
def comitiva(nome, classe, reino="Reino Desconhecido"):

    # A função retorna um TEXTO formatado.
    return f"Nome: {nome} | Classe: {classe} | Reino: {reino}"





# O *armas permite receber uma quantidade indeterminada de itens.
def ch_armas(nome, *armas):

    print(f"\nA Chamada das Armas foi feita para {nome}!")

    # Percorre cada arma recebida.
    for arma in armas:
        print(f"- {arma}")




# O **atributos permite receber vários argumentos nomeados.
# Os argumentos nomeados serão armazenados em um dicionário.
def stat(nome, **atributos):

    print(f"\nNome: {nome}")
    print("Atributos:")

    # .items() permite percorrer o nome do atributo
    # e o seu respectivo valor.
    for atributo, valor in atributos.items():
        print(f"- {atributo}: {valor}")



def jornada(perigos):

    print("\nIniciando a jornada para Mordor...\n")

    # Percorre todos os perigos da lista.
    for perigo in perigos:

        # Se encontrar Nevoeiro, ignora e continua a viagem.
        if perigo == "Nevoeiro":
            print("Nevoeiro encontrado! Ignorando o combate e seguindo viagem...")
            continue

        # Se encontrar um Balrog, interrompe a jornada.
        if perigo == "Balrog":
            print("Balrog encontrado!")
            print("Gandalf avisa: FUJAM!")
            break

        # Para qualquer outro perigo, prepara-se para o combate.
        print(f"Perigo encontrado: {perigo}! Preparem-se para o combate!")





perigos = ["Orcs", "Nevoeiro", "Aranhas", "Balrog", "Uruk-hai"]



# Pedimos os dados do herói.
nome_inp = input("Digite o seu nome: ").capitalize()
classe_inp = input("Digite a sua classe: ").capitalize()

# Aqui o reino é opcional.
# O usuário pode deixar vazio e a função usará
# "Reino Desconhecido" como valor padrão.
reino_inp = input("Digite o seu reino (opcional): ").capitalize()


# Se o usuário não informou o reino, usamos a função
# sem passar o terceiro argumento.
if reino_inp == "":
    registro = comitiva(nome_inp, classe_inp)
else:
    registro = comitiva(nome_inp, classe_inp, reino_inp)

print("\nSeu registro foi feito!")
print(registro)




armas = []

# Pergunta quantas armas o herói deseja levar.
quant_armas = int(input("\nDigite a quantidade de armas que você deseja levar: "))

# Repete a pergunta de acordo com a quantidade escolhida.
for i in range(quant_armas):

    armas_inp = input(f"Digite a arma {i + 1}: ").capitalize()
    armas.append(armas_inp)



ch_armas(nome_inp, *armas)




# Vamos passar os atributos diretamente como argumentos nomeados.
stat(
    nome_inp,
    forca=90,
    durabilidade=40,
    magia=34,
    agilidade=15
)

jornada(perigos)
```
