```python
# AULA 05 | EXERCÍCIO 3

# Monitoramento de TI Objetivo:
# Criar um script para registrar alertas, aplicar configurações dinâmicas e
# processar chamados da infraestrutura.

# Etapa 1:
# Registro de Servidor (global e Docstring) Crie uma variável global e
# uma função que use a palavra-chave global para controlar o total de servidores online.
# Adicione uma Docstring explicando a função, parâmetros e o impacto na variável global.
# Desative/ative 5 servidores e imprima o saldo de servidores ativos.
servidores_online = 5
def servidor(quantidade):
  """
  Altera a quantidade de servidores online.
  Parâmetro: quantidade de servidores.
  Altera a variável global servidores_online.
  """
  global servidores_online
  servidores_online = servidores_online + quantidade
servidor(-2)
servidor(2)
print("\nMONITORAMENTO DE TI")
print("\nServidores ativos:", servidores_online)

# Etapa 2:
# Configuração de Conexao ao Banco de Dados Aninhada
# Crie uma função para gerenciar conexoes e dentro dela, criar uma variável local.
# Crie uma função interna que use nonlocal para alterar a quantidade de conexoes.
# Execução: Chame a função interna 2 vezes e exiba o limite de conexões restantes.
def conexoes():
  limite = 5
  def conectar():
    nonlocal limite
    limite = limite - 1
  conectar()
  conectar()
  print("Conexões restantes:", limite)
conexoes()

# Etapa 3:
# Ajuste Temporário de Cota (Função lambda)
# Crie uma função com lambda para alterar a cota de armazenamento de arquivos em nuvem no servidor.
# Teste: chame a função alterando a cota.
cota = lambda valor: valor + 5
print("Nova cota:", cota(100))
```
