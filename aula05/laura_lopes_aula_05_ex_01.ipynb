```python
pacotes = [12, 0, 25, 40, -1, 18, 50]

for i in pacotes:
  if i == 0:
    print("Pacote nulo")
    continue
  elif i< 0:
    print("\n!!!--Alerta numero negativo--!!!")
    break
  else:
    print(f"Pacote de {i} MB processado com sucesso.")
```
