```python
# AULA 05 | EXERCÍCIO 2

# O Sistema da Comitiva do Anel Objetivo:
# Criar um script para gerenciar os membros da Comitiva e
# preparar seus equipamentos utilizando funções.

# Etapa 1:
# Registro do Herói (Parâmetros com valor padrão e return):
# O Crie uma função que receba o nome e a classe do personagem e, opcionalmente, o seu reino.
# A função deve retornar um texto formatado
# Teste: invoque a função com e sem o parâmetro reino.

def heroi(nome, classe, reino="Sem Reino"):
  return f"Nome: {nome} | Classe: {classe} | Reino: {reino}"

print("REGISTRO DE HERÓIS")
print(heroi("Gandalf", "Mago", "Urano"))
print(heroi("Frodo", "Hobbit"))

# Etapa 2:
# A Chamada das Armas: Crie uma função que receba o nome do herói e uma quantidade indeterminada de itens.
# A função deve imprimir o nome do herói e listar cada item recebido com um marcador (-).
# Teste: chame a função passando 3 itens.

def armas(nome, *itens):
  print(f"Herói: {nome}")
  print("Armas:")

  for item in itens:
    print(f"- {item}")

print("\nCHAMADA DAS ARMAS")
armas("Gandalf", "Cajado", "Espada", "Anel")

# Etapa 3:
# Ficha Detalhada do Personagem:
# Crie uma função para registrar estatísticas de combate dinâmicas.
# A função deve exibir o nome do personagem e percorrer todos os atributos passados
# imprimindo o nome do atributo e o seu valor.
# Teste: chame a função usando argumentos nomeados.

def itens(nome, **itens):
  print(f"Nome: {nome}")
  print("Atributos:")

  for item, valor in itens.items():
    print(f"- {item}: {valor}")

print("\nFICHA DETALHADA DO HERÓI")
itens("Gandalf", Força=80, Agilidade=95)

# Desafio Bônus (Controle de Fluxo):
# A Jornada: Crie um laço que percorra uma lista de perigos no caminho para
# Mordor: Se o perigo for "Nevoeiro": ignorar o combate e seguir viagem.
# Se o perigo for "Balrog": interromper a jornada imediatamente (Gandalf avisa para fugir!).

print("\nA JORNADA")

perigos = ["Orc", "Nevoeiro", "Balrog"]

for perigo in perigos:
  if perigo == "Nevoeiro":
    print("🌫️Nevoeiro no caminho! Vamos seguir com cuidado.")
    continue
  if perigo == "Balrog":
    print("🚨É um Balrog! Gandalf grita: Fujam!")
    break
  print(f"⚠️Perigo! {perigo} no caminho! Preparem-se!")
```
