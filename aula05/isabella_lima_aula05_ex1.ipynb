```python
# AULA 05 | EXERCÍCIO 1

# Sistema de Monitoramento de Tráfego de Rede Objetivo:
# Você está desenvolvendo um script para analisar pacotes de dados que transitam
# por um servidor da empresa. O sistema recebe uma lista com o tamanho dos pacotes
# (em Megabytes) e precisa processá-los individualmente seguindo algumas regras de segurança.

# Dados de Entrada: Utilize a seguinte lista de pacotes recebidos:
pacotes = [12, 0, 25, 40, -1, 18, 50]
# Escreva um programa usando um laço para percorrer a lista de pacotes e aplique as seguintes regras:
# Ignore pacotes com tamanho igual a 0, pois eles não contêm informação relevante.
for pacote in pacotes:
    if pacote == 0:
# Se o sistema encontrar um pacote com valor menor que zero (ex: -1),
# exiba uma mensagem de alerta e interrompa o processamento (pois indica uma falha de segurança).
        continue
    elif pacote < 0:
        print("⚠️ Alerta! Processamento interrompido por uma falha na segurança.")
# Para os demais pacotes, exiba a mensagem: "Pacote de [X] MB processado com sucesso."
        break
    else:
        print(f"✅ Pacote de {pacote} MB processado com sucesso.")
```
