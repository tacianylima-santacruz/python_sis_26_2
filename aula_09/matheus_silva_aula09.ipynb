```python
import os

# --- ATIVIDADE 01 ---
print('--- Iniciando Atividade 01 ---')

# Etapa 1: Inicialização do Log
with open('acessos.log', 'w', encoding='utf-8') as f:
    f.write('--- SISTEMA DE AUDITORIA DE LOGS ---\n')
    f.write('2023-10-27 08:00:00 - SISTEMA - Inicializado\n')

# Etapa 2: Registro de Atividades
eventos = [
    '2023-10-27 08:05:00 - LOGIN - admin\n',
    '2023-10-27 08:10:00 - INFO - Servidor OK\n',
    '2023-10-27 08:15:00 - ERRO - Falha de Conexao\n',
    '2023-10-27 08:20:00 - LOGIN - usuario1\n',
    '2023-10-27 08:25:00 - ERRO - Acesso Negado\n'
]
with open('acessos.log', 'a', encoding='utf-8') as f:
    f.writelines(eventos)

# Etapa 3: Leitura e Inspeção Seletiva
with open('acessos.log', 'r', encoding='utf-8') as f:
    cabecalho = f.readline().strip()
    print(f'Cabeçalho: {cabecalho}')

with open('acessos.log', 'r', encoding='utf-8') as f:
    linhas = f.readlines()
    print(f'Último evento: {linhas[-1].strip()}')

# Etapa 4: Auditoria de Segurança
contagem = 0
palavra_chave = 'LOGIN'  # Escolha LOGIN ou ERRO
print(f'\nAuditoria de Segurança (Buscando por: {palavra_chave}):')
with open('acessos.log', 'r', encoding='utf-8') as f:
    for linha in f:
        if palavra_chave in linha:
            print(linha.strip())
            contagem += 1
print(f'Total de ocorrências de "{palavra_chave}": {contagem}')
```

```python
# --- ATIVIDADE 02 ---
print('\n--- Iniciando Atividade 02 (Resiliência) ---')

# Etapa 1: Tentativa de leitura de arquivo inexistente
try:
    with open('servidor_inexistente.log', 'r', encoding='utf-8') as f:
        conteudo = f.read()
except FileNotFoundError:
    print('[Erro Tratado] O arquivo "servidor_inexistente.log" não foi encontrado.')

# Etapa 2, 3 e 4: Proteção de escrita, decodificação e cleanup com finally
arquivo_seguro = None
try:
    # Tentamos abrir com encoding incorreto para simular erro de decodificação se necessário,
    # ou apenas protegemos a escrita contra falhas de permissão.
    arquivo_seguro = open('acessos.log', 'a', encoding='utf-8')
    arquivo_seguro.write('2023-10-27 08:30:00 - INFO - Fim do expediente\n')
    print('Gravação protegida executada com sucesso.')
except PermissionError:
    print('[Erro Tratado] Sem permissão para escrever no arquivo.')
except UnicodeDecodeError:
    print('[Erro Tratado] Erro de decodificação ao ler/escrever no arquivo.')
except Exception as e:
    print(f'[Erro Inesperado] {e}')
finally:
    if arquivo_seguro and not arquivo_seguro.closed:
        arquivo_seguro.close()
        print('Arquivo de log fechado com segurança no bloco finally.')
```
