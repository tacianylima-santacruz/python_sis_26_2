```python
# AULA 09 | ATIVIDADE 01 | Leitura e escrita Sistema de Auditoria de Logs

# Você faz parte da equipe de segurança e precisa automatizar o monitoramento de acessos a um servidor crítico.
# Você irá criar, atualizar e analisar o arquivo acessos.log.

# Objetivo:
# *Crie um script Python que execute as 4 etapas a seguir:

# Etapa 1: Inicialização do Log (Modo Escrita - w)
# Crie o arquivo acessos.log gravando o cabeçalho do sistema e o primeiro evento de inicialização:

with open("acessos.log", "w", encoding="utf-8") as arquivo:
    arquivo.write("SISTEMA DE AUDITORIA DE ACESSOS\n")
    arquivo.write("2026-10-14 08:30:15 - INFO - Sistema inicializado\n")

print("✅ Arquivo acessos.log criado com sucesso.\n")

# Etapa 2: Registro de Atividades (Modo Append - a)
# Sem apagar o conteúdo existente, adicione pelo menos 5 eventos ao final do arquivo.

with open("acessos.log", "a", encoding="utf-8") as arquivo:
    arquivo.write("2026-10-14 08:31:02 - LOGIN - Usuário admin conectado\n")
    arquivo.write("2026-10-14 08:35:22 - ERRO - Senha incorreta\n")
    arquivo.write("2026-10-14 08:35:40 - LOGIN - Usuário bat conectado\n")
    arquivo.write("2026-10-14 08:40:11 - ERRO - Falha na conexão\n")
    arquivo.write("2026-10-14 08:42:00 - INFO - Conexão restabelecida\n")

print("✅ Cinco eventos adicionados ao arquivo.\n")

# Etapa 3: Leitura e Inspeção Seletiva Abra o arquivo em modo de leitura e resolva:

# 1 - Inspeção do cabeçalho: Usando readline(), leia e exiba apenas a primeira linha do arquivo (o título).

with open("acessos.log", "r", encoding="utf-8") as arquivo:
    print(arquivo.readline().strip())

# 2 - Último evento registrado: Usando readlines(),

with open("acessos.log", "r", encoding="utf-8") as arquivo:
    linhas = arquivo.readlines()

# 3 - obtenha e imprima diretamente a última linha registrada no arquivo.

print(linhas[-1].strip())

# Etapa 4: Auditoria de Segurança Utilizando a iteração direta (for linha in arquivo:):

# 1 - Percorra o arquivo sem carregar tudo na memória de uma só vez.

palavra_chave = "ERRO"
total = 0

with open("acessos.log", "r", encoding="utf-8") as arquivo:
    for linha in arquivo:

        # 2 - Identifique e imprima no terminal apenas os eventos marcados com uma palavra-chave como: LOGIN ou ERRO.

        if palavra_chave in linha:
            print(linha.strip())
            total += linha.count(palavra_chave)

# 3 - Exiba a contagem total dessa palavra chave encontrada ao final.

print(f"\nTotal da palavra-chave {palavra_chave}: {total}")
```

```python
# AULA 09 | ATIVIDADE 02 | Módulo de Resiliência de Arquivos

# Etapa 1: Adicione uma camada de tratamento de erros ao script anterior utilizando blocos try / except / finally, cobrindo os seguintes cenários:
# Tentativa de leitura de arquivo inexistente (FileNotFoundError):
# Tente abrir um arquivo com nome incorreto (ex.: servidor_inexistente.log) antes de ler o arquivo principal.
# Capture FileNotFoundError e mostre uma mensagem amigável ao usuário em vez de deixar o programa quebrar.

try:
    with open("servidor_inexistente.log", "r", encoding="utf-8") as arquivo:
        print(arquivo.read())
except FileNotFoundError:
    print("⚠️ O arquivo servidor_inexistente.log não foi encontrado.\n")

# Etapa 2: Proteção de escrita e permissões (PermissionError):
# Envolva a escrita e o append em blocos protegidos para capturar restrições de escrita no diretório ou em arquivos somente-leitura.

try:
    with open("acessos.log", "w", encoding="utf-8") as arquivo:
        arquivo.write("SISTEMA DE AUDITORIA DE ACESSOS\n")
        arquivo.write("2026-10-14 08:30:15 - INFO - Sistema inicializado\n")

    print("✅ Arquivo acessos.log criado com sucesso.\n")

except PermissionError:
    print("⚠️ Sem permissão para criar ou sobrescrever acessos.log.\n")
except OSError as erro:
    print(f"⚠️ Não foi possível gravar o arquivo: {erro}\n")
else:
    try:
        with open("acessos.log", "a", encoding="utf-8") as arquivo:
            arquivo.write("2026-10-14 08:31:02 - LOGIN - Usuário admin conectado\n")
            arquivo.write("2026-10-14 08:35:22 - ERRO - Senha incorreta\n")
            arquivo.write("2026-10-14 08:35:40 - LOGIN - Usuário bat conectado\n")
            arquivo.write("2026-10-14 08:40:11 - ERRO - Falha na conexão\n")
            arquivo.write("2026-10-14 08:42:00 - INFO - Conexão restabelecida\n")

        print("✅ Cinco eventos adicionados ao arquivo.\n")

    except PermissionError:
        print("⚠️ Sem permissão para adicionar eventos ao arquivo.\n")
    except OSError as erro:
        print(f"⚠️ Não foi possível adicionar os eventos: {erro}\n")

# Etapa 3: Tratamento de decodificação (UnicodeDecodeError):
# Capture erros de decodificação caso o arquivo seja aberto com encoding incompatível.

arquivo = None

try:
    arquivo = open("acessos.log", "r", encoding="utf-8")

    print(arquivo.readline().strip())

    arquivo.seek(0)
    linhas = arquivo.readlines()
    print(linhas[-1].strip())

    arquivo.seek(0)
    palavra_chave = "ERRO"
    total = 0

    print("\n🔎 AUDITORIA DE SEGURANÇA\n")

    for linha in arquivo:
        if palavra_chave in linha:
            print(linha.strip())
            total += linha.count(palavra_chave)

    print(f"\nTotal da palavra-chave {palavra_chave}: {total}")

except FileNotFoundError:
    print("⚠️ O arquivo acessos.log não foi encontrado.")
except PermissionError:
    print("⚠️ Sem permissão para ler acessos.log.")
except UnicodeDecodeError:
    print("⚠️ Não foi possível ler o arquivo com a codificação UTF-8.")
except Exception as erro:
    print(f"⚠️ Ocorreu um erro inesperado: {erro}")

# Etapa 4: Fechamento e cleanup com finally:
# Demonstrar o fechamento seguro de conexões/arquivos mesmo se ocorrer um erro inesperado.

finally:
    if arquivo is not None:
        arquivo.close()
        print("\n✅ Arquivo fechado com segurança.")
```
