# Atividade Prática 01 Aula 09
Atividade Leitura e escrita
Sistema de Auditoria de Logs
Você faz parte da equipe de segurança e precisa automatizar o monitoramento de acessos a um servidor crítico. Você irá criar, atualizar e analisar o arquivo acessos.log.

Objetivo:  
*Crie um script Python que execute as 4 etapas a seguir:

Etapa 1: Inicialização do Log (Modo Escrita - w)
Crie o arquivo acessos.log gravando o cabeçalho do sistema e o primeiro evento de inicialização:

Etapa 2: Registro de Atividades (Modo Append - a)
Sem apagar o conteúdo existente, adicione pelo menos 5 eventos ao final do arquivo.

Etapa 3: Leitura e Inspeção Seletiva
Abra o arquivo em modo de leitura e resolva:

1 - Inspeção do cabeçalho: Usando readline(), leia e exiba apenas a primeira linha do arquivo (o título).

2 - Último evento registrado: Usando readlines(),

3 - obtenha e imprima diretamente a última linha registrada no arquivo.

Etapa 4: Auditoria de Segurança
Utilizando a iteração direta (for linha in arquivo:):

1 - Percorra o arquivo sem carregar tudo na memória de uma só vez.

2 - Identifique e imprima no terminal apenas os eventos marcados com uma palavra-chave como: LOGIN ou ERRO.

3 - Exiba a contagem total dessa palavra chave encontrada ao final.

```python

# ATIVIDADE PRÁTICA 01 - AULA 09
# Sistema de Auditoria de Logs

# ETAPA 1: Inicialização do Log (modo escrita - w)

with open("acessos.log", "w", encoding="utf-8") as arquivo:
    arquivo.write("=== SISTEMA DE AUDITORIA DE ACESSOS ===\n")
    arquivo.write("INICIALIZAÇÃO: Sistema de monitoramento iniciado.\n")


# ETAPA 2: Registro de Atividades (modo append - a)

eventos = [
    "LOGIN: Usuário Nicolas acessou o servidor.\n",
    "LOGIN: Usuário administrador acessou o sistema.\n",
    "ERRO: Tentativa de acesso com senha incorreta.\n",
    "LOGOUT: Usuário Nicolas encerrou a sessão.\n",
    "ERRO: Tentativa de acesso não autorizado.\n"
]

with open("acessos.log", "a", encoding="utf-8") as arquivo:
    arquivo.writelines(eventos)


# ETAPA 3: Leitura e Inspeção Seletiva

with open("acessos.log", "r", encoding="utf-8") as arquivo:

    # 1 - Inspeção do cabeçalho usando readline()
    primeira_linha = arquivo.readline()
    print("=== CABEÇALHO DO SISTEMA ===")
    print(primeira_linha.strip())

    # 2 - Último evento usando readlines()
    arquivo.seek(0)
    linhas = arquivo.readlines()

    print("\n=== ÚLTIMO EVENTO REGISTRADO ===")
    print(linhas[-1].strip())


# ETAPA 4: Auditoria de Segurança

contagem_login = 0
contagem_erro = 0

print("\n=== AUDITORIA DE SEGURANÇA ===")

with open("acessos.log", "r", encoding="utf-8") as arquivo:

    # Leitura linha por linha, sem carregar tudo na memória
    for linha in arquivo:
        if "LOGIN" in linha:
            print(linha.strip())
            contagem_login += 1

        elif "ERRO" in linha:
            print(linha.strip())
            contagem_erro += 1

print("\n=== RESUMO DA AUDITORIA ===")
print("Total de eventos LOGIN:", contagem_login)
print("Total de eventos ERRO:", contagem_erro)
print("Total de eventos identificados:", contagem_login + contagem_erro)
```

# Atividade Prática 02 Aula 09
Módulo de Resiliência de Arquivos

Etapa 1: Adicione uma camada de tratamento de erros ao script anterior utilizando blocos try / except / finally, cobrindo os seguintes cenários:
Tentativa de leitura de arquivo inexistente (FileNotFoundError):
Tente abrir um arquivo com nome incorreto (ex.: servidor_inexistente.log) antes de ler o arquivo principal.
Capture FileNotFoundError e mostre uma mensagem amigável ao usuário em vez de deixar o programa quebrar.

Etapa 2: Proteção de escrita e permissões (PermissionError):
Envolva a escrita e o append em blocos protegidos para capturar restrições de escrita no diretório ou em arquivos somente-leitura.
Etapa 3: Tratamento de decodificação (UnicodeDecodeError):
Capture erros de decodificação caso o arquivo seja aberto com encoding incompatível.

Etapa 4: Fechamento e cleanup com finally:
Demonstrar o fechamento seguro de conexões/arquivos mesmo se ocorrer um erro inesperado.


```python

# ATIVIDADE PRÁTICA 02 - AULA 09
# Módulo de Resiliência de Arquivos

# ETAPA 1: Tratamento de arquivo inexistente

arquivo = None

try:
    arquivo = open("servidor_inexistente.log", "r", encoding="utf-8")
    conteudo = arquivo.read()

except FileNotFoundError:
    print("ERRO: O arquivo solicitado não foi encontrado.")

finally:
    if arquivo is not None and not arquivo.closed:
        arquivo.close()
    print("Verificação de arquivo inexistente finalizada.")


# ETAPA 2: Proteção de escrita e permissões

# Escrita inicial
arquivo = None

try:
    arquivo = open("acessos.log", "w", encoding="utf-8")
    arquivo.write("=== SISTEMA DE AUDITORIA DE ACESSOS ===\n")
    arquivo.write("INICIALIZAÇÃO: Sistema iniciado com sucesso.\n")
    print("\nArquivo criado e cabeçalho gravado com sucesso.")

except PermissionError:
    print("ERRO: Sem permissão para criar ou escrever no arquivo.")

finally:
    if arquivo is not None and not arquivo.closed:
        arquivo.close()
    print("Operação de escrita finalizada.")


# Adicionando eventos ao arquivo
arquivo = None

try:
    arquivo = open("acessos.log", "a", encoding="utf-8")

    eventos = [
        "LOGIN: Usuário Nicolas acessou o servidor.\n",
        "LOGIN: Usuário administrador acessou o sistema.\n",
        "ERRO: Senha incorreta.\n",
        "LOGOUT: Usuário Nicolas encerrou a sessão.\n",
        "ERRO: Tentativa de acesso não autorizado.\n"
    ]

    arquivo.writelines(eventos)
    print("Eventos adicionados com sucesso.")

except PermissionError:
    print("ERRO: Sem permissão para adicionar informações ao arquivo.")

finally:
    if arquivo is not None and not arquivo.closed:
        arquivo.close()
    print("Operação de append finalizada.")


# ETAPA 3: Tratamento de erro de decodificação

arquivo = None

try:
    # UTF-16 é incompatível com o arquivo gravado em UTF-8
    arquivo = open("acessos.log", "r", encoding="utf-16")
    conteudo = arquivo.read()
    print(conteudo)

except UnicodeDecodeError:
    print("\nERRO: Não foi possível decodificar o arquivo.")
    print("Verifique se o encoding utilizado é compatível.")

finally:
    if arquivo is not None and not arquivo.closed:
        arquivo.close()
    print("Verificação de decodificação finalizada.")


# ETAPA 4: Fechamento seguro com finally

arquivo = None

try:
    arquivo = open("acessos.log", "r", encoding="utf-8")

    print("\n=== CONTEÚDO DO ARQUIVO ===")
    for linha in arquivo:
        print(linha.strip())

except FileNotFoundError:
    print("ERRO: Arquivo de auditoria não encontrado.")

except PermissionError:
    print("ERRO: Sem permissão para ler o arquivo.")

except UnicodeDecodeError:
    print("ERRO: Problema na decodificação do arquivo.")

except Exception as erro:
    print("ERRO inesperado:", erro)

finally:
    if arquivo is not None and not arquivo.closed:
        arquivo.close()
        print("\nArquivo fechado com segurança.")
    else:
        print("\nNão havia arquivo aberto para fechar.")

print("\n=== PROGRAMA FINALIZADO ===")
```
