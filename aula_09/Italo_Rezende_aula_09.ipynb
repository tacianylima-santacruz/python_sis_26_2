Atividade Prática 01 - Aula 09

```python


nome_arquivo = "acessos.log"

# Etapa 1: Inicialização do Log (Modo Escrita - 'w')
with open(nome_arquivo, "w", encoding="utf-8") as arquivo:
    arquivo.write("=== SISTEMA DE AUDITORIA DE LOGS - SERVIDOR CRÍTICO ===\n")
    arquivo.write("[2026-11-18 08:00:00] [INIT] Sistema de monitoramento iniciado.\n")

print("Etapa 1 concluída: arquivo 'acessos.log' inicializado.")

# Etapa 2: Registro de Atividades (Modo Append - 'a')
novos_eventos = [
    "[2026-11-18 08:05:12] [LOGIN] Usuário 'admin' autenticado com sucesso (IP: 192.168.1.10).\n",
    "[2026-11-18 08:12:45] [ACESSO] Leitura de configurações do servidor por 'admin'.\n",
    "[2026-11-18 08:20:03] [ERRO] Falha de autenticação para o usuário 'root' (IP: 203.0.113.45).\n",
    "[2026-11-18 08:25:19] [LOGIN] Usuário 'auditor' autenticado com sucesso (IP: 192.168.1.22).\n",
    "[2026-11-18 08:31:50] [ERRO] Tentativa de acesso negada ao banco de dados (IP: 203.0.113.45).\n",
    "[2026-11-18 08:40:00] [LOGOUT] Usuário 'admin' encerrou a sessão.\n",
]

with open(nome_arquivo, "a", encoding="utf-8") as arquivo:
    arquivo.writelines(novos_eventos)

print(f"Etapa 2 concluída: {len(novos_eventos)} eventos adicionados ao log.\n")

# Etapa 3: Leitura e Inspeção Seletiva (Modo Leitura - 'r')
with open(nome_arquivo, "r", encoding="utf-8") as arquivo:
    # 1 - Inspeção do cabeçalho usando readline()
    cabecalho = arquivo.readline()
    print("--- Inspeção do Cabeçalho ---")
    print(cabecalho.strip())

    # 2 e 3 - Último evento registrado usando readlines()
    arquivo.seek(0)
    todas_linhas = arquivo.readlines()
    ultimo_evento = todas_linhas[-1]
    print("\n--- Último Evento Registrado ---")
    print(ultimo_evento.strip())

# Etapa 4: Auditoria de Segurança (Iteração direta linha a linha)
palavra_chave = "ERRO"  # Pode ser alterado para "LOGIN"
contador = 0

print(f"\n--- Auditoria de Segurança (Filtro: {palavra_chave}) ---")
with open(nome_arquivo, "r", encoding="utf-8") as arquivo:
    for linha in arquivo:
        if palavra_chave in linha:
            print(linha.strip())
            contador += 1

print(f"\nTotal de eventos com a palavra-chave '{palavra_chave}': {contador}")
```

Atividade Prática 02 - Aula 09

```python


nome_arquivo = "acessos.log"
arquivo_inexistente = "servidor_inexistente.log"

# ----------------------------------------------------------
# Etapa 1: Tentativa de leitura de arquivo inexistente (FileNotFoundError)
# ----------------------------------------------------------
print("--- Etapa 1: Teste de FileNotFoundError ---")
try:
    with open(arquivo_inexistente, "r", encoding="utf-8") as arq:
        print(arq.read())
except FileNotFoundError:
    print(f"[AVISO] O arquivo '{arquivo_inexistente}' não foi encontrado. Verifique o nome ou o caminho e tente novamente.")

# ----------------------------------------------------------
# Etapa 2: Proteção de escrita e permissões (PermissionError)
# ----------------------------------------------------------
print("\n--- Etapa 2: Escrita e Append protegidos (PermissionError) ---")
try:
    with open(nome_arquivo, "w", encoding="utf-8") as arquivo:
        arquivo.write("=== SISTEMA DE AUDITORIA DE LOGS - SERVIDOR CRÍTICO ===\n")
        arquivo.write("[2026-11-18 08:00:00] [INIT] Sistema de monitoramento iniciado.\n")
    print("Inicialização (modo 'w') realizada com sucesso.")
except PermissionError:
    print(f"[ERRO DE PERMISSÃO] Sem permissão para criar ou sobrescrever '{nome_arquivo}'.")
except OSError as e:
    print(f"[ERRO DE SISTEMA] Falha ao inicializar o arquivo: {e}")

novos_eventos = [
    "[2026-11-18 08:05:12] [LOGIN] Usuário 'admin' autenticado com sucesso (IP: 192.168.1.10).\n",
    "[2026-11-18 08:12:45] [ACESSO] Leitura de configurações do servidor por 'admin'.\n",
    "[2026-11-18 08:20:03] [ERRO] Falha de autenticação para o usuário 'root' (IP: 203.0.113.45).\n",
    "[2026-11-18 08:25:19] [LOGIN] Usuário 'auditor' autenticado com sucesso (IP: 192.168.1.22).\n",
    "[2026-11-18 08:31:50] [ERRO] Tentativa de acesso negada ao banco de dados (IP: 203.0.113.45).\n",
    "[2026-11-18 08:40:00] [LOGOUT] Usuário 'admin' encerrou a sessão.\n",
]

try:
    with open(nome_arquivo, "a", encoding="utf-8") as arquivo:
        arquivo.writelines(novos_eventos)
    print(f"Registro de atividades (modo 'a') concluído: {len(novos_eventos)} eventos gravados.")
except PermissionError:
    print(f"[ERRO DE PERMISSÃO] Sem permissão de escrita (append) em '{nome_arquivo}'. Verifique se o arquivo não é somente-leitura.")
except OSError as e:
    print(f"[ERRO DE SISTEMA] Falha ao adicionar eventos ao arquivo: {e}")

# ----------------------------------------------------------
# Etapa 3: Tratamento de decodificação (UnicodeDecodeError)
# ----------------------------------------------------------
print("\n--- Etapa 3: Teste de UnicodeDecodeError ---")
try:
    # Tenta abrir propositalmente um arquivo UTF-8 (com acentos) usando ASCII
    with open(nome_arquivo, "r", encoding="ascii") as arquivo:
        conteudo = arquivo.read()
except UnicodeDecodeError:
    print(f"[AVISO] Erro de decodificação ao ler '{nome_arquivo}' com encoding 'ascii'. Reabrindo com 'utf-8'...")
    with open(nome_arquivo, "r", encoding="utf-8") as arquivo:
        primeira_linha = arquivo.readline().strip()
        print(f"Leitura recuperada com sucesso! Cabeçalho: {primeira_linha}")

# ----------------------------------------------------------
# Etapa 4: Fechamento seguro e cleanup com finally
# ----------------------------------------------------------
print("\n--- Etapa 4: Auditoria com fechamento seguro no bloco finally ---")
palavra_chave = "ERRO"
contador = 0
arquivo_auditoria = None

try:
    arquivo_auditoria = open(nome_arquivo, "r", encoding="utf-8")
    for linha in arquivo_auditoria:
        if palavra_chave in linha:
            print(linha.strip())
            contador += 1
    print(f"Total de eventos com '{palavra_chave}': {contador}")
except FileNotFoundError:
    print(f"[AVISO] Arquivo '{nome_arquivo}' não encontrado para auditoria.")
except PermissionError:
    print(f"[ERRO DE PERMISSÃO] Sem permissão para ler '{nome_arquivo}'.")
except UnicodeDecodeError:
    print(f"[ERRO DE DECODIFICAÇÃO] Encoding incompatível ao ler '{nome_arquivo}'.")
except Exception as erro_inesperado:
    print(f"[ERRO INESPERADO] Ocorreu uma falha durante a auditoria: {erro_inesperado}")
finally:
    if arquivo_auditoria is not None and not arquivo_auditoria.closed:
        arquivo_auditoria.close()
        print("[CLEANUP] Arquivo fechado com segurança no bloco finally.")
    else:
        print("[CLEANUP] Verificação concluída: nenhum arquivo permaneceu aberto.")
```
