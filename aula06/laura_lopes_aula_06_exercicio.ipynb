```python
#laura_lopes_aula_06_exercicio_03

# Criando o dicionário de contatos
contatos = {
    "Shallan": "99999-1111",
    "Kaladin": "99999-2222",
    "Adolin": "99999-3333"
}

print("Contatos iniciais:", contatos)


# Adicionando um novo contato depois da criação do dicionário
contatos["Dalinar"] = "99999-4444"

print("\nDepois de adicionar Dalinar:", contatos)


# Verificando se Chai existe
if "Chai" not in contatos:
    contatos["Chai"] = "99999-5555"
    print("\nChai não existia. Contato adicionado:")
    print(contatos)
else:
    print("Chai já existe nos contatos.")


# Excluindo um contato
# pop() é usado porque permite definir um valor padrão.
# Assim, não ocorre erro caso o contato não exista.
contatos.pop("Adolin", None)

print("\nDepois de excluir Adolin:", contatos)


# Imprimindo todos os nomes dos contatos
print("\nNomes dos contatos:")

for nome in contatos:
    print(nome)
```

```python
#laura_lopes_aula_06_exercicio_04

# Criando uma lista com alguns livros repetidos
livros = [
    "Duna",
    "Os Relatos da Guerra das Tempestades",
    "Duna",
    "Palavras de Radiancia",
    "MistBorn",
    "Elandris",
    "MistBorn"
]

print("Lista original:")
print(livros)

# Convertendo a lista em um conjunto
# O set remove automaticamente os livros repetidos
livros_unicos = set(livros)

print("\nLivros sem duplicatas:")
print(livros_unicos)
```
