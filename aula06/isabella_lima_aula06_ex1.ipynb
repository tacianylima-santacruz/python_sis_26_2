```python
# AULA 06 | EXERCÍCIO 1 | Animais Favoritos

# Objetivo: Crie uma lista com vários animais para realizar filtros: Filtros:
animais = ["cachorro", "gato", "coelho", "hamster", "cavalo"]
favoritos = ["cachorro", "coelho"]

# Mostre em tela seus animais favoritos;
print("\nANIMAIS FAVORITOS")
print("▪ Meus animais favoritos são:", favoritos)

# Imprima a quantidade de animais da lista e a quantidade de animais favoritos;
print("\nQUANTIDADE DE ANIMAIS EM CADA LISTA")
print("▪ Quantidade de animais:", len(animais))
print("▪ Quantidade de animais favoritos:", len(favoritos))

# Mostre qual é o primeiro animail favorito e o último;
print("\nORDEM DOS ANIMAIS FAVORITOS")
print("▪ Primeiro animal favorito:", favoritos[0])
print("▪ Último animal favorito:", favoritos[-1])

# Remova um dos animais da lista e imprima o resultado;
print("\nANIMAIS REMOVIDOS")
animais.remove("gato")
print("▪ Lista atualizada após a remoção:", animais)

# Adicione um ou mais animais na lista;
print("\nANIMAIS ADICIONADOS")
animais.append("peixe dourado")
animais.append("macaco")
print("▪ Lista atualizada após adição:", animais)

# Verifique quantos animais temos agora;
print("\nNOVA QUANTIDADE DE ANIMAIS")
print("▪ Quantidade de animais atualizadas:", len(animais))

# Troque a ordem dos animais favoritos de acordo com sua preferência e imprima.
print("\nNOVA ORDEM DOS ANIMAIS FAVORITOS")
favoritos = ["cachorro", "macaco", "coelho"]
print("▪ Ordem atualizada dos favoritos:", favoritos)
```
