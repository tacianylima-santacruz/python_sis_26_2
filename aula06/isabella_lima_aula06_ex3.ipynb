```python
# AULA 06 | EXERCÍCIO 3 | Agenda Telefônica (Dicionários)

# Crie um dicionário de contatos com alguns nomes (como chaves) e
# números de telefone (como valores);
contatos = {
    "Isabella": "(41) 91111-1111",
    "Cici": "(41) 92222-2222",
    "Isaque": "(41) 93333-3333"
}
print("\n☎️ AGENDA DE CONTATOS TELEFÔNICOS:")

for nome, telefone in contatos.items():
    print("▫️", nome, "-", telefone)

# Adicione um novo contato ao dicionário sem alterá-lo diretamente (na declaração original);
contatos["Kisame"] = "(41) 94444-4444"
print("\n👤 NOVO CONTATO ADICIONADO À AGENDA:")

for nome, telefone in contatos.items():
    print("▫️", nome, "-", telefone)

# Verifique se o contato "Chai" existe nos contatos.
# Se não, adicione-a com um número de telefone inventado.;
if "Chai" in contatos:
    print("\n✅ O contato Chai foi localizado:")
    print("▫️ Chai -", contatos["Chai"])
else:
    print("\n❌ O contato Chai não foi localizado.")
    contatos["Chai"] = "(41) 95555-5555"
    print("✅ O contato de Chai foi adicionado:", contatos["Chai"])

# Exclua um dos seus contatos do dicionário;
# use um método que não cause um erro se o contato não existir;
contatos.pop("Kisame", None)

# Imprima todos os nomes presentes no dicionário.
print("\n☎️ AGENDA DE CONTATOS TELEFÔNICOS ATUALIZADA:")

for nome, telefone in contatos.items():
    print("▫️", nome, "-", telefone)

# OBS.:Para cada ação imprima o resultado.
```
