```python
# AULA 06 | EXERCÍCIO 2 | O Cardápio Fixo do Axolote (Tuplas)

# Objetivo: Como sabemos, as tuplas são coleções imutáveis,
# sendo o formato ideal para armazenar dados que não devem ser modificados acidentalmente.
# O axolote de estimação tem uma rotina de alimentação muito restrita!

# Crie uma tupla contendo três elementos de tipos variados:
# o horário da refeição (texto),
# o alimento principal (texto) e
# a quantidade de vitaminas (número).
cardapio = ("10:00", "🐛 Larvas de insetos", 2)

# Tente alterar o alimento principal da tupla para "pizza de calabresa".
# Delete a linha que causou o erro.
# cardapio[1] = "pizza de calabresa" (Transformei a linha em comentário em vez de excluí-la.)
# Agora, utilize a técnica de desempacotamento de tuplas para extrair esses três valores e
# salvá-los em três variáveis independentes usando apenas 1 linha de código.
horario, alimento, vitaminas = cardapio

# Imprima uma frase amigável usando as variáveis desempacotadas.
print("Às", horario, "o Axolote come seu café da manhã preferido:\n", alimento, "com", vitaminas, "vitaminas.")
```
