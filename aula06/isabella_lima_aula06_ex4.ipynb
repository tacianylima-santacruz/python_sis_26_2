```python
# AULA 06 | EXERCÍCIO 4 | A Sacola do Clube do Livro (Conjuntos)

# Objetivo: Um conjunto (set) é uma estrutura que, por natureza,
# não permite elementos duplicados,
# sendo útil para remover duplicatas de coleções de dados automaticamente.
# Fui à feira de troca de livros com meus amigos e
# jogamos todos os livros que arrecadamos em uma grande sacola.
# Acabamos pegando várias cópias do mesmo livro!

# Crie uma lista inicial com cerca de 6 ou 7 títulos de livros,
# mas coloque de propósito uns 3 títulos repetidos.
livros = ["It", "O Iluminado", "Carrie", "It", "Misery", "O Iluminado", "Carrie", "Cemitério Maldito", "Christine", "Cujo", "Doutor Sono", "Salem"]

print("\n📚 LIVROS ARRECADADOS:")
for livro in livros:
    print("▪", livro)

# Converta essa sua lista em um Conjunto para forçar o sistema a eliminar
# todas as cópias duplicadas em um passe de mágica.
livros_unicos = set(livros)

# Imprima o conjunto na tela.
print("\n📖 LIVROS NÃO REPETIDOS:")
for livro in livros_unicos:
    print("▪", livro)

# Note que o resultado final perdeu as repetições e
# que os itens provavelmente mudaram de lugar,
# já que os conjuntos não mantêm a ordem de inserção dos elementos!
```
