```python
# Atividade Pratica 02 -O Analisador de Pergaminhos da Terra-Média
# Programa para limpar e analisar o texto do pergaminho

# Pede para o usuario digitar o texto
texto = input("Digite o texto do pergaminho: ")

# Remove os espacos do inicio e do final
texto_limpo = texto.strip()

# Divide o texto em palavras
palavras = texto_limpo.split()

# Conta o total de caracteres do texto
total_caracteres = len(texto_limpo)

# Conta o total de palavras
total_palavras = len(palavras)

# Pega a primeira palavra
primeira_palavra = palavras[0]

# Pega a ultima palavra
ultima_palavra = palavras[-1]

# Verifica se a palavra "Anel" existe no texto
tem_anel = "Anel" in texto_limpo

# Conta quantas vezes a palavra "Anel" aparece
quantidade_anel = texto_limpo.count("Anel")

# Mostra o relatorio
print("\n----- RELATORIO -----")
print(f"Texto limpo: {texto_limpo}")
print(f"Total de caracteres: {total_caracteres}")
print(f"Total de palavras: {total_palavras}")
print(f"Primeira palavra: {primeira_palavra}")
print(f"Ultima palavra: {ultima_palavra}")

# Mostra o resultado do detector de ameacas
print(f"Existe a palavra Anel? {tem_anel}")
print(f"A palavra Anel aparece {quantidade_anel} vezes.")



```

```python
# Atividade Pratica 03 - A Jornada do Personagem

# Dados do personagem
nome = "HarryPotter"
idade = 18
altura = 1.58
bruxo = True

# Calculo 1 - dano do ataque
forca = 20
bonus_espada = 10
dano = forca + bonus_espada

# Calculo 2 - vida depois do ataque
vida = 100
dano_monstro = 30
vida_atual = vida - dano_monstro

# Calculo 3 - pontos de experiencia
experiencia = 50
pontos_ganhos = 25
experiencia_total = experiencia + pontos_ganhos

# Mostra os dados
print(f"Nome: {nome}")
print(f"Idade: {idade}")
print(f"Altura: {altura}")
print(f"Bruxo: {bruxo}")

# Decisao 1 - verifica o dano
if dano > 25:
    print("Harry causou um ataque forte!")
else:
    print("Harry causou um ataque fraco.")

# Decisao 2 - verifica a vida
if vida_atual <= 0:
    print("Game Over!")
else:
    print(f"Harry continua vivo com {vida_atual} de vida.")

# Decisao 3 - verifica a experiencia
if experiencia_total >= 70:
    print("Harry subiu de nivel!")
else:
    print("Harry ainda precisa ganhar mais experiencia.")

# Mostra os calculos
print(f"Dano do ataque: {dano}")
print(f"Experiencia total: {experiencia_total}")
```
