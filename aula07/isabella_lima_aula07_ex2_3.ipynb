```python
# AULA 07 | EXERCÍCIO 2 | Parte 03 - Bugs na Trilha

# Você está escrevendo um programa que calcula a distância total de uma trilha.
# Os dados são flexíveis, e cada segmento da trilha é passado como uma tupla que consiste na distância e na unidade.
# A função deve imprimir o total de todos esses segmentos.
# O programa roda sem travar (não quebra), mas algo sobre esse total parece estar errado...
# Use técnicas de depuração (debugging) para descobrir o que há de errado com este código e conserte o bug:

def total_distance(*segments):
    total = 0
    for length, unit in segments:
         if unit.strip().lower() == "m": # Correção: remove espaços e transforma a unidade em minúscula.
            total += length
         elif unit.strip().lower() == "km":
            total += length * 1000 # Correção: quilômetros devem ser multiplicados por 1000 para converter em metros.

    print("Total distance:", total, "meters")

total_distance(
    (2.5, "km"),
    (400, "M"),
    (1.0, "km"),
    (800, "m"),
    (1.3, "KM"),
    (900, "m "),
    (1.1, "km")
)
```
