```python
'''laura_lopes_aula_07_exercicio-01'''

def nap_time(hours_list):
    total = 0
    counter = 0
    while counter < len(hours_list):
        total += hours_list[counter]
        print(total)
        counter += 1
    return total
cat_naps = [2, 3, 1.5, 4]
print("Total nap time:", nap_time(cat_naps))

#total estava começando a contar do indice[1] ao invez de [0]
#pois contador começõu com valor 1
#ajustado para 0
```

```python
'''laura_lopes_aula_07_exercicio-02'''

def average_meows(logs):
    total_meows_per_hour = 0
    valid_days = 0

    for log in logs:
        nap_hours = 0

        for nap in log["naps"]:
            nap_hours += nap
            print(nap_hours)

        awake_hours = 24 - nap_hours

        if awake_hours > 0:
            total_meows_per_hour += log["meows"] / awake_hours
            valid_days += 1

    return total_meows_per_hour / valid_days


logs = [
    {"day": "Monday", "naps": [1, 2, 5], "meows": 90},
    {"day": "Tuesday", "naps": [4, 3, 2], "meows": 75},
    {"day": "Wednesday", "naps": [8, 6, 9, 1], "meows": 9},
    {"day": "Thursday", "naps": [1, 1, 5, 7], "meows": 38},
    {"day": "Friday", "naps": [3, 3, 2, 8], "meows": 94},
    {"day": "Saturday", "naps": [4, 8, 3], "meows": 64},
    {"day": "Sunday", "naps": [3, 3, 2, 8], "meows": 37}
]

print("Average meows per awake hour:", average_meows(logs))

# ZeroDivisionError
# dividindo por 0 em:
# print("Average meows per awake hour:", average_meows(logs))
# e total_meows_per_hour += log["meows"] / awake_hours
#
# com print(nap_hours) dentro do for eu vi que ele esta verificando
# e somando os valores dos naps de cada dia.
# Na quarta-feira, a soma dos naps chega a 24,
# fazendo awake_hours virar 0, pois 24 - 24 é 0.
#
# O erro acontece porque o programa tenta calcular meows / 0.
# Para evitar o erro, coloquei if awake_hours > 0 para só calcular
# os miados por hora quando o gato teve pelo menos uma hora acordado.
# Também adicionei valid_days para contar somente os dias válidos na média.
```

```python
'''laura_lopes_aula_07_exercicio-02'''

def total_distance(*segments):
    total = 0

    for length, unit in segments:

        # DEBUGGING - Técnica: inspeção/print debugging
        # Mostramos os valores recebidos para descobrir
        # por que alguns segmentos não estão sendo somados.
        print("DEBUG:", length, unit)

        # DEBUGGING - Técnica: normalização dos dados
        # .strip() remove espaços extras, como em "m "
        # .lower() transforma "M", "KM", "Km" etc. em minúsculas.
        unit = unit.strip().lower()

        # DEBUGGING - Bug encontrado:
        # O código original fazia:
        # if unit.lower() == "M":
        #
        # O problema é que .lower() transforma "M" em "m".
        # Portanto, comparar com "M" nunca seria verdadeiro.
        if unit == "m":
            total += length

        # DEBUGGING - Bug encontrado:
        # O código original fazia:
        # total += length / 1000
        #
        # Como queremos o resultado final em METROS,
        # devemos transformar km em metros:
        # 1 km = 1000 metros
        elif unit == "km":
            total += length * 1000

    print("Total distance:", total, "meters")


total_distance(
    (2.5, "km"),
    (400, "M"),
    (1.0, "km"),
    (800, "m"),
    (1.3, "KM"),
    (900, "m "),
    (1.1, "km")
)
```

```python
'''laura_lopes_aula_07_exercicio-03'''

from IPython.display import clear_output

def ticket_type():
    while True:
        clear_output()

        try:
            age = int(input("Enter your age: "))

            if age < 0:
                print("Please enter a valid age.")
                input("Press Enter to try again...")
                continue

            if age < 12:
                print("You get in for free. Lucky you!")
            elif age < 18:
                print("You get a youth discount!")
            elif age < 65:
                print("That's a full-price ticket for you.")
            else:
                print("You get a senior discount!")

            print("Bem Vindo!")
            break

        except ValueError:
            print("Numero invalido")
            input("Digite Enter para tentar de novo...")

ticket_type()
```
