```python
# AULA 07 | EXERCÍCIO 3 | Caça aos Bugs e Tratamento de Erros - Ingressos do museu

# Você está ajudando um museu local a configurar uma máquina de ingressos.
# Você escreveu um programa em Python para ela: o visitante digita a sua idade, e a máquina imprime o tipo de ingresso que ele recebe.
# A lógica é simples:

# Crianças menores de 12 anos têm acesso gratuito;
# Adolescentes menores de 18 anos recebem um desconto para jovens;
# Adultos menores de 65 anos pagam o valor integral;
# Pessoas com 65 anos ou mais recebem um desconto para idosos;
# Porém você deve tratar o código para que independente doque o visitante digite o programa não dê erro.

# Caso seja uma idade válida o programa deve dar as boas vindas. Caso não deve pedir redigitação.

# E independente do usuário digita certo ou errado o sistema deve limpar a tela para aguardar pelo próximo visitante.

def ticket_type():
    while True:
        try:
            age = int(input("Enter your age: ")) # Correção: converte a idade para inteiro.
            if age < 12:
                print("You get in for free. Lucky you!")
            elif age < 18:
                print("You get a youth discount!")
            elif age < 65:
                print("That's a full-price ticket for you.")
            else:
                print("You get a senior discount!")
        except ValueError: # Correção: evita erro se o usuário digitar um valor inválido.
            print("Invalid age. Please try again.")
        else: # Correção: dá boas-vindas quando a idade digitada é válida.
            print("Welcome!")
            break
        finally: # Executado independentemente de ocorrer erro ou não, simulando a limpeza da tela.
            print("Cleaning screen...")

ticket_type()
```
