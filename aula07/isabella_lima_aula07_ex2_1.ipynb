```python
# AULA 07 | EXERCÍCIO 2 | Parte 01 - Caça aos Bugs - Soneca de gato com bugs

# Nós escrevemos um programa para calcular o tempo total que o gato dorme em um dia.
# Mas algo parece estranho...
# O total que a nossa função retorna não corresponde ao tempo real que o gato passou tirando sonecas:
# Mesmo que você identifique o erro neste código logo de cara, pratique usar comandos print,
# um ponto de interrupção (breakpoint) ou o registro de rastreamento (traceback log)
# para descobrir o que está errado e com que tipo de bug estamos lidando.
# Além disso, corrija o problema para que a função retorne o tempo total de soneca correto.

def nap_time(hours_list):
    total = 0
    counter = 0 # Correção: "counter = 1" fazia a contagem começar no segundo item da lista.
    while counter < len(hours_list):
        total += hours_list[counter]
        counter += 1
    return total
cat_naps = [2, 3, 1.5, 4]
print("Total nap time:", nap_time(cat_naps))
```
