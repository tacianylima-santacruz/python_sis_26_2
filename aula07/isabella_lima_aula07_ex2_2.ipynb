```python
# AULA 07 | EXERCÍCIO 2 | Parte 02 - Quanto o seu gato mia?

# Imagine que você escreveu este programa para calcular com que frequência o seu gato mia, em média, por hora acordado.
# Você manteve um registro de quanto ele miou e de quanto ele dormiu a cada dia.
# Você arredondou os tempos de soneca; afinal, você é um desenvolvedor de software, não um veterinário.
# Tudo no seu código parece estar perfeitamente bem. Mas adivinha? O programa trava mesmo assim.
# Hora de depurar (debug)! Use o traceback, declarações print e/ou pontos de interrupção (breakpoints)
# para descobrir o que está causando o travamento e com que tipo de bug você está lidando.
# Pense em uma maneira inteligente de modificar o código para que ele não trave mais:

def average_meows(logs):
    total_meows_per_hour = 0

    for log in logs:
        nap_hours = 0

        for nap in log["naps"]:
            nap_hours += nap

        awake_hours = 24 - nap_hours
        if awake_hours > 0: # Correção: verifica se o gato ficou acordado antes de realizar a divisão.
           total_meows_per_hour += log["meows"] / awake_hours

    return total_meows_per_hour / len(logs)

logs = [
    {"day": "Monday", "naps": [1, 2, 5], "meows": 90},
    {"day": "Tuesday", "naps": [4, 3, 2], "meows": 75},
    {"day": "Wednesday", "naps": [8, 6, 9, 1], "meows": 9},
    {"day": "Thursday", "naps": [1, 1, 5, 7], "meows": 38},
    {"day": "Friday", "naps": [3, 3, 2, 8], "meows": 94},
    {"day": "Saturday", "naps": [4, 8, 3], "meows": 64},
    {"day": "Sunday", "naps": [3, 3, 2, 8], "meows": 37}
]

print("Average meows per awake hour:", average_meows(logs))
```
