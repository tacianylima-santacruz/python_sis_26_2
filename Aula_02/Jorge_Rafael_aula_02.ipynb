```python
# ===== Variáveis do personagem =====
nome = str("Jorge")     # Nome do personagem (texto)
idade = int(28)                    # Idade/nível do personagem (número inteiro)
altura = float(1.82)               # Altura do personagem em metros (número decimal)
arma = True                    # Indica se o personagem possui armadura (booleano

class Personagem:
    """
    Classe que representa a ficha de um personagem.

    Atributos:
        nome (str): Nome do personagem.
        idade (int): Idade ou nível do personagem.
        altura (float): Altura do personagem, em metros.
        armadura (bool): Indica se o personagem possui armadura (True) ou não (False).
        classificacao (str): Classe/categoria do personagem (ex: Guerreiro, Mago, Arqueiro).
    """

    def __init__(self, nome, idade, altura, arma, soldado):
        """Exibe as informações do personagem de forma formatada."""
        self.nome = nome
        self.idade = idade
        self.altura = altura
        self.arma = arma
        self.soldado = soldado

    def exibir_dados(self):
        print(f"Nome: {self.nome}")
        print(f"Idade: {self.idade}")
        print(f"Altura: {self.altura}")
        print(f"Armadura: {'Sim' if self.arma else 'Não'}")
        print(f"Classe: {self.soldado}")

  # Criando o personagem com os dados definidos acima
Jorge = Personagem(nome, idade, altura, arma, soldado = "Marechal")
Jorge.exibir_dados()

# Consultando a documentação da classe
help(Personagem)


# ===== Cadastro dinâmico do personagem/usuário =====

nome_usuario = input("Digite o nome do Personagem:")
idade_usuario = int(input("Digite a idade do Personagem:"))
altura_usuario = float(input("Digite a altura do Personagem:"))
arma_usuario = input("O Personagem possiu armar ? (S/N)".strip().upper()== "S" )

# Convertendo a entrada "S"/"N" em um valor booleano (True/False)

# ===== Exibindo a ficha do personagem =====

print("\n===Ficha do Personagem=== ")
print(f"Nome: {nome_usuario}")
print(f"Idade: {idade_usuario}")
print(f"Altura: {altura_usuario}")
print(f"Armadura: {'Sim' if arma_usuario else 'Não'}")

# ===== Reaproveitando dados do personagem =====
nome_usuario = input("Digite o nome do personagem: ")
idade_usuario = int(input("Digite a idade do personagem: "))
altura_usuario = float(input("Digite a altura do personagem (em metros, ex: 1.75): "))
vida_atual = int(input("Digite a vida atual do personagem: "))
vida_maxima = int(input("Digite a vida máxima do personagem: "))

# ===== Novas variáveis com operações aritméticas =====

# 1. Cálculo de quantos anos faltam para o personagem atingir a maioridade "élfica" (ex: 100 anos)
anos_para_maioridade = 100 - idade_usuario

# 2. Cálculo da porcentagem de vida atual em relação à vida máxima
porcentagem_vida = (vida_atual / vida_maxima) * 100

# 3. Cálculo de quantos "níveis" o personagem já passou, supondo 1 nível a cada 5 anos de idade
niveis_por_idade = idade_usuario // 5   # // é a divisão inteira (descarta a parte decimal)

# ===== Comparações (operadores relacionais) =====
esta_com_vida_baixa = porcentagem_vida < 30          # True se a vida estiver abaixo de 30%
e_maior_de_idade = idade_usuario >= 18               # True se tiver 18 anos ou mais
altura_acima_da_media = altura_usuario > 1.70         # True se for mais alto que 1.70m

# ===== Exibindo os resultados =====
print("\n===== FICHA DO PERSONAGEM =====")
print(f"Nome: {nome_usuario}")
print(f"Idade: {idade_usuario} anos")
print(f"Altura: {altura_usuario} m")
print(f"Vida: {vida_atual}/{vida_maxima}")

print("\n===== CÁLCULOS =====")
print(f"Anos até a maioridade élfica (100 anos): {anos_para_maioridade}")
print(f"Porcentagem de vida atual: {porcentagem_vida:.1f}%")
print(f"Nível estimado pela idade: {niveis_por_idade}")

print("\n===== COMPARAÇÕES =====")
print(f"Está com vida baixa (abaixo de 30%)? {esta_com_vida_baixa}")
print(f"É maior de idade (18+)? {e_maior_de_idade}")
print(f"Altura acima da média (1.70m)? {altura_acima_da_media}")
```
