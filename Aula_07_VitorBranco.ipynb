```python
def contagem_regressiva(tempo):

    if tempo <= 0:
        print("BOOM! O Anel explodiu!")
        return

    print(f"Faltam {tempo} segundos...")

    # Passo Recursivo: A função chama a si mesma
    contagem_regressiva(tempo - 1)

print("Iniciando contagem para a explosão do Anel...")
contagem_regressiva(3)
```

```python
def average_meows(logs):
    total_meows_per_hour = 0
    valid_days = 0  # Vamos contar apenas os dias em que o gato ficou acordado

    for log in logs:
        nap_hours = 0

        for nap in log["naps"]:
            nap_hours += nap

        awake_hours = 24 - nap_hours

        # Só fazemos a divisão se o gato passou alguma hora acordado
        if awake_hours > 0:
            total_meows_per_hour += log["meows"] / awake_hours
            valid_days += 1
        else:
            print(f"Aviso: O gato hibernou na {log['day']}! Ignorando este dia na média.")

    # Protegendo a média final também!
    if valid_days > 0:
        return total_meows_per_hour / valid_days
    else:
        return 0

logs = [
    {"day": "Monday", "naps": [1, 2, 5], "meows": 90},
    {"day": "Tuesday", "naps": [4, 3, 2], "meows": 75},
    {"day": "Wednesday", "naps": [8, 6, 9, 1], "meows": 9},
    {"day": "Thursday", "naps": [1, 1, 5, 7], "meows": 38},
    {"day": "Friday", "naps": [3, 3, 2, 8], "meows": 94},
    {"day": "Saturday", "naps": [4, 8, 3], "meows": 64},
    {"day": "Sunday", "naps": [3, 3, 2, 8], "meows": 37}
]

print("Média de miados por hora acordada:", average_meows(logs))
```

```python
def total_distance(*segments):
    total = 0
    for length, unit in segments:
        # .strip() remove espaços nas pontas e .lower() deixa tudo minúsculo
        clean_unit = unit.strip().lower()

        if clean_unit == "m":
            total += length
        elif clean_unit == "km":
            total += length * 1000 # Para converter km em m, multiplicamos!

    print("Total distance:", total, "meters")

total_distance(
    (2.5, "km"),
    (400, "M"),
    (1.0, "km"),
    (800, "m"),
    (1.3, "KM"),
    (900, "m "),
    (1.1, "km")
)
```
