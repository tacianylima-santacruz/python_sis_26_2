Atividade Prática 03 - Atividade operadores e conversão
Agora que você já sabe capturar dados com input(), converter tipos, fazer cálculos e comparações com operadores relacionais, é hora inserir essa dinâmica no seu próprio personagem. Crie pelo menos 3 novas variáveis com operações aritiméticas e comparações. Mostre o resultado dos seus cálculos e testes usando print()

```python
nick_personagem = input("Digite seu nome: ")
raca_personagem = input("Digite sua raça: ")
classe_personagem = input("Digite sua classe: ")
weapon_personagem = input("Digite sua arma: ")

# Mostrando seu personagem criado
print("Bem-vindo", nick_personagem, "!")
print("Sua raça escolhida foi:", raca_personagem)
print("Sua classe é:", classe_personagem)
print("Sua arma inicial é:", weapon_personagem)
print("Agora, equipe-se e prepare-se para uma longa jornada!")

# Atributos do personagem
vida = int(input("Digite a vida do personagem: "))
ataque = int(input("Digite o ataque do personagem: "))
defesa = int(input("Digite a defesa do personagem: "))
nivel = int(input("Digite o nível do personagem: "))

# Operações aritméticas (3 variáveis)
poder_total = ataque + defesa
vida_restante = vida - 20
ataque_dobrado = ataque * 2

# Comparações (operadores relacionais)
nivel_alto = nivel >= 10
vida_baixa = vida < 30
ataque_maior_defesa = ataque > defesa

# Exibindo os resultados
print("\n--- Resultados ---")
print("Poder total (ataque + defesa):", poder_total)
print("Vida restante após dano de 20:", vida_restante)
print("Ataque dobrado:", ataque_dobrado)

print("\n--- Comparações ---")
print("O personagem é nível 10 ou maior?", nivel_alto)
print("A vida está abaixo de 30?", vida_baixa)
print("O ataque é maior que a defesa?", ataque_maior_defesa)
```
