# Comentários e Docstrings


```python
def calcular_ataque(forca, bonus_arma):
    """
    Calcula o poder de ataque total de um personagem.
    Recebe a força base e o bônus da arma equipada, retornando o valor final do dano.
    """

    # Adicionamos 2 pontos extras de dano por padrão devido ao bônus da classe Guerreiro
    dano_total = forca + bonus_arma + 2

    return dano_total
```

# Operadores Artiméticos

```python
servidores_tipo_1 = 10
servidores_tipo_2 = 5
print(servidores_tipo_1 + servidores_tipo_2)
```

```python
custo_servidor_tipo_1 = 1000.0
custo_servidor_tipo_2 = 500.5
print(custo_servidor_tipo_1 + custo_servidor_tipo_2)
```

```python
media = (custo_servidor_tipo_1 + custo_servidor_tipo_2) / 2
print(media)
```

```python
media = (custo_servidor_tipo_1 + custo_servidor_tipo_2) // 2
print(media)
```

```python
resto = (custo_servidor_tipo_1 + custo_servidor_tipo_2) % 2
print(resto)
```

# Conversão de Tipos

```python
primeiro_numero = input ("Digite um número:")
segundo_numero = input ("Digite outro número:")
print(primeiro_numero + segundo_numero)
```

```python
# O input normal lê como texto. Colocamos o int() em volta para já converter.
nivel_personagem = int(input("Qual o nível atual do seu Personagem? "))

# Agora que é um número (int), podemos fazer contas!
novo_nivel = nivel_personagem + 1

# Como a conta já foi feita, convertemos o novo_nivel de volta para texto (str)
# para conseguirmos juntar (concatenar) com a frase usando o sinal de +
print ("Parabéns! Você ganhou XP e subiu para o nível " + str(novo_nivel) + "!")
```
