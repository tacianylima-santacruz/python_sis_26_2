## Atividade Prática 02 - Atividade input() e print()
Agora permitimos que o usuário cadastre dinamicamente os dados de um novo personagem através de entradas do teclado.

```python
# Cadastro dinâmico do personagem via entrada do usuário
print("--- CADASTRO DE PERSONAGEM ---")
novo_nome = input("Digite o nome do personagem: ")
nova_idade = int(input("Digite a idade: "))
nova_altura = float(input("Digite a altura (ex: 1.75): "))
tem_armadura_str = input("O personagem possui armadura? (S/N): ")
tem_armadura = tem_armadura_str.strip().upper() == 'S'

# Exibição da Ficha do Personagem cadastrado
print("\n=============================")
print("     FICHA DO PERSONAGEM     ")
print("=============================")
print(f"Nome:     {novo_nome}")
print(f"Idade:    {nova_idade} anos")
print(f"Altura:   {nova_altura} m")
print(f"Armadura: {'Sim' if tem_armadura else 'Não'}")
print("=============================")
```
