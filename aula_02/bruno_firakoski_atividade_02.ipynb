# 🎮 Atividade Prática 01: Montando a Ficha de um Personagem

Crie **4 variáveis**, utilizando os **4 tipos de dados** apresentados abaixo:

### 1. 📝 Texto (`str`)

Defina o **nome do seu personagem**.

> Exemplo: `nome = "Arthur"`

### 2. 🔢 Número Inteiro (`int`)

Defina a **idade ou o nível** do personagem.

> Exemplo: `idade = 25`

### 3. 📏 Número Decimal (`float`)

Defina a **altura do personagem**.

> Exemplo: `altura = 1.80`

### 4. 🛡️ Verdadeiro ou Falso (`bool`)

Diga se o personagem **possui uma armadura** ou não.

> Exemplo: `armadura = True`


```python
##Criação de personagem:

# Título do print
print("Character creator:")

# Nome do persongaem, formato padrão string
character_name = input("Type the character's name: ")

# Idade do persongaem, "int" converte o conteúdo do input em INTEIRO
character_age = int(input("Type the character's age: "))

# Altura do personagem, "float" transforma o conteúdo do input em FLOAT
character_height = float(input("Type the character's height: "))

# Criação da varíavel armadura do personagem, atribuindo ela a um input
character_armor = input("""
Armor

[1] The character has armor
[0] The character has no armor

Your choice: """)

# "bool" converte a resposta do input acima para BOOLEAN
armor = bool(int(character_armor))

# Print final:
print(
    f"""
Name: {character_name}
Age: {character_age}
Height: {character_height}
Armor: {armor}
"""
)
```
