```python

# Atividade Prática 02: Cadastro do Personagem

# Solicita os dados ao usuário
nome_personagem = input("Digite o nome do personagem: ")
idade_personagem = int(input("Digite a idade do personagem: "))
altura_personagem = float(input("Digite a altura do personagem em metros: "))
possui_armadura = input("O personagem possui armadura? (sim/não): ")

# Exibe a ficha do personagem
print("\n===== FICHA DO PERSONAGEM =====")
print("Nome:", nome_personagem)
print("Idade:", idade_personagem, "anos")
print("Altura:", altura_personagem, "metros")
print("Possui armadura:", possui_armadura)

print("===============================")
```

```python

# Atividade Prática 03 - Operadores e Conversão

# Cadastro do personagem
nome = input("Digite o nome do personagem: ")
forca = int(input("Digite a força do personagem: "))
moedas = float(input("Quantas moedas o personagem possui? "))

# Cálculo 1: força após receber um bônus
bonus_forca = 10
forca_total = forca + bonus_forca

# Cálculo 2: moedas após comprar uma poção
preco_pocao = 15
moedas_restantes = moedas - preco_pocao

# Cálculo 3: experiência após completar uma missão
experiencia = 100
bonus_experiencia = 50
experiencia_total = experiencia + bonus_experiencia

# Comparação 1: verificar se a força é suficiente
forca_suficiente = forca_total >= 30

# Comparação 2: verificar se pode comprar a poção
pode_comprar = moedas >= preco_pocao

# Comparação 3: verificar se completou a experiência necessária
missao_completa = experiencia_total >= 150

# Exibição dos resultados
print("\n===== FICHA DO PERSONAGEM =====")
print("Nome:", nome)

print("\n===== CÁLCULOS =====")
print("Força total:", forca_total)
print("Moedas restantes:", moedas_restantes)
print("Experiência total:", experiencia_total)

print("\n===== COMPARAÇÕES =====")
print("Força suficiente?", forca_suficiente)
print("Pode comprar a poção?", pode_comprar)
print("Missão completa?", missao_completa)
```
