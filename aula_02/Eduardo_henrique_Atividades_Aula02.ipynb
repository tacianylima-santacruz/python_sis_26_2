```python
class Ourico:
    """Classe que representa um personagem Ourico"""

    def __init__(self, nome, nivel, altura, super_forma):
        self.nome = nome                # nome do personagem
        self.nivel = nivel              # nível do personagem
        self.altura = altura            # altura do personagem em metros
        self.super_forma = super_forma  # se está em super forma (True/False)


# Cadastro dos dados via input()
nome = input("Digite o nome do personagem: ")
nivel = int(input("Digite o nível do personagem: "))
altura = float(input("Digite a altura do personagem (em metros): "))
resposta = input("O personagem está em super forma? (s/n): ")
super_forma = resposta.strip().lower() == "s"

# Criando o personagem com os dados digitados
nazo = Ourico(nome, nivel, altura, super_forma)

# Mostrando a ficha do personagem
print("\n--- Ficha do Personagem ---")
print("Nome:", nazo.nome)
print("Nível:", nazo.nivel)
print("Altura:", nazo.altura, "m")
print("Super forma:", "Sim" if nazo.super_forma else "Não")
```

```python
class Ourico:
    """Classe que representa um personagem Ourico"""

    def __init__(self, nome, nivel, altura, super_forma):
        self.nome = nome
        self.nivel = nivel
        self.altura = altura
        self.super_forma = super_forma


# Cadastro dos dados via input()
nome = input("Digite o nome do personagem: ")
nivel = int(input("Digite o nível do personagem: "))
altura = float(input("Digite a altura do personagem (em metros): "))
resposta = input("O personagem está em super forma? (s/n): ")
super_forma = resposta.strip().lower() == "s"

nazo = Ourico(nome, nivel, altura, super_forma)

print("\n--- Ficha do Personagem ---")
print("Nome:", nazo.nome)
print("Nível:", nazo.nivel)
print("Altura:", nazo.altura, "m")
print("Super forma:", "Sim" if nazo.super_forma else "Não")

# ---------- Atividade Prática 03 ----------

# 1) Operador aritmético: nível dobrado se estiver em super forma
nivel_em_super_forma = nazo.nivel * 2

# 2) Operador aritmético: quantos níveis faltam para chegar ao nível 9999
niveis_faltando = 9999 - nazo.nivel

# 3) Operador aritmético: proporção altura/nível (só de exemplo de divisão)
proporcao_altura_nivel = nazo.altura / nazo.nivel

# 4) Operador relacional: o personagem já é "nível alto" (acima de 100)?
eh_nivel_alto = nazo.nivel > 100

# 5) Operador relacional: o personagem tem altura igual a 1.40m?
tem_altura_padrao = nazo.altura == 1.60

print("\n--- Cálculos e Testes ---")
print("Nível em super forma (dobrado):", nivel_em_super_forma)
print("Níveis faltando para o nível 9999:", niveis_faltando)
print("Proporção altura/nível:", proporcao_altura_nivel)
print("É nível alto (>100)?", eh_nivel_alto)
print("Tem altura padrão (1.40m)?", tem_altura_padrao)
```
