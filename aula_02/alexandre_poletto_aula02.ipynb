```python
#Prática 03

#Dano do personagem

def calcular_dano(nivel, arma):
    base_dano = 0
    if arma == "espada":
        base_dano = 10
    elif arma == "arco":
        base_dano = 8
    elif arma == "machado":
        base_dano = 12

    dano_total = base_dano + (nivel * 2)
    return dano_total

#EXEMPLO
dano_causado = calcular_dano(nivel_novo_personagem, arma_novo_personagem)
print(f"O personagem {novo_personagem} com a arma {arma_novo_personagem} causa {dano_causado} de dano.")
```
