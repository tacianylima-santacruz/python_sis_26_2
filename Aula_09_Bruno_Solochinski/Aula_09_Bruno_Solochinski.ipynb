```python
# Etapa 1: cria o arquivo (w)
with open("acessos.log", "w", encoding="utf-8") as f:
    f.write("=== SISTEMA DE AUDITORIA DE ACESSOS ===\n")
    f.write("INFO - Sistema inicializado\n")

# Etapa 2: adiciona eventos (a)
with open("acessos.log", "a", encoding="utf-8") as f:
    f.write("LOGIN - usuario admin entrou\n")
    f.write("INFO - arquivo config.ini lido\n")
    f.write("ERRO - senha incorreta para usuario joao\n")
    f.write("LOGIN - usuario maria entrou\n")
    f.write("ERRO - acesso negado ao diretorio /root\n")

# Etapa 3: leitura seletiva
with open("acessos.log", "r", encoding="utf-8") as f:
    print(f.readline().strip())        # primeira linha (título)
    print(f.readlines()[-1].strip())   # última linha

# Etapa 4: auditoria linha a linha
total = 0
with open("acessos.log", "r", encoding="utf-8") as f:
    for linha in f:
        if "LOGIN" in linha or "ERRO" in linha:
            print(linha.strip())
            total += 1
print("Total de LOGIN/ERRO:", total)
```

```python
# Etapa 1: arquivo inexistente
try:
    open("servidor_inexistente.log")
except FileNotFoundError:
    print("Arquivo servidor_inexistente.log não encontrado.")

# Etapa 2: escrita e append protegidos
try:
    with open("acessos.log", "w", encoding="utf-8") as f:
        f.write("=== SISTEMA DE AUDITORIA DE ACESSOS ===\n")
    with open("acessos.log", "a", encoding="utf-8") as f:
        f.write("ERRO - autenticação falhou\n")
except PermissionError:
    print("Sem permissão para escrever no arquivo.")

# Etapa 3: encoding incompatível (arquivo tem acento, lido como ascii)
try:
    with open("acessos.log", "r", encoding="ascii") as f:
        f.read()
except UnicodeDecodeError:
    print("Encoding incompatível com o arquivo.")

# Etapa 4: finally garante o fechamento
f = None
try:
    f = open("acessos.log", "r", encoding="utf-8")
    print(f.readline().strip())
except Exception as e:
    print("Erro inesperado:", e)
finally:
    if f:
        f.close()
        print("Arquivo fechado.")
```
