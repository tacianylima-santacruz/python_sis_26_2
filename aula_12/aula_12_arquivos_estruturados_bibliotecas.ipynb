# Monitoramento de recursos

```python
import psutil
import time
```

CPU

```python
# 1. Coleta de Dados de Processamento (CPU)
carga_cpu_pct = psutil.cpu_percent(interval=1.0)
nucleos_logicos = psutil.cpu_count(logical=True)
print(f"Processador (CPU): {carga_cpu_pct}% alocado")
print(f"({nucleos_logicos} núcleos lógicos)")
print(carga_cpu_pct)
print(nucleos_logicos)
```

Memória RAM

```python
# 2. Coleta de Dados de Memória RAM
memoria_ram = psutil.virtual_memory()
total_ram_gb = memoria_ram.total / (1024 ** 3)
disponivel_ram_gb = memoria_ram.available / (1024 ** 3)
uso_ram_pct = memoria_ram.percent
print(f"Memória RAM: {uso_ram_pct}% em uso")
print(f"{disponivel_ram_gb:.2f} GB livres de {total_ram_gb:.2f} GB")
print(memoria_ram)
print(memoria_ram.total)
print(memoria_ram.available)
print(memoria_ram.percent)
```

Armazenamento em disco

```python
# 3. Coleta de Dados de Armazenamento (Disco Principal)
armazenamento_disco = psutil.disk_usage('/')
total_disco_gb = armazenamento_disco.total / (1024 ** 3)
livre_disco_gb = armazenamento_disco.free / (1024 ** 3)
uso_disco_pct = armazenamento_disco.percent
print(f"Armazenamento Disco: {uso_disco_pct}% alocado")
print(f"{livre_disco_gb:.2f} GB livres de {total_disco_gb:.2f} GB")
print(armazenamento_disco)
print(armazenamento_disco.total)
print(armazenamento_disco.free)
print(armazenamento_disco.percent)
```

Tráfego de rede

```python
# 4. Coleta de Tráfego I/O de Rede (Inclusão Obrigatória)
telemetria_rede = psutil.net_io_counters()
bytes_enviados_mb = telemetria_rede.bytes_sent / (1024 ** 2)
bytes_recebidos_mb = telemetria_rede.bytes_recv / (1024 ** 2)
print(f"Tráfego I/O Rede: TX: {bytes_enviados_mb:.2f} MB")
print("RX: {bytes_recebidos_mb:.2f} MB")
print(telemetria_rede)
print(telemetria_rede.bytes_sent)
print(telemetria_rede.bytes_recv)
```

```python
import psutil
import time


def executar_pipeline_telemetria_infraestrutura():
    print("==================================================================")
    print("   PIPELINE DE TELEMETRIA E OBSERVABILIDADE DE INFRAESTRUTURA     ")
    print("==================================================================")

    # 1. Coleta de Dados de Processamento (CPU)
    carga_cpu_pct = psutil.cpu_percent(interval=1.0)
    nucleos_logicos = psutil.cpu_count(logical=True)

    # 2. Coleta de Dados de Memória RAM
    memoria_ram = psutil.virtual_memory()
    total_ram_gb = memoria_ram.total / (1024 ** 3)
    disponivel_ram_gb = memoria_ram.available / (1024 ** 3)
    uso_ram_pct = memoria_ram.percent

    # 3. Coleta de Dados de Armazenamento (Disco Principal)
    armazenamento_disco = psutil.disk_usage('/')
    total_disco_gb = armazenamento_disco.total / (1024 ** 3)
    livre_disco_gb = armazenamento_disco.free / (1024 ** 3)
    uso_disco_pct = armazenamento_disco.percent

    # 4. Coleta de Tráfego I/O de Rede (Inclusão Obrigatória)
    telemetria_rede = psutil.net_io_counters()
    bytes_enviados_mb = telemetria_rede.bytes_sent / (1024 ** 2)
    bytes_recebidos_mb = telemetria_rede.bytes_recv / (1024 ** 2)

    # Exibição Estruturada no Console
    print(f"Processador (CPU)   : {carga_cpu_pct}% alocado ({nucleos_logicos} núcleos lógicos)")
    print(f"Memória RAM         : {uso_ram_pct}% em uso | {disponivel_ram_gb:.2f} GB livres de {total_ram_gb:.2f} GB")
    print(f"Armazenamento Disco : {uso_disco_pct}% alocado | {livre_disco_gb:.2f} GB livres de {total_disco_gb:.2f} GB")
    print(f"Tráfego I/O Rede    : TX: {bytes_enviados_mb:.2f} MB | RX: {bytes_recebidos_mb:.2f} MB")
    print("==================================================================")


# Invocação do agente de telemetria
executar_pipeline_telemetria_infraestrutura()
```
