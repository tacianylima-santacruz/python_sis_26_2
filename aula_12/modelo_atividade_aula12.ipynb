# Atividade Prática 01 Aula 12

Você é responsável pela gestão de ativos de TI e precisa gerar uma rotina de auditoria mensal que automatize a criação do inventário de servidores e converta informações brutas de arquivos de texto em uma planilha Excel para facilitar a visualização gerencial.


Etapa 1: Instalação e Criação.

* Instale e importe a biblioteca openpyxl seguindo a PEP 8.
* Crie uma pasta de trabalho (planilha) com uma aba para o servidores.
* Adicione um cabeçalho:
  * Ex: "Hostname", "IP", "Memória (GB)", "Uso de Disco (%)", "Status".

```python
!pip install openpyxl

from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill

workbook = Workbook()
planilha = workbook.active
planilha.title = "Servidores"

planilha.append([
    "Hostname",
    "IP",
    "Memória (GB)",
    "Uso de Disco (%)",
    "Status"
])

servidores = [
    ["Servidor-01", "192.168.0.1", 16, 65, "Ativo"],
    ["Servidor-02", "192.168.0.2", 32, 85, "Ativo"],
    ["Servidor-03", "192.168.0.3", 8, 45, "Ativo"],
    ["Servidor-04", "192.168.0.4", 16, 92, "Ativo"]
]

for servidor in servidores:
    planilha.append(servidor)

print("Inventário criado com sucesso!")
```

Etapa 2: Edição.

Itere sobre os dados inseridos e, se o Uso de Disco (%) for maior ou igual a 80%, defina a cor da célula ou exiba um alerta no terminal informando que o servidor precisa de expansão de armazenamento.

```python
for celula in planilha[1]:
    celula.font = Font(bold=True)

for linha in range(2, planilha.max_row + 1):
    uso = planilha.cell(row=linha, column=4).value

    if uso >= 80:
        planilha.cell(row=linha, column=4).fill = PatternFill(
            start_color="FF0000",
            end_color="FF0000",
            fill_type="solid"
        )

        nome = planilha.cell(row=linha, column=1).value
        print(f"ALERTA: {nome} precisa de expansão de armazenamento!")

print("Verificação concluída!")
```

Etapa 3: Abertura e Extração do Log

Abra o arquivo de texto log.txt utilizando o bloco with e trate possíveis exceções com try/except (FileNotFoundError). Para cada linha do documento, extraia o nível do evento (INFO, AVISO, ERRO).

(Nota: o arquivo de log foi criado anteriormente nas aulas, podendo ser gerado via script Python ou salvo em um bloco de notas para upload no Colab)


```python
try:
    with open("log.txt", "r", encoding="utf-8") as arquivo:
        linhas = arquivo.readlines()

    dados_logs = []

    for linha in linhas:
        partes = linha.strip().split(" - ", 2)

        if len(partes) == 3:
            dados_logs.append([
                partes[0],
                partes[1],
                partes[2]
            ])

    print("Logs extraídos com sucesso!")
    print(f"Total de registros: {len(dados_logs)}")

except FileNotFoundError:
    print("Arquivo log.txt não encontrado!")
```

Etapa 4: Salvando.

Crie uma nova aba (ou uma planilha separada chamada auditoria_logs.xlsx) contendo as colunas do log: Data/Hora, Tipo de Evento e Mensagem. Preencha-a com os dados extraídos no passo anterior e salve o arquivo final no diretório de execução.

```python
auditoria = Workbook()
aba = auditoria.active
aba.title = "Auditoria de Logs"

aba.append(["Data/Hora", "Tipo de Evento", "Mensagem"])

for registro in dados_logs:
    aba.append(registro)

for celula in aba[1]:
    celula.font = Font(bold=True)

for coluna in aba.columns:
    letra = coluna[0].column_letter
    aba.column_dimensions[letra].width = 35

auditoria.save("auditoria_logs.xlsx")

print("Arquivo auditoria_logs.xlsx salvo com sucesso!")
```

# Atividade Prática 02 Aula 12

Coleta de Métricas de Hardware

Requisitos:  
* Crie scripts em Python que executem as etapas a seguir:


Etapa 1:
Desenvolva um script em Python que utilize a biblioteca psutil para capturar a alocação instantânea de CPU e memória RAM.

```python
!pip install psutil

import psutil
import time
import csv

uso_cpu = psutil.cpu_percent(interval=1)
uso_ram = psutil.virtual_memory().percent

print(f"Uso da CPU: {uso_cpu}%")
print(f"Uso da RAM: {uso_ram}%")
```

Etapa 2:
Implemente um ciclo de amostragem configurado para 5 iterações consecutivas , adotando uma pausa de 1 segundo entre as capturas (time.sleep(1)).

```python
total_iteracoes = 5
intervalo = 1

print("Configuração de coleta:")
print(f"Iterações: {total_iteracoes}")
print(f"Intervalo: {intervalo} segundo")
```

Etapa 3:
Defina o caminho de gravação e crie o arquivo telemetria_sistema.csv. Garanta a gravação explícita do cabeçalho de dados (Iteracao,Uso_CPU_Pct,Uso_RAM_Pct\n) antes do início da coleta.


```python
arquivo_csv = "telemetria_sistema.csv"

with open(arquivo_csv, "w", newline="", encoding="utf-8") as arquivo:
    arquivo.write("Iteracao,Uso_CPU_Pct,Uso_RAM_Pct\n")

print("Arquivo CSV criado com sucesso!")
```

Etapa 4:
Persista incrementalmente cada amostra coletada no arquivo CSV utilizando o gerenciador de contexto with open() configurado obrigatoriamente no modo de anexação ('a').

```python
for iteracao in range(1, total_iteracoes + 1):

    uso_cpu = psutil.cpu_percent(interval=None)
    uso_ram = psutil.virtual_memory().percent

    with open(arquivo_csv, "a", newline="", encoding="utf-8") as arquivo:
        escritor = csv.writer(arquivo)
        escritor.writerow([iteracao, uso_cpu, uso_ram])

    print(
        f"Iteração {iteracao}: "
        f"CPU {uso_cpu}% | RAM {uso_ram}%"
    )

    if iteracao < total_iteracoes:
        time.sleep(intervalo)

print("Coleta finalizada!")
print("Arquivo telemetria_sistema.csv salvo!")
```

# Atividade Prática 03 Aula 12

Coleta de Métricas de Hardware

Requisitos:  
*Crie scripts em Python que executem as etapas a seguir:

Etapa 1:
Efetue a ingestão do arquivo telemetria_sistema.csv utilizando a biblioteca pandas (pd.read_csv()).

```python
import pandas as pd

dados = pd.read_csv("telemetria_sistema.csv")

print("Dados de telemetria:")
print(dados)
```

Etapa 2:
Calcule as médias aritméticas do consumo de CPU e RAM, imprimindo um resumo descritivo no console.

```python
media_cpu = dados["Uso_CPU_Pct"].mean()
media_ram = dados["Uso_RAM_Pct"].mean()

print("RESUMO DO CONSUMO DE HARDWARE")
print("-----------------------------")
print(f"Média de uso da CPU: {media_cpu:.2f}%")
print(f"Média de uso da RAM: {media_ram:.2f}%")

print("\nResumo estatístico:")
print(dados[["Uso_CPU_Pct", "Uso_RAM_Pct"]].describe())
```

Etapa 3:
Desenvolva um gráfico de linha utilizando matplotlib para exibir as curvas de variação do uso de CPU e RAM ao longo das iterações.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(
    dados["Iteracao"],
    dados["Uso_CPU_Pct"],
    marker="o",
    label="Uso da CPU"
)

plt.plot(
    dados["Iteracao"],
    dados["Uso_RAM_Pct"],
    marker="s",
    label="Uso da RAM"
)

plt.title("Variação do Uso de CPU e RAM")
plt.xlabel("Iterações")
plt.ylabel("Uso (%)")

plt.legend()
plt.grid(True)

plt.show()
```
