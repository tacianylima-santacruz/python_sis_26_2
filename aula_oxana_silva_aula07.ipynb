```python
# Funcao que calcula a media de miados por hora acordado
def average_meows(logs):
    total_meows_per_hour = 0

    # Percorre cada dia da lista
    for log in logs:
        nap_hours = 0

        # Soma as horas de sono
        for nap in log["naps"]:
            nap_hours += nap

        # Calcula as horas acordado
        awake_hours = 24 - nap_hours

        # Evita divisao por zero
        if awake_hours > 0:
            total_meows_per_hour += log["meows"] / awake_hours

    # Retorna a media
    return total_meows_per_hour / len(logs)


# Lista com os dados dos dias
logs = [
    {"day": "Monday", "naps": [1, 2, 5], "meows": 90},
    {"day": "Tuesday", "naps": [4, 3, 2], "meows": 75},
    {"day": "Wednesday", "naps": [8, 6, 9, 1], "meows": 9},
    {"day": "Thursday", "naps": [1, 1, 5, 7], "meows": 38},
    {"day": "Friday", "naps": [3, 3, 2, 8], "meows": 94},
    {"day": "Saturday", "naps": [4, 8, 3], "meows": 64},
    {"day": "Sunday", "naps": [3, 3, 2, 8], "meows": 37}
]

# Mostra o resultado
print("Average meows per awake hour:", average_meows(logs))
```

```python
# Funcao que calcula a distancia total
def total_distance(*segments):
    total = 0

    # Percorre cada segmento
    for length, unit in segments:

        # Converte a unidade para minusculo
        # e remove espacos extras
        unit = unit.lower().strip()

        # Se for metros, adiciona diretamente
        if unit == "m":
            total += length

        # Se for quilometros, converte para metros
        elif unit == "km":
            total += length * 1000

    # Mostra o resultado
    print("Total distance:", total, "meters")


# Dados da trilha
total_distance(
    (2.5, "km"),
    (400, "M"),
    (1.0, "km"),
    (800, "m"),
    (1.3, "KM"),
    (900, "m "),
    (1.1, "km")

)
```

```python
def ticket_type():
    age = input("Enter your age: ")
    if age < 12:
        print("You get in for free. Lucky you!")
    elif age < 18:
        print("You get a youth discount!")
    elif age < 65:
        print("That's a full-price ticket for you.")
    else:
        print("You get a senior discount!")

ticket_type()
```

```python
# Funcao que verifica o tipo de ingresso
def ticket_type():

    # Repete ate o usuario digitar uma idade valida
    while True:

        try:
            # Pede a idade do visitante
            age = int(input("Digite sua idade: "))

            # Verifica se a idade e valida
            if age < 0:
                print("Idade invalida. Digite novamente.")
                continue

            # Menores de 12 anos entram de graca
            elif age < 12:
                print("Bem-vindo! Voce entra de graca.")

            # Menores de 18 anos recebem desconto
            elif age < 18:
                print("Bem-vindo! Voce recebe desconto para jovens.")

            # Menores de 65 anos pagam valor integral
            elif age < 65:
                print("Bem-vindo! Voce paga o valor integral.")

            # Pessoas com 65 anos ou mais recebem desconto
            else:
                print("Bem-vindo! Voce recebe desconto para idosos.")

            # Encerra o programa depois de uma idade valida
            break

        # Se o usuario digitar letras ou outro valor invalido
        except ValueError:
            print("Idade invalida. Digite novamente.")

        # Limpa a tela mesmo quando houver erro
        finally:
            import os
            os.system("cls")


# Chama a funcao
ticket_type()
```
