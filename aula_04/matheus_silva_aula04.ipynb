```python
print('--- Atividade Prática 01 Aula 04 - Operadores Relacionais ---')
pontuacao_jogador1 = 150
pontuacao_jogador2 = 120

print('Pontuação do Jogador 1:', pontuacao_jogador1)
print('Pontuação do Jogador 2:', pontuacao_jogador2)
print('O jogador 1 tem mais pontos que o jogador 2?', pontuacao_jogador1 > pontuacao_jogador2)
print('O jogador 1 tem menos pontos que o jogador 2?', pontuacao_jogador1 < pontuacao_jogador2)
print('O jogador 1 tem pontuação maior ou igual à do jogador 2?', pontuacao_jogador1 >= pontuacao_jogador2)
print('O jogador 1 tem pontuação menor ou igual à do jogador 2?', pontuacao_jogador1 <= pontuacao_jogador2)
print('O jogador 1 tem a mesma pontuação do jogador 2?', pontuacao_jogador1 == pontuacao_jogador2)
print('A pontuação do jogador 1 é diferente da do jogador 2?', pontuacao_jogador1 != pontuacao_jogador2)

print('\n--- Atividade Prática 02 Aula 04 - Operadores Lógicos ---')
usuario_cadastrado = 'admin'
senha_cadastrada = '1234'
conta_bloqueada = False

usuario_digitado = 'admin'
senha_digitada = '1234'

credenciais_corretas = (usuario_digitado == usuario_cadastrado) and (senha_digitada == senha_cadastrada)
acesso_liberado = credenciais_corretas and not conta_bloqueada

print('As credenciais (usuário e senha) estão corretas?', credenciais_corretas)
print('A conta está bloqueada?', conta_bloqueada)
print('O acesso final está liberado?', acesso_liberado)

print('\n--- Atividade Prática 03 Aula 04 - If, elif, else ---')
try:
    nota = float(input('Digite a nota do aluno (0 a 10): '))
    if nota >= 9.0:
        print('Título: Aluno Lendário!')
    elif nota >= 7.0:
        print('Título: Aluno Excelente!')
    elif nota >= 5.0:
        print('Título: Aluno Aprovado!')
    else:
        print('Título: Precisa de Recuperação!')
except ValueError:
    print('Por favor, digite um valor numérico válido para a nota.')

print('\n--- Atividade Prática 04 Aula 04 - A Jornada do Personagem ---')
# Configurações iniciais do herói
hp_inicial = 100
dano_recebido = 30 * 4  # Cálculo aritmético 1: multiplicação
hp_final = hp_inicial - dano_recebido  # Cálculo aritmético 2: subtração

moedas_ouro = 50
custo_pocao = 15
moedas_restantes = moedas_ouro - custo_pocao  # Cálculo aritmético 3: subtração

# Testes condicionais para o estado do personagem
print(f'Seu HP final após o combate é: {hp_final}')
if hp_final <= 0:
    print('💀 Game Over! Seu personagem foi derrotado.')
else:
    print('🛡️ Você sobreviveu ao ataque inimigo!')

print(f'Moedas restantes após comprar a poção: {moedas_restantes}')
if moedas_restantes >= 20:
    print('🏨 Você tem moedas suficientes para passar a noite na confortável Taverna Real!')
else:
    print('⛺ Suas moedas acabaram. Você terá que dormir ao relento sob as estrelas.')

if (hp_final < 30) and (moedas_restantes >= custo_pocao):
    print('🧪 Alerta de Perigo! Seu HP está baixo, use suas moedas restantes para comprar mais poções imediatamente.')
else:
    print('⚖️ Sua situação geral está sob controle no momento.')

print('\n--- Atividade Prática 05 Aula 04 - match...case ---')
print('Escolha o seu problema:')
print('1 - Problema com Computador/Hardware')
print('2 - Instalação de Software')
print('3 - Sem acesso à Internet/Rede')
print('4 - Reset de Senha')

opcao = input('Digite o número correspondente à opção desejada: ')

match opcao:
    case '1':
        print('Direcionar para: Manutenção')
    case '2':
        print('Direcionar para: Equipe de Suporte')
    case '3':
        print('Direcionar para: Infraestrutura')
    case '4':
        print('Direcionar para: Atendimento Automático (Envio de link de redefinição)')
    case _:
        print('Opção inválida. Chamado encaminhado para a Triagem Geral.')

print('\n--- Atividade Prática 06 Aula 04 - While e For ---')
total_compra = 0.0

# Loop 'while' ideal para coletar preços até que a flag de parada '0' seja digitada
while True:
    try:
        preco = float(input('Digite o preço do produto (ou 0 para encerrar): '))
        if preco == 0:
            break
        elif preco < 0:
            print('Preço inválido! Digite um valor positivo.')
            continue
        total_compra += preco
    except ValueError:
        print('Por favor, digite um número válido.')

print(f'\nTotal da compra: R$ {total_compra:.2f}')

# Loop 'for' ideal para iterar em uma sequência fixa de parcelas (de 1 a 6 vezes)
print('Opções de Parcelamento:')
for parcelas in range(1, 7):
    valor_parcela = total_compra / parcelas
    print(f'{parcelas}x de R$ {valor_parcela:.2f} sem juros')

# Resposta sobre os laços de repetição:
# O laço "while" foi a melhor escolha para a Etapa 1 porque não sabemos previamente a quantidade de itens que o cliente irá comprar, necessitando de uma condição de parada dinâmica (digitar 0).
# O laço "for" foi a melhor escolha para a Etapa 2 porque temos um número de repetições fixo e conhecido (parcelamentos de 1 a 6 vezes).

```
