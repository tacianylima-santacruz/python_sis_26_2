Atividade 04 - Aula 04

```python
print("A Jornada do Personagem")

nome = input("Digite o nome do personagem: ")

vida = 100
forca = 25
defesa_monstro = 30
ouro = 50
experiencia = 0

bonus_espada = 10
dano_monstro = 45
preco_pocao = 20
preco_taverna = 25
xp_missao = 40

ataque_total = forca + bonus_espada
vida_final = vida - dano_monstro
ouro_restante = ouro - preco_pocao
experiencia_final = experiencia + xp_missao

print()
print(nome, "iniciou sua jornada com", vida, "pontos de vida e", ouro, "moedas de ouro.")

print()
print("Durante a aventura,", nome, "encontrou um monstro!")
print("Ataque total:", ataque_total)
print("Defesa do monstro:", defesa_monstro)

if ataque_total > defesa_monstro:
    print(nome, "conseguiu acertar um ataque crítico no monstro!")
else:
    print(nome, "não conseguiu vencer a defesa do monstro.")

print()
print("O monstro atacou", nome, "causando", dano_monstro, "de dano.")
print("Vida restante:", vida_final)

if vida_final <= 0:
    print("Game Over!", nome, "foi derrotado na batalha.")
else:
    print(nome, "sobreviveu ao ataque e continuou sua jornada.")

print()
print(nome, "comprou uma poção por", preco_pocao, "moedas de ouro.")
print("Ouro restante:", ouro_restante)

if ouro_restante >= preco_taverna:
    print(nome, "ainda tem dinheiro suficiente para dormir na taverna.")
else:
    print(nome, "não tem dinheiro suficiente para pagar a taverna e terá que dormir na rua.")

print()
print(nome, "completou uma missão e ganhou", xp_missao, "pontos de experiência.")
print("Experiência final:", experiencia_final)

if experiencia_final >= 40:
    print(nome, "subiu de nível!")
else:
    print(nome, "ainda precisa treinar mais para subir de nível.")
```

Atividade 06 - Aula 04

```python


total_compra = 0.0

print("--- CAIXA DA LOJA ---")
print("Digite 0 para finalizar a compra.\n")
\
while True:
    preco_produto = float(input("Digite o preço do produto: R$ "))

    if preco_produto == 0:
        break

    if preco_produto < 0:
        print("Preço inválido. Digite um valor positivo.")
    else:
        total_compra += preco_produto
        print(f"Total parcial: R$ {total_compra:.2f}")

print(f"\nTotal da compra: R$ {total_compra:.2f}")

print("\n--- OPÇÕES DE PARCELAMENTO ---")

for quantidade_parcelas in range(1, 13):
    valor_parcela = total_compra / quantidade_parcelas
    print(f"{quantidade_parcelas}x de R$ {valor_parcela:.2f}")

```
