```python
# Personagem
nome = input("Digite o nome do seu herói: ")

# Atributos iniciais
vida = 100
forca = 25
bonus_espada = 15
defesa_monstro = 30
ouro = 50
xp = 0

print(f"\n{nome} entra na floresta sombria...")

# Cálculo 1: dano do ataque = força + bônus da espada
dano = forca + bonus_espada
print(f"{nome} ataca com {dano} de dano (força {forca} + espada {bonus_espada}).")

# Teste 1: o dano vence a defesa do monstro?
if dano > defesa_monstro:
    print("Acerto crítico! O monstro foi derrotado.")
    xp = xp + 40
else:
    print("O ataque foi fraco e o monstro continua de pé.")

# Cálculo 2: o monstro revida e tira vida
vida = vida - 35
print(f"O monstro revida! A vida de {nome} caiu para {vida}.")

# Teste 2: a vida chegou a zero ou menos?
if vida <= 0:
    print("Game Over! O herói não resistiu.")
else:
    print(f"{nome} continua de pé e segue a jornada.")

# Cálculo 3: compra de uma poção na vila
preco_pocao = 20
ouro = ouro - preco_pocao
print(f"{nome} compra uma poção por {preco_pocao} de ouro e fica com {ouro}.")

# Teste 3: o ouro que sobrou paga a taverna?
preco_taverna = 35
if ouro >= preco_taverna:
    print("Há ouro suficiente: o herói dorme na taverna e recupera as forças.")
else:
    print("Falta ouro para a taverna: o herói dorme na rua, debaixo das estrelas.")

print(f"\nFim do capítulo! {nome} terminou com {vida} de vida, {ouro} de ouro e {xp} de XP.")
```

```python
print("=== HELP DESK - TRIAGEM DE CHAMADOS ===")
print("1 - Problema com Computador/Hardware")
print("2 - Instalação de Software")
print("3 - Sem acesso à Internet/Rede")
print("4 - Reset de Senha")

opcao = input("Digite o número do seu problema: ")

match opcao:
    case "1":
        print("Chamado direcionado para: Manutenção.")
    case "2":
        print("Chamado direcionado para: Equipe de Suporte.")
    case "3":
        print("Chamado direcionado para: Infraestrutura.")
    case "4":
        print("Chamado direcionado para: Atendimento Automático. Enviando link de redefinição de senha...")
    case _:
        print("Opção inválida. Chamado encaminhado para a Triagem Geral.")
```

```python
# Etapa 1: passando as compras (while)
total = 0.0
preco = float(input("Digite o preço do produto (0 para finalizar): R$ "))

while preco != 0:
    total = total + preco
    preco = float(input("Digite o preço do produto (0 para finalizar): R$ "))

print(f"\nTotal da compra: R$ {total:.2f}")

# Etapa 2: opções de parcelamento (for)
print("\nOpções de parcelamento:")
for parcelas in range(1, 7):
    valor_parcela = total / parcelas
    print(f"{parcelas}x de R$ {valor_parcela:.2f}")

# Resposta: Na etapa 1 a melhor escolha foi o while, porque não sabemos quantos
# produtos o cliente vai levar; o laço precisa repetir até o caixa digitar 0,
# ou seja, depende de uma condição e não de uma quantidade fixa de repetições.
# Na etapa 2 a melhor escolha foi o for, porque já sabemos exatamente quantas
# vezes repetir (de 1x a 6x), e o range() controla o contador automaticamente.
```
