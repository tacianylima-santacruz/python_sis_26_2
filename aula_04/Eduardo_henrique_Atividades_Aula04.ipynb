```python
vida_atual = 100
forca_ataque = 15
bonus_super_forma = 5
defesa_metal_Sonic = 12
aneis_atual = 120
preco_Light_Speed_Dash = 20
preco_super_forma = 100
experiencia = 0
dano_metal_sonic = 90
# calculos
dano_causado = forca_ataque + bonus_super_forma - defesa_metal_Sonic
aneis_atual = aneis_atual - preco_Light_Speed_Dash
experiencia = experiencia + 50
vida_atual = vida_atual - dano_metal_sonic  # Sonic sofre o ataque do Metal Sonic

# prints mostrando o que aconteceu
print('Dano causado ao Metal sonic', dano_causado)
print('Dano recebido do Metal sonic', dano_metal_sonic)
print('Vida atual de Sonic', vida_atual)
print('Aneis restantes', aneis_atual)
print('Experiencia Atual', experiencia)

# Decisoes da historia
if vida_atual <= 0:
    print('Metal: Sonic dead or alive is M-Mine. Game Over')
elif (forca_ataque + bonus_super_forma) > defesa_metal_Sonic:
    print('Sonic: NOW ILL SHOW YOU!!. You win')
else:
    print('Metal sonic resistiu ao ataque')

if aneis_atual >= preco_Light_Speed_Dash:
    print('Sonic aprendeu Light Speed Dash!!.')
else:
    print('Sonic nao aprendeu Light Speed Dash .')
```

```python
# Passando as compras
total = 0
preco = float(input("Digite o preço do produto (0 para encerrar): "))

while preco != 0:
    total += preco
    preco = float(input("Digite o preço do produto (0 para encerrar): "))

print(f"Total da compra: R$ {total:.2f}")

# Opções de Parcelamento
print("Opções de parcelamento:")
for parcela in range(1, 13):
    valor_parcela = total / parcela
    print(f"{parcela}x de R$ {valor_parcela:.2f}")

# Qual laço de repetição foi a melhor escolha para a etapa 1 e 2 e porque?
# - Etapa 1 (while): Foi a melhor escolha porque não sabemos quantos produtos
#   o cliente vai comprar, dependendo de uma condição de parada (digitar 0).
# - Etapa 2 (for): Foi a melhor escolha porque sabemos exatamente o número
#   de parcelas a exibir (de 1 até 12), sendo um loop com limite definido.
```
