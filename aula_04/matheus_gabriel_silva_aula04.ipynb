```python
# Atividade Prática 01 - Aula 04
# Operadores Relacionais

pontuacao_jogador1 = 85
pontuacao_jogador2 = 92

print(f"Jogador 1: {pontuacao_jogador1} pontos | Jogador 2: {pontuacao_jogador2} pontos\n")

print("O jogador 1 tem a mesma pontuação que o jogador 2?")
print(pontuacao_jogador1 == pontuacao_jogador2)

print("\nO jogador 1 tem uma pontuação diferente do jogador 2?")
print(pontuacao_jogador1 != pontuacao_jogador2)

print("\nA pontuação do jogador 1 é maior que a do jogador 2?")
print(pontuacao_jogador1 > pontuacao_jogador2)

print("\nA pontuação do jogador 1 é menor que a do jogador 2?")
print(pontuacao_jogador1 < pontuacao_jogador2)

print("\nA pontuação do jogador 1 é maior ou igual à do jogador 2?")
print(pontuacao_jogador1 >= pontuacao_jogador2)

print("\nA pontuação do jogador 1 é menor ou igual à do jogador 2?")
print(pontuacao_jogador1 <= pontuacao_jogador2)
```

```python
# Atividade Prática 02 - Aula 04
# Operadores Lógicos - Sistema de Login

# --- Dados cadastrados no sistema ---
usuario_cadastrado = "matheus"
senha_cadastrada = "python123"
conta_bloqueada = False

# --- Dados digitados pelo usuário ---
usuario_digitado = "matheus"
senha_digitada = "python123"

# --- Verificações ---
credenciais_corretas = (usuario_digitado == usuario_cadastrado) and (senha_digitada == senha_cadastrada)

acesso_liberado = credenciais_corretas and (not conta_bloqueada)

print(f"As credenciais estão corretas? {credenciais_corretas}")
print(f"A conta está bloqueada? {conta_bloqueada}")
print(f"O acesso final está liberado? {acesso_liberado}")
```

```python
# Atividade Prática 04 - Aula 04
# A Jornada do Personagem - Operadores e Condicionais

# --- Atributos iniciais do personagem ---
hp_personagem = 30
forca_personagem = 12
bonus_espada = 6
defesa_monstro = 15
ouro_atual = 40
preco_pocao = 25
preco_taverna = 10

# --- Cálculo 1: dano recebido em batalha ---
dano_recebido = 35
hp_personagem = hp_personagem - dano_recebido

# --- Cálculo 2: ouro após comprar a poção ---
ouro_restante = ouro_atual - preco_pocao

# --- Cálculo 3: força total do ataque ---
ataque_total = forca_personagem + bonus_espada

print("===== A JORNADA DO PERSONAGEM =====\n")

# --- Decisão 1: o personagem morreu? ---
if hp_personagem <= 0:
    print("💀 GAME OVER! Seu personagem caiu em batalha.")
else:
    print(f"❤️ Seu personagem sobreviveu com {hp_personagem} de HP.")

# --- Decisão 2: o ataque foi crítico? ---
if ataque_total > defesa_monstro:
    print("⚔️ ACERTO CRÍTICO! Sua espada rompeu a defesa do monstro!")
else:
    print("🛡️ O monstro resistiu ao ataque.")

# --- Decisão 3: dá pra pagar a taverna? ---
if ouro_restante >= preco_taverna:
    print(f"🛌 Com {ouro_restante} moedas de ouro, você paga a taverna e descansa.")
else:
    print(f"🌧️ Com apenas {ouro_restante} moedas de ouro, você passa a noite na rua.")
```

```python
# Atividade Prática 05 - Aula 04
# Sistema de Chamados (Help Desk) com match...case

print("===== SISTEMA DE CHAMADOS - TI =====")
print("1 - Problema com Computador/Hardware")
print("2 - Instalação de Software")
print("3 - Sem acesso à Internet/Rede")
print("4 - Reset de Senha")

opcao = input("Digite o número do seu problema: ")

match opcao:
    case "1":
        print("🔧 Chamado direcionado para: Manutenção")
    case "2":
        print("💻 Chamado direcionado para: Equipe de Suporte")
    case "3":
        print("🌐 Chamado direcionado para: Infraestrutura")
    case "4":
        print("🔑 Chamado direcionado para: Atendimento Automático (link de redefinição enviado)")
    case _:
        print("⚠️ Opção inválida. Chamado encaminhado para a Triagem Geral.")
```

```python
# Atividade Prática 06 - Aula 04
# Sistema de Caixa e Parcelamento (while e for)

# --- Etapa 1: passando as compras (while) ---
total_compra = 0

print("Digite o preço de cada produto. Digite 0 para finalizar a compra.\n")

preco = float(input("Preço do produto: "))

while preco != 0:
    total_compra = total_compra + preco
    preco = float(input("Preço do produto: "))

print(f"\nTotal da compra: R$ {total_compra:.2f}")

# --- Etapa 2: opções de parcelamento (for) ---
print("\n===== OPÇÕES DE PARCELAMENTO =====")

for numero_parcelas in range(1, 7):
    valor_parcela = total_compra / numero_parcelas
    print(f"{numero_parcelas}x de R$ {valor_parcela:.2f}")

# O 'while' foi a melhor escolha para a etapa 1 porque não sabemos de antemão
# quantos produtos o cliente vai comprar — o laço precisa repetir até uma
# condição de parada (preço = 0), e não um número fixo de vezes.
#
# O 'for' foi a melhor escolha para a etapa 2 porque já sabemos exatamente
# quantas opções de parcela queremos mostrar (de 1x a 6x), ou seja, é um
# número de repetições conhecido e fixo.
```
