Atividade Pratica 1

```python
pontuacao1 = int(input("Digite a pontuação do jogador 1: "))
pontuacao2 = int(input("Digite a pontuação do jogador 2: "))

igual = pontuacao1 == pontuacao2
print(f"Comparando se são iguais:", igual)

maior = pontuacao1 > pontuacao2
print(f"Comparando se o jogador 1 fez mais pontos que o jogador 2:", maior)

menor = pontuacao1 < pontuacao2
print(f"Comparando se o jogador 1 fez menos pontos que o jogador 2:", menor)

diferente = pontuacao1 != pontuacao2
print(f"Comparando se a pontuação do jogador 1 é diferente do jogador 2:", diferente)

maior_igual = pontuacao1 >= pontuacao2
print(f"Comparando se o jogador 1 fez mais ou o mesmos pontos que o jogador 2:", maior_igual)

menor_igual = pontuacao1 <= pontuacao2
print(f"Comparando se o jogador 1 fez menos ou o mesmos pontos que o jogador 2:", menor_igual)
```

Atividade 6

```python
total = 0

while True:
  preco = float(input("Digite o preço do produto(Digite 0 para finalizar):"))
  if preco == 0:
    break
  total += preco

print(f"O total a pagar é R${total:.2f}")

for parcelas in range(1, 11):
    valor_parcela = total / parcelas
    print(f"{parcelas}x de R$ {valor_parcela:.2f}")
```
