# Atividade Prática 02 Aula 04 - Operadores Lógicos

Você está criando o sistema de segurança de um aplicativo. Para que o usuário consiga fazer o login com sucesso, algumas regras precisam ser atendidas simultaneamente:

*O nome de usuário digitado deve ser igual ao cadastrado no sistema.

*A senha digitada deve ser igual à cadastrada.

*A conta não pode estar bloqueada.

Crie as variáveis necessárias e escreva expressões lógicas que retornem True ou False para as seguintes verificações, imprimindo os resultados na tela:

*As credenciais (usuário e senha) estão corretas?

*A conta está bloqueada?

*O acesso final está liberado (credenciais corretas E conta NÃO bloqueada)?

```python
#Variáveis cadastradas no sistema
usuario_cadastrado = "admin"
senha_cadastrada = "senha123"
conta_bloqueada = False

#Variáveis de entrada do usuário
usuario_digitado = "admin"
senha_digitada = "senha123"

#Verificação das credenciais
credenciais_corretas = (usuario_digitado == usuario_cadastrado) and (senha_digitada == senha_cadastrada)
print(f"As credenciais estão corretas? {credenciais_corretas}")

#Verificação de bloqueio
print(f"A conta está bloqueada? {conta_bloqueada}")

#Verificação de acesso final
acesso_liberado = credenciais_corretas and (not conta_bloqueada)

print(f"O acesso final está liberado? {acesso_liberado}")
```

# Atividade Prática 03 Aula 04 - If, elif, else

Escreva um programa em Python que:

Pergunte ao usuário qual foi a sua nota (lembre-se de converter para float, pois as notas podem ter decimais).

Utilize as estruturas condicionais if, elif e else para avaliar a nota e exibir na tela o título conquistado pelo aluno, para cada uma das condições abaixo deve-se exibir uma mensagem diferente:

* Nota 9.0 ou mais;

* Nota de 7.0 até 8.9;

* Nota de 5.0 até 6.9;

* Nota abaixo de 5.0;


```python
nota = float(input("Digite a sua nota: "))


if nota >= 9.0:
    print(f"Sua nota foi {nota}. Parabéns!")
elif nota >= 7.0:
    print(f"Sua nota foi {nota}. Muito bem! Mas voce ainda pode melhorar.")
elif nota >= 5.0:
    print(f"Sua nota foi {nota}. Nota abaixo da média. Estude para a recuperação")
else:
    print(f"Sua nota foi {nota}. Nota abaixo da média, se esforçe mais.")
```

# Atividade Prática 04 Aula 04 - A Jornada do Personagem (Operadores e Condicionais)
Agora que você já sabe capturar dados com input(), converter tipos e fazer cálculos, chegou a hora de dar inteligência e consequências para a história do seu personagem usando o comando if!

Anteriormente, usar operadores relacionais (>, <, ==) apenas mostrava True ou False na tela. Agora, você vai usar essas comparações para fazer o computador tomar decisões.

A Missão:
* Crie um pequeno script onde o seu personagem toma ações que envolvem cálculos e decisões.

* Cálculos e Variáveis: Crie pelo menos 3 variáveis que envolvam operações aritméticas (ex: calcular o dano de um ataque, subtrair moedas de ouro após uma compra, somar pontos de experiência).

* Tomada de Decisão (if / else): Baseado nos resultados desses cálculos, crie 3 testes condicionais para verificar o estado do seu personagem.

* Mostre o que aconteceu na história usando o print().

Exemplos de ideias que você pode testar com o if:

* A vida (HP) do personagem chegou a zero ou menos? (Game Over)

* A força somada com o bônus da espada é maior que a defesa do monstro? (Acerto crítico)

* O dinheiro que sobrou após comprar a poção é suficiente para pagar a taverna? (Pode dormir ou vai pra rua)

```python
#Exercício igual ao da aula 03, como faltei na aula 4 copiei o exercício aqui
print("A Jornada do Personagem")

hp_personagem = 100
forca_personagem = 15
dinheiro_personagem = 50

print(f"\nSeu personagem inicia com: {hp_personagem} HP, {forca_personagem} de Força e {dinheiro_personagem} moedas de ouro.")

# Cenário 1:
print("\nUm orc furioso aparece! Ele te ataca!")
dano_monstro = 25
hp_personagem_apos_ataque = hp_personagem - dano_monstro
print(f"Você sofreu {dano_monstro} de dano! Seu HP agora é: {hp_personagem_apos_ataque}")

# Decisão 1: Verificar se o personagem sobreviveu
if hp_personagem_apos_ataque <= 0:
    print("GAME OVER! Seu personagem foi derrotado pelo Orc Furioso.")
    hp_personagem = 0
else:
    print("Você sobreviveu ao ataque do Orc.")
    hp_personagem = hp_personagem_apos_ataque

#Cenário 2:
if hp_personagem > 0:
    print("\nVocê se prepara para contra atacar")
    bonus_espada = 8
    defesa_monstro = 20
    dano_total_personagem = forca_personagem + bonus_espada
    print(f"Sua força com a espada é: {dano_total_personagem}. A defesa do Orc é: {defesa_monstro}.")

    #Decisão 2:
    if dano_total_personagem > defesa_monstro:
        print("Acerto crítico! O Orc Furioso é derrotado!")
        recompensa_ouro = 20
        dinheiro_personagem += recompensa_ouro
        print(f"Você ganhou {recompensa_ouro} moedas de ouro. Total: {dinheiro_personagem}")
    else:
        print("Seu ataque não foi suficiente. O Orc escapou")

#Cenário 3:
print("\nApós a batalha, você encontra um vendedor. Uma poção de vida custa 30 moedas de ouro.")
preco_pocao = 30
dinheiro_apos_compra = dinheiro_personagem - preco_pocao

if dinheiro_personagem >= preco_pocao:
    print(f"Você compra a poção de vida. Sobram {dinheiro_apos_compra} moedas de ouro.")
    dinheiro_personagem = dinheiro_apos_compra
else:
    print("Você não tem dinheiro suficiente para a poção de vida.")

#Decisão 3: Veificar se pode pagar a taverna
custo_taverna = 25
print(f"\nMais tarde, você chega a uma taverna. Um quarto custa {custo_taverna} moedas de ouro.")

if dinheiro_personagem >= custo_taverna:
    print("Você tem dinheiro suficiente")
    dinheiro_personagem -= custo_taverna
    print(f"Moedas restantes: {dinheiro_personagem}")
else:
    print("Você não tem dinheiro para a taverna, terá que dormir na rua")
    print(f"Moedas restantes: {dinheiro_personagem}")
```

# Atividade Prática 05 Aula 04 - match...case
Sistema de Chamados (Help Desk)
Você foi contratado para desenvolver a triagem de um sistema de chamados de TI. O programa deve pedir ao usuário para digitar o número correspondente ao problema que ele está enfrentando. Utilizando a estrutura match...case, direcione o chamado para a equipe correta e exiba uma mensagem confirmando o direcionamento, o menu de escolhas deve ter pelo menos 4 opções.

Exemplos de Opções:
1 - Problema com Computador/Hardware ➞ Direcionar para: Manutenção

2 - Instalação de Software➞ Direcionar para: Equipe de Suporte

3 - Sem acesso à Internet/Rede ➞ Direcionar para: Infraestrutura

4 - Reset de Senha ➞ Direcionar para: Atendimento Automático (Envio de link de redefinição)

Qualquer outra opção ➞ Exibir: Opção inválida. Chamado encaminhado para a Triagem Geral.


```python
#Menu
print("SISTEMA DE CHAMADOS (HELP DESK)")
print("1 - Problema com Computador/Hardware")
print("2 - Instalação de Software")
print("3 - Sem acesso à Internet/Rede")
print("4 - Reset de Senha")

opcao = input("Digite o número correspondente ao problema que está enfrentando: ")

#match...case para direcionar o chamado
match opcao:
    case "1":
        print("Direcionado para manutenção.")
    case "2":
        print("Direcionado para equipe de suporte.")
    case "3":
        print("Direcionado para equipe de infraestrutura.")
    case "4":
        print("Direcionado para o atendimento automático. Um link de redefinição de senha foi enviado para o seu e-mail.")
    case _:
        print("Opção inválida. Chamado encaminhado para a Triagem Geral.")
```

# Atividade Prática 06 Aula 04 - While e For
Sistema de Caixa e Parcelamento
Objetivo: Desenvolver um programa que utilize os dois tipos de laço de repetição (while e for).

Cenário:
Você está programando o sistema de caixa de uma loja. O programa deve funcionar em duas etapas:

Passando as compras: O sistema deve pedir ao caixa para digitar o preço de cada produto. Como não sabemos quantos produtos o cliente vai levar, o programa deve continuar pedindo os preços e somando o total até que o caixa digite o valor 0 (zero), que significa que a compra acabou.

Opções de Parcelamento: Após calcular o total, o sistema deve mostrar ao cliente as opções de parcelamento, mostrando o valor de cada parcela.

Ao final do código, adicione um comentário (#) respondendo: Qual laço de repetição foi a melhor escolha para a etapa 1 e 2 e porque?



```python
#Etapa 1
total_compra = 0.0
print("ETAPA 1: PASSANDO AS COMPRAS")
print("Digite o preço de cada produto. Digite 0 para encerrar a compra.")

while True:
    preco = float(input("Preço do produto: R$ "))
    if preco == 0:
        break
    total_compra += preco

print(f"\nTotal da compra: R$ {total_compra:.2f}")

#Etapa 2
print("\nETAPA 2: OPÇÕES DE PARCELAMENTO")
print("Opções de parcelamento disponíveis (de 1 a 12 vezes sem juros):")

for parcelas in range(1, 13):
    valor_parcela = total_compra / parcelas
    print(f"{parcelas}x de R$ {valor_parcela:.2f}")

# Justificativa sobre a escolha dos laços:
# 1) Na Etapa 1, o loop 'while' foi a melhor escolha porque não sabemos
#    a quantidade de produtos que o cliente vai comprar. Ele repete a verificação
#    até a compra acabar quando o cliente digitar 0.
# 2) Na Etapa 2, o loop 'for' foi a melhor escolha porque já se sabe o número de repetições
#    necessárias (de 1 à 12).
#    Dessa forma, o controle de repetição usando o range() é mais direto
```
