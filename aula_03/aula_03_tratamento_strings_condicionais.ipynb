# Sequência de caracteres


```python
texto_strings = "Um Anel para a todos governar"

# Pegando a primeira letra
primeira_letra = texto_strings[0]

# Pegando a letra 'l' da segunda palavra
letra_seis = texto_strings[6]

# Mostrando que o espaço também conta como caractere!
letra_sete = texto_strings[7]

# Acessando de trás para frente (índice negativo)
ultima_letra = texto_strings[-1]

print(f"Primeira letra: {primeira_letra}")  # Saída: U
print(f"Letra no índice 6: {letra_seis}")    # Saída: l
print(f"Letra no índice 7: {letra_sete}")   # Saída:  (um espaço em branco)
print(f"Última letra: {ultima_letra}")      # Saída: r
```

# Recuperar substrings



```python
texto_strings = "Um Anel para a todos governar"
primeira_palavra = texto_strings[:7]        # Do início até o índice 6
mensagem_principal = texto_strings[12:]     # Do índice 12 até o final
trecho_meio = texto_strings[7:12]           # Do índice 7 ao 11
trecho_sem_indice_definido = texto_strings[:]
print(f"Início: {primeira_palavra}")        # Saída: Um Anel
print(f"Fim: {mensagem_principal}")         # Saída: para a todos governar
print(f"Meio: {trecho_meio}")
print(f"Sem índice definido: {trecho_sem_indice_definido}")
```

# Lower, Upper, Capitalize e Inversão:

```python
titulo = "o Senhor dos Anéis"
grito = "vocês não poderão passar!"
sussurro = "GOLLUM... meu precioso..."

# Transformando em maiúsculas (grito)
print(grito.upper())
# Transformando em minúsculas (sussurro)
print(sussurro.lower())
# Capitalizando (primeira letra maiúscula)
print(titulo.capitalize())

# Invertendo o texto usando fatiamento com passo negativo [::-1]
inscricao = "Um Anel para a todos governar"
texto_invertido = inscricao[::-1]

print(f"Inscrição ao contrário: {texto_invertido}")
# Saída: ranrevog sodot a arap lenA mU
```

# len()


```python
# A função len() conta a quantidade total de caracteres, incluindo espaços!

# Exemplo 1: O nome de um personagem
mago = "Gandalf, o Cinzento"
tamanho_nome = len(mago)

print(f"O nome '{mago}' possui {tamanho_nome} caracteres.")
# Saída: O nome 'Gandalf, o Cinzento' possui 19 caracteres.


# Exemplo 2: A fala icônica do Gollum
fala_gollum = "Meu precioso!"
tamanho_fala = len(fala_gollum)

print(f"A fala tem {tamanho_fala} letras e símbolos.")
# Saída: A fala tem 13 letras e símbolos.

```

# Split

```python
# String recebida com os nomes separados por vírgula
lista_membros = "Frodo, Sam, Aragorn, Legolas, Gimli, Boromir, Gandalf"

# O split() divide a string a cada vírgula e cria uma lista de membros
membros_sociedade = lista_membros.split(", ")

print("Escolhido:")
print(f"- {membros_sociedade[2]}")

```

```python
# Redação enviada pelo membro Pippin
redacao = '''A batalha no Abismo de Helm foi decisiva para a vitoria
de Rohan contra as forças de Saruman.'''

# Sem argumentos, o split() divide por qualquer espaço em branco
palavras = redacao.split()

print("Palavras encontradas:", palavras)
print(f"Total de palavras na redação: {len(palavras)}")
print(f"Total de caracteres na redação: {len(redacao)}")
print(f"Palavra escolhida: {palavras[3]}")
```

# Strip

```python
# A senha secreta da porta está toda em letras minúsculas e sem espaços
senha_secreta = "mellon"
# O jogador digita a senha no teclado (com espaços sobrando
entrada_jogador = "mellon  "
print("Sem tratamento:",senha_secreta == entrada_jogador)
# Sem limpar os dados, a porta continuaria fechada, pois "mellon  " é diferente de "mellon".

senha_tratada = entrada_jogador.strip()
print("Com tratamento:",senha_secreta == senha_tratada)

```

# **As Portas de Durin (a entrada para as minas de Moria)**
A inscrição na porta mágica diz: "Fale, amigo, e entre". A senha correta para abrir a porta é a palavra "amigo" em élfico, que é "mellon".

Imagine que você está criando um jogo de texto e o jogador precisa digitar essa senha. O jogador pode acabar digitando espaços a mais ou digitando letras maiúsculas.

```python
# A senha secreta está toda em letras minúsculas e sem espaços
senha_secreta = "mellon"
# O jogador digita a senha no teclado (com espaços sobrando e letras maiúsculas)
entrada_jogador = "MELLON "
print("Sem tratamento:",senha_secreta == entrada_jogador)
# Sem limpar os dados, a porta continuaria fechada
senha_tratada_parcial = entrada_jogador.strip()
print("Com tratamento parcial:",senha_secreta == senha_tratada_parcial)
# False ainda pois "MELLON" é diferente de "mellon".
# Vamos encadear o .strip() e o .lower() para tratar essa entrada:
senha_tratada = entrada_jogador.strip().lower()
print("Com tratamento completo",senha_secreta == senha_tratada)
```

# Formatando números

```python
nota1 = 8.75
nota2 = 9.30
nota3 = 7.80

# Cálculo da média simples
media = (nota1 + nota2 + nota3) / 3
print(f"A média do aluno Samwise é: {media}")
# Uso do :.2f dentro da f-string para limitar as casas decimais
print(f"A média do aluno Samwise é: {media:.2f}")
```

```python
ouro_arrecadado = 1550000

# O uso de :, adiciona vírgula como separador de milhar (padrão norte-americano/python)
print(f"Total arrecadado para a biblioteca: {ouro_arrecadado:,} moedas de ouro.")

# Ou substituindo por ponto (padrão brasileiro)
formatado_br = f"{ouro_arrecadado:,}".replace(",", ".")
print(f"Formatado em PT-BR: R$ {formatado_br}")
```

```python
# Valor com milhares e centavos
ouro_arrecadado = 1550000.01

# 1. Formatamos no padrão americano: "1,550,000.01"
us_format = f"{ouro_arrecadado:,.2f}"

# 2. Invertemos os símbolos para o padrão PT-BR
br_format = us_format.replace(",", "X").replace(".", ",").replace("X", ".")

print(f"Total arrecadado: R$ {br_format}")
```

```python
import locale

# Configura a localização para o Português do Brasil
locale.setlocale(locale.LC_ALL, 'pt_BR.UTF-8')

# Valor com milhares e centavos

valor = 1550000.01

# O parâmetro grouping=True garante a separação dos milhares
formatado_br = locale.currency(valor, symbol=True, grouping=True)

print(formatado_br)
```

# Condições Lógicas

```python
# O professor Gandalf registra a nota do aluno Bilbo:
nota_texto = "7"   # Tipo: str (String / Texto)
nota_numero = 7    # Tipo: int (Inteiro / Número)

# Comparando se a nota em texto é igual à nota em número:
resultado = (nota_texto == nota_numero)

print("Nota em texto:", type(nota_texto))
print("Nota em número:", type(nota_numero))
print("As duas variáveis são iguais?", resultado)
```

# If/Else

```python
senha_digitada = input("Digite a senha, amigo: ")

# Comparando se o texto digitado é IGUAL ao esperado
if senha_digitada == "mellon":
    print("A porta se abre! Bem-vindo às Minas de Moria.")
else:
    print("Senha incorreta! A porta continua trancada.")
```

If, Elif, Else

```python
nota_aluno = float(input("Digite a nota obtida: "))

if nota_aluno >= 9.0:
    print("Conceito: Mago Supremo (Aprovado com Louvor)")

elif nota_aluno >= 7.0:
    print("Conceito: Guardião (Aprovado)")

elif nota_aluno >= 5.0:
    print("Conceito: Aprendiz em Recuperação (Exame Especial)")

else:
    # Só executa se TODAS as condições anteriores (if e elifs) forem FALSAS
    print("Conceito: Reprovado (Retornar para o Condado)")
```
