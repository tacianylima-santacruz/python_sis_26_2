```python
referencia = "ATGGTGCACCTGACTCCTGAG"

# Entrada e padronização
dna = input("Digite uma fita de DNA com 21 nucleotídeos: ").upper()

print("=" * 45)
print(f"Fita digitada (formatada): {dna}")

# Processamento 1: comparação com a referência, sem if
mutante = dna != referencia
diagnosticos = [
    "✅ Sequência NORMAL (Hemoglobina A)",
    "⚠️ Sequência MUTANTE (Anemia Falciforme - HbS)"
]
diagnostico = diagnosticos[mutante]  # False = índice 0, True = índice 1
print(diagnostico)

# Processamento 2: tradução com .replace() encadeado
proteina = (dna.replace("ATG", "Met-")
               .replace("GTG", "Val-")
               .replace("CAC", "His-")
               .replace("CTG", "Leu-")
               .replace("ACT", "Thr-")
               .replace("CCT", "Pro-")
               .replace("GAG", "Glu")).rstrip("-")
print(f"Proteína traduzida: {proteina}")

# Processamento 3: busca de aminoácido com .find()
aminoacido = input("Digite um aminoácido para buscar (ex: Val ou His): ").strip().capitalize()
posicao = proteina.find(aminoacido)

# Relatório final
print("\n" + "=" * 45)
print("RELATÓRIO FINAL")
print("=" * 45)
print(f"DNA de referência: {referencia}")
print(f"DNA digitado:      {dna}")
print(f"Tamanho da fita:   {len(dna)} nucleotídeos")
print(f"Proteína traduzida: {proteina}")
print(f"Diagnóstico:       {diagnostico}")
print(f"Posição de '{aminoacido}' na proteína: {posicao}")
```

```python
# Entrada de dados (o usuário pode digitar espaços a mais no começo e no fim)
texto = input("Digite o texto do pergaminho: ")

# Limpeza e higienização
texto_limpo = texto.strip()

# Separa o texto em palavras
palavras = texto_limpo.split()

# Palavra procurada pelo Detector de Ameaças
ameaca = "Anel"

# Relatório
print("\n=== RELATÓRIO DO PERGAMINHO ===")
print(f"Texto limpo: '{texto_limpo}'")
print(f"Total de caracteres: {len(texto_limpo)}")
print(f"Total de palavras: {len(palavras)}")
print(f"Primeira palavra: {palavras[0]}")
print(f"Última palavra: {palavras[-1]}")

# Detector de ameaças (ignora maiúsculas e minúsculas)
print("\n=== DETECTOR DE AMEAÇAS ===")
print(f"A palavra '{ameaca}' existe no texto? {ameaca.lower() in texto_limpo.lower()}")
print(f"Quantas vezes '{ameaca}' aparece: {texto_limpo.lower().count(ameaca.lower())}")
```

```python

# A Jornada do Personagem

# Variáveis iniciais
vida = 100
forca = 20
moedas = 50

# Cálculo 1: dano do ataque
bonus_espada = 15
dano = forca + bonus_espada

# Cálculo 2: moedas após comprar uma poção
preco_pocao = 20
moedas = moedas - preco_pocao

# Cálculo 3: vida após receber um ataque
dano_monstro = 40
vida = vida - dano_monstro

# Teste condicional 1: verificar a vida
if vida <= 0:
    print("Game Over! Seu personagem morreu.")
else:
    print("Você sobreviveu ao ataque! Vida:", vida)

# Teste condicional 2: verificar o dano
defesa_monstro = 30

if dano > defesa_monstro:
    print("Ataque poderoso! Você derrotou o monstro!")
else:
    print("O monstro resistiu ao seu ataque!")

# Teste condicional 3: verificar as moedas
preco_taverna = 25

if moedas >= preco_taverna:
    print("Você pode pagar a taverna e descansar!")
else:
    print("Você não tem moedas suficientes para a taverna!")
```
