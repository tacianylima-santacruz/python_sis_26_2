```python
# Atividade Prática 01 - Aula 03
# Analisando código genético - Gene HBB (Hemoglobina)

# --- Entrada de dados ---
fita_digitada = input("Digite a fita de DNA (21 nucleotídeos): ")

# --- Processamento 1: padronização e comparação ---
fita_formatada = fita_digitada.upper()  # deixa tudo maiúsculo
print(f"Fita digitada (formatada): {fita_formatada}")

fita_referencia = "ATGGTGCACCTGACTCCTGAG"  # sequência normal da Hemoglobina A

mutante = fita_formatada != fita_referencia  # True = mutante | False = normal

diagnosticos = [
    "✅ Sequência NORMAL (Hemoglobina A)",
    "⚠️ Sequência MUTANTE (Anemia Falciforme - HbS)"
]

resultado_diagnostico = diagnosticos[mutante]  # usa o booleano como índice (True=1, False=0)

# --- Processamento 2: tradução dos códons em aminoácidos ---
# Obs: a ordem dos .replace() importa aqui! Se "CTG" for trocado antes de "CCT",
# o Python encontra um "CTG" "escondido" dentro do trecho "CCTGAG" (que na verdade
# são os códons CCT + GAG) e embaralha a tradução. Por isso ACT e CCT vêm primeiro.
proteina = fita_formatada.replace("ATG", "Met-")
proteina = proteina.replace("GTG", "Val-")
proteina = proteina.replace("CAC", "His-")
proteina = proteina.replace("ACT", "Thr-")
proteina = proteina.replace("CCT", "Pro-")
proteina = proteina.replace("CTG", "Leu-")
proteina = proteina.replace("GAG", "Glu")

print(f"Proteína traduzida: {proteina}")

# --- Processamento 3: busca de um aminoácido na proteína ---
aminoacido_buscado = input("Digite o aminoácido que deseja buscar (ex: Val, His): ")
posicao_encontrada = proteina.find(aminoacido_buscado)

# --- Relatório final ---
print("\n===== RELATÓRIO FINAL =====")
print(f"DNA de referência: {fita_referencia}")
print(f"DNA digitado: {fita_formatada}")
print(f"Tamanho da fita: {len(fita_formatada)} nucleotídeos")
print(f"Proteína traduzida: {proteina}")
print(f"Diagnóstico: {resultado_diagnostico}")
print(f"Posição do aminoácido '{aminoacido_buscado}': {posicao_encontrada}")
```

```python
# Atividade Prática 02 - Aula 03
# O Analisador de Pergaminhos da Terra-Média

# --- Entrada de dados ---
# Digite a frase com espaços a mais no começo e no fim, ex: "   o Um Anel deve ser destruído   "
texto_pergaminho = input("Digite o texto do pergaminho: ")

# --- Limpeza e higienização ---
texto_limpo = texto_pergaminho.strip()  # remove espaços do início e do fim

# --- Relatório ---
total_caracteres = len(texto_limpo)
palavras = texto_limpo.split()  # separa o texto em uma lista de palavras
total_palavras = len(palavras)
primeira_palavra = palavras[0]
ultima_palavra = palavras[-1]

print("\n===== RELATÓRIO DO PERGAMINHO =====")
print(f"Texto limpo: '{texto_limpo}'")
print(f"Total de caracteres: {total_caracteres}")
print(f"Total de palavras: {total_palavras}")
print(f"Primeira palavra: {primeira_palavra}")
print(f"Última palavra: {ultima_palavra}")

# --- Detector de ameaças ---
palavra_procurada = "Anel"
existe_palavra = palavra_procurada in texto_limpo
quantidade_ocorrencias = texto_limpo.count(palavra_procurada)

print("\n===== DETECTOR DE AMEAÇAS =====")
print(f"A palavra '{palavra_procurada}' foi encontrada no texto? {existe_palavra}")
print(f"Quantidade de vezes que '{palavra_procurada}' aparece: {quantidade_ocorrencias}")
```

```python
# Atividade Prática 03 - A Jornada do Personagem
# Operadores aritméticos + Condicionais (if / else)

# --- Atributos iniciais do personagem ---
hp_personagem = 20
forca_personagem = 15
bonus_espada = 8
defesa_monstro = 20
ouro_atual = 50
xp_atual = 30

# --- Cálculo 1: dano recebido em uma batalha ---
dano_recebido = 25
hp_personagem = hp_personagem - dano_recebido  # HP após o ataque do monstro

# --- Cálculo 2: gasto na loja ---
preco_pocao = 45
ouro_restante = ouro_atual - preco_pocao  # ouro após comprar a poção

# --- Cálculo 3: experiência ganha na missão ---
xp_da_missao = 70
xp_total = xp_atual + xp_da_missao  # soma da experiência

print("===== A JORNADA DO PERSONAGEM =====\n")

# --- Decisão 1: o personagem morreu? ---
if hp_personagem <= 0:
    print("💀 GAME OVER! Seu personagem caiu em batalha.")
else:
    print(f"❤️ Seu personagem sobreviveu com {hp_personagem} de HP.")

# --- Decisão 2: o ataque foi crítico? ---
ataque_total = forca_personagem + bonus_espada
if ataque_total > defesa_monstro:
    print("⚔️ ACERTO CRÍTICO! Sua espada rompeu a defesa do monstro!")
else:
    print("🛡️ O monstro resistiu ao ataque.")

# --- Decisão 3: o personagem consegue pagar a taverna? ---
preco_taverna = 20
if ouro_restante >= preco_taverna:
    print(f"🛌 Com {ouro_restante} moedas de ouro, você paga a taverna e descansa.")
else:
    print(f"🌧️ Com apenas {ouro_restante} moedas de ouro, você passa a noite na rua.")

print(f"\n✨ Experiência total acumulada: {xp_total} XP")
```
