# Atividade Prática 01 Aula 03: Analisando código genético



Gene HBB (Hemoglobina)

**O que é o Gene HBB? **Trecho de DNA responsável por sintetizar a Beta-Hemoglobina, proteína essencial das hemácias para o transporte de oxigênio no sangue.

**Por que este trecho? **É uma sequência real. Mudanças de apenas uma letra neste gene alteram o aminoácido e causam doenças como a Anemia Falciforme.

**Objetivo de Hoje:** Criar um script em Python para traduzir o DNA em aminoácidos e buscar amnoácidos específicos dentro da proteína.

**Entrada:** Sequência de DNA do gene HBB (atggtgcacctgactcctgag).

Solicite que o usuário digite uma fita de DNA com 21 nucleotídeos através da função input().

**Processamento 1:** Padronização para maiúsculas e formatação de texto (.upper(), *, f-strings). E comparando a fita.

Exiba a fita formatada usando f-string.
Defina a sequência de referência normal da Hemoglobina A: "ATGGTGCACCTGACTCCTGAG".

Compare a fita digitada pelo usuário com a fita de referência usando o operador de diferença (!=) para obter um valor Booleano (True se for mutante, False se for normal).

Crie uma lista com as duas opções de diagnóstico:
["✅ Sequência NORMAL (Hemoglobina A)", "⚠️ Sequência MUTANTE (Anemia Falciforme - HbS)"]

Exiba o resultado acessando o índice da lista com a variável booleana (diagnosticos[mutante]), sem utilizar a instrução if.

**Processamento 2: **Comparação e Tradução de trincas (códons) para aminoácidos via manipulação de strings (.replace()).

**Conceitos:** Método .replace() encadeado e concatenação.

Instruções: Crie a variável proteina aplicando .replace() sucessivos na fita de DNA.

Imprima a cadeia de aminoácidos resultante na tela.
Traduza a fita de DNA substituindo as trincas (códons) pelos seus respectivos aminoácidos utilizando o método .replace() encadeado.

Tabela de Códons do Gene HBB:
ATG = Met- | GTG = Val- | CAC = His-CTG = Leu- | ACT = Thr- | CCT = Pro- | GAG = Glu.

**Processamento 3: **Localização de padrões dentro da fita (.find()).

Solicite ao usuário um aminoácido para busca (ex: Val ou His) e localize seu índice na proteína traduzida usando .find().

Imprima o relatório final contendo: DNA de referência, DNA digitado, tamanho da fita (len()), proteína traduzida, resultado do diagnóstico e a posição do aminoácido pesquisado.

```python
print("Análise de Código Genético")

fita_dna = input("Digite uma fita de DNA com 21 nucleotídeos: ")

#Processamento 1
fita_dna_formatada = fita_dna.upper()
print(f"Fita de DNA digitada: {fita_dna_formatada}")

dna_referencia = "ATGGTGCACCTGACTCCTGAG"

#Comparação com a fita de referência
mutante = fita_dna_formatada != dna_referencia

#Lista de dianósticos
diagnosticos = [
    "Sequencia normal",
    "Sequencia mutante"
]

#Diagnóstico sem usar if
print(f"Diagnóstico: {diagnosticos[mutante]}")

#Processamento 2
# Tabela de Códons do Gene HBB: ATG = Met- | GTG = Val- | CAC = His-CTG = Leu- | ACT = Thr- | CCT = Pro- | GAG = Glu.

proteina = fita_dna_formatada.replace("ATG", "Met-")
proteina = proteina.replace("GTG", "Val-")
proteina = proteina.replace("CAC", "His-")
proteina = proteina.replace("CTG", "Leu-")
proteina = proteina.replace("ACT", "Thr-")
proteina = proteina.replace("CCT", "Pro-")
proteina = proteina.replace("GAG", "Glu-")

print(f"Cadeia de aminoácidos resultante: {proteina}")

#Processamento 3
aminoacido_busca = input("Digite um aminoácido para busca: ")

posicao_aminoacido = proteina.find(aminoacido_busca)

#Relatorio Final
print("\n--- Relatório Final ---")
print(f"DNA de Referência: {dna_referencia}")
print(f"DNA Digitado: {fita_dna_formatada}")
print(f"Tamanho da Fita de DNA: {len(fita_dna_formatada)} nucleotídeos")
print(f"Proteína Traduzida: {proteina}")
print(f"Resultado do Diagnóstico: {diagnosticos[mutante]}")

if posicao_aminoacido != -1:
    print(f"Posição do aminoácido '{aminoacido_busca}': {posicao_aminoacido}")
else:
    print(f"O aminoácido '{aminoacido_busca}' não foi encontrado na proteína.")
```

# Atividade Prática 02 Aula 03: O Analisador de Pergaminhos da Terra-Média.


Objetivo: Utilizar variáveis, métodos de manipulação de strings e índices para limpar, analisar e extrair informações de um texto de forma sequencial.

A Missão:
A Sociedade do Anel encontrou um pergaminho antigo, mas o texto está bagunçado! Seu trabalho é criar um programa em Python que receba esse texto, faça a higienização dos dados e gere um relatório.

Passo a Passo (Requisitos do Sistema):

**Entrada de Dados:**

* O programa deve solicitar que o usuário digite o texto do pergaminho.

* Na frase digitada incluia espaços a mais no começo e fim.

**Limpeza e Higienização**:

* Remova os espaços inúteis no início e no final do texto digitado e guarde em uma variável.

**O Relatório**

* O programa deve calcular e exibir na tela:
* O número total de caracteres do texto limpo.
* A quantidade total de palavras presentes no texto.
* A primeira palavra do texto.
* A última palavra do texto.

**Detector de Ameaças**:

* O sistema deve verificar se a palavra p.e. "Anel", ou outra palavra a sua escolha, existe dentro do texto e imprimir o resultado diretamente;
* O sistema deve contar quantas vezes a palavra aparece.


```python
# Entrada de Dados
print("Analisador de Pergaminhos da Terra-Média")
texto_pergaminho = input("Digite o texto do pergaminho (inclua espaços no começo e fim para testar a limpeza): ")

texto_limpo = texto_pergaminho.strip()
print(f"\nTexto original: '{texto_pergaminho}'")
print(f"Texto limpo:    '{texto_limpo}'")

#Relatório

# Número total de caracteres do texto limpo
num_caracteres = len(texto_limpo)
print(f"Número total de caracteres (texto limpo): {num_caracteres}")

# Quantidade total de palavras presentes no texto
palavras = texto_limpo.split()
quantidade_palavras = len(palavras)
print(f"Quantidade total de palavras: {quantidade_palavras}")

# A primeira palavra do texto
primeira_palavra = palavras[0] if palavras else "N/A"
print(f"Primeira palavra do texto: {primeira_palavra}")

# A última palavra do texto
ultima_palavra = palavras[-1] if palavras else "N/A"
print(f"Última palavra do texto: {ultima_palavra}")

# Detector de Ameaças
palavra_chave = input("\nDigite uma palavra para buscar no texto (ex: Anel): ")

existe_palavra = palavra_chave.lower() in texto_limpo.lower()

# Imprime o resultado diretamente
print(f"A palavra '{palavra_chave}' existe no texto? {existe_palavra}")

# Conta quantas vezes a palavra aparece
contagem_palavra = texto_limpo.lower().count(palavra_chave.lower())
print(f"A palavra '{palavra_chave}' aparece {contagem_palavra} vez(es).")
```

Atividade Prática 03 - A Jornada do Personagem (Operadores e Condicionais)


Agora que você já sabe capturar dados com input(), converter tipos e fazer cálculos, chegou a hora de dar inteligência e consequências para a história do seu personagem usando o comando if!

Anteriormente, usar operadores relacionais (>, <, ==) apenas mostrava True ou False na tela. Agora, você vai usar essas comparações para fazer o computador tomar decisões.

A Missão:
* Crie um pequeno script onde o seu personagem toma ações que envolvem cálculos e decisões.

* Cálculos e Variáveis: Crie pelo menos 3 novas variáveis que envolvam operações aritméticas (ex: calcular o dano de um ataque, subtrair moedas de ouro após uma compra, somar pontos de experiência).

* Tomada de Decisão (if / else): Baseado nos resultados desses cálculos, crie 3 testes condicionais para verificar o estado do seu personagem.

* Mostre o que aconteceu na história usando o print().

Exemplos de ideias que você pode testar com o if:

* A vida (HP) do personagem chegou a zero ou menos? (Game Over!)

* A força somada com o bônus da espada é maior que a defesa do monstro? (Acerto crítico!)

* O dinheiro que sobrou após comprar a poção é suficiente para pagar a taverna? (Pode dormir ou vai pra rua!)

```python
print("A Jornada do Personagem")


hp_personagem = 100
forca_personagem = 15
dinheiro_personagem = 50

print(f"\nSeu personagem inicia com: {hp_personagem} HP, {forca_personagem} de Força e {dinheiro_personagem} moedas de ouro.")

# Cenário 1:
print("\nUm orc furioso aparece! Ele te ataca!")
dano_monstro = 25
hp_personagem_apos_ataque = hp_personagem - dano_monstro
print(f"Você sofreu {dano_monstro} de dano! Seu HP agora é: {hp_personagem_apos_ataque}")

# Decisão 1: Verificar se o personagem sobreviveu
if hp_personagem_apos_ataque <= 0:
    print("GAME OVER! Seu personagem foi derrotado pelo Orc Furioso.")
    hp_personagem = 0
else:
    print("Você sobreviveu ao ataque do Orc.")
    hp_personagem = hp_personagem_apos_ataque

#Cenário 2:
if hp_personagem > 0:
    print("\nVocê se prepara para contra atacar")
    bonus_espada = 8
    defesa_monstro = 20
    dano_total_personagem = forca_personagem + bonus_espada
    print(f"Sua força com a espada é: {dano_total_personagem}. A defesa do Orc é: {defesa_monstro}.")

    #Decisão 2:
    if dano_total_personagem > defesa_monstro:
        print("Acerto crítico! O Orc Furioso é derrotado!")
        recompensa_ouro = 20
        dinheiro_personagem += recompensa_ouro
        print(f"Você ganhou {recompensa_ouro} moedas de ouro. Total: {dinheiro_personagem}")
    else:
        print("Seu ataque não foi suficiente. O Orc escapou")

#Cenário 3:
print("\nApós a batalha, você encontra um vendedor. Uma poção de vida custa 30 moedas de ouro.")
preco_pocao = 30
dinheiro_apos_compra = dinheiro_personagem - preco_pocao

if dinheiro_personagem >= preco_pocao:
    print(f"Você compra a poção de vida. Sobram {dinheiro_apos_compra} moedas de ouro.")
    dinheiro_personagem = dinheiro_apos_compra
else:
    print("Você não tem dinheiro suficiente para a poção de vida.")

#Decisão 3: Veificar se pode pagar a taverna
custo_taverna = 25
print(f"\nMais tarde, você chega a uma taverna. Um quarto custa {custo_taverna} moedas de ouro.")

if dinheiro_personagem >= custo_taverna:
    print("Você tem dinheiro suficiente")
    dinheiro_personagem -= custo_taverna
    print(f"Moedas restantes: {dinheiro_personagem}")
else:
    print("Você não tem dinheiro para a taverna, terá que dormir na rua")
    print(f"Moedas restantes: {dinheiro_personagem}")
```
