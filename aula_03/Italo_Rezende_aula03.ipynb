Atividade 01 - Aula 03

```python
referencia = "ATGGTGCACCTGACTCCTGAG"


dna = input("Digite uma fita de DNA com 21 nucleotídeos: ").upper()


print(f"\nFita de DNA digitada: {dna}")


mutante = dna != referencia

diagnosticos = [
    "✅ Sequência NORMAL (Hemoglobina A)",
    "⚠️ Sequência MUTANTE (Anemia Falciforme - HbS)"
]


diagnostico = diagnosticos[mutante]


proteina = dna.replace("ATG", "Met-") \
             .replace("GTG", "Val-") \
             .replace("CAC", "His-") \
             .replace("CTG", "Leu-") \
             .replace("ACT", "Thr-") \
             .replace("CCT", "Pro-") \
             .replace("GAG", "Glu-")


print(f"Proteína traduzida: {proteina}")


aminoacido = input("\nDigite um aminoácido para buscar (ex: Val ou His): ")


posicao = proteina.find(aminoacido)



print("\n========== RELATÓRIO FINAL ==========")
print(f"DNA de referência: {referencia}")
print(f"DNA digitado:      {dna}")
print(f"Tamanho da fita:   {len(dna)} nucleotídeos")
print(f"Proteína traduzida: {proteina}")
print(f"Diagnóstico:        {diagnostico}")
print(f"Aminoácido buscado: {aminoacido}")
print(f"Posição encontrada: {posicao}")
```

Atividade 02 - Aula 03

```python

texto_pergaminho = input("Digite o texto do pergaminho com espaços no começo e no fim: ")

texto_limpo = texto_pergaminho.strip()

palavras = texto_limpo.split()

total_caracteres = len(texto_limpo)
total_palavras = len(palavras)
primeira_palavra = palavras[0]
ultima_palavra = palavras[-1]

palavra_ameaca = "Anel"
ameaca_existe = palavra_ameaca.lower() in texto_limpo.lower()
quantidade_ameacas = texto_limpo.lower().count(palavra_ameaca.lower())

print("\n--- RELATÓRIO DO PERGAMINHO ---")
print("Texto limpo:", texto_limpo)
print("Total de caracteres:", total_caracteres)
print("Total de palavras:", total_palavras)
print("Primeira palavra:", primeira_palavra)
print("Última palavra:", ultima_palavra)

print("\n--- DETECTOR DE AMEAÇAS ---")
print(f'A palavra "{palavra_ameaca}" existe no texto?', ameaca_existe)
print(f'A palavra "{palavra_ameaca}" aparece', quantidade_ameacas, "vez(es).")
```

Atividade 03 - Aula 03

```python

print("--- A Jornada de Arion ---")

vida = 100
forca = 30
defesa_monstro = 40
ouro = 50
experiencia = 80

dano_ataque = forca + 15          # Bônus da espada
vida_restante = vida - 35         # Dano causado pelo monstro
ouro_restante = ouro - 30         # Compra de uma poção
experiencia_total = experiencia + 25  # Experiência recebida

print("\nArion atacou o monstro com", dano_ataque, "pontos de força.")
print("Após a batalha, sua vida ficou em", vida_restante, "pontos.")
print("Após comprar uma poção, restaram", ouro_restante, "moedas.")
print("Experiência total:", experiencia_total)

if vida_restante <= 0:
    print("\nGame Over! Arion foi derrotado.")
else:
    print("\nArion sobreviveu à batalha!")

if dano_ataque > defesa_monstro:
    print("Acerto crítico! O ataque superou a defesa do monstro.")
else:
    print("O monstro conseguiu defender o ataque.")

if ouro_restante >= 20:
    print("Arion tem dinheiro suficiente para dormir na taverna.")
else:
    print("Arion não tem dinheiro para a taverna e dormirá na rua.")
```
