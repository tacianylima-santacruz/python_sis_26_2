### Resolução da Atividade Prática 01: Analisando código genético (Gene HBB)

```python
# 1. Entrada de Dados
dna_usuario = input("Digite uma fita de DNA com exatamente 21 nucleotídeos: ")

# 2. Processamento 1: Padronização, Comparação e Diagnóstico (sem IF)
dna_usuario_formatado = dna_usuario.upper().strip()
referencia = "ATGGTGCACCTGACTCCTGAG"

# Determina se há mutação (True se for diferente, False se for igual)
mutante = (dna_usuario_formatado != referencia)

diagnosticos = [
    "✅ Sequência NORMAL (Hemoglobina A)",
    "⚠️ Sequência MUTANTE (Anemia Falciforme - HbS)"
]

# Exibe o diagnóstico acessando o índice da lista com o booleano (False vira 0, True vira 1)
diagnostico_final = diagnosticos[int(mutante)]

# 3. Processamento 2: Tradução de trincas (códons) para aminoácidos com .replace() encadeado
# ATG = Met- | GTG = Val- | CAC = His- | CTG = Leu- | ACT = Thr- | CCT = Pro- | GAG = Glu.
proteina = dna_usuario_formatado \
    .replace("ATG", "Met-") \
    .replace("GTG", "Val-") \
    .replace("CAC", "His-") \
    .replace("CTG", "Leu-") \
    .replace("ACT", "Thr-") \
    .replace("CCT", "Pro-") \
    .replace("GAG", "Glu")

# 4. Processamento 3: Localização de padrões dentro da fita
amino_busca = input("Digite o aminoácido para busca (ex: Val, His, Met): ")
posicao_busca = proteina.find(amino_busca)

# Relatório Final
print("\n=== RELATÓRIO DO GENE HBB ===")
print(f"DNA de Referência:      {referencia}")
print(f"DNA Digitado:           {dna_usuario_formatado}")
print(f"Tamanho da fita:        {len(dna_usuario_formatado)} nucleotídeos")
print(f"Resultado Diagnóstico:  {diagnostico_final}")
print(f"Proteína Traduzida:     {proteina}")
print(f"Posição do aminoácido '{amino_busca}': {posicao_busca}")
```

### Resolução da Atividade Prática 02: O Analisador de Pergaminhos da Terra-Média

```python
# 1. Entrada de Dados (Exemplo com espaços extras: "   O Anel de Sauron é poderoso   ")
pergaminho_cru = input("Digite o texto do pergaminho (adicione espaços extras no início e fim para testar): ")

# 2. Limpeza e Higienização
pergaminho_limpo = pergaminho_cru.strip()
palavras = pergaminho_limpo.split()

# 3. Métricas do Relatório
total_caracteres = len(pergaminho_limpo)
total_palavras = len(palavras)
primeira_palavra = palavras[0] if total_palavras > 0 else ""
ultima_palavra = palavras[-1] if total_palavras > 0 else ""

# 4. Detector de Ameaças
palavra_alvo = "Anel"
contem_alvo = palavra_alvo in pergaminho_limpo
qtd_alvo = pergaminho_limpo.count(palavra_alvo)

# Exibição do Relatório
print("\n=== RELATÓRIO DO PERGAMINHO ===")
print(f"Texto Limpo: '{pergaminho_limpo}'")
print(f"Número total de caracteres (limpo): {total_caracteres}")
print(f"Quantidade total de palavras: {total_palavras}")
print(f"Primeira palavra: {primeira_palavra}")
print(f"Última palavra: {ultima_palavra}")
print(f"A palavra '{palavra_alvo}' existe no pergaminho? {contem_alvo}")
print(f"Quantidade de vezes que a palavra '{palavra_alvo}' aparece: {qtd_alvo}")
```

### Resolução da Atividade Prática 03: A Jornada do Personagem (Operadores e Condicionais)

```python
# 1. Variáveis Iniciais do Personagem
hp = 100
moedas = 50
forca = 15
espada_bonus = 8
defesa_monstro = 20

# Cálculos e Novas Variáveis
dano_ataque = forca + espada_bonus
moedas_restantes = moedas - 15  # comprou uma poção que custa 15 moedas
experiencia_final = 80 + 30     # ganhou 30 XP após uma batalha

print("\n--- Início da Jornada ---")
print(f"O guerreiro atacou causando {dano_ataque} de dano (Defesa do monstro: {defesa_monstro}).")
print(f"Sobraram {moedas_restantes} moedas de ouro após comprar uma poção.")
print(f"A nova experiência do herói é {experiencia_final} XP.")

# 2. Tomada de Decisão (3 testes condicionais)
print("\n--- Consequências ---")

# Teste 1: Dano contra Defesa
if dano_ataque > defesa_monstro:
    print("⚔️ Acerto crítico! Seu ataque superou a defesa do monstro!")
else:
    print("🛡️ O monstro bloqueou seu golpe com sucesso!")

# Teste 2: Dinheiro para a Taverna
custo_taverna = 40
if moedas_restantes >= custo_taverna:
    print("🍺 Você tem dinheiro suficiente para pagar a taverna e descansar!")
else:
    print("⛺ Dinheiro insuficiente para a taverna. Vai dormir na rua sob o frio!")

# Teste 3: Nível de Vida restante
hp_final = hp - 100 # Simulação de dano sofrido
if hp_final <= 0:
    print("💀 Game Over! Seus pontos de vida chegaram a zero.")
else:
    print("❤️ Você sobreviveu por pouco à batalha!")
```
