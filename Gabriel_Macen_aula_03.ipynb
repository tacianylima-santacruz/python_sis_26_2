```python
dna_referencia = "ATGGTGCACCTGACTCCTGAG"

dna = input("Digite uma fita de DNA com 21 nucleotídeos: ").upper()

mutante = dna != dna_referencia

diagnosticos = [
    "✅ Sequência NORMAL (Hemoglobina A)",
    "⚠️ Sequência MUTANTE (Anemia Falciforme - HbS)"
]

proteina = dna
proteina = proteina.replace("ATG", "Met-")
proteina = proteina.replace("GTG", "Val-")
proteina = proteina.replace("CAC", "His-")
proteina = proteina.replace("CTG", "Leu-")
proteina = proteina.replace("ACT", "Thr-")
proteina = proteina.replace("CCT", "Pro-")
proteina = proteina.replace("GAG", "Glu-")

aminoacido = input("Digite um aminoácido para buscar (ex: Val ou His): ")

posicao = proteina.find(aminoacido)

print("\n========== RELATÓRIO FINAL ==========")
print(f"DNA de referência: {dna_referencia}")
print(f"DNA digitado: {dna}")
print(f"Tamanho da fita: {len(dna)}")
print(f"Proteína traduzida: {proteina}")
print(f"Diagnóstico: {diagnosticos[mutante]}")
print(f"Aminoácido pesquisado: {aminoacido}")
print(f"Posição do aminoácido: {posicao}")
print("=====================================")
```

```python
# ============================================
# ATIVIDADE PRÁTICA 02 - ANALISADOR DE PERGAMINHOS
# ============================================

texto = input("Digite o texto do pergaminho: ")

texto_limpo = texto.strip()

palavras = texto_limpo.split()

total_caracteres = len(texto_limpo)
total_palavras = len(palavras)

primeira_palavra = palavras[0]
ultima_palavra = palavras[-1]

palavra_busca = "Anel"

existe = palavra_busca in texto_limpo
quantidade = texto_limpo.count(palavra_busca)

print("\n========== RELATÓRIO ==========")
print(f"Texto limpo: {texto_limpo}")
print(f"Total de caracteres: {total_caracteres}")
print(f"Total de palavras: {total_palavras}")
print(f"Primeira palavra: {primeira_palavra}")
print(f"Última palavra: {ultima_palavra}")
print(f"A palavra '{palavra_busca}' existe no texto? {existe}")
print(f"Quantidade de vezes que '{palavra_busca}' aparece: {quantidade}")
print("================================")
```

```python
# ============================================
# ATIVIDADE PRÁTICA 03 - JORNADA DO PERSONAGEM
# ============================================

vida = 100
dano_monstro = 35
vida_restante = vida - dano_monstro

forca = 50
bonus_espada = 30
defesa_monstro = 70
ataque_total = forca + bonus_espada

ouro = 100
preco_pocao = 30
preco_taverna = 50
ouro_restante = ouro - preco_pocao

print("\n========== JORNADA DO PERSONAGEM ==========")

if vida_restante <= 0:
    print("Game Over! O personagem ficou sem vida.")
else:
    print(f"O personagem sobreviveu! Vida restante: {vida_restante}")

if ataque_total > defesa_monstro:
    print("Acerto crítico! O ataque superou a defesa do monstro.")
else:
    print("O ataque foi bloqueado pelo monstro.")

if ouro_restante >= preco_taverna:
    print("Você tem ouro suficiente para pagar a taverna.")
else:
    print("Você não tem ouro suficiente para pagar a taverna.")

print("===========================================")
```
