```python
# Sistema da Comitiva do Anel

# Etapa 1 - Registro do Heroi

def registrar_heroi(nome, classe, reino="Reino dos Homens"):
    return f"Heroi: {nome} | Classe: {classe} | Reino: {reino}"

print(registrar_heroi("Aragorn", "Guerreiro"))
print(registrar_heroi("Legolas", "Arqueiro", "Floresta das Trevas"))


# Etapa 2 - A Chamada das Armas

def chamar_armas(nome, *itens):
    print("Heroi:", nome)

    for item in itens:
        print("-", item)

chamar_armas("Aragorn", "Espada", "Escudo", "Armadura")


# Etapa 3 - Ficha Detalhada do Personagem

def ficha_personagem(nome, **atributos):
    print("Personagem:", nome)

    for atributo, valor in atributos.items():
        print(atributo, ":", valor)

ficha_personagem("Aragorn", forca=80, agilidade=95, defesa=90)


# Desafio Bonus - A Jornada

perigos = ["Orcs", "Nevoeiro", "Aranhas", "Balrog", "Orcs"]

for perigo in perigos:

    if perigo == "Nevoeiro":
        print("Nevoeiro encontrado. Ignorar o combate e seguir viagem.")
        continue

    elif perigo == "Balrog":
        print("Balrog encontrado! Gandalf avisa: FUJAM!")
        break

    else:
        print("Perigo encontrado:", perigo, "- Preparar para o combate.")

```

```python
# Monitoramento de TI


# Etapa 1 - Registro de Servidor
# Variavel global

servidores_online = 10


def alterar_servidores(quantidade, acao):
    """
    Altera a quantidade de servidores online.

    Parametros:
    quantidade: quantidade de servidores a alterar.
    acao: define se servidores serao ativados ou desativados.

    Impacto:
    Altera a variavel global servidores_online.
    """

    global servidores_online

    if acao == "ativar":
        servidores_online = servidores_online + quantidade

    elif acao == "desativar":
        servidores_online = servidores_online - quantidade


# Desativando 5 servidores
alterar_servidores(5, "desativar")

# Ativando 5 servidores
alterar_servidores(5, "ativar")

print("Servidores ativos:", servidores_online)


# Etapa 2 - Configuracao de Conexao ao Banco de Dados

def gerenciar_conexoes():
    conexoes = 5

    def usar_conexao():
        nonlocal conexoes
        conexoes = conexoes - 1

    usar_conexao()
    usar_conexao()

    print("Conexoes restantes:", conexoes)


gerenciar_conexoes()


# Etapa 3 - Ajuste Temporario de Cota

def alterar_cota(cota_atual, nova_cota):

    alterar = lambda cota: nova_cota

    return alterar(cota_atual)


cota = 100

cota = alterar_cota(cota, 200)

print("Nova cota de armazenamento:", cota, "GB")
```

```python
# Tradutor de Maquinas
# Binario e Hexadecimal

# Funcao para converter decimal para binario

def decimal_binario(numero):
    if numero < 2:
        return str(numero)

    return decimal_binario(numero // 2) + str(numero % 2)


# Funcao para converter decimal para hexadecimal

def decimal_hexadecimal(numero):
    caracteres = "0123456789ABCDEF"

    if numero < 16:
        return caracteres[numero]

    return decimal_hexadecimal(numero // 16) + caracteres[numero % 16]


# Entrada do usuario

numero = int(input("Digite um numero decimal: "))


# Resultados

print("Numero decimal:", numero)
print("Binario:", decimal_binario(numero))
print("Hexadecimal:", decimal_hexadecimal(numero))
```
