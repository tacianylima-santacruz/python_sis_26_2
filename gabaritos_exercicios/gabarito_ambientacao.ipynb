# 👑 O Castelo de Camelot 👑


AIVIDADE PRÁTICA DE PROGRAMAÇÃO - INTRODUÇÃO AO PYTHON

HISTÓRIA:
Você e seu grupo de cavaleiros chegaram aos portões do Castelo de Camelot. Para ganhar a permissão de entrar, o guarda do castelo exige que você resolva um
problema de lógica e matemática usando a linguagem Python.

Mas atenção: se você errar as regras de escrita da comunidade Python, o código vai quebrar e vocês serão lançados no "Fosso do Desespero"!

--------------------------------------------------------------------------------
REGRAS IMPORTANTES DA COMUNIDADE PYTHON:
1. Nomes de variáveis NÃO podem ter espaços.
2. Use o padrão snake_case para criar suas variáveis (letras minúsculas separadas por sublinhado, ex: preco_do_bacon).

--------------------------------------------------------------------------------

SUA MISSÃO: Escreva um programa em Python que resolva os três passos abaixo:

PASSO 1: Aritmética das Maçãs:
O Rei Arthur possui 4 caixas de maçãs estocadas. Cada caixa contém exatamente 12 maçãs dentro.
* Crie as variáveis necessárias e calcule o total de maçãs.
* Guarde o resultado final em uma variável chamada: total_macas

PASSO 2: A Divisão dos Cavaleiros:
O Rei decidiu dividir essas maçãs igualmente entre os seus 5 cavaleiros mais leais.
* Use um operador de divisão para descobrir quantas maçãs inteiras cada cavaleiro vai ganhar (não queremos maçãs cortadas em pedaços!).
* Guarde o resultado em uma variável chamada: macas_por_cavaleiro

PASSO 3: O Teste do Portão
O portão do castelo só vai abrir se o número de maçãs que cada cavaleiro recebeu for MAIOR QUE 8.
* Crie uma variável chamada: portao_aberto
* Faça ela receber o resultado da comparação lógica
* Use o comando print() para mostrar o valor de portao_aberto na tela.

------------------------------------------------------

DESAFIO EXTRA:
O que vai acontecer com o portão se o Rei Arthur decidir convidar mais 1 amigo, passando o total de cavaleiros de 5 para 6? O portão vai imprimir True ou False? Por quê?
________________________________________________________________________________


```python
quantidade_caixas = 4
macas_por_caixa = 12
total_macas = quantidade_caixas * macas_por_caixa

# Missão 2: Divisão Inteira
total_cavaleiros = 5
macas_por_cavaleiro = total_macas // total_cavaleiros

# Missão 3: Operadores Lógicos e de Comparação
# O Python vai avaliar se o resultado é maior que 8 (True)
portao_aberto = (macas_por_cavaleiro > 8)

# Mostrando o resultado na tela
print("Quantidade de maçãs por cavaleiro:")
print(macas_por_cavaleiro)

print("O portão do castelo se abriu?")
print(portao_aberto)
```

# Gestão em Tecnologia da Informação
# Exemplo:


```python
import pandas as pd
import plotly.express as px

# Lista de consumo em Terabytes para cada um dos 12 meses
consumo = [10, 11, 12, 14, 15, 15, 16, 18, 19, 21, 22, 25]

# Lista de custos de licenças em Reais para cada um dos 12 meses
licencas = [5000, 5000, 5200, 5200, 5200, 5500, 5500, 5500, 6000, 6000, 6000, 6500]

# Estrutura do dicionário usando as variáveis acima
dados_auditoria = {
    "Mes": ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun", "Jul", "Ago", "Set", "Out", "Nov", "Dez"],
    "Consumo_TB": consumo,
    "Custo_Licencas": licencas
}


import pandas as pd
import plotly.express as px

# Criando o DataFrame
df = pd.DataFrame(dados_auditoria)

# Definição de premissas financeiras
PRECO_POR_TB = 150.00

# Processamento de dados e cálculo de custos
df["Custo_Infraestrutura"] = df["Consumo_TB"] * PRECO_POR_TB
df["Custo_Total"] = df["Custo_Infraestrutura"] + df["Custo_Licencas"]
df = df.round(2)

# Exibição da tabela no terminal
print("\n--- RELATÓRIO CONSOLIDADO DE AUDITORIA DE TI ---")
print(df.to_string(index=False))

# Geração do Dashboard Interativo com Plotly
fig = px.line(
    df,
    x="Mes",
    y=["Custo_Infraestrutura", "Custo_Licencas", "Custo_Total"],
    title="Auditoria e Evolução Mensal de Custos de TI",
    labels={"value": "Valor (R$)", "Mes": "Mês", "variable": "Categoria"},
    markers=True,
)

fig.update_layout(
    hovermode="x unified",
    legend_title_text="Legenda de Custos",
    yaxis_tickprefix="R$ ",
)

fig.show()

```
