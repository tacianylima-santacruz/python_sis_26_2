```python
#Atividade 01

# ===== ANIMAIS FAVORITOS =====

# Lista principal com vários animais
animais = ["cachorro", "gato", "elefante", "leão", "papagaio", "cavalo", "tartaruga"]

# Lista separada apenas com os animais favoritos (subconjunto da lista principal)
animais_favoritos = ["gato", "leão", "cavalo"]

print("=== LISTA COMPLETA DE ANIMAIS ===")
print(animais)
print()

# ===== FILTRO 1: Mostrar os animais favoritos =====
print("=== FILTRO 1: MEUS ANIMAIS FAVORITOS ===")
print(animais_favoritos)
print()

# ===== FILTRO 2: Quantidade de animais e de favoritos =====
print("=== FILTRO 2: QUANTIDADES ===")
print(f"Quantidade total de animais na lista: {len(animais)}")
print(f"Quantidade de animais favoritos: {len(animais_favoritos)}")
print()

# ===== FILTRO 3: Primeiro e último animal favorito =====
print("=== FILTRO 3: PRIMEIRO E ÚLTIMO FAVORITO ===")
print(f"Primeiro animal favorito: {animais_favoritos[0]}")
print(f"Último animal favorito: {animais_favoritos[-1]}")
print()

# ===== FILTRO 4: Remover um animal da lista =====
print("=== FILTRO 4: REMOVENDO UM ANIMAL ===")
animal_removido = animais.pop(1)  # remove "gato" (índice 1) da lista completa
print(f"Animal removido: {animal_removido}")
print(f"Lista atualizada: {animais}")
print()

# ===== FILTRO 5: Adicionar um ou mais animais =====
print("=== FILTRO 5: ADICIONANDO NOVOS ANIMAIS ===")
animais.append("golfinho")
animais.extend(["coelho", "tigre"])
print(f"Lista atualizada: {animais}")
print()

# ===== FILTRO 6: Quantidade de animais agora =====
print("=== FILTRO 6: QUANTIDADE ATUAL ===")
print(f"Agora a lista tem {len(animais)} animais.")
print()

# ===== FILTRO 7: Trocar a ordem dos favoritos =====
print("=== FILTRO 7: NOVA ORDEM DE PREFERÊNCIA ===")
print(f"Ordem antes: {animais_favoritos}")
animais_favoritos[0], animais_favoritos[2] = animais_favoritos[2], animais_favoritos[0]
print(f"Ordem depois: {animais_favoritos}")

#Atividade 02

# ===== O CARDÁPIO FIXO DO AXOLOTE (TUPLAS) =====

# ===== ATIVIDADE 1: Criando a tupla com tipos variados =====
refeicao_axolote = ("08:00", "minhocas", 50)

print("=== CARDÁPIO ATUAL DO AXOLOTE ===")
print(refeicao_axolote)
print()

# ===== ATIVIDADE 2: Tentando alterar a tupla =====
# A linha abaixo causaria erro, pois tuplas são IMUTÁVEIS:
# refeicao_axolote[1] = "pizza de calabresa"
#
# Erro gerado: TypeError: 'tuple' object does not support item assignment
#
# Como o erro acontece, a linha foi removida/comentada acima.
# Isso comprova que, uma vez criada, uma tupla não pode ter seus valores alterados.

# Desempacotamento da tupla em três variáveis independentes (1 linha de código)
horario, alimento, quantidade_vitaminas = refeicao_axolote

# ===== ATIVIDADE 3: Frase amigável usando as variáveis desempacotadas =====
print("=== HORA DA REFEIÇÃO ===")
print(f"Às {horario}, o axolote come {alimento} com {quantidade_vitaminas} vitaminas.")

#Atividade 03
# ===== AGENDA DE CONTATOS (DICIONÁRIOS) =====

# ===== ATIVIDADE 1: Criando o dicionário de contatos =====
contatos = {
    "Ana": "9999-1234",
    "Bruno": "9888-5678",
    "Carla": "9777-4321"
}

print("=== CONTATOS INICIAIS ===")
print(contatos)
print()

# ===== ATIVIDADE 2: Adicionando um novo contato sem alterar a declaração original =====
contatos["Diego"] = "9666-9012"

print("=== APÓS ADICIONAR NOVO CONTATO (Diego) ===")
print(contatos)
print()

# ===== ATIVIDADE 3: Verificando se "Chai" existe, e adicionando se não existir =====
if "Chai" not in contatos:
    contatos["Chai"] = "9555-3344"
    print(">> 'Chai' não existia nos contatos. Adicionado com sucesso!")
else:
    print(">> 'Chai' já está nos contatos.")

print("=== APÓS VERIFICAÇÃO DE 'Chai' ===")
print(contatos)
print()

# ===== ATIVIDADE 4: Excluindo um contato sem gerar erro se não existir =====
contatos.pop("Bruno", None)  # remove "Bruno"; se não existisse, não geraria erro
contatos.pop("Roberto", None)  # tentativa de remover alguém que não existe - sem erro

print("=== APÓS EXCLUSÃO DE CONTATO(S) ===")
print(contatos)
print()

# ===== ATIVIDADE 5: Imprimindo todos os nomes presentes no dicionário =====
print("=== NOMES DE TODOS OS CONTATOS ===")
for nome in contatos.keys():
    print(nome)

#Atividade 04

# ===== A SACOLA DO CLUBE DO LIVRO (CONJUNTOS) =====

# ===== ATIVIDADE 1: Lista inicial com títulos repetidos =====
livros_sacola = ["Duna", "O Hobbit", "Duna", "Harry Potter", "O Hobbit", "1984", "Percy Jackson"]

print("=== LISTA ORIGINAL (COM DUPLICATAS) ===")
print(livros_sacola)
print(f"Quantidade de livros na sacola (com repetições): {len(livros_sacola)}")
print()

# ===== ATIVIDADE 2: Convertendo a lista em conjunto para remover duplicatas =====
livros_unicos = set(livros_sacola)

# ===== ATIVIDADE 3: Imprimindo o conjunto final =====
print("=== CONJUNTO SEM DUPLICATAS ===")
print(livros_unicos)
print(f"Quantidade de títulos únicos: {len(livros_unicos)}")

#Atividade 05

# ===== BANCO DE DADOS DE BIBLIOTECA (COMBINAÇÃO DE TIPOS) =====

# Estrutura combinada: dicionário principal (chave = título)
# contendo outros dicionários e listas aninhadas como valores
biblioteca = {
    "Jane Eyre": {
        "autores": [
            {"nome": "Charlotte Brontë", "ano_nascimento": 1816, "nacionalidade": "Britânica"}
        ],
        "num_paginas": 659,
        "ano_publicacao": 1847,
        "generos": ["Ficção Vitoriana", "Romance de Formação (Bildungsroman)", "Histórico"],
        "avaliacao": 4.6,
        "disponivel": True,
        "emprestado_para": None
    },
    "Python Illustrated": {
        "autores": [
            {"nome": "Zia van Putten", "ano_nascimento": 2023, "nacionalidade": "Holandesa"},
            {"nome": "Maaike van Putten", "ano_nascimento": 1991, "nacionalidade": "Holandesa"},
            {"nome": "Imke van Putten", "ano_nascimento": 1997, "nacionalidade": "Holandesa"}
        ],
        "num_paginas": 432,
        "ano_publicacao": 2026,
        "generos": ["Informativo", "Educação"],
        "avaliacao": 5,
        "disponivel": False,
        "emprestado_para": "Jacob"
    },
    "Se os gatos desaparecessem do mundo": {
        "autores": [
            {"nome": "Genki Kawamura", "ano_nascimento": None, "nacionalidade": None}
        ],
        "num_paginas": 144,
        "ano_publicacao": 2012,
        "generos": ["Ficção Literária", "Realismo Mágico", "Ficção Filosófica"],
        "avaliacao": 4.2,
        "disponivel": False,
        "emprestado_para": "Chai"
    }
}

# ===== TAREFA 1: Listar todos os títulos da biblioteca =====
print("=== TAREFA 1: TÍTULOS DA BIBLIOTECA ===")
for titulo in biblioteca.keys():
    print(titulo)
print()

# ===== TAREFA 2: "Romance" é um gênero de "Python Illustrated"? =====
print("=== TAREFA 2: VERIFICANDO GÊNERO ===")
tem_romance = "Romance" in biblioteca["Python Illustrated"]["generos"]
print(f"'Romance' é um gênero de 'Python Illustrated'? {tem_romance}")
print()

# ===== TAREFA 3: Nacionalidade do autor de Jane Eyre =====
print("=== TAREFA 3: NACIONALIDADE DO AUTOR ===")
nacionalidade_autora = biblioteca["Jane Eyre"]["autores"][0]["nacionalidade"]
print(f"A autora de 'Jane Eyre' é de nacionalidade: {nacionalidade_autora}")
print()

# ===== TAREFA 4: Chai devolveu o livro - atualizar o dicionário =====
print("=== TAREFA 4: DEVOLUÇÃO DO LIVRO ===")
biblioteca["Se os gatos desaparecessem do mundo"]["disponivel"] = True
biblioteca["Se os gatos desaparecessem do mundo"]["emprestado_para"] = None
print("Livro 'Se os gatos desaparecessem do mundo' atualizado após devolução:")
print(biblioteca["Se os gatos desaparecessem do mundo"])
print()

# ===== TAREFA 5: Python Illustrated está disponível? Quem pegou emprestado? =====
print("=== TAREFA 5: DISPONIBILIDADE DE 'PYTHON ILLUSTRATED' ===")
disponivel = biblioteca["Python Illustrated"]["disponivel"]
print(f"'Python Illustrated' está disponível? {disponivel}")

if not disponivel:
    emprestado_para = biblioteca["Python Illustrated"]["emprestado_para"]
    print(f"O livro está emprestado para: {emprestado_para}")
```
