```python
# Atividade Prática 02 - Caça aos Bugs
# Soneca de gato com bugs

def nap_time(hours_list):
    total = 0
    counter = 0

    # Correção: o contador começava em 1, ignorando
    # o primeiro valor da lista. Agora começa em 0.
    while counter < len(hours_list):
        total += hours_list[counter]
        counter += 1

    return total


cat_naps = [2, 3, 1.5, 4]

print("Total nap time:", nap_time(cat_naps))
```

```python
# Atividade Prática 02 - Parte 02
# Quanto o seu gato mia?

def average_meows(logs):
    total_meows_per_hour = 0

    for log in logs:
        nap_hours = 0

        for nap in log["naps"]:
            nap_hours += nap

        awake_hours = 24 - nap_hours

        # Correção: evita divisão por zero caso o gato
        # durma 24 horas ou mais.
        if awake_hours > 0:
            total_meows_per_hour += log["meows"] / awake_hours

    return total_meows_per_hour / len(logs)


logs = [
    {"day": "Monday", "naps": [1, 2, 5], "meows": 90},
    {"day": "Tuesday", "naps": [4, 3, 2], "meows": 75},
    {"day": "Wednesday", "naps": [8, 6, 9, 1], "meows": 9},
    {"day": "Thursday", "naps": [1, 1, 5, 7], "meows": 38},
    {"day": "Friday", "naps": [3, 3, 2, 8], "meows": 94},
    {"day": "Saturday", "naps": [4, 8, 3], "meows": 64},
    {"day": "Sunday", "naps": [3, 3, 2, 8], "meows": 37}
]

print("Average meows per awake hour:", average_meows(logs))
```

```python
# Atividade Prática 02 - Parte 03
# Bugs na Trilha

def total_distance(*segments):
    total = 0

    for length, unit in segments:
        # Correção: lower() transforma "M" em "m",
        # então a comparação deve ser feita com "m".
        unit = unit.strip().lower()

        if unit == "m":
            total += length

        elif unit == "km":
            # Correção: km para metros é multiplicado por 1000.
            total += length * 1000

    print("Total distance:", total, "meters")


total_distance(
    (2.5, "km"),
    (400, "M"),
    (1.0, "km"),
    (800, "m"),
    (1.3, "KM"),
    (900, "m "),
    (1.1, "km")
)
```

```python
import os

def limpar_tela():
    os.system("cls" if os.name == "nt" else "clear")


def ticket_type():
    while True:
        try:
            age = int(input("Digite sua idade: "))

            if age < 0:
                print("Idade inválida. Digite novamente.")
                continue

            print("\nBem-vindo ao museu!")

            if age < 12:
                print("Você entra de graça. Sortudo!")
            elif age < 18:
                print("Você recebe um desconto para jovens!")
            elif age < 65:
                print("Você paga o valor integral.")
            else:
                print("Você recebe um desconto para idosos.")

            break

        except ValueError:
            print("Entrada inválida. Digite uma idade válida.")


ticket_type()

input("\nPressione ENTER para continuar...")
limpar_tela()
```
