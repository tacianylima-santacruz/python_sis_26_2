```python
# Imports
import os
import xml.etree.ElementTree as ET

from openpyxl import Workbook
```

```python
# Amostras
os.makedirs("notas_xml", exist_ok=True)

notas_amostra = [
    {"num": "1001", "data": "2026-10-10", "fornecedor": "Tech Solutions LTDA", "cnpj": "12.345.678/0001-90", "total": "4500.00"},
    {"num": "1002", "data": "2026-10-12", "fornecedor": "Cloud Cloud Servicos", "cnpj": "98.765.432/0001-11", "total": "1280.50"},
    {"num": "1003", "data": "2026-10-15", "fornecedor": "Hardware &amp; Redes SA", "cnpj": "45.123.890/0001-55", "total": "8900.00"},
]

for n in notas_amostra:
    conteudo = f"""<?xml version="1.0" encoding="UTF-8"?>
<NFe>
    <infNFe>
        <ide>
            <nNF>{n['num']}</nNF>
            <dhEmi>{n['data']}T14:30:00-03:00</dhEmi>
        </ide>
        <emit>
            <CNPJ>{n['cnpj']}</CNPJ>
            <xNome>{n['fornecedor']}</xNome>
        </emit>
        <total>
            <ICMSTot>
                <vNF>{n['total']}</vNF>
            </ICMSTot>
        </total>
    </infNFe>
</NFe>
"""
    with open(f"notas_xml/nfe_{n['num']}.xml", "w", encoding="utf-8") as f:
        f.write(conteudo)

print("Amostras criadas na pasta 'notas_xml'!")
```

```python
# Varredura
arquivos = []
for nome in os.listdir("notas_xml"):
    if nome.endswith(".xml"):
        arquivos.append(nome)
print(arquivos)
```

```python
# Extração
notas = []
for nome in arquivos:
    try:
        raiz = ET.parse(f"notas_xml/{nome}").getroot()
        numero = raiz.find(".//nNF").text
        data = raiz.find(".//dhEmi").text[:10]
        fornecedor = raiz.find(".//xNome").text
        cnpj = raiz.find(".//CNPJ").text
        valor = float(raiz.find(".//vNF").text)
        notas.append([numero, data, fornecedor, cnpj, valor])
    except Exception:
        print(f"Erro ao ler {nome}")
```

```python
# Planilha
wb = Workbook()
ws = wb.active
ws.append(["Número", "Data", "Fornecedor", "CNPJ", "Valor Total"])
for n in notas:
    ws.append([n[0], n[1], n[2], n[3], f"{n[4]:.2f}"])
wb.save("notas_fiscais.xlsx")
print(f"{len(notas)} notas salvas.")
```
