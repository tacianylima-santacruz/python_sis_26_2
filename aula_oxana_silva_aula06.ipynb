```python
# Animais Favoritos

# Lista de animais
animais = ["cachorro", "gato", "coelho", "cavalo", "tartaruga"]

# 1. Mostrar os animais favoritos
print("Meus animais favoritos:", animais)

# 2. Mostrar a quantidade de animais
print("Quantidade de animais:", len(animais))

# 3. Mostrar o primeiro e o ultimo animal favorito
print("Primeiro animal favorito:", animais[0])
print("Ultimo animal favorito:", animais[-1])

# 4. Remover um animal da lista
animais.remove("cavalo")
print("Lista depois de remover um animal:", animais)

# 5. Adicionar animais na lista
animais.append("hamster")
animais.append("passaro")
print("Lista depois de adicionar animais:", animais)

# 6. Verificar quantos animais temos agora
print("Quantidade de animais agora:", len(animais))

# 7. Trocar a ordem dos animais
animais.reverse()
print("Animais em nova ordem:", animais)
```

```python
# Atividade Pratica 03 Aula 06
# Agenda Telefonica - Dicionarios

# 1. Criando o dicionario de contatos
contatos = {
    "Ana": "99999-1111",
    "Joao": "99999-2222",
    "Maria": "99999-3333"
}

print("1. Contatos iniciais:", contatos)


# 2. Adicionando um novo contato
contatos["Pedro"] = "99999-4444"

print("2. Novo contato adicionado:", contatos)


# 3. Verificando se Chai existe
if "Chai" not in contatos:
    contatos["Chai"] = "99999-5555"
    print("3. Chai nao existia e foi adicionada:", contatos)
else:
    print("3. Chai ja existe:", contatos)


# 4. Excluindo um contato
contatos.pop("Joao", None)

print("4. Depois de excluir Joao:", contatos)


# 5. Imprimindo todos os nomes
print("5. Nomes presentes no dicionario:")

for nome in contatos:
    print(nome)
```

```python
# A Sacola do Clube do Livro
# Conjuntos

# 1. Criando a lista com livros repetidos
livros = [
    "Duna",
    "O Hobbit",
    "Harry Potter",
    "Duna",
    "O Hobbit",
    "Senhor dos Aneis",
    "Duna"
]

print("Lista com livros repetidos:")
print(livros)


# 2. Convertendo a lista para um conjunto
livros_unicos = set(livros)

print("Conjunto sem livros repetidos:")
print(livros_unicos)


# 3. Mostrando o resultado final
print("Quantidade de livros diferentes:", len(livros_unicos))
```
