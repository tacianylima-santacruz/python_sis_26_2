```python
# ETAPA 1 - IMPORTAÇÕES E PADRÃO PEP 8

import csv
import glob
import os
import xml.etree.ElementTree as ET

import pandas as pd
```

```python
# ETAPA 2 - COLETA E VARREDURA DE XMLS

os.makedirs("notas_xml", exist_ok=True)

notas_amostra = [
    {
        "num": "1001",
        "data": "2026-10-10",
        "fornecedor": "Tech Solutions LTDA",
        "cnpj": "12.345.678/0001-90",
        "total": "4500.00"
    },
    {
        "num": "1002",
        "data": "2026-10-12",
        "fornecedor": "Cloud Cloud Servicos",
        "cnpj": "98.765.432/0001-11",
        "total": "1280.50"
    },
    {
        "num": "1003",
        "data": "2026-10-15",
        "fornecedor": "Hardware e Redes SA",
        "cnpj": "45.123.890/0001-55",
        "total": "8900.00"
    }
]

for nota in notas_amostra:
    conteudo = f"""<?xml version="1.0" encoding="UTF-8"?>
<nota_fiscal>
    <numero>{nota["num"]}</numero>
    <data_emissao>{nota["data"]}T14:30:00-03:00</data_emissao>
    <emitente>
        <cnpj>{nota["cnpj"]}</cnpj>
        <nome>{nota["fornecedor"]}</nome>
    </emitente>
    <valor_total>{nota["total"]}</valor_total>
</nota_fiscal>
"""

    with open(
        f"notas_xml/nfe_{nota['num']}.xml",
        "w",
        encoding="utf-8"
    ) as arquivo:
        arquivo.write(conteudo)

print("Amostras criadas na pasta 'notas_xml'!")
```

```python
# Varredura dos arquivos XML

arquivos_xml = glob.glob("notas_xml/*.xml")

print("\nArquivos XML encontrados:")

for arquivo in arquivos_xml:
    print("-", arquivo)

print("\nTotal de arquivos encontrados:", len(arquivos_xml))
```

```python
# ETAPA 3 - EXTRAÇÃO DOS DADOS

dados_notas = []

for arquivo in arquivos_xml:

    try:
        arvore = ET.parse(arquivo)
        raiz = arvore.getroot()

        numero = raiz.findtext("numero")
        data_emissao = raiz.findtext("data_emissao")
        fornecedor = raiz.findtext("emitente/nome")
        cnpj = raiz.findtext("emitente/cnpj")
        valor_total = raiz.findtext("valor_total")

        dados_notas.append({
            "Numero da Nota": numero,
            "Data de Emissao": data_emissao,
            "Fornecedor": fornecedor,
            "CNPJ": cnpj,
            "Valor Total": float(valor_total)
        })

        print("XML processado:", arquivo)

    except ET.ParseError:
        print("Erro ao ler o XML:", arquivo)

    except Exception as erro:
        print("Erro ao processar o arquivo:", arquivo)
        print("Detalhes:", erro)

print("\nDados extraídos com sucesso!")

for nota in dados_notas:
    print(nota)



```

```python
# ETAPA 4 - GERAÇÃO DA PLANILHA

tabela_notas = pd.DataFrame(dados_notas)

print("Tabela consolidada:")
display(tabela_notas)


# Formatando o valor monetário

tabela_notas["Valor Total"] = tabela_notas["Valor Total"].apply(
    lambda valor: f"{valor:.2f}"
)

display(tabela_notas)

# Gerando o arquivo CSV

tabela_notas.to_csv(
    "notas_fiscais.csv",
    index=False,
    sep=";",
    encoding="utf-8-sig"
)

print("Arquivo 'notas_fiscais.csv' criado com sucesso!")

# Gerando o arquivo XLSX

tabela_notas.to_excel(
    "notas_fiscais.xlsx",
    index=False
)

print("Arquivo 'notas_fiscais.xlsx' criado com sucesso!")
```
