```python
#Atividade 07

#Atividade01

#o erro estava na variável (nome_heroi_cidade)
nome_heroi = "Batman"
print("O herói que vai salvar a cidade é o " + nome_heroi)


#Atividade02
idade_pet = 5
# CORREÇÃO: o operador + não concatena str com int (TypeError). Converti a idade
# para texto com str() antes de juntar as partes da mensagem.
mensagem = "A idade do meu cachorro é " +str(idade_pet)+" anos."
print(mensagem)

#Atividade03

print("Calculadora de Dobro!")
# Simulando alguém que digitou a palavra 'dez' em vez do número 10
numero_digitado = "dez"

# CORREÇÃO: int("dez") gera ValueError, pois "dez" não é um número em formato
# de dígitos. Usei try/except para tratar a entrada inválida sem quebrar o programa.
try:
    dobro = int(numero_digitado) * 2
    print("O dobro é:", dobro)
except ValueError:
    print("Entrada inválida! Digite apenas números, como 10 (e não 'dez').")


#Atividade04

cat_attributes = {
    "name": "Zia",
    "age": 1,
    "color": "grey"
}

# Correção: a chave "breed" não existia no dicionário.
# Adicionamos a raça da gata antes de tentar acessá-la.
cat_attributes["breed"] = "Siamês"

raca_gato = cat_attributes["breed"]

print("A raça da gata é:", raca_gato)


#Atividade05

some_numbers = [10, 20, 30]

# Correção: a lista possui apenas 3 números.
# Os índices começam em 0, então o último elemento está no índice 2.
quarto_numero = some_numbers[2]

print("O número escolhido foi:", quarto_numero)

#Atividade06

tesouro_moedas = 100
aventureiros_sobreviventes = 0

print("Calculando a divisão do tesouro...")

# Correção: não é possível dividir por zero.
# Verificamos se existem aventureiros antes de realizar a divisão.
if aventureiros_sobreviventes > 0:
    moedas_por_pessoa = tesouro_moedas / aventureiros_sobreviventes
    print("Cada aventureiro vai receber:", moedas_por_pessoa)
else:
    print("Não há aventureiros sobreviventes para dividir o tesouro.")


#Atividade07
print("=== Sistema do Zoológico ===")

# horário, alimento, quantidade
dieta_axolote = ("08:00", "minhocas", 50)

print("O axolote enjoou de minhocas. Vamos alterar a dieta para pizza!")

# Correção: tuplas são imutáveis. Por isso, criamos uma nova
# tupla com o alimento alterado em vez de tentar modificar a original.
dieta_axolote = ("08:00", "pizza de calabresa", 50)

print("Nova dieta:", dieta_axolote)


#Atvidade08

def contagem_regressiva(tempo):
    print(f"Faltam {tempo} segundos...")

    # Correção: adicionamos uma condição de parada.
    # A função só chama a si mesma enquanto o tempo for maior que 0.
    if tempo > 0:
        contagem_regressiva(tempo - 1)


print("Iniciando contagem para a explosão do Anel...")
contagem_regressiva(3)




#Atividade02

#Atividade01

def nap_time(hours_list):
    total = 0
    counter = 0  # Correção: começamos no índice 0 para incluir o primeiro cochilo.

    while counter < len(hours_list):
        total += hours_list[counter]
        counter += 1

    return total


cat_naps = [2, 3, 1.5, 4]

print("Total nap time:", nap_time(cat_naps))


#Atividade02

def average_meows(logs):
    total_meows_per_hour = 0
    valid_days = 0

    for log in logs:
        nap_hours = 0

        for nap in log["naps"]:
            nap_hours += nap

        awake_hours = 24 - nap_hours

        # Correção: se o gato dormiu 24 horas ou mais,
        # não podemos dividir os miados por zero.
        if awake_hours > 0:
            total_meows_per_hour += log["meows"] / awake_hours
            valid_days += 1

    # Correção: usamos apenas os dias que possuem horas acordadas.
    return total_meows_per_hour / valid_days


logs = [
    {"day": "Monday", "naps": [1, 2, 5], "meows": 90},
    {"day": "Tuesday", "naps": [4, 3, 2], "meows": 75},
    {"day": "Wednesday", "naps": [8, 6, 9, 1], "meows": 9},
    {"day": "Thursday", "naps": [1, 1, 5, 7], "meows": 38},
    {"day": "Friday", "naps": [3, 3, 2, 8], "meows": 94},
    {"day": "Saturday", "naps": [4, 8, 3], "meows": 64},
    {"day": "Sunday", "naps": [3, 3, 2, 8], "meows": 37}
]

print("Average meows per awake hour:", average_meows(logs))


#Atividade03

def total_distance(*segments):
    total = 0

    for length, unit in segments:
        # Correção: strip() remove espaços extras e lower()
        # transforma a unidade em minúscula para facilitar a comparação.
        unit = unit.strip().lower()

        if unit == "m":
            total += length

        # Correção: como o total será exibido em metros,
        # quilômetros devem ser multiplicados por 1000.
        elif unit == "km":
            total += length * 1000

    print("Total distance:", total, "meters")


total_distance(
    (2.5, "km"),
    (400, "M"),
    (1.0, "km"),
    (800, "m"),
    (1.3, "KM"),
    (900, "m "),
    (1.1, "km")
)


#Atividade04
import os


def limpar_tela():
    # Correção: limpa a tela depois de cada atendimento,
    # independentemente de a idade ser válida ou inválida.
    os.system("cls" if os.name == "nt" else "clear")


def ticket_type():
    while True:
        try:
            age = int(input("Enter your age: "))

            # Correção: verificamos se a idade é válida.
            if age < 0:
                print("Invalid age. Please enter a valid age.")
                limpar_tela()
                continue

            # Idade válida: damos as boas-vindas.
            print("Welcome!")

            if age < 12:
                print("You get in for free. Lucky you!")
            elif age < 18:
                print("You get a youth discount!")
            elif age < 65:
                print("That's a full-price ticket for you.")
            else:
                print("You get a senior discount!")

            limpar_tela()
            break

        except ValueError:
            # Correção: se o visitante digitar algo que não seja
            # um número inteiro, evitamos que o programa quebre.
            print("Invalid input. Please enter your age.")
            limpar_tela()


ticket_type()
```
