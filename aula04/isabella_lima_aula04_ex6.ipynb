```python
# AULA 04 | EXERCÍCIO 6 | WHILE E FOR

# Sistema de Caixa e Parcelamento:
# Objetivo: Desenvolver um programa que utilize os dois tipos de laço de repetição (while e for).
# O sistema deve pedir ao caixa para digitar o preço de cada produto.
# O programa deve continuar somando os preços até que seja digitado 0.
total = 0

preco = float(input("Digite o preço do produto ou 0 para finalizar: R$ "))

while preco != 0:
    total = total + preco
    preco = float(input("Digite o preço do próximo produto ou 0 para finalizar: R$ "))

print(f"\nTotal da compra: R$ {total:.2f}")

# Opções de Parcelamento:
# Após calcular o total, o sistema deve mostrar ao cliente as opções de parcelamento,
# mostrando o valor de cada parcela.
print("\n=== OPÇÕES DE PARCELAMENTO ===")

for parcela in range(1, 6):
    valor_parcela = total / parcela
    print(f"{parcela}x de R$ {valor_parcela:.2f}")

# Na etapa 1, o while é adequado, pois a quantidade de itens da compra pode variar.
# Na etapa 2, o for é adequado, pois o parcelamento possui uma sequência definida.
```
