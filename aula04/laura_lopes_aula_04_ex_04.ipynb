```python
nome = input("A sua jornada começou, me diga qual o seu nome? ")
print(f"Maravilhoso {nome}, agora escolha a sua classe:")
es_classe = int(input("mago digite 1, lutador digite 2, arqueiro digite 3: "))

if es_classe == 1:
    lv = 1
    exp = 1
    classe = "Mago"
    hp_base = 20 * lv
    mana_base = 40 * lv
    dano_mag_base = 25 * lv
    dano_fis_base = 15 * lv
    gold = 15
elif es_classe == 2:
    lv = 1
    exp = 1
    classe = "Lutador"
    hp_base = 50 * lv
    mana_base = 15 * lv
    dano_mag_base = 15 * lv
    dano_fis_base = 25 * lv
    gold = 15
elif es_classe == 3:
    lv = 1
    exp = 1
    classe = "Arqueiro"
    hp_base = 30 * lv
    mana_base = 30 * lv
    dano_mag_base = 20 * lv
    dano_fis_base = 20 * lv
    gold = 15
else:
    print("Escolha uma opção valida")


vamp_hp = 15 * lv
vamp_dano = 10 * lv
vamp_exp = 10 * lv
vamp_drop = 5

esf = False
esm = False

print("\n--------------------------------")
print("Otimo seu personagem foi criando")
print(f"Nome: {nome}")
print(f"Classe: {classe}")
print(f"Level : {lv}")
print(f"Vida: {hp_base}")
print(f"Mana: {mana_base}")
print(f"Gold: {gold}")

print("\nOh não um vampiro apareceu! tente ataca-lo")
act = int(input("Digite 1 para ataque magico e 2 para ataque fisico: "))

if act == 1:
    esm = True
    dano = dano_fis_base * esf + dano_mag_base * esm
    vamp_hp -= dano
    print(f"Você deu {dano}, de dano")
    print(f"A vida do vampiro agora é: {vamp_hp}")
elif act == 2:
    esf = True
    dano = dano_fis_base * esf + dano_mag_base * esm
    vamp_hp -= dano
    print(f"Você deu {dano}, de dano")
    print(f"A vida do vampiro agora é: {vamp_hp}")
else:
    print("Digite um valor valido")


# Primeira decisão: Verificar se o vampiro foi derrotado
if vamp_hp <= 0:
    print("\nParabens você derrotou o seu primeiro inimigo!")
    gold += vamp_drop
    exp += vamp_exp
    print(f"Você ganhou {vamp_drop} de Gold e {vamp_exp} de XP")
    print(f"Agora você tem {gold} de Gold")
else:
    print(f"\nO vampiro ainda está vivo com {vamp_hp} de HP!")
    print(f"Ele contra-ataca e você perde {vamp_dano // 2} de vida!")
    hp_base -= (vamp_dano // 2)
    print(f"Sua vida agora é: {hp_base}")

# Segunda decisão: Verificar se o personagem morreu
if hp_base <= 0:
    print("\n💀 GAME OVER! Seu personagem morreu!")
    print("Sua jornada chegou ao fim...")
    exit()
else:
    print(f"\nVocê sobreviveu com {hp_base} de vida!")

# Terceira decisão: Verificar se subiu de nível
lv_up = 10 * lv
print(f"XP: {exp}")
print(f"XP necessário para subir: {lv_up}")

if exp >= lv_up:
    lv += 1
    print(f"\n🎉 Parabens você subiu de Level! Agora você é nível: {lv}")
    # Recalculando os atributos com o novo level
    if es_classe == 1:
        hp_base = 20 * lv
        mana_base = 40 * lv
        dano_mag_base = 25 * lv
        dano_fis_base = 15 * lv
    elif es_classe == 2:
        hp_base = 50 * lv
        mana_base = 15 * lv
        dano_mag_base = 15 * lv
        dano_fis_base = 25 * lv
    elif es_classe == 3:
        hp_base = 30 * lv
        mana_base = 30 * lv
        dano_mag_base = 20 * lv
        dano_fis_base = 20 * lv
    print(f"Vida: {hp_base}")
    print(f"Mana: {mana_base}")
    print(f"Dano Mágico: {dano_mag_base}")
    print(f"Dano Físico: {dano_fis_base}")
else:
    print(f"\nFalta {lv_up - exp} de XP para subir de nível!")

# Quarta decisão: O que fazer com o gold após a batalha
print("\n--- Loja da Vila ---")
print(f"Você tem {gold} de gold")
print("1. Comprar poção de vida (custa 10 gold)")
print("2. Comprar poção de mana (custa 8 gold)")
print("3. Ir para a taverna descansar (custa 5 gold)")
print("4. Guardar o dinheiro e continuar jornada")

loja = int(input("Escolha uma opção: "))

# Quinta decisão: Verificar se tem gold suficiente para comprar
if loja == 1:
    if gold >= 10:
        gold -= 10
        hp_base += 20
        print(f"\nVocê comprou uma poção de vida! +20 de HP")
        print(f"HP atual: {hp_base}")
        print(f"Gold restante: {gold}")
    else:
        print(f"\nGold insuficiente! Você tem {gold}, precisa de 10")

elif loja == 2:
    if gold >= 8:
        gold -= 8
        mana_base += 15
        print(f"\nVocê comprou uma poção de mana! +15 de MANA")
        print(f"Mana atual: {mana_base}")
        print(f"Gold restante: {gold}")
    else:
        print(f"\nGold insuficiente! Você tem {gold}, precisa de 8")

elif loja == 3:
    if gold >= 5:
        gold -= 5
        hp_base += 10
        print(f"\nVocê descansou na taverna! +10 de HP")
        print(f"HP atual: {hp_base}")
        print(f"Gold restante: {gold}")
    else:
        print(f"\nGold insuficiente! Você tem {gold}, precisa de 5")
        print("Você dormiu na rua... -5 de HP por não descansar bem!")
        hp_base -= 5
        print(f"HP atual: {hp_base}")

elif loja == 4:
    print("\nVocê guardou seu dinheiro e continuou a jornada!")
else:
    print("Opção inválida!")



print("\nFim da jornada! ")
```
