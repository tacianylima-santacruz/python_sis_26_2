```python
nota = float(input("Digite a sua nota: "))

if nota >= 9.0:
  print("\nOtimo trabalho aluno A++")
elif nota >= 7.0:
  print("\nBom trabalho aluno nota B")
elif nota >= 5.0:
  print("\nAprovado, mas por pouco")
else:
  print("\nReprovado")

```
