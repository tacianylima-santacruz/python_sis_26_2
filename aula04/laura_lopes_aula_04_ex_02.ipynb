```python
print("O nome de usuário digitado deve ser igual ao cadastrado no sistema.")
user_salvo = "admin"
pswrd_salvo = "123"
block = False

user = input("Digite o nome de usuario ")
pswrd = input("Digite a senha ")

credenc_user = user == user_salvo
credenc_pswrd = pswrd == pswrd_salvo
credenc = credenc_pswrd and credenc_user
acess = credenc and not block

print(f"\nAs credenciais (usuário e senha) estão corretas? {credenc}")
print(f"A conta está bloqueada? {block}")
print(f"O acesso final está liberado (credenciais corretas E conta NÃO bloqueada)? {acess}")
```
