```python
# AULA 04 | EXERCÍCIO 3 | IF, ELIF, ELSE

# Pergunte ao usuário qual foi a sua nota
# (lembre-se de converter para float, pois as notas podem ter decimais).
nota = float(input("Digite sua nota: "))

# Utilize as estruturas condicionais if, elif e else para avaliar a nota e
# exibir na tela o título conquistado pelo aluno,
# para cada uma das condições abaixo deve-se exibir uma mensagem diferente:
# Nota 9.0 ou mais;
if nota >= 9:
    print(f"Sua nota foi {nota}, portanto você está aprovado com sucesso!")
# Nota de 7.0 até 8.9;
elif nota >= 7:
    print(f"Sua nota foi {nota}, portanto você está aprovado!")
# Nota de 5.0 até 6.9;
elif nota >= 5:
    print(f"Sua nota foi {nota}, portanto você está em recuperação!")
# Nota abaixo de 5.0;
else:
    print(f"Sua nota foi {nota}, portanto você está reprovado!")
```
