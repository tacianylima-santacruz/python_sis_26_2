```python

print("Sistema de Caixa e Parcelamento")
print(" ")

precos =[]
preco = 1
preco_total = 0

while preco != 0:
    preco = float(input("Digite o preço do produto: "))

    if preco != 0:
        precos.append(preco)

for i in precos:
  preco_total += i

parcelamento_10 = preco_total/10
parcelamento_15 = preco_total/15
parcelamento_20 = preco_total/20

print(f"\nValor total: {preco_total:.2f}")

print("\nOpções de parcelamento:")
print(f"10 parcelas, valor por parcela: {parcelamento_10:.2f}")
print(f"15 parcelas, valor por parcela: {parcelamento_15:.2f}")
print(f"20 parcelas, valor por parcela: {parcelamento_20:.2f}")


# O while foi a melhor escolha para a etapa 1 porque não sabemos
# quantos produtos serão cadastrados. O laço continua até o usuário
# digitar 0.
#O for foi a melhor escolha para a etapa 2 porque podemos
# percorrer a lista de preços para somar todos os valores.
```
