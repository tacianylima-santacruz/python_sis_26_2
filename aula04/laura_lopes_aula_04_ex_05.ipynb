```python
print("  SISTEMA DE CHAMADOS - HELP DESK  ")

print("\n Menu de Opcoes:")
print("1 - Problema com Computador/Hardware")
print("2 - Instalacao de Software")
print("3 - Sem acesso a Internet/Rede")
print("4 - Reset de Senha")
print("5 - Problema com Impressora")
print("6 - Virus/Malware")
print("0 - Sair do sistema")


opcao = int(input("\n Digite o numero do seu problema: "))

match opcao:
  case 1:
    print("\n Chamado direcionado para: Manutencao")
    print(" Equipe de hardware ira verificar seu computador.")
  case 2:
    print("\n Chamado direcionado para: Equipe de Suporte")
    print(" Especialistas em software ajudarao na instalacao.")
  case 3:
    print("\n Chamado direcionado para: Infraestrutura")
    print(" Equipe de rede ira verificar sua conexao.")
  case 4:
    print("\n Chamado direcionado para: Atendimento Automatico")
    print(" Link de redefinicao de senha enviado para seu e-mail.")
  case 5:
    print("\n Chamado direcionado para: Equipe de Impressao")
    print(" Especialistas em impressao irao resolver seu problema.")
  case 6:
    print("\n Chamado direcionado para: Equipe de Seguranca")
    print(" Especialistas em seguranca irao remover o virus.")
  case 0:
     print("\n Saindo do sistema...")
  case _:
     print("\n  Opcao invalida!")
     print(" Chamado encaminhado para a Triagem Geral.")
```
