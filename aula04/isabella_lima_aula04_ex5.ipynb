```python
# AULA 04 | EXERCÍCIO 5 | MATCH...CASE

# Sistema de Chamados (Help Desk)
# Você foi contratado para desenvolver a triagem de um sistema de chamados de TI.
# O programa deve pedir ao usuário para digitar o número correspondente ao problema que ele está enfrentando.
# Utilizando a estrutura match...case, direcione o chamado para a equipe correta e
# exiba uma mensagem confirmando o direcionamento, o menu de escolhas deve ter pelo menos 4 opções.
print("=== SISTEMA DE CHAMADOS DE TI ===")
print("\nBem-vindo ao Help Desk!")
print("\nSelecione abaixo o número do problema que você está enfrentando:")
print("\n1 | Problema com Computador/Hardware")
print("2 | Instalação de Software")
print("3 | Sem acesso à Internet/Rede")
print("4 | Reset de Senha")
print("5 | Outro problema")

opcao = int(input("\nDigite o número da opção: "))

numero_chamado = 2026001

match opcao:
  case 1:
    print(f"Chamado Nº {numero_chamado} direcionado para Manutenção.")

  case 2:
    print(f"Chamado Nº {numero_chamado} direcionado para Equipe de Suporte.")

  case 3:
    print(f"Chamado Nº {numero_chamado} direcionado para Infraestrutura.")

  case 4:
    print(f"Chamado Nº {numero_chamado} direcionado para Atendimento Automático.")

  case 5:
    problema = input("\nFaça uma pequena descrição do problema: ")
    print("\nDescrição registrada com sucesso!")
    print(f"\nChamado Nº {numero_chamado} direcionado para Atendimento Humano.")

  case _:
    print("Opção inválida. Selecione uma das opções disponíveis.")
```
