```python
pl1 = int(input())
pl2 = int(input())

igual = pl1 == pl2
diferente = pl1 != pl2
maior = pl1 > pl2
menor = pl1 < pl2
maior_igual = pl1 >=pl2
menor_igual = pl1 <= pl2

print(f"O jogador 1 tem a mesma pontuação que o jogador 2? {igual}")
print(f"O jogador 1 tem a pontuação diferente que o jogador 2? {diferente}")
print(f"O jogador 1 tem a maior pontuação que o jogador 2? {maior}")
print(f"O jogador 1 tem a menor pontuação que o jogador 2? {menor}")
print(f"O jogador 1 tem a maior ou igual pontuação que o jogador 2? {maior_igual}")
print(f"O jogador 1 tem a menor ou igual pontuação que o jogador 2? {menor_igual}")
```
