####Aritmetica das maçãs:

```python
caixa_1 = 12
caixa_2 = 12
caixa_3 = 12
caixa_4 = 12

total_macas = caixa_1 + caixa_2 + caixa_3 + caixa_4

macas_por_cavaleiro = total_macas // 5

portao_aberto = macas_por_cavaleiro >= 8

print (portao_aberto)
```

```python
macas_por_caixa = 12
qtd_caixas = 4
total_cavaleiros = 6

total_macas = macas_por_caixa * qtd_caixas
macas_por_cavaleiro = total_macas // total_cavaleiros

portao_aberto = macas_por_cavaleiro >= 8

print(portao_aberto)
```
