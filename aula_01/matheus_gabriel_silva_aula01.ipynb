# **Atividade Prática 01: Montando a Ficha de um Personagem**

Crie 4 variáveis, utilizando os 4 tipos de dados:
Texto (str): Ex: Nome do seu personagem.

Número Inteiro (int): Ex: Defina a idade ou nível dele.

Número Decimal (float): Ex: Defina a altura do personagem.

Verdadeiro ou Falso (boolean): Ex: Diga se ele possui uma armadura (True) ou não (False).

```python
# # Trabalhando com 'int'
valor = int(input('Escreva um número de 1 a 10: '))
print(valor)

# # Trabalhando com 'str'
Nome = str(input('Escreva seu nome Completo: '))
print(Nome)

# #Trabalhando com 'float'
num = float(input('Escreva quebrado para dividir por 12'))
print(num/12)

# Trabalhando com 'bool'
senha = bool(input('Voce tem senha nos seus aplicativos? (1 - sim | 0 - Não)'))
if senha == 0:
  print('Você tem senha!!!')
else:
  print('Você não tem senha !')
```
