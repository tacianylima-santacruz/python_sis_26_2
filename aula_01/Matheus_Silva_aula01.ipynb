```python
# 1. Texto (str) - Sempre entre aspas
nome_personagem = "Aragorn"

# 2. Número Inteiro (int) - Sem casas decimais
nivel = 15

# 3. Número Decimal (float) - Usa ponto no lugar da vírgula
altura = 1.85

# 4. Verdadeiro ou Falso (bool) - Começa com letra maiúscula
possui_armadura = True

# Exibindo a ficha na tela
print("--- FICHA DO PERSONAGEM ---")
print("Nome:", nome_personagem)
print("Nível:", nivel)
print("Altura:", altura, "m")
print("Possui armadura?", possui_armadura)
```
