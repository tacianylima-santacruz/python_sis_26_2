### ***Aula 1***

Atividade Camelot

```python
rei_arthur = 4
total_macas = rei_arthur * 12
print(total_macas)

cavaleiros = 5
total_macas_cavaleiros = int(total_macas / cavaleiros)
print('O total de maças por cavaleiros é: ', total_macas_cavaleiros)

portao_aberto = 8
print(portao_aberto < total_macas_cavaleiros,'= liberado')

adicional = cavaleiros + 1
resultado = adicional // total_macas
final = (resultado > portao_aberto)
print(final)
```
