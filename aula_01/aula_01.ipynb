```python
def calcular_imposto(valor_produto):
    # Comentário comum: Apenas um lembrete interno para quem está programando
    # Alíquota fixa estabelecida pelo setor fiscal em 2026.
    """
    Docstring: Documentação oficial acessível via help() ou IDE.

    Calcula o imposto aplicável sobre o valor de um produto.

    Parâmetros:
        valor_produto (float): O preço base do produto.

    Retorna:
        float: O valor total do imposto calculado.
    """
    ALIQUOTA = 0.15
    return valor_produto * ALIQUOTA
```

```python
# --- 1. Entrada de dados via input() ---
# Por padrão, TODO valor retornado pelo input() é do tipo texto (str)
qtd_servidores_texto = input("Digite a quantidade de servidores: ")

# Mostra o tipo original retornado pelo input
print("Tipo do dado lido pelo input():", type(qtd_servidores_texto))
# Saída: <class 'str'>


# --- 2. Conversão de tipo (Casting) ---
# Convertendo o texto para um número inteiro (int)
qtd_servidores = int(qtd_servidores_texto)

# Mostra o tipo após a conversão
print("Tipo após a conversão com int():", type(qtd_servidores))
# Saída: <class 'int'>


# --- 3. Exemplo prático de saída com print() e f-string ---
# Agora podemos realizar operações matemáticas com a variável convertida
total_servidores = qtd_servidores + 5

print(f"\nQuantidade informada: {qtd_servidores}")
print(f"Total com bônus de 5: {total_servidores}")
```

```python
# O usuário vai digitar um valor (exemplo: 5)
qtd_texto = input("Digite um número: ")

# Tentando multiplicar a entrada por 10
resultado = qtd_texto * 5

print(f"O resultado é: {resultado}")
```

```python
# O usuário vai digitar dois valores (exemplo: 5 e 10)
texto1 = input("Digite o primeiro número: ")
texto2 = input("Digite o segundo número: ")

# Juntando (concatenando) as duas entradas de texto
resultado = texto1 + texto2

print(f"O resultado é: {resultado}")
```
