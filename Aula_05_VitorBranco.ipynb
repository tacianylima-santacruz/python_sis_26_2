```python
pacotes = [12, 0, 25, 40, -1, 18, 50]

for pacote in pacotes:
    if pacote == 0:
        continue

    if pacote < 0:
        print("ALERTA: pacote com tamanho negativo detectado!")
        break

    print(f"Pacote de {pacote} MB processado com sucesso.")
```

```python

def registrar_heroi(nome, classe, reino="Desconhecido"):
    return f"Herói: {nome} | Classe: {classe} | Reino: {reino}"


nome = input("Digite o nome do herói: ")
classe = input("Digite a classe do herói: ")
reino = input("Digite o reino (ou deixe vazio): ")

if reino == "":
    print(registrar_heroi(nome, classe))
else:
    print(registrar_heroi(nome, classe, reino))

def chamar_armas(nome, *itens):
    print(f"\nEquipamentos de {nome}:")

    for item in itens:
        print(f"- {item}")


item1 = input("Digite o primeiro item: ")
item2 = input("Digite o segundo item: ")
item3 = input("Digite o terceiro item: ")

chamar_armas(nome, item1, item2, item3)

def ficha_personagem(nome, classe, reino, **atributos):
    print(f"\n===== FICHA DO PERSONAGEM =====")
    print(f"Nome: {nome}")
    print(f"Classe: {classe}")
    print(f"Reino: {reino}")

    print("\nAtributos:")

    for atributo, valor in atributos.items():
        print(f"{atributo}: {valor}")


forca = input("Digite a força: ")
agilidade = input("Digite a agilidade: ")
resistencia = input("Digite a resistência: ")

ficha_personagem(
    nome,
    classe,
    reino,
    forca=forca,
    agilidade=agilidade,
    resistencia=resistencia
)
```

```python

servidores_online = 0


def alterar_servidores(quantidade):
    """
    Ativa ou desativa servidores e atualiza o total de servidores online.

    Parâmetros:
        quantidade (int): Quantidade de servidores a ativar ou desativar.

    Impacto na variável global:
        Modifica a variável global servidores_online.
    """
    global servidores_online
    servidores_online += quantidade


alterar_servidores(5)
alterar_servidores(-5)

print("Saldo de servidores ativos:", servidores_online)



def gerenciar_conexoes():
    limite_conexoes = 10

    def usar_conexao():
        nonlocal limite_conexoes
        limite_conexoes -= 1

    usar_conexao()
    usar_conexao()

    print("Limite de conexões restantes:", limite_conexoes)


gerenciar_conexoes()


cota_armazenamento = 100

alterar_cota = lambda cota_atual, nova_cota: nova_cota

cota_armazenamento = alterar_cota(cota_armazenamento, 150)

print("Nova cota de armazenamento:", cota_armazenamento, "GB")

```
