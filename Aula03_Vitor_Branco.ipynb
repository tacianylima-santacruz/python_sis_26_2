```python

dna = input("Digite uma fita de DNA com 21 nucleotídeos: ").upper()

# Sequência de referência da Hemoglobina A
dna_referencia = "ATGGTGCACCTGACTCCTGAG"

# Comparação das sequências
mutante = dna != dna_referencia

# Diagnósticos
diagnosticos = [
    "✅ Sequência NORMAL (Hemoglobina A)",
    "⚠️ Sequência MUTANTE (Anemia Falciforme - HbS)"
]

# Tradução da fita de DNA para aminoácidos
proteina = dna.replace("ATG", "Met-") \
             .replace("GTG", "Val-") \
             .replace("CAC", "His-") \
             .replace("CTG", "Leu-") \
             .replace("ACT", "Thr-") \
             .replace("CCT", "Pro-") \
             .replace("GAG", "Glu-")

# Busca de aminoácido
aminoacido = input("Digite um aminoácido para buscar (ex: Val ou His): ")

posicao = proteina.find(aminoacido)

# Relatório final
print("\n========== RELATÓRIO GENÉTICO ==========")
print(f"DNA de referência: {dna_referencia}")
print(f"DNA digitado: {dna}")
print(f"Tamanho da fita: {len(dna)} nucleotídeos")
print(f"Proteína traduzida: {proteina}")
print(f"Diagnóstico: {diagnosticos[mutante]}")
print(f"Aminoácido pesquisado: {aminoacido}")
print(f"Posição do aminoácido: {posicao}")
```

```python


# Entrada do texto
pergaminho = input("Digite o texto do pergaminho: ")

# Limpeza dos espaços no início e no final
texto_limpo = pergaminho.strip()

# Separação das palavras
palavras = texto_limpo.split()

# Informações do texto
total_caracteres = len(texto_limpo)
total_palavras = len(palavras)
primeira_palavra = palavras[0]
ultima_palavra = palavras[-1]

# Detector de ameaças
palavra_ameaca = "Anel"
existe_ameaca = palavra_ameaca in texto_limpo
quantidade_ameaca = texto_limpo.count(palavra_ameaca)

# Relatório
print("\n========== RELATÓRIO DO PERGAMINHO ==========")
print(f"Texto limpo: {texto_limpo}")
print(f"Total de caracteres: {total_caracteres}")
print(f"Total de palavras: {total_palavras}")
print(f"Primeira palavra: {primeira_palavra}")
print(f"Última palavra: {ultima_palavra}")

print("\n========== DETECTOR DE AMEAÇAS ==========")
print(f"A palavra '{palavra_ameaca}' existe no texto? {existe_ameaca}")
print(f"Quantidade de vezes que aparece: {quantidade_ameaca}")
```
