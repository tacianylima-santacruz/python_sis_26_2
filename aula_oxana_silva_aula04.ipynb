```python
#Pontuacao dos jogadores

pontuacao_jogador1 = 80
pontuacao_jogador2 = 70

print("Jogador 1 e maior que o Jogador 2:", pontuacao_jogador1 > pontuacao_jogador2)
print("Jogador 1 e menor que o Jogador 2:", pontuacao_jogador1 < pontuacao_jogador2)
print("Jogador 1 e igual ao Jogador 2:", pontuacao_jogador1 == pontuacao_jogador2)
print("Jogador 1 e diferente do Jogador 2:", pontuacao_jogador1 != pontuacao_jogador2)
print("Jogador 1 e maior ou igual ao Jogador 2:", pontuacao_jogador1 >= pontuacao_jogador2)
print("Jogador 1 e menor ou igual ao Jogador 2:", pontuacao_jogador1 <= pontuacao_jogador2)
```

```python
# Sistema de seguranca

usuario_cadastrado = "Oxana"
senha_cadastrada = "1234"

usuario_digitado = "Oxana"
senha_digitada = "1234"

conta_bloqueada = False

print("As credenciais estao corretas:", usuario_digitado == usuario_cadastrado and senha_digitada == senha_cadastrada)

print("A conta esta bloqueada:", conta_bloqueada)

print("O acesso final esta liberado:", usuario_digitado == usuario_cadastrado and senha_digitada == senha_cadastrada and not conta_bloqueada)
```

```python
# A Jornada do Personagem

vida = 100
dano = 35
vida_restante = vida - dano

forca = 20
bonus_espada = 15
forca_total = forca + bonus_espada

ouro = 100
preco_pocao = 30
ouro_restante = ouro - preco_pocao

print("O personagem entrou em uma batalha!")

if vida_restante <= 0: print("Game Over! O personagem perdeu toda a vida.")
else: print("O personagem sobreviveu com", vida_restante, "de vida.")

if forca_total >= 30: print("Ataque forte! O personagem acertou o monstro.")
else: print("O ataque foi fraco e o personagem nao conseguiu derrotar o monstro.")

if ouro_restante >= 50: print("O personagem tem dinheiro suficiente para dormir na taverna.")
else: print("O personagem nao tem dinheiro suficiente e vai dormir na rua.")

```

```python
# Sistema de Chamados Help Desk

print("=== SISTEMA DE CHAMADOS ===")
print("1 - Problema com Computador ou Hardware")
print("2 - Instalacao de Software")
print("3 - Sem acesso a Internet ou Rede")
print("4 - Reset de Senha")

opcao = int(input("Digite o numero do problema: "))

match opcao:
    case 1:
        print("Chamado direcionado para: Manutencao")
    case 2:
        print("Chamado direcionado para: Equipe de Suporte")
    case 3:
        print("Chamado direcionado para: Infraestrutura")
    case 4:
        print("Chamado direcionado para: Atendimento Automatico")
    case _:
        print("Opcao invalida. Chamado encaminhado para a Triagem Geral.")



```

```python
total = 0

print("=== CAIXA DA LOJA ===")

while True:
    preco = float(input("Digite o preco do produto (0 para finalizar): "))

    if preco == 0:
        break

    total = total + preco

print("Total da compra: R$", total)

print("=== OPCOES DE PARCELAMENTO ===")

for parcelas in range(1, 6):
    valor_parcela = total / parcelas
    print(parcelas, "x de R$", round(valor_parcela, 2))

# O while foi melhor para a etapa 1 porque nao sabemos quantos produtos serao comprados.
# O for foi melhor para a etapa 2 porque sabemos que vamos mostrar de 1 ate 5 parcelas.

```
