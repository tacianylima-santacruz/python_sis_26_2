```python
# Etapa 1
config = {
    "cluster": "cluster-producao",
    "so": "Ubuntu 22.04",
    "servidores": [
        {"nome": "srv-01", "ip": "192.168.0.10", "portas": [22, 80, 443]},
        {"nome": "srv-02", "ip": "192.168.0.11", "portas": [22, 3306]},
        {"nome": "srv-03", "ip": "192.168.0.12", "portas": [22, 8080]},
    ],
}
```

```python
# Etapa 2
import json

with open("cluster_config.json", "w") as f:
    json.dump(config, f, indent=4)
```

```python
# Etapa 3
import csv

with open("servidores_cluster.csv", "w", newline="") as f:
    w = csv.writer(f, delimiter=";")
    w.writerow(["Servidor", "IP", "Portas"])
    for s in config["servidores"]:
        w.writerow([s["nome"], s["ip"], ",".join(map(str, s["portas"]))])
```

```python
# Etapa 1 - Atv 2
!pip install openpyxl
import openpyxl
from openpyxl.styles import PatternFill
```

```python
# Etapa 2
wb = openpyxl.Workbook()
ws = wb.active
ws.title = "Servidores"
ws.append(["Hostname", "IP", "Memória (GB)", "Uso de Disco (%)", "Status"])
```

```python
# Etapa 3
servidores = [
    ("srv-web-01", "10.0.0.1", 16, 45, "Ativo"),
    ("srv-db-01", "10.0.0.2", 64, 85, "Ativo"),
    ("srv-app-01", "10.0.0.3", 32, 80, "Ativo"),
    ("srv-bkp-01", "10.0.0.4", 8, 30, "Inativo"),
]
for s in servidores:
    ws.append(s)
```

```python
# Etapa 4
vermelho = PatternFill("solid", fgColor="FF0000")
for linha in ws.iter_rows(min_row=2):
    if linha[3].value >= 80:
        linha[3].fill = vermelho
        print(f"{linha[0].value}: precisa de expansão de armazenamento")
wb.save("inventario_servidores.xlsx")
```

```python
# Log de exemplo (opcional)
with open("log.txt", "w", encoding="utf-8") as f:
    f.write("2026-11-18 10:00:00 - INFO - Sistema iniciado\n")
    f.write("2026-11-18 10:05:12 - AVISO - Uso de disco alto\n")
    f.write("2026-11-18 10:09:40 - ERRO - Falha na conexão com o banco\n")
```

```python
# Etapa 1
try:
    with open("log.txt", "r", encoding="utf-8") as f:
        linhas = f.readlines()
except FileNotFoundError:
    print("log.txt não encontrado.")
    linhas = []
```

```python
# Etapa 2
dados = []
for linha in linhas:
    data, nivel, msg = linha.strip().split(" - ", 2)
    dados.append([data, nivel, msg])
```

```python
# Etapa 3
import openpyxl

wb = openpyxl.Workbook()
ws = wb.active
ws.append(["Data/Hora", "Tipo de Evento", "Mensagem"])
```

```python
# Etapa 4
for d in dados:
    ws.append(d)
wb.save("auditoria_logs.xlsx")
```
