```python
# Atividade 1

# ETAPA 1 - INICIALIZAÇÃO DO LOG

with open("acessos.log", "w", encoding="utf-8") as arquivo:
    arquivo.write("SISTEMA DE AUDITORIA DE ACESSOS\n")
    arquivo.write("2026-11-01 08:00:00 | SISTEMA | INICIALIZACAO\n")

print("Arquivo acessos.log criado com sucesso!")


# ETAPA 2 - REGISTRO DE ATIVIDADES

eventos = [
    "2026-11-01 08:15:00 | LOGIN | usuario=Laura\n",
    "2026-11-01 08:30:00 | LOGIN | usuario=Admin\n",
    "2026-11-01 09:10:00 | ERRO | senha_invalida\n",
    "2026-11-01 09:45:00 | LOGIN | usuario=Laura\n",
    "2026-11-01 10:20:00 | ERRO | acesso_negado\n"
]

with open("acessos.log", "a", encoding="utf-8") as arquivo:
    for evento in eventos:
        arquivo.write(evento)

print("Eventos adicionados ao arquivo com sucesso!")

# ETAPA 3 - LEITURA E INSPEÇÃO SELETIVA

with open("acessos.log", "r", encoding="utf-8") as arquivo:
    primeira_linha = arquivo.readline()

    print("Cabeçalho do arquivo:")
    print(primeira_linha.strip())

# Obtendo o último evento registrado

with open("acessos.log", "r", encoding="utf-8") as arquivo:
    linhas = arquivo.readlines()

    ultima_linha = linhas[-1]

    print("\nÚltimo evento registrado:")
    print(ultima_linha.strip())


# ETAPA 4 - AUDITORIA DE SEGURANÇA

palavra_chave = "LOGIN"
contador = 0

with open("acessos.log", "r", encoding="utf-8") as arquivo:
    print("Eventos encontrados com a palavra-chave LOGIN:")

    for linha in arquivo:
        if palavra_chave in linha:
            print(linha.strip())
            contador += 1

print("\nTotal de eventos LOGIN:", contador)
```

```python
# Atividade 2

# ETAPA 1 - TRATAMENTO DE ARQUIVO INEXISTENTE

try:
    with open("servidor_inexistente.log", "r", encoding="utf-8") as arquivo:
        conteudo = arquivo.read()

except FileNotFoundError:
    print("Arquivo não encontrado. Verifique o nome do arquivo.")

# ETAPA 2 - PROTEÇÃO DE ESCRITA E PERMISSÕES

try:
    with open("acessos.log", "w", encoding="utf-8") as arquivo:
        arquivo.write("SISTEMA DE AUDITORIA DE ACESSOS\n")
        arquivo.write(
            "2026-11-01 08:00:00 | SISTEMA | INICIALIZACAO\n"
        )

    print("Arquivo criado com sucesso!")

except PermissionError:
    print("Erro: não foi possível escrever no arquivo.")


# Proteção do modo append

try:
    with open("acessos.log", "a", encoding="utf-8") as arquivo:
        arquivo.write(
            "2026-11-01 08:15:00 | LOGIN | usuario=Laura\n"
        )

    print("Evento adicionado com sucesso!")

except PermissionError:
    print("Erro: não foi possível adicionar informações ao arquivo.")



# ETAPA 3 - TRATAMENTO DE DECODIFICAÇÃO

try:
    with open("acessos.log", "r", encoding="ascii") as arquivo:
        conteudo = arquivo.read()

    print(conteudo)

except UnicodeDecodeError:
    print("Erro: não foi possível decodificar o arquivo com este encoding.")



# ETAPA 4 - FECHAMENTO E CLEANUP

arquivo = None

try:
    arquivo = open("acessos.log", "r", encoding="utf-8")

    conteudo = arquivo.read()

    print("Arquivo lido com sucesso!")
    print(conteudo)

except FileNotFoundError:
    print("Erro: arquivo não encontrado.")

except Exception as erro:
    print("Ocorreu um erro inesperado:", erro)

finally:
    if arquivo is not None:
        arquivo.close()
        print("Arquivo fechado com segurança.")
    else:
        print("Não foi necessário fechar o arquivo.")
```
