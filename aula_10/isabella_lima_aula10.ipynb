```python
# AULA 10 | ATIVIDADE 01 | Exportador de Configurações

# Você precisa criar uma rotina para exportar as configurações de infraestrutura de TI em dois formatos estruturados:
# JSON (para a aplicação carregar) e CSV (para o banco de dados).

# Requisitos:
# *Crie um script Python que execute as etapas a seguir:

import csv
import json

# Etapa 1: Estrutura em Memória. Crie um dicionário representando as configurações de um cluster de servidores
# (ex.: nome do cluster, versão do SO, e uma lista com 3 servidores com IP e portas liberadas).

cluster = {
    "nome": "cluster_empresa",
    "versao_so": "Ubuntu Server 24.04",
    "servidores": [
        {
            "nome": "servidor_web",
            "ip": "192.168.1.10",
            "portas": [80, 443],
        },
        {
            "nome": "servidor_banco",
            "ip": "192.168.1.20",
            "portas": [3306],
        },
        {
            "nome": "servidor_backup",
            "ip": "192.168.1.30",
            "portas": [22, 873],
        },
    ],
}

print("\n🖥️ CONFIGURAÇÕES DO CLUSTER\n")
print(f"Nome: {cluster['nome']}")
print(f"Sistema operacional: {cluster['versao_so']}\n")

for servidor in cluster["servidores"]:
    print(f"Servidor: {servidor['nome']}")
    print(f"IP: {servidor['ip']}")
    print(f"Portas: {servidor['portas']}\n")

# Etapa 2: Exportação JSON.
# Salve esse dicionário em um arquivo cluster_config.json de forma indentada e legível.

with open("cluster_config.json", "w", encoding="utf-8") as arquivo:
    json.dump(cluster, arquivo, indent=4, ensure_ascii=False)

print("✅ Arquivo JSON salvo: cluster_config.json\n")

# Etapa 3: Exportação CSV.
# A partir da lista de servidores do seu dicionário, grave um arquivo servidores_cluster.csv com as colunas:
# Servidor, IP, Portas (utilizando ; como delimitador).

with open(
    "servidores_cluster.csv", "w", newline="", encoding="utf-8"
) as arquivo:
    escritor = csv.writer(arquivo, delimiter=";")
    escritor.writerow(["Servidor", "IP", "Portas"])

    for servidor in cluster["servidores"]:
        portas = ", ".join(str(porta) for porta in servidor["portas"])
        escritor.writerow([servidor["nome"], servidor["ip"], portas])

print("✅ Arquivo CSV salvo: servidores_cluster.csv")
```

```python
# AULA 10 | ATIVIDADE 02 | Relatório em Planilha Inventário de Servidores

# Você é responsável pela gestão de ativos de TI e
# precisa gerar uma planilha com o inventário dos servidores virtuais para a auditoria mensal.

# Requisitos:
# *Crie scripts em Python que executem as etapas a seguir:

# Etapa 1: Instalação.
# Instale e importe a biblioteca openpyxl seguindo a PEP 8.

%pip install -q openpyxl

from openpyxl import Workbook
from openpyxl.styles import PatternFill

# Etapa 2: Crie uma planilha.
# Crie uma pasta de trabalho (planilha) com uma aba para o servidores.
# Adicione um cabeçalho: Ex: "Hostname", "IP", "Memória (GB)", "Uso de Disco (%)", "Status".

planilha = Workbook()
aba = planilha.active
aba.title = "Servidores"

aba.append([
    "Hostname", "IP", "Memória (GB)", "Uso de Disco (%)", "Status"
])

# Etapa 3: Inserção de dados.
# Insira pelo menos 4 servidores com dados simulados.

servidores = [
    ["servidor_web", "192.168.1.10", 8, 65, "Ativo"],
    ["servidor_banco", "192.168.1.20", 16, 85, "Ativo"],
    ["servidor_backup", "192.168.1.30", 32, 92, "Ativo"],
    ["servidor_teste", "192.168.1.40", 4, 80, "Inativo"],
]

for servidor in servidores:
    aba.append(servidor)

# Etapa 4: Avisos.
# Se o Uso de Disco (%) for maior ou igual a 80%, defina a cor da célula ou
# exiba um alerta no terminal informando que o servidor precisa de expansão de armazenamento.
# Salve o arquivo.

cor_alerta = PatternFill(
    start_color="FFFF00",
    end_color="FFFF00",
    fill_type="solid",
)

print("\n📋 INVENTÁRIO DE SERVIDORES\n")

for linha in aba.iter_rows(min_row=2):
    hostname = linha[0].value
    ip = linha[1].value
    memoria = linha[2].value
    uso_disco = linha[3].value
    status = linha[4].value

    print(f"🖥️ Servidor: {hostname}")
    print(f"IP: {ip}")
    print(f"Memória: {memoria} GB")
    print(f"Uso de disco: {uso_disco}%")
    print(f"Status: {status}")

    if uso_disco >= 80:
        linha[3].fill = cor_alerta
        print("⚠️ Precisa de expansão de armazenamento.")

    print()

nome_arquivo = "isabella_lima_aula10_ex2.xlsx"
planilha.save(nome_arquivo)

print("✅ PLANILHA SALVA COM SUCESSO")
print(f"Arquivo: {nome_arquivo}")

```

```python
# AULA 10 | ATIVIDADE 03 | Conversor de Log para Planilha Arquivo de texto para excel

# Na aula anterior, trabalhamos com o arquivo de texto log.txt
# Precisamos converter essas informações brutas em uma planilha Excel para facilitar a visualização gerencial. .

# Requisitos:
# *Crie scripts em Python que executem as etapas a seguir:

%pip install -q openpyxl

from openpyxl import Workbook

# Etapa 1: log.txt.
# Abra o arquivo log.txt utilizando o bloco with e trate possíveis exceções com try/except (FileNotFoundError).

linhas = []
arquivo_encontrado = False

try:
    with open("log.txt", "r", encoding="utf-8") as arquivo:
        linhas = arquivo.readlines()

    arquivo_encontrado = True
    print("✅ Arquivo log.txt lido com sucesso.\n")

except FileNotFoundError:
    print("⚠️ Arquivo log.txt não encontrado. Envie o arquivo para o Colab.")

# Etapa 2: Extração de dados.
# Para cada linha de log, extraia o nível do evento (INFO, AVISO, ERRO).

dados_logs = []

for linha in linhas:
    if linha.strip():
        data_hora, tipo_evento, mensagem = linha.strip().split(" - ", 2)
        dados_logs.append([data_hora, tipo_evento, mensagem])

        print(f"Data/Hora: {data_hora}")
        print(f"Tipo de evento: {tipo_evento}")
        print(f"Mensagem: {mensagem}\n")

# Etapa 3: Criação de Planilha
# Crie uma nova planilha auditoria_logs.xlsx contendo as colunas do log: Data/Hora, Tipo de Evento e Mensagem.

planilha = Workbook()
aba = planilha.active
aba.title = "Logs"

aba.append(["Data/Hora", "Tipo de Evento", "Mensagem"])

# Etapa 4: Salvando.
# Preencha a planilha e salve-a no diretório de execução.

if arquivo_encontrado:
    for registro in dados_logs:
        aba.append(registro)

    planilha.save("auditoria_logs.xlsx")

    print("✅ PLANILHA SALVA COM SUCESSO")
    print("Arquivo: auditoria_logs.xlsx")
    print(f"Eventos registrados: {len(dados_logs)}")
```
