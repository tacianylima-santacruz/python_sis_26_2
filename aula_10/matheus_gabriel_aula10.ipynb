```python
import json
import csv
import os

# =====================================================================
# Atividade Prática 01 Aula 10 - Exportador de Configurações
# =====================================================================

print("--- Executando Atividade Prática 01 ---")

# Etapa 1: Estrutura em Memória
config_cluster = {
    "nome_cluster": "Cluster-Producao-01",
    "versao_so": "Ubuntu 22.04 LTS",
    "servidores": [
        {"nome": "srv-web-01", "ip": "192.168.1.10", "portas": "80,443"},
        {"nome": "srv-db-01", "ip": "192.168.1.20", "portas": "5432"},
        {"nome": "srv-cache-01", "ip": "192.168.1.30", "portas": "6379"}
    ]
}
print("Dicionário em memória criado com sucesso.")

# Etapa 2: Exportação JSON
with open('cluster_config.json', 'w', encoding='utf-8') as f:
    json.dump(config_cluster, f, indent=4, ensure_ascii=False)
print("Arquivo 'cluster_config.json' exportado.")

# Etapa 3: Exportação CSV
with open('servidores_cluster.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f, delimiter=';')
    writer.writerow(['Servidor', 'IP', 'Portas'])
    for s in config_cluster['servidores']:
        writer.writerow([s['nome'], s['ip'], s['portas']])
print("Arquivo 'servidores_cluster.csv' exportado.\n")

# =====================================================================
# Atividade Prática 02 Aula 10 - Relatório em Planilha Inventário
# =====================================================================

print("--- Executando Atividade Prática 02 ---")

# Etapa 1: Instalação e Importação
# Nota: No Google Colab o openpyxl já vem instalado por padrão.
from openpyxl import Workbook
from openpyxl.styles import PatternFill

# Etapa 2: Crie uma planilha
wb_inventario = Workbook()
ws_inventario = wb_inventario.active
ws_inventario.title = "Servidores"
ws_inventario.append(["Hostname", "IP", "Memória (GB)", "Uso de Disco (%)", "Status"])

# Etapa 3: Inserção de dados (4 servidores virtuais simulados)
dados_servidores = [
    ["srv-app-01", "10.0.0.5", 16, 45, "Ativo"],
    ["srv-backup", "10.0.0.9", 32, 85, "Ativo"], # Alerta: >= 80%
    ["srv-proxy-01", "10.0.0.12", 8, 30, "Ativo"],
    ["srv-dev-01", "10.0.0.50", 12, 90, "Ativo"]  # Alerta: >= 80%
]

for linha in dados_servidores:
    ws_inventario.append(linha)

# Etapa 4: Avisos e salvamento
# O cabeçalho está na linha 1, os dados começam na linha 2
fill_alerta = PatternFill(start_color="FFC7CE", end_color="FFC7CE", fill_type="solid")

for row in range(2, ws_inventario.max_row + 1):
    hostname = ws_inventario.cell(row=row, column=1).value
    uso_disco = ws_inventario.cell(row=row, column=4).value

    if uso_disco >= 80:
        print(f"[ALERTA] Servidor {hostname} está com {uso_disco}% de uso de disco e precisa de expansão!")
        # Define cor de fundo vermelha clara para a célula de uso de disco
        ws_inventario.cell(row=row, column=4).fill = fill_alerta

wb_inventario.save('inventario_servidores.xlsx')
print("Planilha 'inventario_servidores.xlsx' gerada.\n")

# =====================================================================
# Atividade Prática 03 Aula 10 - Conversor de Log para Planilha
# =====================================================================

print("--- Executando Atividade Prática 03 ---")

# Criando um arquivo log.txt simulado para a atividade rodar sem erros
conteudo_log_simulado = """2023-10-25 10:15:30 - INFO - Sistema inicializado com sucesso.
2023-10-25 10:16:02 - AVISO - Uso de memória acima de 75%.
2023-10-25 10:17:15 - ERRO - Falha de conexão com o banco de dados.
2023-10-25 10:18:00 - INFO - Conexão restabelecida.
"""
with open('log.txt', 'w', encoding='utf-8') as f:
    f.write(conteudo_log_simulado)

# Etapa 1 e 2: Leitura do log e extração de dados
linhas_extraidas = []
try:
    with open('log.txt', 'r', encoding='utf-8') as f_log:
        for linha in f_log:
            linha = linha.strip()
            if not linha:
                continue

            # Separando por ' - ' baseado no padrão comum do log
            partes = linha.split(' - ')
            if len(partes) >= 3:
                data_hora = partes[0]
                tipo_evento = partes[1] # INFO, AVISO, ERRO
                mensagem = partes[2]
                linhas_extraidas.append([data_hora, tipo_evento, mensagem])
except FileNotFoundError:
    print("Erro: O arquivo log.txt não foi encontrado.")

# Etapa 3 e 4: Criação e Salvamento da Planilha de Auditoria
if linhas_extraidas:
    wb_logs = Workbook()
    ws_logs = wb_logs.active
    ws_logs.title = "Logs de Auditoria"

    # Cabeçalho
    ws_logs.append(["Data/Hora", "Tipo de Evento", "Mensagem"])

    # Preenchimento
    for l in linhas_extraidas:
        ws_logs.append(l)

    wb_logs.save('auditoria_logs.xlsx')
    print("Planilha 'auditoria_logs.xlsx' gerada a partir do log com sucesso!\n")
else:
    print("Nenhum dado de log extraído para salvar.\n")
```
