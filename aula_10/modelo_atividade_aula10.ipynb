# Atividade Prática 01 Aula 10

Exportador de Configurações

Você precisa criar uma rotina para exportar as configurações de infraestrutura de TI em dois formatos estruturados: JSON (para a aplicação carregar) e CSV (para o banco de dados).

Requisitos:  
*Crie um script Python que execute as  etapas a seguir:
Etapa 1: Estrutura em Memória.
Crie um dicionário representando as configurações de um cluster de servidores (ex.: nome do cluster, versão do SO, e uma lista com 3 servidores com IP e portas liberadas).

```python

# Etapa 1: Estrutura em memória

cluster = {
    "nome_cluster": "cluster_principal",
    "versao_so": "Ubuntu 22.04",
    "servidores": [
        {
            "servidor": "Servidor 1",
            "ip": "192.168.1.10",
            "portas": [22, 80, 443]
        },
        {
            "servidor": "Servidor 2",
            "ip": "192.168.1.11",
            "portas": [22, 3306]
        },
        {
            "servidor": "Servidor 3",
            "ip": "192.168.1.12",
            "portas": [22, 8080]
        }
    ]
}

print("Configurações do cluster criadas com sucesso!")
print(cluster)
```

Etapa 2: Exportação JSON.
Salve esse dicionário em um arquivo cluster_config.json de forma indentada e legível.



```python

# Etapa 2: Exportação JSON

import json

with open("cluster_config.json", "w", encoding="utf-8") as arquivo:
    json.dump(cluster, arquivo, indent=4, ensure_ascii=False)

print("Arquivo cluster_config.json criado com sucesso!")
```

Etapa 3: Exportação CSV.
A partir da lista de servidores do seu dicionário, grave um arquivo servidores_cluster.csv com as colunas: Servidor, IP, Portas (utilizando ; como delimitador).


```python

# Etapa 3: Exportação CSV

import csv

with open("servidores_cluster.csv", "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo, delimiter=";")

    escritor.writerow(["Servidor", "IP", "Portas"])

    for servidor in cluster["servidores"]:
        portas = ", ".join(str(porta) for porta in servidor["portas"])

        escritor.writerow([
            servidor["servidor"],
            servidor["ip"],
            portas
        ])

print("Arquivo servidores_cluster.csv criado com sucesso!")
```

# Atividade Prática 02 Aula 10

Relatório em Planilha
Inventário de Servidores
Você é responsável pela gestão de ativos de TI e precisa gerar uma planilha com o inventário dos servidores virtuais para a auditoria mensal.

Requisitos:  
*Crie scripts em Python que executem as etapas a seguir:

Etapa 1: Instalação.
 Instale e importe a biblioteca openpyxl seguindo a PEP 8.

```python
!pip install openpyxl

from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill
```

Etapa 2: Crie uma planilha.
Crie uma pasta de trabalho (planilha) com uma aba para o servidores.
Adicione um cabeçalho:
   Ex: "Hostname", "IP", "Memória (GB)", "Uso de Disco (%)", "Status".

```python
workbook = Workbook()

planilha = workbook.active
planilha.title = "Servidores"

cabecalho = [
    "Hostname",
    "IP",
    "Memória (GB)",
    "Uso de Disco (%)",
    "Status"
]

planilha.append(cabecalho)

for celula in planilha[1]:
    celula.font = Font(bold=True)
    celula.fill = PatternFill(
        start_color="ADD8E6",
        end_color="ADD8E6",
        fill_type="solid"
    )

print("Planilha criada com sucesso!")
```

Etapa 3: Inserção de dados.
Insira pelo menos 4 servidores com dados simulados.

```python
servidores = [
    ["Servidor-01", "192.168.0.1", 16, 65, "Ativo"],
    ["Servidor-02", "192.168.0.2", 32, 85, "Ativo"],
    ["Servidor-03", "192.168.0.3", 8, 45, "Ativo"],
    ["Servidor-04", "192.168.0.4", 16, 92, "Ativo"]
]

for servidor in servidores:
    planilha.append(servidor)

print("Servidores inseridos com sucesso!")
```

Etapa 4: Avisos.
Se o Uso de Disco (%) for maior ou igual a 80%, defina a cor da célula ou exiba um alerta no terminal informando que o servidor precisa de expansão de armazenamento.
 Salve o arquivo.

```python
for linha in range(2, planilha.max_row + 1):

    uso_disco = planilha.cell(row=linha, column=4).value

    if uso_disco >= 80:
        planilha.cell(row=linha, column=4).fill = PatternFill(
            start_color="FF0000",
            end_color="FF0000",
            fill_type="solid"
        )

        print(
            f"ALERTA: {planilha.cell(row=linha, column=1).value} "
            "precisa de expansão de armazenamento!"
        )

# Ajustando largura das colunas
for coluna in planilha.columns:
    letra = coluna[0].column_letter
    planilha.column_dimensions[letra].width = 22

workbook.save("inventario_servidores.xlsx")

print("Arquivo inventario_servidores.xlsx salvo com sucesso!")
```

# Atividade Prática 03 Aula 10

Conversor de Log para Planilha
Arquivo de texto para excel
Na aula anterior, trabalhamos com o arquivo de texto log.txt Precisamos converter essas informações brutas em uma planilha Excel para facilitar a visualização gerencial. .

Requisitos:  
*Crie scripts em Python que executem as etapas a seguir:

Etapa 1: log.txt.
Abra o arquivo log.txt utilizando o bloco with e trate possíveis exceções com try/except (FileNotFoundError).


```python
try:
    with open("log.txt", "r", encoding="utf-8") as arquivo:
        linhas_log = arquivo.readlines()

    print("Arquivo log.txt lido com sucesso!")

except FileNotFoundError:
    print("Arquivo não encontrado. Criando arquivo de exemplo...")

    with open("log.txt", "w", encoding="utf-8") as arquivo:
        arquivo.write("""2026-10-14 08:30:15 - INFO - Sistema inicializado com sucesso
2026-10-14 08:31:02 - INFO - Usuario admin realizou login
2026-10-14 08:35:22 - AVISO - Tentativa de acesso com senha incorreta
2026-10-14 08:40:11 - ERRO - Falha ao conectar ao banco de dados
2026-10-14 08:42:00 - INFO - Conexao reestabelecida
""")

    with open("log.txt", "r", encoding="utf-8") as arquivo:
        linhas_log = arquivo.readlines()
```

Etapa 2: Extração de dados.

Para cada linha de log, extraia o nível do evento (INFO, AVISO, ERRO).

```python
dados_logs = []

for linha in linhas_log:
    partes = linha.strip().split(" - ", 2)

    if len(partes) == 3:
        data_hora = partes[0]
        tipo_evento = partes[1]
        mensagem = partes[2]

        dados_logs.append([
            data_hora,
            tipo_evento,
            mensagem
        ])

print("Dados extraídos:")
for dado in dados_logs:
    print(dado)
```

Etapa 3: Criação de Planilha

Crie uma nova planilha auditoria_logs.xlsx contendo as colunas do log: Data/Hora, Tipo de Evento e Mensagem.

```python
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill

workbook = Workbook()
planilha = workbook.active
planilha.title = "Auditoria de Logs"

planilha.append([
    "Data/Hora",
    "Tipo de Evento",
    "Mensagem"
])

for celula in planilha[1]:
    celula.font = Font(bold=True)
    celula.fill = PatternFill(
        start_color="ADD8E6",
        end_color="ADD8E6",
        fill_type="solid"
    )

print("Planilha criada!")
```

Etapa 4: Salvando.

Preencha a planilha e salve-a no diretório de execução.

```python
for dado in dados_logs:
    planilha.append(dado)

for coluna in planilha.columns:
    letra = coluna[0].column_letter
    planilha.column_dimensions[letra].width = 35

workbook.save("auditoria_logs.xlsx")

print("Arquivo auditoria_logs.xlsx salvo com sucesso!")
print(f"Total de registros: {len(dados_logs)}")
```
