# Atividade Prática 01 Aula 10

Exportador de Configurações

Você precisa criar uma rotina para exportar as configurações de infraestrutura de TI em dois formatos estruturados: JSON (para a aplicação carregar) e CSV (para o banco de dados).

Requisitos:  
*Crie um script Python que execute as  etapas a seguir:
Etapa 1: Estrutura em Memória.
Crie um dicionário representando as configurações de um cluster de servidores (ex.: nome do cluster, versão do SO, e uma lista com 3 servidores com IP e portas liberadas).

```python

# Etapa 1: Estrutura em memória

cluster = {
    "nome_cluster": "cluster_principal",
    "versao_so": "Ubuntu 22.04",
    "servidores": [
        {
            "servidor": "Servidor 1",
            "ip": "192.168.1.10",
            "portas": [22, 80, 443]
        },
        {
            "servidor": "Servidor 2",
            "ip": "192.168.1.11",
            "portas": [22, 3306]
        },
        {
            "servidor": "Servidor 3",
            "ip": "192.168.1.12",
            "portas": [22, 8080]
        }
    ]
}

print("Configurações do cluster criadas com sucesso!")
print(cluster)
```

Etapa 2: Exportação JSON.
Salve esse dicionário em um arquivo cluster_config.json de forma indentada e legível.



```python

# Etapa 2: Exportação JSON

import json

with open("cluster_config.json", "w", encoding="utf-8") as arquivo:
    json.dump(cluster, arquivo, indent=4, ensure_ascii=False)

print("Arquivo cluster_config.json criado com sucesso!")
```

Etapa 3: Exportação CSV.
A partir da lista de servidores do seu dicionário, grave um arquivo servidores_cluster.csv com as colunas: Servidor, IP, Portas (utilizando ; como delimitador).


```python

# Etapa 3: Exportação CSV

import csv

with open("servidores_cluster.csv", "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo, delimiter=";")

    escritor.writerow(["Servidor", "IP", "Portas"])

    for servidor in cluster["servidores"]:
        portas = ", ".join(str(porta) for porta in servidor["portas"])

        escritor.writerow([
            servidor["servidor"],
            servidor["ip"],
            portas
        ])

print("Arquivo servidores_cluster.csv criado com sucesso!")
```
