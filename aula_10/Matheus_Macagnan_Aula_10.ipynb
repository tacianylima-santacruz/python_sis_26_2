```python

import csv
import json

Cluster = [
    ["Nome", "Versao S.O", "Servidores", "Portas"],
    ["Servidor 1", "Windows", "192.168.1.11", ["20", "50", "300"]],
    ["Servidor 2", "Windows", "192.168.1.10", ["30", "100", "500"]],
    ["Servidor 3", "Windows", "192.168.1.12", ["50", "150", "700"]]
]

# Exportação completa para CSV
with open("Cluster.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(Cluster)

# Exportação completa para JSON
with open("Cluster.json", "w", encoding="utf-8") as arquivo:
    json.dump(Cluster, arquivo, indent=4, ensure_ascii=False)

# Exportação dos servidores para um CSV separado
with open("servidores_cluster.csv", "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo, delimiter=";")

    # Cabeçalho
    escritor.writerow(["Nome", "Servidores", "Portas"])

    # Começa no índice 1 para pular o cabeçalho da lista
    for servidor in Cluster[1:]:
        escritor.writerow([
            servidor[0],           # Nome do servidor
            servidor[2],           # Endereço IP
            "/".join(servidor[3])  # Portas
        ])

print("Arquivos CSV e JSON criados com sucesso!")
```

```python

#Importando a classe WorkBook que re representa um arquivo do excel
from openpyxl import Workbook

#Importando a classe para poder pintar o fundo das celular no Excel
from openpyxl.styles import PatternFill

# Etapa 2: Criar a planilha
#criando um arquivo no Excel Vazio e armazenando em planilha
planilha = Workbook()

#Pega a aba que vem por padrao na planilha e guarda na variavel
aba = planilha.active
#Renomeia a aba
aba.title = "Servidores"

#Criando uma lista, cada linha e uma coluna (A,B,C,D,E)
cabecalho = [
    "Hostname",
    "IP",
    "Memória (GB)",
    "Uso de Disco (%)",
    "Status"
]

#Adiciona uma linha nova no fim da aba, virando a linha 1
aba.append(cabecalho)

# Etapa 3: Inserir os dados
servidores = [
    ["SRV-WEB01", "192.168.1.10", 8, 65, "Ativo"],
    ["SRV-DB01", "192.168.1.20", 16, 85, "Ativo"],
    ["SRV-APP01", "192.168.1.30", 8, 92, "Ativo"],
    ["SRV-BKP01", "192.168.1.40", 4, 45, "Manutenção"]
]

#Usando for para percorrer a lista de servidores e adicionar
# em cada linha da planilha
for servidor in servidores:
    aba.append(servidor)

# Etapa 4: Criar os alertas
#Define o estilo de preenchimento que será usado nas células
cor_alerta = PatternFill(
    fill_type="solid",
    fgColor="FF0000"
)


for linha in range(2, aba.max_row + 1): # Laço de repetição para passar poelas linhas no excel
    uso_disco = aba.cell(row=linha, column=4).value #Le a coluna 4 e guarda na variavel
    hostname = aba.cell(row=linha, column=1).value #Mesma ideia que a de cima

    # if para saber quando deve ser pintadoa  linha
    if uso_disco >= 80:
        aba.cell(row=linha, column=4).fill = cor_alerta #Pinta a celula da cor definida

        print(
            f"ALERTA: O servidor {hostname} "
            "precisa de expansão de armazenamento!"
        )

# Salvar o arquivo
planilha.save("/content/inventario_servidores.xlsx")

print("Planilha criada com sucesso!")
```

```python

from openpyxl import Workbook

#Try é um bloco de tratamento de erros
try:
    # Etapa 1: Abrir o arquivo de log
    #Abrindo o arquivo, r significa leitura(read)
    #whit fecha o arquivo quando o bloco termina
    #as ar
    with open("log.txt", "r", encoding="utf-8") as arquivo:
        linhas = arquivo.readlines()

    # Etapa 2: Criar a planilha e o cabeçalho
    planilha = Workbook()
    aba = planilha.active
    aba.title = "Auditoria"

    aba.append(["Data/Hora", "Tipo de Evento", "Mensagem"])

    # Etapa 3: Extrair os dados de cada linha
    for linha in linhas:
        linha = linha.strip()

        if not linha:
            continue

        partes = linha.split(maxsplit=2)

        if len(partes) < 3:
            print(f"Linha ignorada por estar incompleta: {linha}")
            continue

        data_hora = partes[0] + " " + partes[1]
        restante = partes[2].split(maxsplit=1)

        if len(restante) < 2:
            print(f"Linha ignorada por estar incompleta: {linha}")
            continue

        tipo_evento = restante[0]
        mensagem = restante[1]

        if tipo_evento not in ["INFO", "AVISO", "ERRO"]:
            print(f"Tipo de evento desconhecido: {tipo_evento}")
            continue

        aba.append([data_hora, tipo_evento, mensagem])

    # Etapa 4: Salvar a planilha
    planilha.save("/content/auditoria_logs.xlsx")

    print("Planilha auditoria_logs.xlsx criada com sucesso!")

except FileNotFoundError:
    print("Erro: o arquivo log.txt não foi encontrado.")

except PermissionError:
    print("Erro: não foi possível acessar ou salvar um arquivo.")
```
