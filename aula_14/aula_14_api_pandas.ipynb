Normalizando Dados

```python
import requests
import pandas as pd

# Exemplo com múltiplas cidades usando o mesmo endpoint da WeatherAPI
cidades = ["Araucaria", "Curitiba", "Londrina", "Maringa", "Ponta Grossa"]
lista_dados = []

for cidade in cidades:
    parametros = {"key": api_key, "q": cidade, "lang": "pt"}
    try:
        resposta = requests.get(link_api, params=parametros, timeout=10)
        resposta.raise_for_status()
        lista_dados.append(resposta.json())
    except requests.exceptions.RequestException as erro:
        print(f"Erro ao consultar {cidade}: {erro}")

# Normalizando os sub-dicionários (location e current) em colunas planas
df_clima = pd.json_normalize(lista_dados, sep="_")

# Visualizando as principais colunas geradas
print(display(df_clima[["location_name", "location_localtime", "current_temp_c", "current_humidity"]]))
# Visualizar tabela transferida para Pandas
df_clima.head()
#df_clima.info()
```

Tratamento, tipagem e agregação pós-API

```python
# 1. Selecionando e renomeando colunas de interesse
colunas = {
    "location_name": "Cidade",
    "location_localtime": "Data_Hora",
    "current_temp_c": "Temperatura_C",
    "current_humidity": "Umidade_Pct",
    "current_condition_text": "Condicao"
}
df_tratado = df_clima[list(colunas.keys())].rename(columns=colunas).copy()

# 2. Removendo duplicatas e tratando nulos
df_tratado = df_tratado.drop_duplicates(subset=["Cidade"])

# 3. Conversão explícita de tipos (pós-API)
df_tratado["Data_Hora"] = pd.to_datetime(df_tratado["Data_Hora"], errors="coerce")
df_tratado["Temperatura_C"] = pd.to_numeric(df_tratado["Temperatura_C"], errors="coerce")
df_tratado["Umidade_Pct"] = pd.to_numeric(df_tratado["Umidade_Pct"], errors="coerce")

# 4. Agregação simples: média de temperatura por condição climática
resumo_clima = df_tratado.groupby("Condicao")["Temperatura_C"].mean().reset_index()
print(resumo_clima)
```

Plotagem

```python
import matplotlib.pyplot as plt

# Configurando o gráfico com boas práticas visuais
plt.figure(figsize=(9, 5))
plt.bar(df_tratado["Cidade"], df_tratado["Temperatura_C"], color="#E91E63")

plt.title("Temperatura Atual nas Cidades do Paraná (°C)", fontsize=14)
plt.xlabel("Cidade", fontsize=12)
plt.ylabel("Temperatura (°C)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()

# Exportando o resultado tratado para auditoria
df_tratado.to_csv("clima_parana_tratado.csv", index=False, encoding="utf-8")
print("Arquivo clima_parana_tratado.csv exportado com sucesso!")
```
