```python
# AULA 03 | EXERCÍCIO 1

# Processamento 1
# Defina a sequência de referência normal da Hemoglobina A
referencia = "ATGGTGCACCTGACTCCTGAG"
# Fita digitada pelo usuário
dna = input("Digite a fita de DNA com 21 nucleotídeos: ").upper()
# Compare a fita digitada pelo usuário com a fita de referência
mutante = dna == referencia
# Crie uma lista com as duas opções de diagnóstico
diagnosticos = [
    "✅ Sequência NORMAL (Hemoglobina A)",
    "⚠️ Sequência MUTANTE (Anemia Falciforme - HbS)"]
# Exiba o resultado
print(diagnosticos[mutante])

# Processamento 2
# Tabela de Códons do Gene HBB
# DNA substituindo as trincas (códons) pelos seus respectivos aminoácidos
proteina = dna.replace("ATG", "Met")
# Crie a variável proteina
proteina = proteina.replace("GTG", "Val")
proteina = proteina.replace("CAC", "His")
proteina = proteina.replace("CCT", "Pro")
proteina = proteina.replace("CTG", "Leu")
proteina = proteina.replace("ACT", "Thr")
proteina = proteina.replace("GAG", "Glu")
# Imprima a cadeia de aminoácidos resultante na tela
print(f"Proteína traduzida: {proteina}")

# Processamento 3
# Solicite ao usuário um aminoácido
aminoacido = input("Digite um aminoácido para buscar: ")
# Localize seu índice na proteína traduzida
posicao = proteina.find(aminoacido)

# Imprima o relatório final
print("\n=======================================================")
print("RELATÓRIO FINAL")
print("=======================================================")
print(f"DNA de referência: {referencia}")
print(f"DNA digitado: {dna}")
print(f"Tamanho da fita: {len(dna)}")
print(f"Proteína traduzida: {proteina}")
print(f"Diagnóstico: {diagnosticos[mutante]}")
print(f"Posição de {aminoacido}: {posicao}")
print("=======================================================")
```
