```python
# AULA 03 | EXERCÍCIO 2

# Entrada de Dados:
# O programa deve solicitar que o usuário digite o texto do pergaminho.
# Na frase digitada incluia espaços a mais no começo e fim.
texto = input("Digite o texto do pergaminho com espaços no começo e no final: ")

# Limpeza e Higienização:
# Remova os espaços inúteis no início e no final do texto digitado e guarde em uma variável.
texto_limpo = texto.strip()
palavras = texto_limpo.split()

# Detector de Ameaças:
# O sistema deve verificar se a palavra p.e. "Anel", ou outra palavra a sua escolha,
# existe dentro do texto e imprimir o resultado diretamente;
# O sistema deve contar quantas vezes a palavra aparece.
ameaca = "Anel"
print("A palavra Anel existe no texto?", ameaca in texto_limpo)
print("Quantidade de vezes que Anel aparece:", texto_limpo.count(ameaca))

# Relatório
# O programa deve calcular e exibir na tela:
# O número total de caracteres do texto limpo.
# A quantidade total de palavras presentes no texto.
# A primeira palavra do texto.
# A última palavra do texto.
print("Total de caracteres:", len(texto_limpo))
print("Total de palavras:", len(palavras))
print("Primeira palavra:", palavras[0])
print("Última palavra:", palavras[-1])
```
