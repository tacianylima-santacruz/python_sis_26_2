```python
print("A Sociedade do Anel encontrou um pergaminho antigo, mas o texto está bagunçado!")
print("Seu trabalho é criar um programa em Python que receba esse texto, faça a higienização dos dados e gere um relatório.")

txt = input("Digite o texto do pergaminho: ")

# Limpeza do texto
limpa_txt = txt.strip().lower()

# Separando as palavras
palavras = limpa_txt.split()

# Contagens
chr_count = len(limpa_txt)
plvr_count = len(palavras)

# Primeira e última palavra
first = palavras[0]
last = palavras[-1]

# Detector de ameaças
ameaca = "Anel".lower()

print(f"\nTexto limpo: {limpa_txt}")
print(f"O número total de caracteres do texto limpo: {chr_count}")
print(f"A quantidade total de palavras presentes no texto: {plvr_count}")
print(f"A primeira palavra do texto: {first}")
print(f"A última palavra do texto: {last}")

if ameaca in palavras:
    print(f"A palavra '{ameaca}' foi encontrada!")
else:
    print(f"A palavra '{ameaca}' não foi encontrada.")

print(f"A palavra '{ameaca}' aparece {palavras.count(ameaca)} vez(es).")
```
