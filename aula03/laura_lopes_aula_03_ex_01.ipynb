```python
print("---------Teste Beta-Hemoglobina------------")

dna = input("Digite uma fita de DNA com 21 nucleotídeos: ").upper() #.upper() muda frase para maiusculo

ref_hemoglobina = "ATGGTGCACCTGACTCCTGAG" #Define a referencia

comparacao_fita = ref_hemoglobina == dna #Guarda 0= False/1 = True

lst_diagn = ["✅ Sequência NORMAL (Hemoglobina A)", "⚠️ Sequência MUTANTE (Anemia Falciforme - HbS)"] #indice [0,1]

diagn = lst_diagn[comparacao_fita - 1] #Pega valor de comparacao_fita - 1 e usa como indice para lista


proteina = dna.replace("ATG","Met") #trocao primeiro valor pelo segundo
proteina = proteina.replace("GTG","Val")
proteina = proteina.replace("CAC","His")
proteina = proteina.replace("CCT","Pro")
proteina = proteina.replace("CTG","Leu")
proteina = proteina.replace("ACT","Thr")
proteina = proteina.replace("GAG","Glu")

busca_dna = input("Digite um aminoácido para busca (ex: Val ou His): ")

busca_amin=str(proteina.find(busca_dna)) #Muda para string e .find acha em qual posição a var está

fita=str(len(dna)) #Acha o tamanho da var

print("\n---------------------------------------")
print(f"DNA de referência: {ref_hemoglobina}")
print(f"DNA digitado: {dna}")
print(f"Tamanho da fita: {fita}")
print(f"Proteína traduzida: {proteina}")
print(f"O diagnostico é: {diagn}")
print(f"Posição do aminoácido pesquisado: {busca_amin}")
```
